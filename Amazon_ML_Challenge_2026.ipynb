```python
import pandas as pd
import os
from collections import Counter

base = "/content/amazon_ml_challenge/student_resource/dataset"

source_files = [
    "train/train_source1.tsv",
    "train/train_source2.tsv",
    "train/train_source3.tsv"
]

for file in source_files:

    path = os.path.join(base, file)

    total = 0
    missing_name = 0
    missing_address = 0
    missing_country = 0

    country_counts = Counter()
    name_counts = Counter()
    address_counts = Counter()

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=["entity_id", "business_name", "business_address", "country"],
        chunksize=200_000
    ):

        total += len(chunk)

        missing_name += chunk["business_name"].isna().sum()
        missing_address += chunk["business_address"].isna().sum()
        missing_country += chunk["country"].isna().sum()

        country_counts.update(
            chunk["country"].dropna().astype(str)
        )

        # Only count non-empty names/addresses
        names = chunk["business_name"].dropna().astype(str).str.strip()
        addresses = chunk["business_address"].dropna().astype(str).str.strip()

        name_counts.update(names)
        address_counts.update(addresses)

    print("\n" + "=" * 70)
    print(file)

    print("Total rows:", f"{total:,}")

    print("\nMissing values:")
    print("Business name :", f"{missing_name:,}",
          f"({missing_name / total * 100:.2f}%)")

    print("Business address:", f"{missing_address:,}",
          f"({missing_address / total * 100:.2f}%)")

    print("Country:", f"{missing_country:,}",
          f"({missing_country / total * 100:.2f}%)")

    print("\nTop countries:")
    for country, count in country_counts.most_common(15):
        print(f"{country}: {count:,}")

    print("\nTop duplicate business names:")
    for name, count in name_counts.most_common(10):
        if count > 1:
            print(f"{name} -> {count:,}")

    print("\nTop duplicate addresses:")
    for address, count in address_counts.most_common(10):
        if count > 1:
            print(f"{address} -> {count:,}")
```

```python
s1_ids = set(gt_sample["source1_entity_id"])

s2_ids = set()
s3_ids = set()

for value in gt_sample["matched_entity_ids"].fillna(""):
    for entity_id in value.split(","):
        entity_id = entity_id.strip()

        if entity_id.startswith("S2-"):
            s2_ids.add(entity_id)
        elif entity_id.startswith("S3-"):
            s3_ids.add(entity_id)

print("Source 1 IDs:", len(s1_ids))
print("Source 2 IDs:", len(s2_ids))
print("Source 3 IDs:", len(s3_ids))
```

```python
def retrieve_records(path, wanted_ids, chunksize=200_000):

    wanted_ids = set(wanted_ids)
    results = []

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ],
        chunksize=chunksize
    ):

        found = chunk[chunk["entity_id"].isin(wanted_ids)]

        if len(found) > 0:
            results.append(found)

            wanted_ids -= set(found["entity_id"])

        if not wanted_ids:
            break

    if results:
        return pd.concat(results, ignore_index=True)

    return pd.DataFrame(columns=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ])


s1_path = os.path.join(base, "train/train_source1.tsv")
s2_path = os.path.join(base, "train/train_source2.tsv")
s3_path = os.path.join(base, "train/train_source3.tsv")

s1_records = retrieve_records(s1_path, s1_ids)
s2_records = retrieve_records(s2_path, s2_ids)
s3_records = retrieve_records(s3_path, s3_ids)

print("Retrieved:")
print("S1:", len(s1_records))
print("S2:", len(s2_records))
print("S3:", len(s3_records))
```

# BASELINE SUBMISSION V1

```python
import os
import pandas as pd
import re
import unicodedata

BASE = "/content/amazon_ml_challenge/student_resource"

TRAIN = os.path.join(BASE, "dataset", "train")
TEST = os.path.join(BASE, "dataset", "test")

OUTPUT = os.path.join(BASE, "output")
os.makedirs(OUTPUT, exist_ok=True)

print("BASE:", BASE)
print("OUTPUT:", OUTPUT)
```

```python
def normalize_text(x):
    if pd.isna(x):
        return ""

    x = str(x).lower().strip()
    x = unicodedata.normalize("NFKC", x)

    # punctuation → space
    x = re.sub(r"[^\w\s]", " ", x)

    # multiple spaces → one
    x = re.sub(r"\s+", " ", x)

    return x.strip()


def normalize_name(x):
    return normalize_text(x)


def normalize_address(x):
    return normalize_text(x)


def normalize_country(x):
    return normalize_text(x)


print(normalize_name("Anchor Staking Inc."))
print(normalize_address("2330 CENTRAL AVENUE, KANSAS CITY, KS"))
```

```python
from collections import defaultdict

def build_indexes(path, chunksize=200_000):

    address_index = defaultdict(list)
    name_index = defaultdict(list)

    total = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ],
        chunksize=chunksize
    ):

        total += len(chunk)

        names = chunk["business_name"].map(normalize_name)
        addresses = chunk["business_address"].map(normalize_address)
        countries = chunk["country"].map(normalize_country)

        for entity_id, name, address, country in zip(
            chunk["entity_id"],
            names,
            addresses,
            countries
        ):

            # Strong address key
            if address:
                key = (country, address)
                address_index[key].append(entity_id)

            # Name key
            if name:
                key = (country, name)
                name_index[key].append(entity_id)

        print(f"Processed {total:,} rows...", end="\r")

    print(f"\nFinished: {total:,} rows")

    return address_index, name_index
```

```python
print("Building Source 2 index...")

s2_address, s2_name = build_indexes(
    os.path.join(TEST, "test_source2.tsv")
)

print("\nBuilding Source 3 index...")

s3_address, s3_name = build_indexes(
    os.path.join(TEST, "test_source3.tsv")
)

print("\nIndexes ready!")
```

```python
s1_test = pd.read_csv(
    os.path.join(TEST, "test_source1.tsv"),
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ]
)

print(s1_test.shape)
display(s1_test.head())
```

```python
def get_candidates(row):

    entity_id = row["entity_id"]

    name = normalize_name(row["business_name"])
    address = normalize_address(row["business_address"])
    country = normalize_country(row["country"])

    candidates = set()

    # ------------------------------------------------
    # RULE 1: exact normalized address + country
    # ------------------------------------------------

    if address:
        key = (country, address)

        candidates.update(
            s2_address.get(key, [])
        )

        candidates.update(
            s3_address.get(key, [])
        )

    # ------------------------------------------------
    # RULE 2: exact normalized name + country
    # ------------------------------------------------

    if name:
        key = (country, name)

        candidates.update(
            s2_name.get(key, [])
        )

        candidates.update(
            s3_name.get(key, [])
        )

    return sorted(candidates)
```

```python
results = []
candidate_results = []

for i, row in s1_test.iterrows():

    candidates = get_candidates(row)

    candidate_results.append({
        "source1_entity_id": row["entity_id"],
        "candidate_entity_ids": ",".join(candidates)
    })

    # For V1, use candidates directly as matches.
    results.append({
        "source1_entity_id": row["entity_id"],
        "matched_entity_ids": ",".join(candidates)
    })

    if (i + 1) % 100000 == 0:
        print(f"Processed {i+1:,} / {len(s1_test):,}")
```

```python
print("Loop completed!")
print("Results:", len(results))
print("Candidates:", len(candidate_results))
```

```python
matching_df = pd.DataFrame(results)
candidate_df = pd.DataFrame(candidate_results)

matching_path = os.path.join(
    OUTPUT,
    "matching_results.tsv"
)

candidate_path = os.path.join(
    OUTPUT,
    "candidate_pairs.tsv"
)

matching_df.to_csv(
    matching_path,
    sep="\t",
    index=False,
    encoding="utf-8"
)

candidate_df.to_csv(
    candidate_path,
    sep="\t",
    index=False,
    encoding="utf-8"
)

print("Files created successfully!")
print(matching_path)
print(candidate_path)
```

```python
def get_conservative_matches(row):

    name = normalize_name(row["business_name"])
    address = normalize_address(row["business_address"])
    country = normalize_country(row["country"])

    matches = set()

    # Rule 1: exact normalized address + country
    if address:
        key = (country, address)

        matches.update(s2_address.get(key, []))
        matches.update(s3_address.get(key, []))

    # Rule 2: exact normalized name + country,
    # only when the name is unique within that source
    if name:
        key = (country, name)

        s2_matches = s2_name.get(key, [])
        s3_matches = s3_name.get(key, [])

        if len(s2_matches) == 1:
            matches.add(s2_matches[0])

        if len(s3_matches) == 1:
            matches.add(s3_matches[0])

    return sorted(matches)
```

```python
sample_test = s1_test.head(10000)

sample_counts = []

for _, row in sample_test.iterrows():
    matches = get_conservative_matches(row)
    sample_counts.append(len(matches))

sample_counts = pd.Series(sample_counts)

print("10,000-record conservative test")
print("=" * 60)
print("With matches:", (sample_counts > 0).sum())
print("Without matches:", (sample_counts == 0).sum())
print("Average matches:", sample_counts.mean())
print("Median:", sample_counts.median())
print("Maximum:", sample_counts.max())

print("\nPercentiles:")
print(sample_counts.quantile([0.50, 0.75, 0.90, 0.95, 0.99]))
```

```python
def generate_candidates(row):

    name = normalize_name(row["business_name"])
    address = normalize_address(row["business_address"])
    country = normalize_country(row["country"])

    candidates = set()

    # Block 1: country + exact normalized address
    if address:
        key = (country, address)

        candidates.update(s2_address.get(key, []))
        candidates.update(s3_address.get(key, []))

    # Block 2: country + normalized name
    if name:
        key = (country, name)

        candidates.update(s2_name.get(key, []))
        candidates.update(s3_name.get(key, []))

    return sorted(candidates)
```

```python
from rapidfuzz.fuzz import ratio

def pair_features(s1, candidate):

    name1 = normalize_name(s1["business_name"])
    name2 = normalize_name(candidate["business_name"])

    addr1 = normalize_address(s1["business_address"])
    addr2 = normalize_address(candidate["business_address"])

    country1 = normalize_country(s1["country"])
    country2 = normalize_country(candidate["country"])

    return {
        "name_exact": int(name1 == name2),
        "address_exact": int(addr1 == addr2),
        "country_exact": int(country1 == country2),

        "name_similarity": ratio(name1, name2) / 100,
        "address_similarity": ratio(addr1, addr2) / 100,

        "name_length_diff": abs(len(name1) - len(name2)),
        "address_length_diff": abs(len(addr1) - len(addr2)),

        "name_token_overlap": len(
            set(name1.split()) & set(name2.split())
        ) / max(
            1,
            len(set(name1.split()) | set(name2.split()))
        ),

        "address_token_overlap": len(
            set(addr1.split()) & set(addr2.split())
        ) / max(
            1,
            len(set(addr1.split()) | set(addr2.split()))
        )
    }
```

```python
import re
import unicodedata

def normalize_advanced(text):
    if pd.isna(text):
        return ""

    text = str(text).lower().strip()
    text = unicodedata.normalize("NFKC", text)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste ",
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
name = normalize_advanced(row["business_name"])
address = normalize_advanced(row["business_address"])
country = normalize_country(row["country"])
```

```python
def build_advanced_indexes(path, chunksize=200_000):
    name_index = {}
    address_index = {}

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=["entity_id", "business_name", "business_address", "country"],
        chunksize=chunksize
    ):
        for _, row in chunk.iterrows():
            entity_id = row["entity_id"]
            country = normalize_country(row["country"])

            name = normalize_advanced(row["business_name"])
            address = normalize_advanced(row["business_address"])

            if name:
                key = (country, name)
                name_index.setdefault(key, []).append(entity_id)

            if address:
                key = (country, address)
                address_index.setdefault(key, []).append(entity_id)

    return address_index, name_index
```

```python
s2_adv_address, s2_adv_name = build_advanced_indexes(
    f"{TRAIN}/train_source2.tsv"
)

s3_adv_address, s3_adv_name = build_advanced_indexes(
    f"{TRAIN}/train_source3.tsv"
)

print("S2 advanced indexes:", len(s2_adv_name), len(s2_adv_address))
print("S3 advanced indexes:", len(s3_adv_name), len(s3_adv_address))
```

```python
import pandas as pd

train_s1_10k = pd.read_csv(
    f"{TRAIN}/train_source1.tsv",
    sep="\t",
    usecols=["entity_id", "business_name", "business_address", "country"],
    nrows=10000
)

print("Training S1:", train_s1_10k.shape)
train_s1_10k.head()
```

```python
import re
import unicodedata
import pandas as pd

def normalize_text(x):
    if pd.isna(x):
        return ""
    x = str(x).lower().strip()
    x = unicodedata.normalize("NFKC", x)
    x = re.sub(r"[^\w\s]", " ", x)
    x = re.sub(r"\s+", " ", x)
    return x.strip()

def normalize_country(x):
    return normalize_text(x)

def normalize_advanced(text):
    if pd.isna(text):
        return ""

    text = str(text).lower().strip()
    text = unicodedata.normalize("NFKC", text)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()

print("Normalization functions ready.")
```

```python
from collections import defaultdict

def build_s1_key_maps(s1_df):
    name_map = defaultdict(list)
    address_map = defaultdict(list)

    for _, row in s1_df.iterrows():
        sid = row["entity_id"]
        country = normalize_country(row["country"])

        name = normalize_advanced(row["business_name"])
        address = normalize_advanced(row["business_address"])

        if name:
            name_map[(country, name)].append(sid)

        if address:
            address_map[(country, address)].append(sid)

    return name_map, address_map


s1_name_map, s1_address_map = build_s1_key_maps(train_s1_10k)

print("S1 name keys:", len(s1_name_map))
print("S1 address keys:", len(s1_address_map))
```

```python
from collections import defaultdict

def collect_candidates_from_source(path, s1_name_map, s1_address_map):
    candidates = defaultdict(set)

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=["entity_id", "business_name", "business_address", "country"],
        chunksize=200_000
    ):
        for _, row in chunk.iterrows():

            eid = row["entity_id"]
            country = normalize_country(row["country"])

            name = normalize_advanced(row["business_name"])
            address = normalize_advanced(row["business_address"])

            # Advanced name match
            if name:
                key = (country, name)
                for sid in s1_name_map.get(key, []):
                    candidates[sid].add(eid)

            # Advanced address match
            if address:
                key = (country, address)
                for sid in s1_address_map.get(key, []):
                    candidates[sid].add(eid)

    return candidates


print("Scanning Source 2...")
cand_s2 = collect_candidates_from_source(
    f"{TRAIN}/train_source2.tsv",
    s1_name_map,
    s1_address_map
)

print("Source 2 finished.")
print("S1 records with S2 candidates:", len(cand_s2))

print("Scanning Source 3...")
cand_s3 = collect_candidates_from_source(
    f"{TRAIN}/train_source3.tsv",
    s1_name_map,
    s1_address_map
)

print("Source 3 finished.")
print("S1 records with S3 candidates:", len(cand_s3))
```

# DAY 1 — STRONG MODEL PIPELINE
## Step 1: Rebuild training validation

```python
# ============================================
# DAY 1 — STRONG MODEL PIPELINE
# STEP 1: Load 10,000 training S1 records
# ============================================

import pandas as pd
import numpy as np
from collections import defaultdict

TRAIN_S1 = "/content/amazon_ml_challenge/student_resource/dataset/train/train_source1.tsv"
TRAIN_S2 = "/content/amazon_ml_challenge/student_resource/dataset/train/train_source2.tsv"
TRAIN_S3 = "/content/amazon_ml_challenge/student_resource/dataset/train/train_source3.tsv"
GROUND_TRUTH = "/content/amazon_ml_challenge/student_resource/dataset/train/train_ground_truth.tsv"

# Load only 10,000 Source 1 records
train_s1_10k = pd.read_csv(
    TRAIN_S1,
    sep="\t",
    usecols=["entity_id", "business_name", "business_address", "country"],
    nrows=10000
)

print("Training S1 shape:", train_s1_10k.shape)
print(train_s1_10k.head())

# Build advanced blocking maps
s1_name_map = defaultdict(list)
s1_address_map = defaultdict(list)

for _, row in train_s1_10k.iterrows():

    sid = row["entity_id"]
    country = normalize_country(row["country"])

    name = normalize_advanced(row["business_name"])
    address = normalize_advanced(row["business_address"])

    if name:
        s1_name_map[(country, name)].append(sid)

    if address:
        s1_address_map[(country, address)].append(sid)

print("\nName keys:", len(s1_name_map))
print("Address keys:", len(s1_address_map))
print("S1 records:", len(train_s1_10k))
```

```python
# ============================================
# STEP 2: Generate training candidate pairs
# ============================================

from collections import defaultdict

def generate_training_candidates(source_path, source_name):
    candidates = []
    candidate_records = []

    print(f"\nScanning {source_name}...")

    for chunk_no, chunk in enumerate(
        pd.read_csv(
            source_path,
            sep="\t",
            usecols=[
                "entity_id",
                "business_name",
                "business_address",
                "country"
            ],
            chunksize=200_000
        )
    ):

        # Advanced normalization
        chunk["norm_name"] = chunk["business_name"].map(normalize_advanced)
        chunk["norm_address"] = chunk["business_address"].map(normalize_advanced)
        chunk["norm_country"] = chunk["country"].map(normalize_country)

        for row in chunk.itertuples(index=False):

            eid = row.entity_id
            country = row.norm_country
            name = row.norm_name
            address = row.norm_address

            matched_s1 = set()

            # Name blocking
            if name:
                matched_s1.update(
                    s1_name_map.get((country, name), [])
                )

            # Address blocking
            if address:
                matched_s1.update(
                    s1_address_map.get((country, address), [])
                )

            # Store only actual candidate pairs
            for sid in matched_s1:

                candidates.append((sid, eid))

                candidate_records.append({
                    "source1_entity_id": sid,
                    "candidate_entity_id": eid,
                    "business_name": row.business_name,
                    "business_address": row.business_address,
                    "country": row.country,
                    "source": source_name
                })

        if (chunk_no + 1) % 5 == 0:
            print(
                f"{source_name}: processed {(chunk_no + 1) * 200000:,} rows | "
                f"candidates: {len(candidates):,}"
            )

    print(f"{source_name} candidates: {len(candidates):,}")

    return candidates, candidate_records


# Source 2
train_candidates_s2, train_records_s2 = generate_training_candidates(
    TRAIN_S2,
    "S2"
)

# Source 3
train_candidates_s3, train_records_s3 = generate_training_candidates(
    TRAIN_S3,
    "S3"
)

print("\n================================")
print("CANDIDATE GENERATION COMPLETE")
print("================================")

print("S2 candidates:", len(train_candidates_s2))
print("S3 candidates:", len(train_candidates_s3))
print("Total candidates:", len(train_candidates_s2) + len(train_candidates_s3))
```

```python
# ============================================
# STEP 3: Build candidate dataframe + labels
# ============================================

print("Building candidate dataframe...")

# Combine S2 + S3 candidate records
train_candidate_df = pd.DataFrame(
    train_records_s2 + train_records_s3
)

print("Candidate dataframe shape:", train_candidate_df.shape)

# Load only the ground-truth columns
gt = pd.read_csv(
    GROUND_TRUTH,
    sep="\t",
    usecols=["source1_entity_id", "matched_entity_ids"]
)

print("Ground truth shape:", gt.shape)
print(gt.head())

# Keep only the 10,000 S1 records we are validating
s1_ids = set(train_s1_10k["entity_id"])

gt_10k = gt[gt["source1_entity_id"].isin(s1_ids)].copy()

print("\nGround truth rows for our 10k S1:", len(gt_10k))

# Convert ground truth into a dictionary:
# S1 ID -> set of correct S2/S3 IDs
ground_truth_map = {}

for _, row in gt_10k.iterrows():

    sid = row["source1_entity_id"]
    matches = row["matched_entity_ids"]

    if pd.isna(matches) or str(matches).strip() == "":
        ground_truth_map[sid] = set()
    else:
        ground_truth_map[sid] = set(
            str(matches).split(",")
        )

# Create binary labels
train_candidate_df["label"] = train_candidate_df.apply(
    lambda row: int(
        row["candidate_entity_id"]
        in ground_truth_map.get(row["source1_entity_id"], set())
    ),
    axis=1
)

print("\n================================")
print("LABELING COMPLETE")
print("================================")

print("Total candidate pairs:", len(train_candidate_df))
print("Positive pairs:", int(train_candidate_df["label"].sum()))
print("Negative pairs:", int((train_candidate_df["label"] == 0).sum()))

print("\nPositive percentage:",
      round(train_candidate_df["label"].mean() * 100, 3), "%")
```

```python
# ============================================
# STEP 4: Build 9 ML features
# ============================================

from difflib import SequenceMatcher

print("Preparing feature dataframe...")

# Rename candidate columns for clarity
train_candidate_df = train_candidate_df.rename(
    columns={
        "business_name": "candidate_business_name",
        "business_address": "candidate_business_address",
        "country": "candidate_country"
    }
)

# Add Source 1 information
s1_feature_df = train_s1_10k[
    ["entity_id", "business_name", "business_address", "country"]
].rename(
    columns={
        "entity_id": "source1_entity_id",
        "business_name": "s1_business_name",
        "business_address": "s1_business_address",
        "country": "s1_country"
    }
)

train_candidate_df = train_candidate_df.merge(
    s1_feature_df,
    on="source1_entity_id",
    how="left"
)

print("Merged dataframe shape:", train_candidate_df.shape)


# --------------------------------------------
# Helper functions
# --------------------------------------------

def safe_text(x):
    if pd.isna(x):
        return ""
    return str(x)


def similarity(a, b):
    a = safe_text(a)
    b = safe_text(b)

    if not a or not b:
        return 0.0

    return SequenceMatcher(None, a, b).ratio()


def token_overlap(a, b):
    a = set(safe_text(a).split())
    b = set(safe_text(b).split())

    if not a or not b:
        return 0.0

    return len(a & b) / len(a | b)


# --------------------------------------------
# Normalize text
# --------------------------------------------

train_candidate_df["s1_name_norm"] = (
    train_candidate_df["s1_business_name"]
    .map(normalize_advanced)
)

train_candidate_df["candidate_name_norm"] = (
    train_candidate_df["candidate_business_name"]
    .map(normalize_advanced)
)

train_candidate_df["s1_address_norm"] = (
    train_candidate_df["s1_business_address"]
    .map(normalize_advanced)
)

train_candidate_df["candidate_address_norm"] = (
    train_candidate_df["candidate_business_address"]
    .map(normalize_advanced)
)

train_candidate_df["s1_country_norm"] = (
    train_candidate_df["s1_country"]
    .map(normalize_country)
)

train_candidate_df["candidate_country_norm"] = (
    train_candidate_df["candidate_country"]
    .map(normalize_country)
)


# --------------------------------------------
# 1. Exact features
# --------------------------------------------

train_candidate_df["name_exact"] = (
    train_candidate_df["s1_name_norm"]
    == train_candidate_df["candidate_name_norm"]
).astype(int)

train_candidate_df["address_exact"] = (
    train_candidate_df["s1_address_norm"]
    == train_candidate_df["candidate_address_norm"]
).astype(int)

train_candidate_df["country_exact"] = (
    train_candidate_df["s1_country_norm"]
    == train_candidate_df["candidate_country_norm"]
).astype(int)


# --------------------------------------------
# 2. Similarity features
# --------------------------------------------

print("Calculating name similarity...")

train_candidate_df["name_similarity"] = [
    similarity(a, b)
    for a, b in zip(
        train_candidate_df["s1_name_norm"],
        train_candidate_df["candidate_name_norm"]
    )
]

print("Calculating address similarity...")

train_candidate_df["address_similarity"] = [
    similarity(a, b)
    for a, b in zip(
        train_candidate_df["s1_address_norm"],
        train_candidate_df["candidate_address_norm"]
    )
]


# --------------------------------------------
# 3. Token overlap
# --------------------------------------------

print("Calculating token overlap...")

train_candidate_df["name_token_overlap"] = [
    token_overlap(a, b)
    for a, b in zip(
        train_candidate_df["s1_name_norm"],
        train_candidate_df["candidate_name_norm"]
    )
]

train_candidate_df["address_token_overlap"] = [
    token_overlap(a, b)
    for a, b in zip(
        train_candidate_df["s1_address_norm"],
        train_candidate_df["candidate_address_norm"]
    )
]


# --------------------------------------------
# 4. Length differences
# --------------------------------------------

train_candidate_df["name_length_diff"] = (
    train_candidate_df["s1_name_norm"].str.len()
    -
    train_candidate_df["candidate_name_norm"].str.len()
).abs()

train_candidate_df["address_length_diff"] = (
    train_candidate_df["s1_address_norm"].str.len()
    -
    train_candidate_df["candidate_address_norm"].str.len()
).abs()


# --------------------------------------------
# Final feature list
# --------------------------------------------

FEATURES = [
    "name_exact",
    "address_exact",
    "country_exact",
    "name_similarity",
    "address_similarity",
    "name_token_overlap",
    "address_token_overlap",
    "name_length_diff",
    "address_length_diff"
]

print("\n================================")
print("FEATURE ENGINEERING COMPLETE")
print("================================")

print("Features:", FEATURES)
print("Feature dataframe shape:", train_candidate_df.shape)

print("\nFeature preview:")
print(train_candidate_df[FEATURES + ["label"]].head())

print("\nMissing values:")
print(train_candidate_df[FEATURES].isna().sum())
```

```python
# ============================================
# STEP 5: Train CatBoost + Entity-level F0.5
# ============================================

from catboost import CatBoostClassifier
from sklearn.metrics import precision_score, recall_score
import numpy as np

# --------------------------------------------
# 1. Split by Source-1 entity
# --------------------------------------------

all_s1_ids = train_s1_10k["entity_id"].unique()

rng = np.random.RandomState(42)
rng.shuffle(all_s1_ids)

split_point = int(len(all_s1_ids) * 0.80)

train_s1_ids = set(all_s1_ids[:split_point])
val_s1_ids = set(all_s1_ids[split_point:])

train_df = train_candidate_df[
    train_candidate_df["source1_entity_id"].isin(train_s1_ids)
].copy()

val_df = train_candidate_df[
    train_candidate_df["source1_entity_id"].isin(val_s1_ids)
].copy()

print("Training S1 entities:", len(train_s1_ids))
print("Validation S1 entities:", len(val_s1_ids))

print("Training candidate pairs:", len(train_df))
print("Validation candidate pairs:", len(val_df))


# --------------------------------------------
# 2. Prepare X / y
# --------------------------------------------

X_train = train_df[FEATURES]
y_train = train_df["label"]

X_val = val_df[FEATURES]
y_val = val_df["label"]


# --------------------------------------------
# 3. Train CatBoost
# --------------------------------------------

model = CatBoostClassifier(
    iterations=300,
    depth=7,
    learning_rate=0.08,
    loss_function="Logloss",
    eval_metric="Logloss",
    random_seed=42,
    verbose=False,
    allow_writing_files=False
)

print("\nTraining CatBoost...")

model.fit(
    X_train,
    y_train,
    eval_set=(X_val, y_val),
    early_stopping_rounds=50,
    verbose=False
)

print("Training complete.")
print("Best iteration:", model.get_best_iteration())


# --------------------------------------------
# 4. Validation probabilities
# --------------------------------------------

val_df["probability"] = model.predict_proba(X_val)[:, 1]


# --------------------------------------------
# 5. Entity-level Macro F0.5
# --------------------------------------------

def entity_macro_f05(
    val_candidates,
    threshold,
    ground_truth_map,
    validation_s1_ids
):

    scores = []
    all_true = []
    all_pred = []

    for sid in validation_s1_ids:

        # True matches
        true_set = ground_truth_map.get(sid, set())

        # Candidate predictions above threshold
        rows = val_candidates[
            (val_candidates["source1_entity_id"] == sid) &
            (val_candidates["probability"] >= threshold)
        ]

        pred_set = set(rows["candidate_entity_id"])

        tp = len(true_set & pred_set)
        fp = len(pred_set - true_set)
        fn = len(true_set - pred_set)

        # Special singleton/no-match case
        if len(true_set) == 0 and len(pred_set) == 0:
            score = 1.0

        else:
            precision = (
                tp / (tp + fp)
                if (tp + fp) > 0
                else 0.0
            )

            recall = (
                tp / (tp + fn)
                if (tp + fn) > 0
                else 0.0
            )

            if precision == 0 and recall == 0:
                score = 0.0
            else:
                score = (
                    1.25 * precision * recall
                    / (0.25 * precision + recall)
                )

        scores.append(score)

        all_true.extend(true_set)
        all_pred.extend(pred_set)

    macro_f05 = np.mean(scores)

    return macro_f05


# --------------------------------------------
# 6. Test threshold = 0.50
# --------------------------------------------

threshold = 0.50

macro_f05 = entity_macro_f05(
    val_df,
    threshold,
    ground_truth_map,
    val_s1_ids
)

pred_binary = (
    val_df["probability"] >= threshold
).astype(int)

precision = precision_score(
    y_val,
    pred_binary,
    zero_division=0
)

recall = recall_score(
    y_val,
    pred_binary,
    zero_division=0
)

print("\n================================")
print("VALIDATION RESULTS")
print("================================")

print(f"Threshold       : {threshold}")
print(f"Pair Precision  : {precision:.6f}")
print(f"Pair Recall     : {recall:.6f}")
print(f"Macro F0.5      : {macro_f05:.6f}")
```

```python
# ============================================
# STEP 5B: Measure BLOCKING / CANDIDATE RECALL
# ============================================

total_true_matches = 0
covered_true_matches = 0

entity_results = []

for sid in val_s1_ids:

    # All real matches from ground truth
    true_set = ground_truth_map.get(sid, set())

    # Candidates generated by our blocking
    candidate_set = set(
        val_df[
            val_df["source1_entity_id"] == sid
        ]["candidate_entity_id"]
    )

    covered = true_set & candidate_set

    total_true_matches += len(true_set)
    covered_true_matches += len(covered)

    entity_results.append({
        "source1_entity_id": sid,
        "true_matches": len(true_set),
        "candidates": len(candidate_set),
        "covered_true_matches": len(covered),
        "missed_true_matches": len(true_set - candidate_set)
    })


candidate_recall = (
    covered_true_matches / total_true_matches
    if total_true_matches > 0 else 0
)

entity_results_df = pd.DataFrame(entity_results)

print("================================")
print("BLOCKING RECALL")
print("================================")

print("Validation S1 entities:", len(val_s1_ids))
print("Total true matches:", total_true_matches)
print("True matches found in candidates:", covered_true_matches)
print("True matches missed by blocking:",
      total_true_matches - covered_true_matches)

print(
    f"\nCandidate Recall: {candidate_recall:.6f}"
)

print(
    f"Candidate Recall: {candidate_recall * 100:.2f}%"
)

print("\nEntities with missed true matches:",
      (entity_results_df["missed_true_matches"] > 0).sum())

print("\nEntities with zero candidates:",
      (entity_results_df["candidates"] == 0).sum())
```

```python
# ============================================
# STEP 6: Analyze MISSED TRUE MATCHES
# ============================================

# Get validation S1 records
val_s1_df = train_s1_10k[
    train_s1_10k["entity_id"].isin(val_s1_ids)
].copy()

# Create lookup for S1 information
s1_lookup = val_s1_df.set_index("entity_id").to_dict("index")

# Collect the actual true-match IDs that blocking missed
missed_pairs = []

for sid in val_s1_ids:

    true_set = ground_truth_map.get(sid, set())

    candidate_set = set(
        val_df[
            val_df["source1_entity_id"] == sid
        ]["candidate_entity_id"]
    )

    missed = true_set - candidate_set

    for eid in missed:
        missed_pairs.append((sid, eid))

print("Missed true pairs:", len(missed_pairs))

# We need to retrieve the actual Source 2/3 records
missed_ids = set(eid for _, eid in missed_pairs)

print("Unique missed entity IDs:", len(missed_ids))

# ------------------------------------------------
# Scan Source 2 and Source 3 to retrieve them
# ------------------------------------------------

missed_records = {}

def collect_missed_records(path, source_name):

    print(f"\nScanning {source_name} for missed true records...")

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ],
        chunksize=200_000
    ):

        found = chunk[
            chunk["entity_id"].isin(missed_ids)
        ]

        for _, row in found.iterrows():

            missed_records[row["entity_id"]] = {
                "business_name": row["business_name"],
                "business_address": row["business_address"],
                "country": row["country"],
                "source": source_name
            }

    print(
        f"{source_name}: retrieved "
        f"{sum(1 for x in missed_records.values() if x['source'] == source_name):,}"
    )


collect_missed_records(TRAIN_S2, "S2")
collect_missed_records(TRAIN_S3, "S3")

print("\nRetrieved missed records:", len(missed_records))


# ------------------------------------------------
# Build comparison table
# ------------------------------------------------

analysis_rows = []

for sid, eid in missed_pairs:

    if sid not in s1_lookup:
        continue

    if eid not in missed_records:
        continue

    s1 = s1_lookup[sid]
    target = missed_records[eid]

    s1_name = normalize_advanced(s1["business_name"])
    target_name = normalize_advanced(target["business_name"])

    s1_address = normalize_advanced(s1["business_address"])
    target_address = normalize_advanced(target["business_address"])

    analysis_rows.append({
        "source1_entity_id": sid,
        "matched_entity_id": eid,
        "s1_name": s1["business_name"],
        "matched_name": target["business_name"],
        "s1_address": s1["business_address"],
        "matched_address": target["business_address"],
        "s1_country": s1["country"],
        "matched_country": target["country"],
        "name_exact_after_normalization": int(
            s1_name == target_name and s1_name != ""
        ),
        "address_exact_after_normalization": int(
            s1_address == target_address and s1_address != ""
        )
    })

missed_analysis_df = pd.DataFrame(analysis_rows)

print("\n================================")
print("MISSED MATCH ANALYSIS")
print("================================")

print("Rows:", len(missed_analysis_df))

print(
    "\nName exact:",
    missed_analysis_df[
        "name_exact_after_normalization"
    ].sum()
)

print(
    "Address exact:",
    missed_analysis_df[
        "address_exact_after_normalization"
    ].sum()
)

print("\nSample missed matches:")
display(
    missed_analysis_df.head(20)
)
```

```python
# ============================================
# STEP 7: Test additional blocking keys
# ============================================

import re
from collections import Counter


def get_tokens(text):
    text = normalize_advanced(text)
    return set(
        t for t in text.split()
        if len(t) >= 4
    )


def get_numbers(text):
    if pd.isna(text):
        return set()
    return set(re.findall(r"\d+", str(text)))


def get_postal_codes(text):
    if pd.isna(text):
        return set()

    text = str(text)

    # US ZIP / ZIP+4
    us = set(re.findall(r"\b\d{5}(?:-\d{4})?\b", text))

    # Indian PIN
    india = set(re.findall(r"\b[1-9]\d{5}\b", text))

    return us | india


# --------------------------------------------
# Calculate key overlap for missed matches
# --------------------------------------------

results = []

for _, row in missed_analysis_df.iterrows():

    name1 = normalize_advanced(row["s1_name"])
    name2 = normalize_advanced(row["matched_name"])

    addr1 = normalize_advanced(row["s1_address"])
    addr2 = normalize_advanced(row["matched_address"])

    name_tokens_1 = get_tokens(name1)
    name_tokens_2 = get_tokens(name2)

    addr_tokens_1 = get_tokens(addr1)
    addr_tokens_2 = get_tokens(addr2)

    numbers_1 = get_numbers(addr1)
    numbers_2 = get_numbers(addr2)

    postal_1 = get_postal_codes(addr1)
    postal_2 = get_postal_codes(addr2)

    results.append({

        # Name token overlap
        "name_token_overlap":
            len(name_tokens_1 & name_tokens_2) > 0,

        # Two or more common name tokens
        "name_2plus_tokens":
            len(name_tokens_1 & name_tokens_2) >= 2,

        # Address token overlap
        "address_token_overlap":
            len(addr_tokens_1 & addr_tokens_2) > 0,

        # Two or more address tokens
        "address_2plus_tokens":
            len(addr_tokens_1 & addr_tokens_2) >= 2,

        # Shared number
        "shared_address_number":
            len(numbers_1 & numbers_2) > 0,

        # Shared postal code
        "shared_postal":
            len(postal_1 & postal_2) > 0,

        # Combined signals
        "name_or_address_token":
            (
                len(name_tokens_1 & name_tokens_2) > 0
                or
                len(addr_tokens_1 & addr_tokens_2) > 0
            ),

        "strong_combined":
            (
                len(name_tokens_1 & name_tokens_2) >= 1
                and
                len(addr_tokens_1 & addr_tokens_2) >= 1
            )
    })


blocking_analysis = pd.DataFrame(results)

print("================================")
print("MISSED MATCH RECOVERY ANALYSIS")
print("================================")

for col in blocking_analysis.columns:

    count = blocking_analysis[col].sum()
    percentage = count / len(blocking_analysis) * 100

    print(
        f"{col:25s} : "
        f"{count:5d} / {len(blocking_analysis)} "
        f"({percentage:.2f}%)"
    )
```

```python
# ============================================
# STEP 8: Build token-frequency statistics
# ============================================

from collections import Counter
import re

name_token_freq = Counter()
address_token_freq = Counter()

def useful_tokens(text):
    text = normalize_advanced(text)

    tokens = set(
        t for t in text.split()
        if len(t) >= 4 and not t.isdigit()
    )

    return tokens


def scan_token_frequency(path, source_name):

    print(f"\nScanning {source_name} for token frequencies...")

    processed = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "business_name",
            "business_address"
        ],
        chunksize=200_000
    ):

        for name in chunk["business_name"]:
            for token in useful_tokens(name):
                name_token_freq[token] += 1

        for address in chunk["business_address"]:
            for token in useful_tokens(address):
                address_token_freq[token] += 1

        processed += len(chunk)

        if processed % 1_000_000 == 0:
            print(
                f"{source_name}: "
                f"{processed:,} rows processed"
            )


scan_token_frequency(TRAIN_S2, "S2")
scan_token_frequency(TRAIN_S3, "S3")


print("\n================================")
print("TOKEN FREQUENCY COMPLETE")
print("================================")

print("Unique name tokens:", len(name_token_freq))
print("Unique address tokens:", len(address_token_freq))

print("\nMost common NAME tokens:")
print(name_token_freq.most_common(20))

print("\nMost common ADDRESS tokens:")
print(address_token_freq.most_common(20))
```

```python
# ============================================
# STEP 9: Test rare-token thresholds
# ============================================

def get_token_set(text):
    text = normalize_advanced(text)

    return set(
        token
        for token in text.split()
        if len(token) >= 4 and not token.isdigit()
    )


# Build token sets for the missed true pairs
missed_token_rows = []

for _, row in missed_analysis_df.iterrows():

    s1_name_tokens = get_token_set(row["s1_name"])
    match_name_tokens = get_token_set(row["matched_name"])

    s1_addr_tokens = get_token_set(row["s1_address"])
    match_addr_tokens = get_token_set(row["matched_address"])

    missed_token_rows.append({
        "name_shared":
            s1_name_tokens & match_name_tokens,

        "address_shared":
            s1_addr_tokens & match_addr_tokens
    })


# --------------------------------------------
# Test different rarity thresholds
# --------------------------------------------

thresholds = [5, 10, 20, 50, 100, 250, 500, 1000]

print("================================")
print("RARE TOKEN RECOVERY")
print("================================")

print(
    f"{'Threshold':>10} | "
    f"{'Name':>8} | "
    f"{'Address':>10} | "
    f"{'Either':>8}"
)

print("-" * 50)

for threshold in thresholds:

    name_recovered = 0
    address_recovered = 0
    either_recovered = 0

    for row in missed_token_rows:

        name_ok = any(
            name_token_freq.get(token, 10**9)
            <= threshold
            for token in row["name_shared"]
        )

        address_ok = any(
            address_token_freq.get(token, 10**9)
            <= threshold
            for token in row["address_shared"]
        )

        if name_ok:
            name_recovered += 1

        if address_ok:
            address_recovered += 1

        if name_ok or address_ok:
            either_recovered += 1

    print(
        f"{threshold:>10} | "
        f"{name_recovered:>5} "
        f"({name_recovered / len(missed_token_rows) * 100:5.1f}%) | "
        f"{address_recovered:>5} "
        f"({address_recovered / len(missed_token_rows) * 100:5.1f}%) | "
        f"{either_recovered:>5} "
        f"({either_recovered / len(missed_token_rows) * 100:5.1f}%)"
    )
```

```python
# ============================================
# STEP 10: Estimate candidate volume
# ============================================

def get_rare_tokens(text, freq_counter, threshold):
    tokens = get_token_set(text)

    return {
        token
        for token in tokens
        if freq_counter.get(token, 10**9) <= threshold
    }


# Build the 10k S1 token sets
s1_token_stats = []

for _, row in train_s1_10k.iterrows():

    name_tokens = get_rare_tokens(
        row["business_name"],
        name_token_freq,
        1000
    )

    address_tokens = get_rare_tokens(
        row["business_address"],
        address_token_freq,
        1000
    )

    s1_token_stats.append({
        "entity_id": row["entity_id"],
        "name_tokens": name_tokens,
        "address_tokens": address_tokens
    })


# --------------------------------------------
# Estimate volume for different thresholds
# --------------------------------------------

thresholds = [50, 100, 250, 500, 1000]

print("================================")
print("ESTIMATED CANDIDATE VOLUME")
print("================================")

for threshold in thresholds:

    estimated_name = 0
    estimated_address = 0

    for _, row in train_s1_10k.iterrows():

        name_tokens = get_rare_tokens(
            row["business_name"],
            name_token_freq,
            threshold
        )

        address_tokens = get_rare_tokens(
            row["business_address"],
            address_token_freq,
            threshold
        )

        estimated_name += sum(
            name_token_freq[token]
            for token in name_tokens
        )

        estimated_address += sum(
            address_token_freq[token]
            for token in address_tokens
        )

    estimated_total = estimated_name + estimated_address

    print(
        f"\nThreshold {threshold}:"
    )
    print(
        f"  Name candidate upper estimate    : "
        f"{estimated_name:,}"
    )
    print(
        f"  Address candidate upper estimate : "
        f"{estimated_address:,}"
    )
    print(
        f"  Combined upper estimate          : "
        f"{estimated_total:,}"
    )
```

```python
# ============================================
# STEP 11: Build rare-token indexes
# ============================================

from collections import defaultdict

RARE_THRESHOLD = 250

name_token_index_s2 = defaultdict(set)
address_token_index_s2 = defaultdict(set)

name_token_index_s3 = defaultdict(set)
address_token_index_s3 = defaultdict(set)


def add_tokens_to_index(
    text,
    entity_id,
    index,
    frequency_counter,
    threshold
):
    tokens = get_token_set(text)

    for token in tokens:

        if frequency_counter.get(token, 10**9) <= threshold:
            index[token].add(entity_id)


def build_rare_indexes(
    path,
    source_name,
    name_index,
    address_index
):

    print(f"\nBuilding rare-token indexes for {source_name}...")

    processed = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_name",
            "business_address"
        ],
        chunksize=200_000
    ):

        for row in chunk.itertuples(index=False):

            add_tokens_to_index(
                row.business_name,
                row.entity_id,
                name_index,
                name_token_freq,
                RARE_THRESHOLD
            )

            add_tokens_to_index(
                row.business_address,
                row.entity_id,
                address_index,
                address_token_freq,
                RARE_THRESHOLD
            )

        processed += len(chunk)

        if processed % 1_000_000 == 0:
            print(
                f"{source_name}: "
                f"{processed:,} rows"
            )

    print(
        f"{source_name} indexes complete."
    )


build_rare_indexes(
    TRAIN_S2,
    "S2",
    name_token_index_s2,
    address_token_index_s2
)

build_rare_indexes(
    TRAIN_S3,
    "S3",
    name_token_index_s3,
    address_token_index_s3
)


print("\n================================")
print("RARE TOKEN INDEXES COMPLETE")
print("================================")

print(
    "S2 name tokens:",
    len(name_token_index_s2)
)

print(
    "S2 address tokens:",
    len(address_token_index_s2)
)

print(
    "S3 name tokens:",
    len(name_token_index_s3)
)

print(
    "S3 address tokens:",
    len(address_token_index_s3)
)
```

```python
# ============================================
# STEP 12: Improved Blocking V2
# Rare tokens <= 250 + Country
# ============================================

RARE_THRESHOLD_V2 = 250

# --------------------------------------------
# Build country-aware token maps for
# validation S1 records
# --------------------------------------------

val_name_token_map = defaultdict(set)
val_address_token_map = defaultdict(set)

val_exact_name_map = defaultdict(set)
val_exact_address_map = defaultdict(set)

for _, row in val_s1_df.iterrows():

    sid = row["entity_id"]
    country = normalize_country(row["country"])

    name = normalize_advanced(row["business_name"])
    address = normalize_advanced(row["business_address"])

    # Exact blocking
    if name:
        val_exact_name_map[(country, name)].add(sid)

    if address:
        val_exact_address_map[(country, address)].add(sid)

    # Rare name tokens
    for token in get_token_set(name):

        if name_token_freq.get(token, 10**9) <= RARE_THRESHOLD_V2:
            val_name_token_map[(country, token)].add(sid)

    # Rare address tokens
    for token in get_token_set(address):

        if address_token_freq.get(token, 10**9) <= RARE_THRESHOLD_V2:
            val_address_token_map[(country, token)].add(sid)


print("S1 exact name keys:",
      len(val_exact_name_map))

print("S1 exact address keys:",
      len(val_exact_address_map))

print("S1 rare name keys:",
      len(val_name_token_map))

print("S1 rare address keys:",
      len(val_address_token_map))


# --------------------------------------------
# Generate V2 candidates
# --------------------------------------------

v2_candidates = defaultdict(set)


def scan_for_v2_candidates(path, source_name):

    print(f"\nScanning {source_name} for V2 candidates...")

    processed = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ],
        chunksize=200_000
    ):

        for row in chunk.itertuples(index=False):

            eid = row.entity_id
            country = normalize_country(row.country)

            name = normalize_advanced(row.business_name)
            address = normalize_advanced(row.business_address)

            matched_s1 = set()

            # --------------------------------
            # Exact name
            # --------------------------------
            if name:
                matched_s1.update(
                    val_exact_name_map.get(
                        (country, name),
                        set()
                    )
                )

            # --------------------------------
            # Exact address
            # --------------------------------
            if address:
                matched_s1.update(
                    val_exact_address_map.get(
                        (country, address),
                        set()
                    )
                )

            # --------------------------------
            # Rare name tokens
            # --------------------------------
            for token in get_token_set(name):

                if name_token_freq.get(
                    token,
                    10**9
                ) <= RARE_THRESHOLD_V2:

                    matched_s1.update(
                        val_name_token_map.get(
                            (country, token),
                            set()
                        )
                    )

            # --------------------------------
            # Rare address tokens
            # --------------------------------
            for token in get_token_set(address):

                if address_token_freq.get(
                    token,
                    10**9
                ) <= RARE_THRESHOLD_V2:

                    matched_s1.update(
                        val_address_token_map.get(
                            (country, token),
                            set()
                        )
                    )

            # Store candidate pairs
            for sid in matched_s1:
                v2_candidates[sid].add(eid)

        processed += len(chunk)

        if processed % 1_000_000 == 0:
            print(
                f"{source_name}: "
                f"{processed:,} rows processed"
            )


scan_for_v2_candidates(TRAIN_S2, "S2")
scan_for_v2_candidates(TRAIN_S3, "S3")


# --------------------------------------------
# Calculate statistics
# --------------------------------------------

total_v2_candidates = sum(
    len(x)
    for x in v2_candidates.values()
)

covered_true_matches_v2 = 0
total_true_matches_v2 = 0
zero_candidate_entities_v2 = 0
missed_entities_v2 = 0

for sid in val_s1_ids:

    true_set = ground_truth_map.get(
        sid,
        set()
    )

    candidate_set = v2_candidates.get(
        sid,
        set()
    )

    total_true_matches_v2 += len(true_set)

    covered_true_matches_v2 += len(
        true_set & candidate_set
    )

    if len(candidate_set) == 0:
        zero_candidate_entities_v2 += 1

    if len(true_set - candidate_set) > 0:
        missed_entities_v2 += 1


v2_recall = (
    covered_true_matches_v2 /
    total_true_matches_v2
    if total_true_matches_v2 > 0
    else 0
)


print("\n================================")
print("V2 BLOCKING RESULTS")
print("================================")

print(
    "Total candidate pairs:",
    f"{total_v2_candidates:,}"
)

print(
    "Total true matches:",
    f"{total_true_matches_v2:,}"
)

print(
    "True matches found:",
    f"{covered_true_matches_v2:,}"
)

print(
    "True matches missed:",
    f"{total_true_matches_v2 - covered_true_matches_v2:,}"
)

print(
    f"\nCandidate Recall: {v2_recall:.6f}"
)

print(
    f"Candidate Recall: {v2_recall * 100:.2f}%"
)

print(
    "Entities with zero candidates:",
    zero_candidate_entities_v2
)

print(
    "Entities with missed true matches:",
    missed_entities_v2
)
```

```python
# ============================================
# STEP 13: Test threshold 500 on V2 misses
# ============================================

v2_missed_pairs = []

for sid in val_s1_ids:

    true_set = ground_truth_map.get(sid, set())

    candidate_set = v2_candidates.get(
        sid,
        set()
    )

    missed = true_set - candidate_set

    for eid in missed:
        v2_missed_pairs.append((sid, eid))


print("V2 missed true pairs:", len(v2_missed_pairs))


# Build lookup for the missed target records
missed_target_lookup = missed_records


recovered_by_500 = 0
recovered_name_500 = 0
recovered_address_500 = 0

still_missed_500 = []

for sid, eid in v2_missed_pairs:

    if eid not in missed_target_lookup:
        continue

    # S1 record
    s1 = s1_lookup[sid]

    # Target S2/S3 record
    target = missed_target_lookup[eid]

    # Normalize
    s1_name = normalize_advanced(
        s1["business_name"]
    )

    target_name = normalize_advanced(
        target["business_name"]
    )

    s1_address = normalize_advanced(
        s1["business_address"]
    )

    target_address = normalize_advanced(
        target["business_address"]
    )

    # Shared name tokens
    name_shared = (
        get_token_set(s1_name)
        &
        get_token_set(target_name)
    )

    # Shared address tokens
    address_shared = (
        get_token_set(s1_address)
        &
        get_token_set(target_address)
    )

    name_hit = any(
        name_token_freq.get(token, 10**9) <= 500
        for token in name_shared
    )

    address_hit = any(
        address_token_freq.get(token, 10**9) <= 500
        for token in address_shared
    )

    if name_hit:
        recovered_name_500 += 1

    if address_hit:
        recovered_address_500 += 1

    if name_hit or address_hit:
        recovered_by_500 += 1
    else:
        still_missed_500.append((sid, eid))


print("\n================================")
print("THRESHOLD 500 RECOVERY")
print("================================")

print(
    "V2 missed pairs:",
    len(v2_missed_pairs)
)

print(
    "Potentially recovered by name:",
    recovered_name_500
)

print(
    "Potentially recovered by address:",
    recovered_address_500
)

print(
    "Potentially recovered by either:",
    recovered_by_500
)

print(
    "Still missed:",
    len(still_missed_500)
)

print(
    "\nPotential recovery:",
    round(
        recovered_by_500 /
        len(v2_missed_pairs) * 100,
        2
    ),
    "%"
)

# Estimated overall recall if all these are recovered
estimated_recall_500 = (
    covered_true_matches_v2 +
    recovered_by_500
) / total_true_matches_v2

print(
    "\nEstimated candidate recall after 500:",
    round(estimated_recall_500 * 100, 2),
    "%"
)
```

```python
# ============================================
# STEP 14: Analyze remaining V2 misses
# ============================================

from difflib import SequenceMatcher
import re

def text_similarity(a, b):
    a = normalize_advanced(a)
    b = normalize_advanced(b)

    if not a or not b:
        return 0.0

    return SequenceMatcher(None, a, b).ratio()


def extract_numbers(text):
    if pd.isna(text):
        return set()

    return set(re.findall(r"\d+", str(text)))


diagnostic_rows = []

for sid, eid in v2_missed_pairs:

    if sid not in s1_lookup or eid not in missed_records:
        continue

    s1 = s1_lookup[sid]
    target = missed_records[eid]

    s1_name = normalize_advanced(s1["business_name"])
    target_name = normalize_advanced(target["business_name"])

    s1_address = normalize_advanced(s1["business_address"])
    target_address = normalize_advanced(target["business_address"])

    name_tokens_1 = get_token_set(s1_name)
    name_tokens_2 = get_token_set(target_name)

    addr_tokens_1 = get_token_set(s1_address)
    addr_tokens_2 = get_token_set(target_address)

    shared_name = name_tokens_1 & name_tokens_2
    shared_addr = addr_tokens_1 & addr_tokens_2

    nums_1 = extract_numbers(s1_address)
    nums_2 = extract_numbers(target_address)

    shared_numbers = nums_1 & nums_2

    name_sim = text_similarity(
        s1_name,
        target_name
    )

    address_sim = text_similarity(
        s1_address,
        target_address
    )

    diagnostic_rows.append({

        "source1_entity_id": sid,
        "matched_entity_id": eid,

        "shared_name_tokens": len(shared_name),
        "shared_address_tokens": len(shared_addr),

        "shared_address_numbers": len(shared_numbers),

        "name_similarity": name_sim,
        "address_similarity": address_sim,

        "name_token_any": len(shared_name) >= 1,
        "name_token_2plus": len(shared_name) >= 2,

        "address_token_any": len(shared_addr) >= 1,
        "address_token_2plus": len(shared_addr) >= 2,

        "address_number_shared": len(shared_numbers) >= 1
    })


v2_miss_diagnostic = pd.DataFrame(
    diagnostic_rows
)

print("================================")
print("V2 MISS DIAGNOSTICS")
print("================================")

print(
    "Remaining missed pairs:",
    len(v2_miss_diagnostic)
)

print("\nRecovery signals:")

signals = [
    "name_token_any",
    "name_token_2plus",
    "address_token_any",
    "address_token_2plus",
    "address_number_shared"
]

for signal in signals:

    count = v2_miss_diagnostic[signal].sum()

    print(
        f"{signal:25s}: "
        f"{count:4d} "
        f"({count / len(v2_miss_diagnostic) * 100:.2f}%)"
    )


print("\nSimilarity thresholds:")

for threshold in [0.60, 0.70, 0.75, 0.80, 0.85, 0.90]:

    name_count = (
        v2_miss_diagnostic["name_similarity"]
        >= threshold
    ).sum()

    address_count = (
        v2_miss_diagnostic["address_similarity"]
        >= threshold
    ).sum()

    either_count = (
        (v2_miss_diagnostic["name_similarity"] >= threshold)
        |
        (v2_miss_diagnostic["address_similarity"] >= threshold)
    ).sum()

    print(
        f"{threshold:.2f}  | "
        f"name: {name_count:4d} | "
        f"address: {address_count:4d} | "
        f"either: {either_count:4d}"
    )


print("\nSimilarity statistics:")

print(
    v2_miss_diagnostic[
        ["name_similarity", "address_similarity"]
    ].describe()
)
```

```python
# ============================================
# STEP 15: Test address-number + token signal
# ============================================

address_combo_results = []

for _, row in v2_miss_diagnostic.iterrows():

    sid = row["source1_entity_id"]
    eid = row["matched_entity_id"]

    s1 = s1_lookup[sid]
    target = missed_records[eid]

    s1_address = normalize_advanced(
        s1["business_address"]
    )

    target_address = normalize_advanced(
        target["business_address"]
    )

    # Address numbers
    nums1 = extract_numbers(s1_address)
    nums2 = extract_numbers(target_address)

    shared_numbers = nums1 & nums2

    # Address tokens
    tokens1 = get_token_set(s1_address)
    tokens2 = get_token_set(target_address)

    shared_tokens = tokens1 & tokens2

    # Rare shared address tokens
    rare_shared_tokens = {
        token
        for token in shared_tokens
        if address_token_freq.get(
            token,
            10**9
        ) <= 250
    }

    address_combo_results.append({

        "number_and_any_token":
            len(shared_numbers) > 0
            and len(shared_tokens) > 0,

        "number_and_rare_token":
            len(shared_numbers) > 0
            and len(rare_shared_tokens) > 0,

        "number_only":
            len(shared_numbers) > 0
    })


combo_df = pd.DataFrame(
    address_combo_results
)

print("================================")
print("ADDRESS COMBINATION ANALYSIS")
print("================================")

for col in combo_df.columns:

    count = combo_df[col].sum()

    print(
        f"{col:25s}: "
        f"{count:4d} / {len(combo_df)} "
        f"({count / len(combo_df) * 100:.2f}%)"
    )
```

```python
from collections import Counter

def extract_numbers(text):
    if not text:
        return set()
    return set(re.findall(r'\b\d+\b', str(text)))

number_freq = Counter()

for path in [TRAIN_S2, TRAIN_S3]:
    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=["business_address"],
        chunksize=200000
    ):
        for addr in chunk["business_address"].dropna():
            nums = extract_numbers(normalize_advanced(addr))
            number_freq.update(nums)

print("Unique address numbers:", len(number_freq))

print("\nMost common address numbers:")
for num, count in number_freq.most_common(30):
    print(num, count)
```

```python
thresholds = [10, 25, 50, 100, 250, 500, 1000]

print("================================")
print("ADDRESS NUMBER FREQUENCY ANALYSIS")
print("================================")

for threshold in thresholds:
    recovered = 0

    for _, row in v2_miss_diagnostic.iterrows():
        sid = row["source1_entity_id"]
        eid = row["matched_entity_id"]

        s1 = s1_lookup[sid]
        target = missed_records[eid]

        nums1 = extract_numbers(
            normalize_advanced(s1["business_address"])
        )
        nums2 = extract_numbers(
            normalize_advanced(target["business_address"])
        )

        shared = nums1 & nums2

        if any(number_freq.get(n, 10**9) <= threshold for n in shared):
            recovered += 1

    print(
        f"threshold <= {threshold:4d}: "
        f"{recovered:4d} / 2298 "
        f"({recovered / 2298 * 100:.2f}%)"
    )
```

```python
import pandas as pd
import re
import unicodedata
from collections import defaultdict

def norm(x):
    if pd.isna(x):
        return ""
    x = str(x).lower().strip()
    x = unicodedata.normalize("NFKC", x)
    x = re.sub(r"[^\w\s]", " ", x)
    x = re.sub(r"\s+", " ", x)
    return x.strip()

# -----------------------------
# Build Source 1 indexes
# -----------------------------

print("Loading Source 1...")

s1 = pd.read_csv(
    TEST_S1,
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ]
)

print("Source 1 rows:", len(s1))

address_map = defaultdict(list)
name_map = defaultdict(list)

for row in s1.itertuples(index=False):

    sid = row.entity_id
    country = norm(row.country)

    name = norm(row.business_name)
    address = norm(row.business_address)

    if address:
        address_map[(country, address)].append(sid)

    if name:
        name_map[(country, name)].append(sid)

# Only unique name keys
unique_name_map = {
    key: ids[0]
    for key, ids in name_map.items()
    if len(ids) == 1
}

print("Address keys:", len(address_map))
print("Unique name keys:", len(unique_name_map))

# -----------------------------
# Generate matches
# -----------------------------

matches = defaultdict(list)

def process_source(path, source_name):

    print(f"\nProcessing {source_name}...")

    total = 0
    address_count = 0
    name_count = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ],
        chunksize=200000
    ):

        for row in chunk.itertuples(index=False):

            total += 1

            eid = row.entity_id
            country = norm(row.country)

            name = norm(row.business_name)
            address = norm(row.business_address)

            found = set()

            # Exact address + country
            if address:
                ids = address_map.get(
                    (country, address),
                    []
                )

                if ids:
                    found.update(ids)
                    address_count += len(ids)

            # Unique exact name + country
            if not found and name:
                sid = unique_name_map.get(
                    (country, name)
                )

                if sid is not None:
                    found.add(sid)
                    name_count += 1

            for sid in found:
                matches[sid].append(eid)

        print(
            f"{source_name}: processed {total:,} | "
            f"S1 matched: {len(matches):,}"
        )

    print(
        f"{source_name} COMPLETE | "
        f"address matches: {address_count:,} | "
        f"name matches: {name_count:,}"
    )


process_source(TEST_S2, "Source 2")
process_source(TEST_S3, "Source 3")

# Remove duplicates
for sid in matches:
    matches[sid] = sorted(set(matches[sid]))

print("\n================================")
print("SUBMISSION #2 CANDIDATES")
print("================================")
print("S1 entities:", len(s1))
print("S1 with matches:", len(matches))
print("S1 without matches:", len(s1) - len(matches))
```

```python
# ============================================
# SUBMISSION #3
# Add exact-name matches where S1 has <= 2
# entities with that same country + name
# ============================================

matches_v3 = defaultdict(list)

# Start with Submission #2 matches
for sid, ids in matches.items():
    matches_v3[sid] = list(ids)

# Name keys with exactly 2 S1 entities
name2_map = {
    key: ids
    for key, ids in name_map.items()
    if len(ids) == 2
}

print("Name keys with exactly 2 S1 entities:", len(name2_map))

def process_duplicate_names(path, source_name):

    print(f"\nProcessing {source_name}...")

    total = 0
    added_links = 0
    matched_rows = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_name",
            "country"
        ],
        chunksize=200000
    ):

        for row in chunk.itertuples(index=False):

            total += 1

            country = norm(row.country)
            name = norm(row.business_name)

            if not name:
                continue

            ids = name2_map.get((country, name))

            if ids:
                matched_rows += 1

                for sid in ids:
                    if row.entity_id not in matches_v3[sid]:
                        matches_v3[sid].append(row.entity_id)
                        added_links += 1

        print(
            f"{source_name}: processed {total:,} | "
            f"new links: {added_links:,}"
        )

    print(
        f"{source_name} COMPLETE | "
        f"new links: {added_links:,}"
    )


process_duplicate_names(TEST_S2, "Source 2")
process_duplicate_names(TEST_S3, "Source 3")

# Remove duplicates and sort
for sid in matches_v3:
    matches_v3[sid] = sorted(set(matches_v3[sid]))

print("\n================================")
print("SUBMISSION #3 SUMMARY")
print("================================")

print("S1 entities:", len(s1))
print("S1 with matches:", len(matches_v3))
print("S1 without matches:", len(s1) - len(matches_v3))

total_links = sum(len(x) for x in matches_v3.values())

print("Total matched links:", total_links)
```

```python
# ============================================
# STEP 18B-1
# Rebuild validation foundation
# ============================================

import pandas as pd
import numpy as np
import re
import unicodedata
from collections import defaultdict, Counter

BASE = "/content/amazon_ml_challenge/student_resource"

TRAIN_S1 = f"{BASE}/dataset/train/train_source1.tsv"
TRAIN_S2 = f"{BASE}/dataset/train/train_source2.tsv"
TRAIN_S3 = f"{BASE}/dataset/train/train_source3.tsv"
GROUND_TRUTH = f"{BASE}/dataset/train/train_ground_truth.tsv"

# --------------------------------------------
# Normalization
# --------------------------------------------

def normalize_advanced(text):
    if pd.isna(text):
        return ""

    text = str(text).lower().strip()
    text = unicodedata.normalize("NFKC", text)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()


def normalize_country(x):
    return normalize_advanced(x)


def get_token_set(text):
    if not text:
        return set()
    return set(str(text).split())


def extract_numbers(text):
    if not text:
        return set()
    return set(re.findall(r"\b\d+\b", str(text)))


# --------------------------------------------
# Load first 10,000 training Source 1
# --------------------------------------------

print("Loading first 10,000 Source 1 rows...")

train_s1_10k = pd.read_csv(
    TRAIN_S1,
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ],
    nrows=10000
)

print("Source 1:", train_s1_10k.shape)

# --------------------------------------------
# Proper entity-level split
# Same split used in our earlier validation
# --------------------------------------------

s1_train = train_s1_10k.iloc[:8000].copy()
s1_val = train_s1_10k.iloc[8000:].copy()

print("Training S1 entities:", len(s1_train))
print("Validation S1 entities:", len(s1_val))

# --------------------------------------------
# Load ground truth for these 10,000 S1
# --------------------------------------------

print("\nLoading ground truth...")

gt = pd.read_csv(
    GROUND_TRUTH,
    sep="\t"
)

gt_10k = gt[
    gt["source1_entity_id"].isin(
        set(train_s1_10k["entity_id"])
    )
].copy()

print("Ground-truth rows:", len(gt_10k))

# --------------------------------------------
# Create lookups
# --------------------------------------------

s1_lookup = (
    train_s1_10k
    .set_index("entity_id")
    .to_dict("index")
)

# Ground truth lookup
gt_lookup = dict(
    zip(
        gt_10k["source1_entity_id"],
        gt_10k["matched_entity_ids"]
    )
)

# Normalize ground-truth values to sets
def parse_match_ids(x):
    if pd.isna(x):
        return set()

    x = str(x).strip()

    if not x:
        return set()

    return {
        item.strip()
        for item in x.split(",")
        if item.strip()
    }


gt_lookup = {
    sid: parse_match_ids(value)
    for sid, value in gt_lookup.items()
}

print("\nGround-truth validation entities:", len(gt_lookup))

# --------------------------------------------
# Validation S1 IDs
# --------------------------------------------

val_ids = set(s1_val["entity_id"])

print("Validation S1 IDs:", len(val_ids))

# --------------------------------------------
# Basic verification
# --------------------------------------------

assert len(train_s1_10k) == 10000
assert len(s1_train) == 8000
assert len(s1_val) == 2000

print("\n================================")
print("FOUNDATION READY")
print("================================")
print("10k S1 loaded        :", len(train_s1_10k))
print("Train entities       :", len(s1_train))
print("Validation entities  :", len(s1_val))
print("Ground truth rows    :", len(gt_10k))
print("Validation IDs       :", len(val_ids))
```

```python
# ============================================
# STEP 18B-2
# Rebuild V2 candidate maps
# ============================================

# Maps from validation S1
name_map_val = defaultdict(list)
address_map_val = defaultdict(list)

for row in s1_val.itertuples(index=False):

    sid = row.entity_id
    country = normalize_country(row.country)

    name = normalize_advanced(row.business_name)
    address = normalize_advanced(row.business_address)

    if name:
        name_map_val[(country, name)].append(sid)

    if address:
        address_map_val[(country, address)].append(sid)

# Build rare-token indexes from FULL Source 2 + Source 3.
# We need token frequencies first.

print("Building full training token frequencies...")

name_token_freq = Counter()
address_token_freq = Counter()

for path, source_name in [
    (TRAIN_S2, "Source 2"),
    (TRAIN_S3, "Source 3")
]:

    print(f"\nScanning {source_name}...")

    rows = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=["business_name", "business_address"],
        chunksize=200000
    ):

        for name in chunk["business_name"].dropna():
            tokens = get_token_set(
                normalize_advanced(name)
            )
            name_token_freq.update(tokens)

        for address in chunk["business_address"].dropna():
            tokens = get_token_set(
                normalize_advanced(address)
            )
            address_token_freq.update(tokens)

        rows += len(chunk)

        if rows % 1000000 == 0:
            print(f"  processed {rows:,}")

print("\n================================")
print("TOKEN FREQUENCIES READY")
print("================================")
print("Name tokens:", len(name_token_freq))
print("Address tokens:", len(address_token_freq))

print("\nMost common address tokens:")
for token, count in address_token_freq.most_common(20):
    print(f"{token:20s} {count:,}")

# --------------------------------------------
# Build rare-token maps
# Same V2 cutoff: <=250
# --------------------------------------------

RARE_LIMIT = 250

rare_name_map = defaultdict(list)
rare_address_map = defaultdict(list)

print("\nBuilding rare-token indexes...")

for path, source_name in [
    (TRAIN_S2, "Source 2"),
    (TRAIN_S3, "Source 3")
]:

    print(f"Indexing {source_name}...")

    rows = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ],
        chunksize=200000
    ):

        for row in chunk.itertuples(index=False):

            eid = row.entity_id
            country = normalize_country(row.country)

            name_tokens = get_token_set(
                normalize_advanced(row.business_name)
            )

            address_tokens = get_token_set(
                normalize_advanced(row.business_address)
            )

            for token in name_tokens:
                if name_token_freq.get(token, 10**9) <= RARE_LIMIT:
                    rare_name_map[
                        (country, token)
                    ].append(eid)

            for token in address_tokens:
                if address_token_freq.get(token, 10**9) <= RARE_LIMIT:
                    rare_address_map[
                        (country, token)
                    ].append(eid)

        rows += len(chunk)

        if rows % 1000000 == 0:
            print(f"  indexed {rows:,}")

print("\n================================")
print("V2 BLOCKING INDEXES READY")
print("================================")
print("Rare name keys:", len(rare_name_map))
print("Rare address keys:", len(rare_address_map))
```

```python
# ============================================
# STEP 18B-3
# Rebuild V2 candidate pairs
# ============================================

candidate_pairs_v2 = []

def scan_candidates(path, source_name):

    print(f"\nScanning {source_name}...")

    rows = 0
    added = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ],
        chunksize=200000
    ):

        for row in chunk.itertuples(index=False):

            eid = row.entity_id
            country = normalize_country(row.country)

            name = normalize_advanced(row.business_name)
            address = normalize_advanced(row.business_address)

            found_s1 = set()

            # -----------------------------
            # V2 BLOCK 1: exact name
            # -----------------------------
            if name:
                found_s1.update(
                    name_map_val.get(
                        (country, name),
                        []
                    )
                )

            # -----------------------------
            # V2 BLOCK 2: exact address
            # -----------------------------
            if address:
                found_s1.update(
                    address_map_val.get(
                        (country, address),
                        []
                    )
                )

            # -----------------------------
            # V2 BLOCK 3: rare name token
            # -----------------------------
            name_tokens = get_token_set(name)

            for token in name_tokens:

                if name_token_freq.get(
                    token, 10**9
                ) <= RARE_LIMIT:

                    found_s1.update(
                        rare_name_map.get(
                            (country, token),
                            []
                        )
                    )

            # -----------------------------
            # V2 BLOCK 4: rare address token
            # -----------------------------
            address_tokens = get_token_set(address)

            for token in address_tokens:

                if address_token_freq.get(
                    token, 10**9
                ) <= RARE_LIMIT:

                    found_s1.update(
                        rare_address_map.get(
                            (country, token),
                            []
                        )
                    )

            # Store only actual candidate pairs
            for sid in found_s1:
                candidate_pairs_v2.append(
                    (sid, eid)
                )

            added += len(found_s1)
            rows += 1

        print(
            f"{source_name}: "
            f"{rows:,} rows | "
            f"{added:,} candidate links"
        )

    print(
        f"{source_name} COMPLETE | "
        f"candidate links: {added:,}"
    )


scan_candidates(TRAIN_S2, "Source 2")
scan_candidates(TRAIN_S3, "Source 3")

# --------------------------------------------
# Convert to DataFrame and deduplicate
# --------------------------------------------

v2_candidates = pd.DataFrame(
    candidate_pairs_v2,
    columns=[
        "source1_entity_id",
        "matched_entity_id"
    ]
).drop_duplicates()

print("\n================================")
print("V2 CANDIDATE SET")
print("================================")
print("Candidate pairs:", len(v2_candidates))
print(
    "S1 entities with candidates:",
    v2_candidates["source1_entity_id"].nunique()
)

# --------------------------------------------
# Calculate candidate recall
# --------------------------------------------

candidate_lookup = defaultdict(set)

for row in v2_candidates.itertuples(index=False):
    candidate_lookup[
        row.source1_entity_id
    ].add(row.matched_entity_id)

total_true = 0
found_true = 0
missed_true = 0
zero_candidate_entities = 0

for sid in val_ids:

    true_ids = gt_lookup.get(sid, set())

    candidates = candidate_lookup.get(
        sid,
        set()
    )

    if len(candidates) == 0:
        zero_candidate_entities += 1

    total_true += len(true_ids)

    found = true_ids & candidates

    found_true += len(found)
    missed_true += len(true_ids - candidates)

print("\n================================")
print("V2 RECALL VERIFICATION")
print("================================")
print("Validation S1 entities:", len(val_ids))
print("Total true matches:", total_true)
print("True matches found:", found_true)
print("True matches missed:", missed_true)

print(
    "Candidate recall:",
    f"{found_true / total_true * 100:.2f}%"
)

print(
    "Zero-candidate S1 entities:",
    zero_candidate_entities
)
```

```python
# ============================================
# RAM-SAFE VALIDATION SETUP
# ============================================

from google.colab import drive
drive.mount("/content/drive")

import pandas as pd
import numpy as np
import re
import unicodedata
from collections import defaultdict, Counter

BASE = "/content/amazon_ml_challenge/student_resource"

TRAIN_S1 = f"{BASE}/dataset/train/train_source1.tsv"
TRAIN_S2 = f"{BASE}/dataset/train/train_source2.tsv"
TRAIN_S3 = f"{BASE}/dataset/train/train_source3.tsv"
GROUND_TRUTH = f"{BASE}/dataset/train/train_ground_truth.tsv"

# --------------------------------------------
# Normalization
# --------------------------------------------

def normalize_advanced(text):
    if pd.isna(text):
        return ""

    text = str(text).lower().strip()
    text = unicodedata.normalize("NFKC", text)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()


def normalize_country(text):
    return normalize_advanced(text)


def get_token_set(text):
    if not text:
        return set()
    return set(str(text).split())


def extract_numbers(text):
    if not text:
        return set()
    return set(re.findall(r"\b\d+\b", str(text)))


# --------------------------------------------
# Load only 10,000 Source-1 rows
# --------------------------------------------

print("Loading 10,000 Source-1 rows...")

s1_10k = pd.read_csv(
    TRAIN_S1,
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ],
    nrows=10000
)

# Exact same entity-level split
s1_train = s1_10k.iloc[:8000].copy()
s1_val = s1_10k.iloc[8000:].copy()

val_ids = set(s1_val["entity_id"])

# --------------------------------------------
# Ground truth
# --------------------------------------------

print("Loading ground truth...")

gt = pd.read_csv(
    GROUND_TRUTH,
    sep="\t",
    usecols=[
        "source1_entity_id",
        "matched_entity_ids"
    ]
)

gt = gt[
    gt["source1_entity_id"].isin(
        set(s1_10k["entity_id"])
    )
].copy()


def parse_match_ids(value):
    if pd.isna(value):
        return set()

    value = str(value).strip()

    if not value:
        return set()

    return {
        x.strip()
        for x in value.split(",")
        if x.strip()
    }


gt_lookup = {
    row.source1_entity_id:
        parse_match_ids(row.matched_entity_ids)
    for row in gt.itertuples(index=False)
}

# --------------------------------------------
# Tiny S1 validation indexes
# --------------------------------------------

s1_val_name_map = defaultdict(list)
s1_val_address_map = defaultdict(list)

for row in s1_val.itertuples(index=False):

    sid = row.entity_id
    country = normalize_country(row.country)

    name = normalize_advanced(row.business_name)
    address = normalize_advanced(row.business_address)

    if name:
        s1_val_name_map[
            (country, name)
        ].append(sid)

    if address:
        s1_val_address_map[
            (country, address)
        ].append(sid)

print("\n================================")
print("RAM-SAFE SETUP COMPLETE")
print("================================")
print("S1 total loaded       :", len(s1_10k))
print("S1 validation         :", len(s1_val))
print("Ground truth rows     :", len(gt))
print("Validation IDs        :", len(val_ids))
print("Name keys             :", len(s1_val_name_map))
print("Address keys          :", len(s1_val_address_map))
```

```python
# ============================================
# STEP 2 — RAM-SAFE TOKEN FREQUENCIES
# Only count tokens that occur in the 2,000
# validation Source-1 records.
# ============================================

print("Building validation token vocabulary...")

val_name_tokens = set()
val_address_tokens = set()

for row in s1_val.itertuples(index=False):

    name = normalize_advanced(row.business_name)
    address = normalize_advanced(row.business_address)

    val_name_tokens.update(get_token_set(name))
    val_address_tokens.update(get_token_set(address))

print("Validation name tokens:", len(val_name_tokens))
print("Validation address tokens:", len(val_address_tokens))

# --------------------------------------------
# Count only those tokens in Source 2 + Source 3
# --------------------------------------------

name_freq_val = Counter()
address_freq_val = Counter()

for path, source_name in [
    (TRAIN_S2, "Source 2"),
    (TRAIN_S3, "Source 3")
]:

    print(f"\nScanning {source_name}...")

    rows = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "business_name",
            "business_address"
        ],
        chunksize=200000
    ):

        for value in chunk["business_name"].dropna():

            tokens = get_token_set(
                normalize_advanced(value)
            )

            for token in tokens:
                if token in val_name_tokens:
                    name_freq_val[token] += 1

        for value in chunk["business_address"].dropna():

            tokens = get_token_set(
                normalize_advanced(value)
            )

            for token in tokens:
                if token in val_address_tokens:
                    address_freq_val[token] += 1

        rows += len(chunk)

        if rows % 1_000_000 == 0:
            print(f"  processed {rows:,}")

print("\n================================")
print("RELEVANT TOKEN FREQUENCIES READY")
print("================================")

print("Name frequencies stored:", len(name_freq_val))
print("Address frequencies stored:", len(address_freq_val))

print("\nMost common relevant address tokens:")

for token, count in address_freq_val.most_common(20):
    print(f"{token:20s} {count:,}")

print("\nMost common relevant name tokens:")

for token, count in name_freq_val.most_common(20):
    print(f"{token:20s} {count:,}")
```

```python
# ============================================
# STEP 3 — RAM-SAFE V2 BLOCKING INDEX
#
# Index ONLY the 2,000 validation S1 entities.
# We do NOT build indexes containing S2/S3 rows.
# ============================================

RARE_LIMIT = 250

# --------------------------------------------
# Tiny inverted indexes:
# token -> validation S1 IDs
# --------------------------------------------

rare_name_to_s1 = defaultdict(set)
rare_address_to_s1 = defaultdict(set)

# Exact indexes already exist:
# s1_val_name_map
# s1_val_address_map

# --------------------------------------------
# Build rare-token indexes from validation S1
# --------------------------------------------

for row in s1_val.itertuples(index=False):

    sid = row.entity_id

    name = normalize_advanced(row.business_name)
    address = normalize_advanced(row.business_address)

    name_tokens = get_token_set(name)
    address_tokens = get_token_set(address)

    # Name token is useful only if its frequency
    # in S2+S3 is <= 250
    for token in name_tokens:

        if name_freq_val.get(token, 10**9) <= RARE_LIMIT:
            rare_name_to_s1[token].add(sid)

    # Same idea for address tokens
    for token in address_tokens:

        if address_freq_val.get(token, 10**9) <= RARE_LIMIT:
            rare_address_to_s1[token].add(sid)


print("================================")
print("RAM-SAFE V2 INDEX READY")
print("================================")

print("Exact name keys:",
      len(s1_val_name_map))

print("Exact address keys:",
      len(s1_val_address_map))

print("Rare name token keys:",
      len(rare_name_to_s1))

print("Rare address token keys:",
      len(rare_address_to_s1))

# Show a few rare tokens
print("\nSample rare address tokens:")

shown = 0

for token in sorted(
    rare_address_to_s1,
    key=lambda x: address_freq_val.get(x, 10**9)
):

    print(
        f"{token:25s} "
        f"freq={address_freq_val.get(token, 0):6d} "
        f"S1={len(rare_address_to_s1[token])}"
    )

    shown += 1

    if shown >= 15:
        break
```

```python
# ============================================
# STEP 3B — COUNTRY-AWARE RARE TOKEN INDEX
# ============================================

rare_name_to_s1 = defaultdict(set)
rare_address_to_s1 = defaultdict(set)

for row in s1_val.itertuples(index=False):

    sid = row.entity_id
    country = normalize_country(row.country)

    name = normalize_advanced(row.business_name)
    address = normalize_advanced(row.business_address)

    # Name tokens
    for token in get_token_set(name):

        if name_freq_val.get(token, 10**9) <= RARE_LIMIT:
            rare_name_to_s1[
                (country, token)
            ].add(sid)

    # Address tokens
    for token in get_token_set(address):

        if address_freq_val.get(token, 10**9) <= RARE_LIMIT:
            rare_address_to_s1[
                (country, token)
            ].add(sid)

print("================================")
print("COUNTRY-AWARE V2 INDEX READY")
print("================================")
print("Rare name keys:", len(rare_name_to_s1))
print("Rare address keys:", len(rare_address_to_s1))
```

```python
# ============================================
# STEP 4 — RAM-SAFE V2 CANDIDATE GENERATION
# ============================================

# One small set per validation S1 entity.
candidate_lookup_v2 = {
    sid: set()
    for sid in val_ids
}

def process_validation_candidates(path, source_name):

    print(f"\nScanning {source_name}...")

    rows = 0
    links_added = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ],
        chunksize=200000
    ):

        for row in chunk.itertuples(index=False):

            eid = row.entity_id
            country = normalize_country(row.country)

            name = normalize_advanced(row.business_name)
            address = normalize_advanced(row.business_address)

            found_s1 = set()

            # --------------------------------
            # BLOCK 1: exact name + country
            # --------------------------------
            if name:
                found_s1.update(
                    s1_val_name_map.get(
                        (country, name),
                        []
                    )
                )

            # --------------------------------
            # BLOCK 2: exact address + country
            # --------------------------------
            if address:
                found_s1.update(
                    s1_val_address_map.get(
                        (country, address),
                        []
                    )
                )

            # --------------------------------
            # BLOCK 3: rare name token + country
            # --------------------------------
            for token in get_token_set(name):

                if name_freq_val.get(
                    token, 10**9
                ) <= RARE_LIMIT:

                    found_s1.update(
                        rare_name_to_s1.get(
                            (country, token),
                            set()
                        )
                    )

            # --------------------------------
            # BLOCK 4: rare address token + country
            # --------------------------------
            for token in get_token_set(address):

                if address_freq_val.get(
                    token, 10**9
                ) <= RARE_LIMIT:

                    found_s1.update(
                        rare_address_to_s1.get(
                            (country, token),
                            set()
                        )
                    )

            # --------------------------------
            # Store only for our 2,000 S1
            # --------------------------------
            for sid in found_s1:

                candidate_lookup_v2[sid].add(eid)
                links_added += 1

            rows += 1

        if rows % 1_000_000 == 0:
            current_links = sum(
                len(x)
                for x in candidate_lookup_v2.values()
            )

            print(
                f"{source_name}: "
                f"{rows:,} rows | "
                f"candidate links stored: {current_links:,}"
            )

    total_links = sum(
        len(x)
        for x in candidate_lookup_v2.values()
    )

    print(
        f"{source_name} COMPLETE | "
        f"rows: {rows:,} | "
        f"candidate links stored: {total_links:,}"
    )


process_validation_candidates(
    TRAIN_S2,
    "Source 2"
)

process_validation_candidates(
    TRAIN_S3,
    "Source 3"
)

# ============================================
# V2 candidate statistics
# ============================================

candidate_counts = np.array([
    len(candidate_lookup_v2[sid])
    for sid in val_ids
])

total_candidate_pairs = int(
    candidate_counts.sum()
)

zero_candidate_entities = int(
    (candidate_counts == 0).sum()
)

print("\n================================")
print("V2 CANDIDATE STATISTICS")
print("================================")

print("Validation S1:", len(val_ids))
print("Candidate pairs:", total_candidate_pairs)
print(
    "S1 with candidates:",
    int((candidate_counts > 0).sum())
)
print(
    "S1 with zero candidates:",
    zero_candidate_entities
)

print(
    "Average candidates/S1:",
    round(candidate_counts.mean(), 2)
)

print(
    "Median candidates/S1:",
    int(np.median(candidate_counts))
)

print(
    "95th percentile:",
    int(np.percentile(candidate_counts, 95))
)

print(
    "99th percentile:",
    int(np.percentile(candidate_counts, 99))
)

print(
    "Maximum candidates:",
    int(candidate_counts.max())
)

# ============================================
# Candidate recall
# ============================================

total_true_matches = 0
found_true_matches = 0

for sid in val_ids:

    true_ids = gt_lookup.get(
        sid,
        set()
    )

    candidates = candidate_lookup_v2.get(
        sid,
        set()
    )

    total_true_matches += len(true_ids)

    found_true_matches += len(
        true_ids & candidates
    )

missed_true_matches = (
    total_true_matches -
    found_true_matches
)

candidate_recall = (
    found_true_matches /
    total_true_matches
    if total_true_matches
    else 0
)

print("\n================================")
print("V2 RECALL")
print("================================")

print(
    "Total true matches:",
    total_true_matches
)

print(
    "True matches found:",
    found_true_matches
)

print(
    "True matches missed:",
    missed_true_matches
)

print(
    "Candidate recall:",
    f"{candidate_recall * 100:.2f}%"
)
```

```python
# ============================================
# STEP 5 — V3 ADDRESS MULTI-TOKEN BLOCK
#
# Add:
#   country
#   + at least 2 shared address tokens
#   + at least 1 shared token with frequency <=2500
#
# We keep the existing V2 candidates and add
# only new candidates from this block.
# ============================================

V3_RARE_LIMIT = 2500

# --------------------------------------------
# Store normalized address token sets for
# validation S1 only
# --------------------------------------------

val_address_tokens = {}
val_country = {}

for row in s1_val.itertuples(index=False):

    sid = row.entity_id

    val_country[sid] = normalize_country(row.country)

    val_address_tokens[sid] = get_token_set(
        normalize_advanced(row.business_address)
    )

# --------------------------------------------
# Build token -> validation S1 IDs
# ONLY for tokens that can participate in V3
# --------------------------------------------

address_token_to_s1_v3 = defaultdict(set)

for sid in val_ids:

    country = val_country[sid]
    tokens = val_address_tokens[sid]

    for token in tokens:

        if address_freq_val.get(
            token, 10**9
        ) <= V3_RARE_LIMIT:

            address_token_to_s1_v3[
                (country, token)
            ].add(sid)

print("================================")
print("V3 ADDRESS INDEX READY")
print("================================")
print(
    "Rare address token keys:",
    len(address_token_to_s1_v3)
)

# --------------------------------------------
# Start from verified V2 candidates
# --------------------------------------------

candidate_lookup_v3 = {
    sid: set(candidates)
    for sid, candidates in candidate_lookup_v2.items()
}

new_links = 0

def process_v3_source(path, source_name):

    global new_links

    print(f"\nScanning {source_name}...")

    rows = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_address",
            "country"
        ],
        chunksize=200000
    ):

        for row in chunk.itertuples(index=False):

            eid = row.entity_id

            country = normalize_country(row.country)

            address = normalize_advanced(
                row.business_address
            )

            source_tokens = get_token_set(address)

            if not source_tokens:
                rows += 1
                continue

            # --------------------------------
            # Get possible S1 entities through
            # rare address tokens
            # --------------------------------

            possible_s1 = set()

            for token in source_tokens:

                if address_freq_val.get(
                    token, 10**9
                ) <= V3_RARE_LIMIT:

                    possible_s1.update(
                        address_token_to_s1_v3.get(
                            (country, token),
                            set()
                        )
                    )

            # --------------------------------
            # Require at least 2 shared
            # address tokens
            # --------------------------------

            for sid in possible_s1:

                s1_tokens = val_address_tokens[sid]

                shared = source_tokens & s1_tokens

                if len(shared) >= 2:

                    # At least one shared token
                    # must satisfy the rarity rule
                    rare_shared = any(
                        address_freq_val.get(
                            token,
                            10**9
                        ) <= V3_RARE_LIMIT
                        for token in shared
                    )

                    if rare_shared:

                        if eid not in candidate_lookup_v3[sid]:

                            candidate_lookup_v3[sid].add(eid)
                            new_links += 1

            rows += 1

        if rows % 1_000_000 == 0:

            print(
                f"{source_name}: "
                f"{rows:,} rows | "
                f"new V3 links: {new_links:,}"
            )

    print(
        f"{source_name} COMPLETE | "
        f"new V3 links: {new_links:,}"
    )


process_v3_source(
    TRAIN_S2,
    "Source 2"
)

process_v3_source(
    TRAIN_S3,
    "Source 3"
)

# --------------------------------------------
# V3 statistics
# --------------------------------------------

v3_counts = np.array([
    len(candidate_lookup_v3[sid])
    for sid in val_ids
])

v3_total_candidates = int(
    v3_counts.sum()
)

v3_zero = int(
    (v3_counts == 0).sum()
)

# --------------------------------------------
# Recall
# --------------------------------------------

v3_total_true = 0
v3_found_true = 0

for sid in val_ids:

    true_ids = gt_lookup.get(
        sid,
        set()
    )

    candidates = candidate_lookup_v3.get(
        sid,
        set()
    )

    v3_total_true += len(true_ids)

    v3_found_true += len(
        true_ids & candidates
    )

v3_missed = (
    v3_total_true -
    v3_found_true
)

v3_recall = (
    v3_found_true /
    v3_total_true
)

print("\n================================")
print("V3 RESULTS")
print("================================")

print("V2 candidate pairs:", 212682)
print("New V3 links:", new_links)
print("V3 candidate pairs:", v3_total_candidates)

print(
    "Average candidates/S1:",
    round(v3_counts.mean(), 2)
)

print(
    "Median candidates/S1:",
    int(np.median(v3_counts))
)

print(
    "P95:",
    int(np.percentile(v3_counts, 95))
)

print(
    "P99:",
    int(np.percentile(v3_counts, 99))
)

print(
    "Maximum:",
    int(v3_counts.max())
)

print(
    "Zero-candidate S1:",
    v3_zero
)

print("\nTrue matches:", v3_total_true)
print("Found:", v3_found_true)
print("Missed:", v3_missed)

print(
    "Candidate recall:",
    f"{v3_recall * 100:.2f}%"
)

print(
    "Recall improvement:",
    f"{(v3_recall - 0.7313) * 100:+.2f} percentage points"
)
```

```python
# ============================================
# STEP 6 — V3 ADDRESS BLOCK, RARITY <= 1000
# ============================================

V3_TEST_LIMIT = 1000

# --------------------------------------------
# Build token -> validation S1 index
# --------------------------------------------

address_token_to_s1_1000 = defaultdict(set)

for sid in val_ids:

    country = val_country[sid]
    tokens = val_address_tokens[sid]

    for token in tokens:

        if address_freq_val.get(
            token, 10**9
        ) <= V3_TEST_LIMIT:

            address_token_to_s1_1000[
                (country, token)
            ].add(sid)

print("Rare address token keys:",
      len(address_token_to_s1_1000))

# --------------------------------------------
# Start from V2
# --------------------------------------------

candidate_lookup_1000 = {
    sid: set(candidates)
    for sid, candidates in candidate_lookup_v2.items()
}

new_links_1000 = 0


def process_v3_1000(path, source_name):

    global new_links_1000

    print(f"\nScanning {source_name}...")

    rows = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_address",
            "country"
        ],
        chunksize=200000
    ):

        for row in chunk.itertuples(index=False):

            eid = row.entity_id

            country = normalize_country(row.country)

            address = normalize_advanced(
                row.business_address
            )

            source_tokens = get_token_set(address)

            if not source_tokens:
                rows += 1
                continue

            possible_s1 = set()

            for token in source_tokens:

                if address_freq_val.get(
                    token, 10**9
                ) <= V3_TEST_LIMIT:

                    possible_s1.update(
                        address_token_to_s1_1000.get(
                            (country, token),
                            set()
                        )
                    )

            for sid in possible_s1:

                shared = (
                    source_tokens &
                    val_address_tokens[sid]
                )

                if len(shared) >= 2:

                    if any(
                        address_freq_val.get(
                            token,
                            10**9
                        ) <= V3_TEST_LIMIT
                        for token in shared
                    ):

                        if eid not in candidate_lookup_1000[sid]:

                            candidate_lookup_1000[sid].add(eid)
                            new_links_1000 += 1

            rows += 1

        if rows % 1_000_000 == 0:
            print(
                f"{source_name}: "
                f"{rows:,} rows | "
                f"new links: {new_links_1000:,}"
            )

    print(
        f"{source_name} COMPLETE | "
        f"new links: {new_links_1000:,}"
    )


process_v3_1000(
    TRAIN_S2,
    "Source 2"
)

process_v3_1000(
    TRAIN_S3,
    "Source 3"
)

# --------------------------------------------
# Statistics
# --------------------------------------------

counts_1000 = np.array([
    len(candidate_lookup_1000[sid])
    for sid in val_ids
])

total_candidates_1000 = int(
    counts_1000.sum()
)

zero_1000 = int(
    (counts_1000 == 0).sum()
)

total_true_1000 = 0
found_true_1000 = 0

for sid in val_ids:

    true_ids = gt_lookup.get(
        sid,
        set()
    )

    candidates = candidate_lookup_1000[sid]

    total_true_1000 += len(true_ids)

    found_true_1000 += len(
        true_ids & candidates
    )

recall_1000 = (
    found_true_1000 /
    total_true_1000
)

print("\n================================")
print("V3 <=1000 RESULTS")
print("================================")

print(
    "Candidate pairs:",
    total_candidates_1000
)

print(
    "New links:",
    new_links_1000
)

print(
    "Average candidates/S1:",
    round(counts_1000.mean(), 2)
)

print(
    "Median:",
    int(np.median(counts_1000))
)

print(
    "P95:",
    int(np.percentile(counts_1000, 95))
)

print(
    "P99:",
    int(np.percentile(counts_1000, 99))
)

print(
    "Maximum:",
    int(counts_1000.max())
)

print(
    "Zero-candidate S1:",
    zero_1000
)

print(
    "True matches found:",
    found_true_1000
)

print(
    "True matches missed:",
    total_true_1000 -
    found_true_1000
)

print(
    "Candidate recall:",
    f"{recall_1000 * 100:.2f}%"
)
```

```python
# ============================================
# STEP 7
# Add:
# country
# + shared address number
# + shared address token
# + number frequency <= 1000
#
# Starting from V3 <=1000.
# ============================================

NUMBER_LIMIT = 1000

# --------------------------------------------
# Step 1: Find address numbers used by
# validation S1
# --------------------------------------------

val_address_numbers = {}

validation_numbers = set()

for sid in val_ids:

    address = normalize_advanced(
        s1_val.loc[
            s1_val["entity_id"] == sid,
            "business_address"
        ].iloc[0]
    )

    nums = extract_numbers(address)

    val_address_numbers[sid] = nums
    validation_numbers.update(nums)

print(
    "Validation address numbers:",
    len(validation_numbers)
)

# --------------------------------------------
# Step 2: Count only relevant numbers
# across Source 2 + Source 3
# --------------------------------------------

number_freq_val = Counter()

for path, source_name in [
    (TRAIN_S2, "Source 2"),
    (TRAIN_S3, "Source 3")
]:

    print(f"\nCounting numbers in {source_name}...")

    rows = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=["business_address"],
        chunksize=200000
    ):

        for value in chunk["business_address"].dropna():

            nums = extract_numbers(
                normalize_advanced(value)
            )

            for num in nums:

                if num in validation_numbers:
                    number_freq_val[num] += 1

        rows += len(chunk)

        if rows % 1_000_000 == 0:
            print(
                f"  processed {rows:,}"
            )

print("\nRelevant number frequencies:",
      len(number_freq_val))

# --------------------------------------------
# Step 3: Build country + number -> S1
# index, only for numbers <= 1000
# --------------------------------------------

number_to_s1 = defaultdict(set)

for sid in val_ids:

    country = val_country[sid]

    for num in val_address_numbers[sid]:

        if number_freq_val.get(
            num,
            10**9
        ) <= NUMBER_LIMIT:

            number_to_s1[
                (country, num)
            ].add(sid)

print(
    "Country-number keys:",
    len(number_to_s1)
)

# --------------------------------------------
# Step 4: Start from V3 <=1000
# --------------------------------------------

candidate_lookup_num = {
    sid: set(candidates)
    for sid, candidates in candidate_lookup_1000.items()
}

new_number_links = 0


def process_number_block(path, source_name):

    global new_number_links

    print(f"\nScanning {source_name}...")

    rows = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_address",
            "country"
        ],
        chunksize=200000
    ):

        for row in chunk.itertuples(index=False):

            eid = row.entity_id

            country = normalize_country(
                row.country
            )

            address = normalize_advanced(
                row.business_address
            )

            source_numbers = extract_numbers(
                address
            )

            if not source_numbers:
                rows += 1
                continue

            source_tokens = get_token_set(
                address
            )

            if not source_tokens:
                rows += 1
                continue

            # Possible S1 entities from shared
            # country + address number
            possible_s1 = set()

            for num in source_numbers:

                if number_freq_val.get(
                    num,
                    10**9
                ) <= NUMBER_LIMIT:

                    possible_s1.update(
                        number_to_s1.get(
                            (country, num),
                            set()
                        )
                    )

            # Require BOTH:
            # 1. shared number
            # 2. shared address token
            for sid in possible_s1:

                shared_numbers = (
                    source_numbers &
                    val_address_numbers[sid]
                )

                if not shared_numbers:
                    continue

                shared_tokens = (
                    source_tokens &
                    val_address_tokens[sid]
                )

                if not shared_tokens:
                    continue

                if eid not in candidate_lookup_num[sid]:

                    candidate_lookup_num[sid].add(eid)
                    new_number_links += 1

            rows += 1

        if rows % 1_000_000 == 0:

            print(
                f"{source_name}: "
                f"{rows:,} rows | "
                f"new links: {new_number_links:,}"
            )

    print(
        f"{source_name} COMPLETE | "
        f"new links: {new_number_links:,}"
    )


process_number_block(
    TRAIN_S2,
    "Source 2"
)

process_number_block(
    TRAIN_S3,
    "Source 3"
)

# --------------------------------------------
# Final statistics
# --------------------------------------------

number_counts = np.array([
    len(candidate_lookup_num[sid])
    for sid in val_ids
])

total_number_candidates = int(
    number_counts.sum()
)

zero_number = int(
    (number_counts == 0).sum()
)

found_number = 0
total_number_true = 0

for sid in val_ids:

    true_ids = gt_lookup.get(
        sid,
        set()
    )

    candidates = candidate_lookup_num[sid]

    total_number_true += len(true_ids)

    found_number += len(
        true_ids & candidates
    )

recall_number = (
    found_number /
    total_number_true
)

print("\n================================")
print("V3 + NUMBER BLOCK RESULTS")
print("================================")

print(
    "Previous V3 <=1000 candidates:",
    531100
)

print(
    "New links:",
    new_number_links
)

print(
    "Final candidate pairs:",
    total_number_candidates
)

print(
    "Average candidates/S1:",
    round(number_counts.mean(), 2)
)

print(
    "Median:",
    int(np.median(number_counts))
)

print(
    "P95:",
    int(np.percentile(number_counts, 95))
)

print(
    "P99:",
    int(np.percentile(number_counts, 99))
)

print(
    "Maximum:",
    int(number_counts.max())
)

print(
    "Zero-candidate S1:",
    zero_number
)

print(
    "True matches found:",
    found_number
)

print(
    "True matches missed:",
    total_number_true -
    found_number
)

print(
    "Candidate recall:",
    f"{recall_number * 100:.2f}%"
)
```

```python
# ============================================
# STEP 8 — TEST PAIR FEATURE ENGINE
# ============================================

from difflib import SequenceMatcher

def similarity(a, b):
    if not a or not b:
        return 0.0
    return SequenceMatcher(None, a, b).ratio()


def pair_features(s1_row, s2_row):

    name1 = normalize_advanced(s1_row["business_name"])
    name2 = normalize_advanced(s2_row["business_name"])

    addr1 = normalize_advanced(s1_row["business_address"])
    addr2 = normalize_advanced(s2_row["business_address"])

    country1 = normalize_country(s1_row["country"])
    country2 = normalize_country(s2_row["country"])

    name_tokens1 = get_token_set(name1)
    name_tokens2 = get_token_set(name2)

    addr_tokens1 = get_token_set(addr1)
    addr_tokens2 = get_token_set(addr2)

    shared_name = name_tokens1 & name_tokens2
    shared_addr = addr_tokens1 & addr_tokens2

    nums1 = extract_numbers(addr1)
    nums2 = extract_numbers(addr2)

    shared_numbers = nums1 & nums2

    name_union = name_tokens1 | name_tokens2
    addr_union = addr_tokens1 | addr_tokens2

    name_jaccard = (
        len(shared_name) / len(name_union)
        if name_union else 0.0
    )

    address_jaccard = (
        len(shared_addr) / len(addr_union)
        if addr_union else 0.0
    )

    return {
        "name_exact": int(
            bool(name1) and name1 == name2
        ),

        "address_exact": int(
            bool(addr1) and addr1 == addr2
        ),

        "country_exact": int(
            country1 == country2
        ),

        "name_similarity": similarity(
            name1, name2
        ),

        "address_similarity": similarity(
            addr1, addr2
        ),

        "name_token_overlap": (
            len(shared_name) /
            max(1, min(
                len(name_tokens1),
                len(name_tokens2)
            ))
        ),

        "address_token_overlap": (
            len(shared_addr) /
            max(1, min(
                len(addr_tokens1),
                len(addr_tokens2)
            ))
        ),

        "name_jaccard": name_jaccard,

        "address_jaccard": address_jaccard,

        "shared_name_token_count": len(
            shared_name
        ),

        "shared_address_token_count": len(
            shared_addr
        ),

        "shared_address_number_count": len(
            shared_numbers
        ),

        "name_length_diff": abs(
            len(name1) - len(name2)
        ),

        "address_length_diff": abs(
            len(addr1) - len(addr2)
        ),

        "name_length_ratio": (
            min(len(name1), len(name2)) /
            max(1, max(len(name1), len(name2)))
        ),

        "address_length_ratio": (
            min(len(addr1), len(addr2)) /
            max(1, max(len(addr1), len(addr2)))
        )
    }


print("Feature engine created successfully.")
```

```python
# ============================================
# STEP 9 — TEST FEATURES ON REAL CANDIDATES
# ============================================

# Take 10 validation S1 entities that have candidates
sample_sids = [
    sid for sid in val_ids
    if len(candidate_lookup_1000.get(sid, set())) > 0
][:10]

# Collect 1 candidate from each
sample_pairs = []

for sid in sample_sids:

    candidate_ids = list(
        candidate_lookup_1000[sid]
    )

    if candidate_ids:
        sample_pairs.append(
            (sid, candidate_ids[0])
        )

print("Sample candidate pairs:", len(sample_pairs))

# --------------------------------------------
# We need the actual Source-2/3 records.
# Collect only these candidate entity IDs.
# --------------------------------------------

needed_eids = {
    eid
    for _, eid in sample_pairs
}

candidate_records = {}

for path, source_name in [
    (TRAIN_S2, "Source 2"),
    (TRAIN_S3, "Source 3")
]:

    print(f"Scanning {source_name} for sample records...")

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ],
        chunksize=200000
    ):

        found = chunk[
            chunk["entity_id"].isin(needed_eids)
        ]

        for row in found.itertuples(index=False):

            candidate_records[
                row.entity_id
            ] = {
                "entity_id": row.entity_id,
                "business_name": row.business_name,
                "business_address": row.business_address,
                "country": row.country
            }

        if len(candidate_records) == len(needed_eids):
            break

    if len(candidate_records) == len(needed_eids):
        break

print(
    "Candidate records found:",
    len(candidate_records),
    "/",
    len(needed_eids)
)

# --------------------------------------------
# Calculate features
# --------------------------------------------

feature_rows = []

for sid, eid in sample_pairs:

    s1_row = s1_10k[
        s1_10k["entity_id"] == sid
    ].iloc[0].to_dict()

    s2_row = candidate_records[eid]

    features = pair_features(
        s1_row,
        s2_row
    )

    features["source1_entity_id"] = sid
    features["matched_entity_id"] = eid

    feature_rows.append(features)

sample_features = pd.DataFrame(
    feature_rows
)

print("\n================================")
print("FEATURE TEST")
print("================================")

print(
    sample_features.to_string(
        index=False
    )
)

print("\nShape:", sample_features.shape)

print(
    "\nMissing values:",
    sample_features.isna().sum().sum()
)
```

```python
# ============================================
# STEP 10 — TRAINING-SIDE V3 <=1000 BLOCKER
# ============================================

# IMPORTANT:
# This uses ONLY the 8,000 training S1 entities.
# The 2,000 validation entities remain untouched.

train_ids = set(
    s1_train["entity_id"]
)

print("Training S1 entities:", len(train_ids))

# --------------------------------------------
# Country / normalized fields for training S1
# --------------------------------------------

train_country = {}
train_address_tokens = {}
train_name_tokens = {}

for row in s1_train.itertuples(index=False):

    sid = row.entity_id

    train_country[sid] = normalize_country(
        row.country
    )

    train_address_tokens[sid] = get_token_set(
        normalize_advanced(
            row.business_address
        )
    )

    train_name_tokens[sid] = get_token_set(
        normalize_advanced(
            row.business_name
        )
    )

# --------------------------------------------
# Exact maps
# --------------------------------------------

train_name_map = defaultdict(list)
train_address_map = defaultdict(list)

for row in s1_train.itertuples(index=False):

    sid = row.entity_id
    country = train_country[sid]

    name = normalize_advanced(
        row.business_name
    )

    address = normalize_advanced(
        row.business_address
    )

    if name:
        train_name_map[
            (country, name)
        ].append(sid)

    if address:
        train_address_map[
            (country, address)
        ].append(sid)

# --------------------------------------------
# Rare address token index
# <=1000
# --------------------------------------------

TRAIN_RARE_LIMIT = 1000

train_address_token_to_s1 = defaultdict(set)

for sid in train_ids:

    country = train_country[sid]

    for token in train_address_tokens[sid]:

        if address_freq_val.get(
            token,
            10**9
        ) <= TRAIN_RARE_LIMIT:

            train_address_token_to_s1[
                (country, token)
            ].add(sid)

print("\n================================")
print("TRAINING BLOCK INDEX READY")
print("================================")

print(
    "Exact name keys:",
    len(train_name_map)
)

print(
    "Exact address keys:",
    len(train_address_map)
)

print(
    "Rare address keys:",
    len(train_address_token_to_s1)
)
```

```python
# ============================================
# STEP 10B — TRAINING-SPECIFIC TOKEN FREQUENCY
# ============================================

print("Building training address-token vocabulary...")

train_address_vocab = set()

for row in s1_train.itertuples(index=False):

    address = normalize_advanced(
        row.business_address
    )

    train_address_vocab.update(
        get_token_set(address)
    )

print(
    "Training address tokens:",
    len(train_address_vocab)
)

# --------------------------------------------
# Count only tokens appearing in TRAIN S1
# across full Source 2 + Source 3
# --------------------------------------------

train_address_freq = Counter()

for path, source_name in [
    (TRAIN_S2, "Source 2"),
    (TRAIN_S3, "Source 3")
]:

    print(f"\nScanning {source_name}...")

    rows = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=["business_address"],
        chunksize=200000
    ):

        for value in chunk["business_address"].dropna():

            tokens = get_token_set(
                normalize_advanced(value)
            )

            for token in tokens:

                if token in train_address_vocab:
                    train_address_freq[token] += 1

        rows += len(chunk)

        if rows % 1_000_000 == 0:
            print(
                f"  processed {rows:,}"
            )

print("\n================================")
print("TRAINING FREQUENCIES READY")
print("================================")

print(
    "Training frequencies stored:",
    len(train_address_freq)
)

print("\nMost common training-relevant address tokens:")

for token, count in train_address_freq.most_common(20):
    print(
        f"{token:20s} {count:,}"
    )
```

```python
# ============================================
# STEP 10C — TRAINING-SPECIFIC NAME FREQUENCY
# ============================================

print("Building training name-token vocabulary...")

train_name_vocab = set()

for row in s1_train.itertuples(index=False):

    name = normalize_advanced(
        row.business_name
    )

    train_name_vocab.update(
        get_token_set(name)
    )

print(
    "Training name tokens:",
    len(train_name_vocab)
)

# --------------------------------------------
# Count only tokens appearing in TRAIN S1
# across full Source 2 + Source 3
# --------------------------------------------

train_name_freq = Counter()

for path, source_name in [
    (TRAIN_S2, "Source 2"),
    (TRAIN_S3, "Source 3")
]:

    print(f"\nScanning {source_name}...")

    rows = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=["business_name"],
        chunksize=200000
    ):

        for value in chunk["business_name"].dropna():

            tokens = get_token_set(
                normalize_advanced(value)
            )

            for token in tokens:

                if token in train_name_vocab:
                    train_name_freq[token] += 1

        rows += len(chunk)

        if rows % 1_000_000 == 0:
            print(
                f"  processed {rows:,}"
            )

print("\n================================")
print("TRAINING NAME FREQUENCIES READY")
print("================================")

print(
    "Training name frequencies stored:",
    len(train_name_freq)
)

print("\nMost common training-relevant name tokens:")

for token, count in train_name_freq.most_common(20):
    print(
        f"{token:20s} {count:,}"
    )
```

```python
# ============================================
# STEP 11 — TRAINING V3 <=1000 INDEX
# ============================================

TRAIN_RARE_NAME_LIMIT = 250
TRAIN_ADDRESS_LIMIT = 1000

# --------------------------------------------
# Exact + rare-name index
# --------------------------------------------

train_rare_name_to_s1 = defaultdict(set)

# Address token index for V3 <=1000
train_address_token_to_s1 = defaultdict(set)

for row in s1_train.itertuples(index=False):

    sid = row.entity_id
    country = train_country[sid]

    # -----------------------------
    # Name
    # -----------------------------

    for token in train_name_tokens[sid]:

        if train_name_freq.get(
            token,
            10**9
        ) <= TRAIN_RARE_NAME_LIMIT:

            train_rare_name_to_s1[
                (country, token)
            ].add(sid)

    # -----------------------------
    # Address
    # -----------------------------

    for token in train_address_tokens[sid]:

        if train_address_freq.get(
            token,
            10**9
        ) <= TRAIN_ADDRESS_LIMIT:

            train_address_token_to_s1[
                (country, token)
            ].add(sid)

print("================================")
print("TRAINING V3 INDEX READY")
print("================================")

print(
    "Exact name keys:",
    len(train_name_map)
)

print(
    "Exact address keys:",
    len(train_address_map)
)

print(
    "Rare name keys:",
    len(train_rare_name_to_s1)
)

print(
    "Address <=1000 keys:",
    len(train_address_token_to_s1)
)
```

```python
# ============================================
# STEP 12 — BUILD CONTROLLED ML TRAINING SET
#
# Training only:
# 8,000 S1 entities
#
# Keep:
#   - ALL candidate positives
#   - controlled number of candidate negatives
#
# This avoids a giant RAM-heavy pair dataset.
# ============================================

import random

random.seed(42)

NEGATIVES_PER_POSITIVE = 8
MIN_NEGATIVES_IF_LOW_POSITIVES = 50000

# --------------------------------------------
# Ground truth for TRAINING S1
# --------------------------------------------

train_gt = {
    sid: gt_lookup.get(sid, set())
    for sid in train_ids
}

# --------------------------------------------
# Store discovered positive pairs
# --------------------------------------------

positive_pairs = []

# We will temporarily keep a controlled
# reservoir of negative pairs.
negative_reservoir = []

# Maximum negatives will be determined
# after we know positive count.
# Start with a generous cap.
MAX_NEGATIVE_RESERVOIR = 200000

rng = random.Random(42)

total_candidate_pairs_seen = 0
positive_pairs_seen = 0

def add_negative(pair):

    global negative_reservoir

    # Reservoir sampling
    n = len(negative_reservoir)

    if n < MAX_NEGATIVE_RESERVOIR:

        negative_reservoir.append(pair)

    else:

        j = rng.randint(
            0,
            total_candidate_pairs_seen - positive_pairs_seen - 1
        )

        if j < MAX_NEGATIVE_RESERVOIR:
            negative_reservoir[j] = pair


def process_training_source(path, source_name):

    global total_candidate_pairs_seen
    global positive_pairs_seen

    print(f"\nScanning {source_name}...")

    rows = 0
    local_candidates = 0
    local_positives = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ],
        chunksize=200000
    ):

        for row in chunk.itertuples(index=False):

            eid = row.entity_id

            country = normalize_country(
                row.country
            )

            name = normalize_advanced(
                row.business_name
            )

            address = normalize_advanced(
                row.business_address
            )

            found_s1 = set()

            # -----------------------------
            # Exact name + country
            # -----------------------------

            if name:

                found_s1.update(
                    train_name_map.get(
                        (country, name),
                        []
                    )
                )

            # -----------------------------
            # Exact address + country
            # -----------------------------

            if address:

                found_s1.update(
                    train_address_map.get(
                        (country, address),
                        []
                    )
                )

            # -----------------------------
            # Rare name <=250
            # -----------------------------

            for token in get_token_set(name):

                if train_name_freq.get(
                    token,
                    10**9
                ) <= TRAIN_RARE_NAME_LIMIT:

                    found_s1.update(
                        train_rare_name_to_s1.get(
                            (country, token),
                            set()
                        )
                    )

            # -----------------------------
            # Address token <=1000
            # -----------------------------

            for token in get_token_set(address):

                if train_address_freq.get(
                    token,
                    10**9
                ) <= TRAIN_ADDRESS_LIMIT:

                    found_s1.update(
                        train_address_token_to_s1.get(
                            (country, token),
                            set()
                        )
                    )

            # -----------------------------
            # Label candidate pairs
            # -----------------------------

            for sid in found_s1:

                pair = (sid, eid)

                total_candidate_pairs_seen += 1
                local_candidates += 1

                true_ids = train_gt.get(
                    sid,
                    set()
                )

                if eid in true_ids:

                    positive_pairs.append(pair)

                    positive_pairs_seen += 1
                    local_positives += 1

                else:

                    add_negative(pair)

            rows += 1

        if rows % 1_000_000 == 0:

            print(
                f"{source_name}: "
                f"{rows:,} rows | "
                f"candidates: {local_candidates:,} | "
                f"positives: {local_positives:,}"
            )

    print(
        f"{source_name} COMPLETE | "
        f"candidates: {local_candidates:,} | "
        f"positives: {local_positives:,}"
    )


process_training_source(
    TRAIN_S2,
    "Source 2"
)

process_training_source(
    TRAIN_S3,
    "Source 3"
)

# --------------------------------------------
# Remove duplicate positives
# --------------------------------------------

positive_pairs = list(
    set(positive_pairs)
)

# --------------------------------------------
# Select negatives
# --------------------------------------------

target_negatives = max(
    len(positive_pairs) * NEGATIVES_PER_POSITIVE,
    MIN_NEGATIVES_IF_LOW_POSITIVES
)

target_negatives = min(
    target_negatives,
    len(negative_reservoir)
)

negative_pairs = negative_reservoir[
    :target_negatives
]

print("\n================================")
print("ML TRAINING CANDIDATES")
print("================================")

print(
    "Total candidate pairs seen:",
    total_candidate_pairs_seen
)

print(
    "Unique positive pairs:",
    len(positive_pairs)
)

print(
    "Negative reservoir:",
    len(negative_reservoir)
)

print(
    "Negatives selected:",
    len(negative_pairs)
)

print(
    "Total ML pairs:",
    len(positive_pairs) +
    len(negative_pairs)
)

print(
    "Positive rate:",
    round(
        len(positive_pairs) /
        max(
            1,
            len(positive_pairs) +
            len(negative_pairs)
        ) * 100,
        2
    ),
    "%"
)
```

```python
# ============================================
# STEP 13 — RETRIEVE ONLY ML PAIR RECORDS
# ============================================

# Combine the pairs
ml_pairs = (
    [(sid, eid, 1) for sid, eid in positive_pairs] +
    [(sid, eid, 0) for sid, eid in negative_pairs]
)

print("ML pairs:", len(ml_pairs))

# --------------------------------------------
# Needed Source-2/3 entity IDs
# --------------------------------------------

needed_eids = {
    eid
    for _, eid, _ in ml_pairs
}

print(
    "Unique Source-2/3 records needed:",
    len(needed_eids)
)

# --------------------------------------------
# Lookup for S1 training records
# --------------------------------------------

s1_train_lookup = (
    s1_train
    .set_index("entity_id")
    .to_dict("index")
)

# --------------------------------------------
# Retrieve only required S2/S3 records
# --------------------------------------------

ml_record_lookup = {}

for path, source_name in [
    (TRAIN_S2, "Source 2"),
    (TRAIN_S3, "Source 3")
]:

    print(f"\nScanning {source_name}...")

    rows = 0
    found_before = len(ml_record_lookup)

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ],
        chunksize=200000
    ):

        found = chunk[
            chunk["entity_id"].isin(
                needed_eids
            )
        ]

        for row in found.itertuples(index=False):

            ml_record_lookup[
                row.entity_id
            ] = {
                "entity_id": row.entity_id,
                "business_name": row.business_name,
                "business_address": row.business_address,
                "country": row.country
            }

        rows += len(chunk)

        if rows % 1_000_000 == 0:
            print(
                f"  processed {rows:,} | "
                f"records found: "
                f"{len(ml_record_lookup):,}"
            )

    print(
        f"{source_name} complete | "
        f"records found so far: "
        f"{len(ml_record_lookup):,}"
    )

print("\n================================")
print("ML RECORD RETRIEVAL")
print("================================")

print(
    "Required records:",
    len(needed_eids)
)

print(
    "Retrieved records:",
    len(ml_record_lookup)
)

print(
    "Missing records:",
    len(needed_eids - set(ml_record_lookup))
)

assert (
    len(needed_eids) ==
    len(ml_record_lookup)
)

print("\nAll ML pair records retrieved successfully.")
```

```python
# ============================================
# STEP 14 — BUILD ML FEATURE MATRIX
# ============================================

print("Building feature matrix...")

feature_rows = []

for i, (sid, eid, label) in enumerate(ml_pairs):

    s1_row = s1_train_lookup[sid]
    s2_row = ml_record_lookup[eid]

    features = pair_features(
        s1_row,
        s2_row
    )

    features["source1_entity_id"] = sid
    features["matched_entity_id"] = eid
    features["label"] = label

    feature_rows.append(features)

    if (i + 1) % 25000 == 0:
        print(
            f"Processed {i + 1:,} / "
            f"{len(ml_pairs):,}"
        )

ml_features = pd.DataFrame(feature_rows)

print("\n================================")
print("ML FEATURE MATRIX")
print("================================")

print(
    "Shape:",
    ml_features.shape
)

print(
    "Positive labels:",
    int(ml_features["label"].sum())
)

print(
    "Negative labels:",
    int(
        (ml_features["label"] == 0).sum()
    )
)

print(
    "Missing feature values:",
    int(
        ml_features.isna().sum().sum()
    )
)

print("\nFeature columns:")

for col in ml_features.columns:
    print(" -", col)
```

```python
# ============================================
# STEP 15 — SAVE + ENTITY-LEVEL ML SPLIT
# ============================================

import os

ML_DIR = "/content/drive/My Drive/Amazon-ML-Challenge-2026/ml"

os.makedirs(ML_DIR, exist_ok=True)

# --------------------------------------------
# Save feature matrix immediately
# --------------------------------------------

feature_path = (
    f"{ML_DIR}/ml_features_v3_1000.pkl"
)

ml_features.to_pickle(feature_path)

print("Feature matrix saved:")
print(feature_path)

# --------------------------------------------
# Entity-level split
# --------------------------------------------

all_training_s1_ids = list(
    s1_train["entity_id"]
)

# Fixed deterministic split
model_train_ids = set(
    all_training_s1_ids[:6000]
)

model_val_ids = set(
    all_training_s1_ids[6000:]
)

print("\n================================")
print("ENTITY-LEVEL SPLIT")
print("================================")

print(
    "Model-train S1:",
    len(model_train_ids)
)

print(
    "Model-validation S1:",
    len(model_val_ids)
)

# --------------------------------------------
# Split feature rows
# --------------------------------------------

train_mask = ml_features[
    "source1_entity_id"
].isin(model_train_ids)

val_mask = ml_features[
    "source1_entity_id"
].isin(model_val_ids)

ml_train = ml_features[
    train_mask
].copy()

ml_val = ml_features[
    val_mask
].copy()

print("\nFeature rows:")

print(
    "Training rows:",
    len(ml_train)
)

print(
    "Validation rows:",
    len(ml_val)
)

print(
    "Training positives:",
    int(ml_train["label"].sum())
)

print(
    "Validation positives:",
    int(ml_val["label"].sum())
)

# --------------------------------------------
# Verify NO entity leakage
# --------------------------------------------

train_entities = set(
    ml_train["source1_entity_id"]
)

val_entities = set(
    ml_val["source1_entity_id"]
)

overlap = train_entities & val_entities

print("\nEntity overlap:", len(overlap))

assert len(overlap) == 0
assert len(model_train_ids) == 6000
assert len(model_val_ids) == 2000

print("\n================================")
print("SPLIT VERIFIED")
print("================================")
print("No Source-1 entity leakage.")
```

```python
# ============================================
# STEP 16 — CATBOOST V1
# ============================================

from catboost import CatBoostClassifier
from sklearn.metrics import (
    precision_score,
    recall_score,
    fbeta_score
)
import os

DROP_COLUMNS = [
    "source1_entity_id",
    "matched_entity_id",
    "label"
]

FEATURE_COLUMNS = [
    col for col in ml_train.columns
    if col not in DROP_COLUMNS
]

X_train = ml_train[FEATURE_COLUMNS]
y_train = ml_train["label"]

X_val = ml_val[FEATURE_COLUMNS]
y_val = ml_val["label"]

print("Features:", len(FEATURE_COLUMNS))
print("Training:", X_train.shape)
print("Validation:", X_val.shape)

# --------------------------------------------
# Train CatBoost
# --------------------------------------------

model_v1 = CatBoostClassifier(
    iterations=500,
    depth=7,
    learning_rate=0.06,
    loss_function="Logloss",
    eval_metric="Logloss",
    random_seed=42,
    verbose=100,
    allow_writing_files=False
)

model_v1.fit(
    X_train,
    y_train,
    eval_set=(X_val, y_val),
    use_best_model=True
)

print("\n================================")
print("MODEL TRAINING COMPLETE")
print("================================")

print(
    "Best iteration:",
    model_v1.get_best_iteration()
)

# --------------------------------------------
# Save model immediately
# --------------------------------------------

MODEL_DIR = (
    "/content/drive/My Drive/"
    "Amazon-ML-Challenge-2026/ml"
)

os.makedirs(MODEL_DIR, exist_ok=True)

MODEL_PATH = (
    f"{MODEL_DIR}/catboost_v1_v3_1000.cbm"
)

model_v1.save_model(MODEL_PATH)

print("Model saved:")
print(MODEL_PATH)

# --------------------------------------------
# Validation probabilities
# --------------------------------------------

val_prob = model_v1.predict_proba(
    X_val
)[:, 1]

print(
    "\nProbability range:",
    float(val_prob.min()),
    "to",
    float(val_prob.max())
)

# --------------------------------------------
# Pair-level threshold sweep
# --------------------------------------------

thresholds = [
    0.30, 0.35, 0.40, 0.45,
    0.50, 0.55, 0.60, 0.65,
    0.70, 0.75, 0.80, 0.85,
    0.90, 0.95
]

pair_results = []

for threshold in thresholds:

    pred = (
        val_prob >= threshold
    ).astype(int)

    precision = precision_score(
        y_val,
        pred,
        zero_division=0
    )

    recall = recall_score(
        y_val,
        pred,
        zero_division=0
    )

    f05 = fbeta_score(
        y_val,
        pred,
        beta=0.5,
        zero_division=0
    )

    pair_results.append({
        "threshold": threshold,
        "precision": precision,
        "recall": recall,
        "f0.5": f05,
        "predicted_pairs": int(pred.sum())
    })

pair_results_df = pd.DataFrame(
    pair_results
)

print("\n================================")
print("PAIR-LEVEL THRESHOLD SWEEP")
print("================================")

print(
    pair_results_df.to_string(index=False)
)

# --------------------------------------------
# Entity-level F0.5
# --------------------------------------------

def entity_f05(
    feature_df,
    probabilities,
    threshold,
    true_lookup
):

    temp = feature_df[
        [
            "source1_entity_id",
            "matched_entity_id"
        ]
    ].copy()

    temp["probability"] = probabilities

    predicted = temp[
        temp["probability"] >= threshold
    ]

    scores = []

    for sid in model_val_ids:

        true_ids = true_lookup.get(
            sid,
            set()
        )

        predicted_ids = set(
            predicted.loc[
                predicted["source1_entity_id"] == sid,
                "matched_entity_id"
            ]
        )

        tp = len(
            true_ids & predicted_ids
        )

        fp = len(
            predicted_ids - true_ids
        )

        fn = len(
            true_ids - predicted_ids
        )

        if tp == 0:

            score = (
                1.0
                if fp == 0 and fn == 0
                else 0.0
            )

        else:

            precision = tp / (tp + fp)
            recall = tp / (tp + fn)

            beta2 = 0.25

            score = (
                (1 + beta2)
                * precision
                * recall
                / (
                    beta2 * precision
                    + recall
                )
            )

        scores.append(score)

    return float(np.mean(scores))


entity_results = []

for threshold in thresholds:

    score = entity_f05(
        ml_val,
        val_prob,
        threshold,
        gt_lookup
    )

    predicted_entities = (
        ml_val.loc[
            val_prob >= threshold,
            "source1_entity_id"
        ].nunique()
    )

    entity_results.append({
        "threshold": threshold,
        "entity_f0.5": score,
        "S1_with_predicted_match":
            predicted_entities
    })

entity_results_df = pd.DataFrame(
    entity_results
)

print("\n================================")
print("ENTITY-LEVEL F0.5")
print("================================")

print(
    entity_results_df.to_string(
        index=False
    )
)

best_row = entity_results_df.loc[
    entity_results_df["entity_f0.5"].idxmax()
]

print("\n================================")
print("BEST LOCAL THRESHOLD")
print("================================")

print(
    "Threshold:",
    float(best_row["threshold"])
)

print(
    "Entity F0.5:",
    float(best_row["entity_f0.5"])
)

print(
    "S1 with predicted match:",
    int(
        best_row[
            "S1_with_predicted_match"
        ]
    )
)
```

```python
# ============================================
# STEP 17 — FINE THRESHOLD SEARCH
# ============================================

fine_thresholds = [
    0.54, 0.56, 0.58,
    0.60, 0.62, 0.64,
    0.66, 0.68, 0.70
]

fine_results = []

for threshold in fine_thresholds:

    score = entity_f05(
        ml_val,
        val_prob,
        threshold,
        gt_lookup
    )

    pred_mask = val_prob >= threshold

    predicted_pairs = int(
        pred_mask.sum()
    )

    predicted_entities = (
        ml_val.loc[
            pred_mask,
            "source1_entity_id"
        ].nunique()
    )

    fine_results.append({
        "threshold": threshold,
        "entity_f0.5": score,
        "predicted_pairs": predicted_pairs,
        "S1_with_match": predicted_entities
    })

fine_results_df = pd.DataFrame(
    fine_results
)

print("\n================================")
print("FINE THRESHOLD SEARCH")
print("================================")

print(
    fine_results_df.to_string(
        index=False
    )
)

best_fine = fine_results_df.loc[
    fine_results_df["entity_f0.5"].idxmax()
]

print("\n================================")
print("FINE BEST THRESHOLD")
print("================================")

print(
    "Threshold:",
    float(best_fine["threshold"])
)

print(
    "Entity F0.5:",
    float(best_fine["entity_f0.5"])
)

print(
    "Predicted pairs:",
    int(best_fine["predicted_pairs"])
)

print(
    "S1 with match:",
    int(best_fine["S1_with_match"])
)
```

```python
# ============================================
# STEP 18 — SCORE V3 <=2500 WITH CATBOOST
# ============================================

# We keep the already-trained model_v1.
# We do NOT retrain anything here.

V3_LIMIT = 2500
SCORE_THRESHOLD = 0.60
BATCH_SIZE = 10000

print("Preparing V3 <=2500 scoring index...")

# --------------------------------------------
# Build country + address-token index
# for <=2500
# --------------------------------------------

address_token_to_s1_2500 = defaultdict(set)

for sid in val_ids:

    country = val_country[sid]

    for token in val_address_tokens[sid]:

        if address_freq_val.get(
            token,
            10**9
        ) <= V3_LIMIT:

            address_token_to_s1_2500[
                (country, token)
            ].add(sid)

print(
    "V3 <=2500 address keys:",
    len(address_token_to_s1_2500)
)

# --------------------------------------------
# Predicted matches accumulated per S1
# --------------------------------------------

predicted_by_s1_2500 = {
    sid: set()
    for sid in val_ids
}

total_candidates_scored = 0
total_positive_predictions = 0

feature_batch = []
pair_batch = []

def score_batch():

    global feature_batch
    global pair_batch
    global total_positive_predictions

    if not feature_batch:
        return

    batch_df = pd.DataFrame(
        feature_batch
    )

    probs = model_v1.predict_proba(
        batch_df[FEATURE_COLUMNS]
    )[:, 1]

    keep = probs >= SCORE_THRESHOLD

    for i in np.where(keep)[0]:

        sid, eid = pair_batch[i]

        predicted_by_s1_2500[
            sid
        ].add(eid)

    total_positive_predictions += int(
        keep.sum()
    )

    feature_batch.clear()
    pair_batch.clear()


def process_v3_2500_source(
    path,
    source_name
):

    global total_candidates_scored

    print(
        f"\nScanning {source_name}..."
    )

    rows = 0
    source_candidates = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ],
        chunksize=200000
    ):

        for row in chunk.itertuples(
            index=False
        ):

            eid = row.entity_id

            country = normalize_country(
                row.country
            )

            name = normalize_advanced(
                row.business_name
            )

            address = normalize_advanced(
                row.business_address
            )

            source_address_tokens = (
                get_token_set(address)
            )

            found_s1 = set()

            # =================================
            # V2: exact name + country
            # =================================

            if name:

                found_s1.update(
                    s1_val_name_map.get(
                        (country, name),
                        []
                    )
                )

            # =================================
            # V2: exact address + country
            # =================================

            if address:

                found_s1.update(
                    s1_val_address_map.get(
                        (country, address),
                        []
                    )
                )

            # =================================
            # V2: rare name <=250
            # =================================

            for token in get_token_set(name):

                if name_freq_val.get(
                    token,
                    10**9
                ) <= 250:

                    found_s1.update(
                        rare_name_to_s1.get(
                            (country, token),
                            set()
                        )
                    )

            # =================================
            # V3: address <=2500
            #
            # Require:
            #   >=2 shared address tokens
            #   AND
            #   >=1 shared token <=2500
            # =================================

            possible_s1 = set()

            for token in source_address_tokens:

                if address_freq_val.get(
                    token,
                    10**9
                ) <= V3_LIMIT:

                    possible_s1.update(
                        address_token_to_s1_2500.get(
                            (country, token),
                            set()
                        )
                    )

            for sid in possible_s1:

                shared_tokens = (
                    source_address_tokens
                    &
                    val_address_tokens[sid]
                )

                if len(shared_tokens) >= 2:

                    rare_shared = any(
                        address_freq_val.get(
                            token,
                            10**9
                        ) <= V3_LIMIT
                        for token in shared_tokens
                    )

                    if rare_shared:

                        found_s1.add(sid)

            # =================================
            # Build features for each candidate
            # =================================

            for sid in found_s1:

                # Retrieve S1 row
                s1_row = s1_val_lookup[sid]

                s2_row = {
                    "entity_id": eid,
                    "business_name":
                        row.business_name,
                    "business_address":
                        row.business_address,
                    "country":
                        row.country
                }

                features = pair_features(
                    s1_row,
                    s2_row
                )

                feature_batch.append(
                    features
                )

                pair_batch.append(
                    (sid, eid)
                )

                total_candidates_scored += 1
                source_candidates += 1

                if len(feature_batch) >= BATCH_SIZE:
                    score_batch()

            rows += 1

        # Finish any remaining batch
        score_batch()

        if rows % 1_000_000 == 0:

            print(
                f"{source_name}: "
                f"{rows:,} rows | "
                f"candidates: "
                f"{source_candidates:,} | "
                f"predicted: "
                f"{total_positive_predictions:,}"
            )

    print(
        f"{source_name} COMPLETE | "
        f"candidates: "
        f"{source_candidates:,} | "
        f"predicted: "
        f"{total_positive_predictions:,}"
    )


# --------------------------------------------
# Need validation S1 lookup
# --------------------------------------------

s1_val_lookup = (
    s1_val
    .set_index("entity_id")
    .to_dict("index")
)

# --------------------------------------------
# Score Source 2 + Source 3
# --------------------------------------------

process_v3_2500_source(
    TRAIN_S2,
    "Source 2"
)

process_v3_2500_source(
    TRAIN_S3,
    "Source 3"
)

# --------------------------------------------
# Calculate entity-level F0.5
# --------------------------------------------

entity_scores_2500 = []

for sid in val_ids:

    true_ids = gt_lookup.get(
        sid,
        set()
    )

    predicted_ids = predicted_by_s1_2500[
        sid
    ]

    tp = len(
        true_ids & predicted_ids
    )

    fp = len(
        predicted_ids - true_ids
    )

    fn = len(
        true_ids - predicted_ids
    )

    if tp == 0:

        score = (
            1.0
            if fp == 0 and fn == 0
            else 0.0
        )

    else:

        precision = tp / (
            tp + fp
        )

        recall = tp / (
            tp + fn
        )

        beta2 = 0.25

        score = (
            1.25
            * precision
            * recall
            / (
                beta2 * precision
                + recall
            )
        )

    entity_scores_2500.append(
        score
    )

final_f05_2500 = float(
    np.mean(entity_scores_2500)
)

entities_with_prediction = sum(
    bool(predicted_by_s1_2500[sid])
    for sid in val_ids
)

print("\n================================")
print("V3 <=2500 + CATBOOST")
print("================================")

print(
    "Candidates scored:",
    total_candidates_scored
)

print(
    "Expected V3 candidates:",
    1316993
)

print(
    "Predicted positive pairs:",
    total_positive_predictions
)

print(
    "S1 with predicted matches:",
    entities_with_prediction
)

print(
    "Entity F0.5:",
    final_f05_2500
)

print("\n================================")
print("COMPARISON")
print("================================")

print(
    "V3 <=1000:",
    "0.889883"
)

print(
    "V3 <=2500:",
    final_f05_2500
)

print(
    "Difference:",
    final_f05_2500 - 0.889883
)
```

```python
import pandas as pd
import re
from collections import defaultdict

TEST_S1 = "/content/amazon_ml_challenge/student_resource/dataset/test/test_source1.tsv"

print("Loading 10,000 test S1 rows...")

sample_s1 = pd.read_csv(
    TEST_S1,
    sep="\t",
    dtype=str,
    keep_default_na=False,
    nrows=10000
)

print("Rows:", len(sample_s1))

# ---------------------------------------------------------
# Build normalized fields ONCE
# ---------------------------------------------------------

print("Normalizing...")

sample_s1["norm_country"] = (
    sample_s1["country"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.strip()
)

sample_s1["norm_name"] = (
    sample_s1["business_name"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.strip()
    .str.replace(r"[^\w\s]", " ", regex=True)
    .str.replace(r"\s+", " ", regex=True)
)

sample_s1["norm_address"] = (
    sample_s1["business_address"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.strip()
    .str.replace(r"[^\w\s]", " ", regex=True)
    .str.replace(r"\s+", " ", regex=True)
)

# ---------------------------------------------------------
# Build indexes
# ---------------------------------------------------------

name_index = defaultdict(list)
address_index = defaultdict(list)
address_token_index = defaultdict(list)

address_freq = defaultdict(int)

for row in sample_s1.itertuples(index=False):

    sid = row.entity_id
    country = row.norm_country
    name = row.norm_name
    address = row.norm_address

    if name:
        name_index[(country, name)].append(sid)

    if address:
        address_index[(country, address)].append(sid)

    tokens = set(address.split())

    for token in tokens:
        if len(token) >= 2:
            address_freq[(country, token)] += 1

for row in sample_s1.itertuples(index=False):

    sid = row.entity_id
    country = row.norm_country
    tokens = set(row.norm_address.split())

    for token in tokens:
        if (
            len(token) >= 2
            and address_freq[(country, token)] <= 1000
        ):
            address_token_index[
                (country, token)
            ].append(sid)

print("Indexes ready.")

# ---------------------------------------------------------
# Estimate candidate volume from 1,000 Source 2 + 1,000 S3
# ---------------------------------------------------------

total_candidates = 0
rows_checked = 0

for path in [
    "/content/amazon_ml_challenge/student_resource/dataset/test/test_source2.tsv",
    "/content/amazon_ml_challenge/student_resource/dataset/test/test_source3.tsv"
]:

    chunk = pd.read_csv(
        path,
        sep="\t",
        dtype=str,
        keep_default_na=False,
        nrows=1000
    )

    for row in chunk.itertuples(index=False):

        country = str(row.country).lower().strip()

        name = (
            str(row.business_name)
            .lower()
            .strip()
        )

        name = re.sub(
            r"[^\w\s]",
            " ",
            name
        )

        name = re.sub(
            r"\s+",
            " ",
            name
        ).strip()

        address = (
            str(row.business_address)
            .lower()
            .strip()
        )

        address = re.sub(
            r"[^\w\s]",
            " ",
            address
        )

        address = re.sub(
            r"\s+",
            " ",
            address
        ).strip()

        candidates = set()

        # Exact name
        candidates.update(
            name_index.get(
                (country, name),
                []
            )
        )

        # Exact address
        candidates.update(
            address_index.get(
                (country, address),
                []
            )
        )

        # Address token block
        counts = defaultdict(int)

        for token in set(address.split()):

            if (
                len(token) >= 2
                and address_freq.get(
                    (country, token),
                    999999
                ) <= 1000
            ):

                for sid in address_token_index.get(
                    (country, token),
                    []
                ):
                    counts[sid] += 1

        for sid, count in counts.items():

            if count >= 2:
                candidates.add(sid)

        total_candidates += len(candidates)
        rows_checked += 1

avg_candidates = (
    total_candidates / rows_checked
)

estimated_total = (
    avg_candidates * 4_887_273
    +
    avg_candidates * 5_082_316
)

print("\n==============================")
print("TEST CANDIDATE ESTIMATE")
print("==============================")
print(
    "Rows sampled:",
    rows_checked
)
print(
    "Candidates:",
    f"{total_candidates:,}"
)
print(
    "Average candidates/source row:",
    round(avg_candidates, 2)
)
print(
    "Rough estimated total:",
    f"{estimated_total:,.0f}"
)
```

```python
# ============================================================
# SUBMISSION 3 — FAST V1.1
# ============================================================

import csv
from pathlib import Path

TEST_S2 = "/content/amazon_ml_challenge/student_resource/dataset/test/test_source2.tsv"
TEST_S3 = "/content/amazon_ml_challenge/student_resource/dataset/test/test_source3.tsv"

OUTPUT_DIR = Path(
    "/content/amazon_ml_challenge/student_resource/output"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

MATCH_FILE = OUTPUT_DIR / "matching_results_submission3.tsv"
CAND_FILE = OUTPUT_DIR / "candidate_pairs_submission3.tsv"

# ------------------------------------------------------------
# Create empty result structures
# ------------------------------------------------------------

predictions = {}

for sid in test_s1["entity_id"]:
    predictions[sid] = []

candidate_count = 0

# ------------------------------------------------------------
# Stream Source 2 + Source 3
# ------------------------------------------------------------

for source_name, path in [
    ("S2", TEST_S2),
    ("S3", TEST_S3)
]:

    print("\nProcessing", source_name)

    processed = 0
    matched_rows = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        dtype=str,
        keep_default_na=False,
        chunksize=100_000
    ):

        for row in chunk.itertuples(index=False):

            country = normalize_country(row.country)
            name = normalize_advanced(row.business_name)
            address = normalize_advanced(row.business_address)

            candidates = set()

            # ------------------------------------------------
            # Exact address + country
            # ------------------------------------------------

            if address:

                candidates.update(
                    address_index.get(
                        (country, address),
                        []
                    )
                )

            # ------------------------------------------------
            # Exact UNIQUE name + country
            # ------------------------------------------------

            if name:

                sid = name_index.get(
                    (country, name)
                )

                if sid:
                    candidates.add(sid)

            # ------------------------------------------------
            # Save candidates + predictions
            # ------------------------------------------------

            if candidates:

                candidate_count += len(candidates)

                matched_rows += 1

                for sid in candidates:

                    predictions[sid].append(
                        row.entity_id
                    )

            processed += 1

        print(
            source_name,
            "| processed:",
            f"{processed:,}",
            "| candidate pairs:",
            f"{candidate_count:,}"
        )

print("\n================================")
print("V1.1 GENERATION COMPLETE")
print("================================")
print(
    "Total candidate pairs:",
    f"{candidate_count:,}"
)
```

```python
# ============================================================
# SUBMISSION 3 — CATBOOST SCORING
# V1.1 candidates + CatBoost threshold 0.60
# ============================================================

from collections import defaultdict

THRESHOLD = 0.60

print("Preparing candidate lookup...")

# Reverse candidate map:
# source2/source3 entity_id -> list of S1 IDs
candidate_to_s1 = defaultdict(list)

for sid, candidate_ids in predictions.items():
    for cid in candidate_ids:
        candidate_to_s1[cid].append(sid)

print(
    "Unique candidate source IDs:",
    f"{len(candidate_to_s1):,}"
)

# S1 lookup
s1_lookup = test_s1.set_index(
    "entity_id"
).to_dict("index")

# Final predictions
final_predictions = defaultdict(list)

scored = 0
positive = 0

# ------------------------------------------------------------
# Stream Source 2 and Source 3
# ------------------------------------------------------------

for source_name, path in [
    ("S2", TEST_S2),
    ("S3", TEST_S3)
]:

    print("\nScoring", source_name)

    processed = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        dtype=str,
        keep_default_na=False,
        chunksize=50_000
    ):

        feature_rows = []
        pair_ids = []

        for row in chunk.itertuples(index=False):

            cid = row.entity_id

            s1_ids = candidate_to_s1.get(cid)

            if not s1_ids:
                processed += 1
                continue

            # Convert namedtuple to dictionary
            source_row = row._asdict()

            for sid in s1_ids:

                s1_row = s1_lookup[sid]

                feature_rows.append(
                    pair_features(
                        s1_row,
                        source_row
                    )
                )

                pair_ids.append(
                    (sid, cid)
                )

            processed += 1

        # ----------------------------------------------------
        # Score this chunk
        # ----------------------------------------------------

        if feature_rows:

            X = pd.DataFrame(
                feature_rows
            )[FEATURE_COLUMNS]

            probs = model_v1.predict_proba(
                X
            )[:, 1]

            for (sid, cid), prob in zip(
                pair_ids,
                probs
            ):

                if prob >= THRESHOLD:

                    final_predictions[sid].append(cid)
                    positive += 1

            scored += len(pair_ids)

        if processed % 500_000 < 50_000:
            print(
                source_name,
                "| processed:",
                f"{processed:,}",
                "| scored:",
                f"{scored:,}",
                "| predicted:",
                f"{positive:,}"
            )

    print(
        source_name,
        "COMPLETE | scored:",
        f"{scored:,}",
        "| predicted:",
        f"{positive:,}"
    )

print("\n========================================")
print("CATBOOST SCORING COMPLETE")
print("========================================")
print("Candidates scored:", f"{scored:,}")
print("Predicted positive pairs:", f"{positive:,}")
print(
    "S1 with predicted matches:",
    sum(
        1 for v in final_predictions.values()
        if v
    )
)
```

```python
# ============================================================
# V2 TEST-SCALE CANDIDATE ESTIMATE
# ============================================================

import pandas as pd
import re
from collections import defaultdict

TEST_S1 = "/content/amazon_ml_challenge/student_resource/dataset/test/test_source1.tsv"
TEST_S2 = "/content/amazon_ml_challenge/student_resource/dataset/test/test_source2.tsv"
TEST_S3 = "/content/amazon_ml_challenge/student_resource/dataset/test/test_source3.tsv"

print("Loading Test S1 sample...")

s1_sample = pd.read_csv(
    TEST_S1,
    sep="\t",
    dtype=str,
    keep_default_na=False,
    nrows=10000
)

# ------------------------------------------------------------
# Normalize
# ------------------------------------------------------------

s1_sample["country_n"] = s1_sample["country"].map(normalize_country)
s1_sample["name_n"] = s1_sample["business_name"].map(normalize_advanced)
s1_sample["address_n"] = s1_sample["business_address"].map(normalize_advanced)

# ------------------------------------------------------------
# Token frequencies within sampled S1
# ------------------------------------------------------------

name_freq = defaultdict(int)
address_freq = defaultdict(int)

for row in s1_sample.itertuples(index=False):

    country = row.country_n

    for token in set(get_token_set(row.name_n)):
        if len(token) >= 2:
            name_freq[(country, token)] += 1

    for token in set(get_token_set(row.address_n)):
        if len(token) >= 2:
            address_freq[(country, token)] += 1

# ------------------------------------------------------------
# Build V2 indexes
# ------------------------------------------------------------

exact_name = defaultdict(list)
exact_address = defaultdict(list)

rare_name = defaultdict(list)
rare_address = defaultdict(list)

for row in s1_sample.itertuples(index=False):

    sid = row.entity_id
    country = row.country_n

    if row.name_n:
        exact_name[
            (country, row.name_n)
        ].append(sid)

    if row.address_n:
        exact_address[
            (country, row.address_n)
        ].append(sid)

    for token in set(get_token_set(row.name_n)):

        if (
            len(token) >= 2
            and name_freq.get(
                (country, token),
                999999
            ) <= 250
        ):
            rare_name[
                (country, token)
            ].append(sid)

    for token in set(get_token_set(row.address_n)):

        if (
            len(token) >= 2
            and address_freq.get(
                (country, token),
                999999
            ) <= 250
        ):
            rare_address[
                (country, token)
            ].append(sid)

print("V2 indexes ready.")

# ------------------------------------------------------------
# Sample Source 2 + Source 3
# ------------------------------------------------------------

TOTAL_CANDIDATES = 0
ROWS = 0

SAMPLES_PER_SOURCE = 5000

for source_name, path in [
    ("S2", TEST_S2),
    ("S3", TEST_S3)
]:

    print("\nSampling", source_name)

    chunk = pd.read_csv(
        path,
        sep="\t",
        dtype=str,
        keep_default_na=False,
        nrows=SAMPLES_PER_SOURCE
    )

    source_candidates = 0

    for row in chunk.itertuples(index=False):

        country = normalize_country(row.country)
        name = normalize_advanced(row.business_name)
        address = normalize_advanced(row.business_address)

        candidates = set()

        # ----------------------------------------------------
        # 1. Exact name + country
        # ----------------------------------------------------

        if name:
            candidates.update(
                exact_name.get(
                    (country, name),
                    []
                )
            )

        # ----------------------------------------------------
        # 2. Exact address + country
        # ----------------------------------------------------

        if address:
            candidates.update(
                exact_address.get(
                    (country, address),
                    []
                )
            )

        # ----------------------------------------------------
        # 3. Rare name token <=250
        # ----------------------------------------------------

        for token in set(
            get_token_set(name)
        ):

            if (
                len(token) >= 2
                and name_freq.get(
                    (country, token),
                    999999
                ) <= 250
            ):

                candidates.update(
                    rare_name.get(
                        (country, token),
                        []
                    )
                )

        # ----------------------------------------------------
        # 4. Rare address token <=250
        # ----------------------------------------------------

        for token in set(
            get_token_set(address)
        ):

            if (
                len(token) >= 2
                and address_freq.get(
                    (country, token),
                    999999
                ) <= 250
            ):

                candidates.update(
                    rare_address.get(
                        (country, token),
                        []
                    )
                )

        source_candidates += len(candidates)

    TOTAL_CANDIDATES += source_candidates
    ROWS += len(chunk)

    print(
        source_name,
        "rows:",
        len(chunk),
        "| candidates:",
        f"{source_candidates:,}",
        "| avg:",
        round(
            source_candidates / len(chunk),
            2
        )
    )

# ------------------------------------------------------------
# Estimate full test size
# ------------------------------------------------------------

AVG_PER_SOURCE_ROW = (
    TOTAL_CANDIDATES / ROWS
)

FULL_TEST_ROWS = (
    4_887_273 + 5_082_316
)

ESTIMATED_TOTAL = (
    AVG_PER_SOURCE_ROW *
    FULL_TEST_ROWS
)

print("\n========================================")
print("V2 TEST-SCALE ESTIMATE")
print("========================================")

print(
    "Average candidates/source row:",
    round(AVG_PER_SOURCE_ROW, 2)
)

print(
    "Estimated total candidate pairs:",
    f"{ESTIMATED_TOTAL:,.0f}"
)

print(
    "Estimated million candidates:",
    round(
        ESTIMATED_TOTAL / 1_000_000,
        2
    )
)
```

```python
import pandas as pd
import numpy as np
import re
import unicodedata
from collections import defaultdict

# ============================================================
# SETTINGS
# ============================================================

S1_SAMPLE_SIZE = 10000
SOURCE_SAMPLE_SIZE = 5000
RARE_LIMIT = 50

# ============================================================
# NORMALIZATION
# ============================================================

def normalize_text(x):
    if pd.isna(x):
        return ""

    x = str(x).lower().strip()
    x = unicodedata.normalize("NFKC", x)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        x = x.replace(old, new)

    x = re.sub(r"[^\w\s]", " ", x)
    x = re.sub(r"\s+", " ", x)

    return x.strip()


def get_tokens(x):
    return set(x.split()) if x else set()


# ============================================================
# LOAD
# ============================================================

ROOT = "/content/amazon_ml_challenge/student_resource/dataset"

S1_PATH = f"{ROOT}/test/test_source1.tsv"
S2_PATH = f"{ROOT}/test/test_source2.tsv"
S3_PATH = f"{ROOT}/test/test_source3.tsv"

print("Loading test data...")

s1 = pd.read_csv(S1_PATH, sep="\t")
s2 = pd.read_csv(S2_PATH, sep="\t")
s3 = pd.read_csv(S3_PATH, sep="\t")

print("S1:", len(s1))
print("S2:", len(s2))
print("S3:", len(s3))


# ============================================================
# SAMPLE S1
# ============================================================

s1_sample = s1.sample(
    n=min(S1_SAMPLE_SIZE, len(s1)),
    random_state=42
).copy()

s1_sample["name_n"] = s1_sample["business_name"].map(normalize_text)
s1_sample["addr_n"] = s1_sample["business_address"].map(normalize_text)
s1_sample["country_n"] = s1_sample["country"].map(normalize_text)

print("\nS1 sample:", len(s1_sample))


# ============================================================
# BUILD EXACT BLOCKS + TOKEN FREQUENCIES
# ============================================================

print("\nBuilding V4 indexes...")

name_country = defaultdict(set)
address_country = defaultdict(set)

name_token_freq = defaultdict(int)
addr_token_freq = defaultdict(int)

for _, r in s1_sample.iterrows():

    sid = r["entity_id"]
    country = r["country_n"]

    name = r["name_n"]
    addr = r["addr_n"]

    # Exact name + country
    if name:
        name_country[(name, country)].add(sid)

    # Exact address + country
    if addr:
        address_country[(addr, country)].add(sid)

    # Name token frequencies
    for t in get_tokens(name):
        name_token_freq[t] += 1

    # Address token frequencies
    for t in get_tokens(addr):
        addr_token_freq[t] += 1


# ============================================================
# RARE TOKENS
# ============================================================

rare_name_tokens = {
    t for t, f in name_token_freq.items()
    if f <= RARE_LIMIT
}

rare_addr_tokens = {
    t for t, f in addr_token_freq.items()
    if f <= RARE_LIMIT
}

print("Rare name tokens:", len(rare_name_tokens))
print("Rare address tokens:", len(rare_addr_tokens))


# ============================================================
# TOKEN -> S1 ENTITY INDEX
# ============================================================

name_token_index = defaultdict(set)
addr_token_index = defaultdict(set)

for _, r in s1_sample.iterrows():

    sid = r["entity_id"]

    for t in get_tokens(r["name_n"]):
        if t in rare_name_tokens:
            name_token_index[t].add(sid)

    for t in get_tokens(r["addr_n"]):
        if t in rare_addr_tokens:
            addr_token_index[t].add(sid)


# ============================================================
# S1 LOOKUP
# ============================================================

s1_lookup = s1_sample.set_index("entity_id").to_dict("index")


# ============================================================
# V4 BLOCKER
# ============================================================

def v4_candidates(row):

    name = normalize_text(row["business_name"])
    addr = normalize_text(row["business_address"])
    country = normalize_text(row["country"])

    name_tokens = get_tokens(name)
    addr_tokens = get_tokens(addr)

    candidates = set()

    # --------------------------------------------------------
    # RULE 1
    # Exact normalized name + country
    # --------------------------------------------------------

    candidates.update(
        name_country.get((name, country), set())
    )

    # --------------------------------------------------------
    # RULE 2
    # Exact normalized address + country
    # --------------------------------------------------------

    candidates.update(
        address_country.get((addr, country), set())
    )

    # --------------------------------------------------------
    # RULE 3
    # At least 2 rare shared name tokens
    # + same country
    # --------------------------------------------------------

    rare_name = name_tokens & rare_name_tokens

    if rare_name:

        token_hits = defaultdict(int)

        for token in rare_name:

            for sid in name_token_index.get(token, []):

                srow = s1_lookup[sid]

                if srow["country_n"] == country:
                    token_hits[sid] += 1

        for sid, count in token_hits.items():

            if count >= 2:
                candidates.add(sid)

    # --------------------------------------------------------
    # RULE 4
    # At least 2 rare shared address tokens
    # + same country
    # --------------------------------------------------------

    rare_addr = addr_tokens & rare_addr_tokens

    if rare_addr:

        token_hits = defaultdict(int)

        for token in rare_addr:

            for sid in addr_token_index.get(token, []):

                srow = s1_lookup[sid]

                if srow["country_n"] == country:
                    token_hits[sid] += 1

        for sid, count in token_hits.items():

            if count >= 2:
                candidates.add(sid)

    return candidates


# ============================================================
# SAMPLE SOURCE ROWS
# ============================================================

s2_sample = s2.sample(
    n=min(SOURCE_SAMPLE_SIZE, len(s2)),
    random_state=42
).copy()

s3_sample = s3.sample(
    n=min(SOURCE_SAMPLE_SIZE, len(s3)),
    random_state=42
).copy()


# ============================================================
# ESTIMATION
# ============================================================

def estimate_candidates(source_df, source_name):

    counts = []

    for i, (_, row) in enumerate(source_df.iterrows()):

        c = v4_candidates(row)
        counts.append(len(c))

        if (i + 1) % 1000 == 0:
            print(
                f"{source_name}: "
                f"{i+1}/{len(source_df)}"
            )

    counts = np.array(counts)

    print("\n" + source_name)
    print("-" * 40)
    print("Sample rows :", len(counts))
    print("Average     :", round(counts.mean(), 2))
    print("Median      :", round(np.median(counts), 2))
    print("P90         :", round(np.percentile(counts, 90), 2))
    print("P95         :", round(np.percentile(counts, 95), 2))
    print("P99         :", round(np.percentile(counts, 99), 2))
    print("Maximum     :", int(counts.max()))

    return counts


print("\n========================================")
print("V4 TEST-SCALE ESTIMATE")
print("========================================")

c2 = estimate_candidates(s2_sample, "S2")
c3 = estimate_candidates(s3_sample, "S3")


# ============================================================
# FULL TEST ESTIMATE
# ============================================================

avg_candidates = (
    c2.mean() + c3.mean()
) / 2

estimated_total = avg_candidates * (
    len(s2) + len(s3)
)

print("\n========================================")
print("FINAL V4 ESTIMATE")
print("========================================")
print(
    "Average candidates/source row:",
    round(avg_candidates, 2)
)
print(
    "Estimated total candidates:",
    f"{estimated_total:,.0f}"
)
print(
    "Estimated million candidates:",
    round(estimated_total / 1e6, 2)
)
print("========================================")
```

```python
# ============================================================
# V4 VALIDATION RECALL
# ============================================================

print("Testing V4 blocker on validation...")

# IMPORTANT:
# This assumes these already exist from our earlier validation:
#   s1_val_lookup
#   gt_lookup
#
# and that s1_val_lookup contains the validation S1 rows.

# ------------------------------------------------------------
# Build V4 indexes from VALIDATION S1 only
# ------------------------------------------------------------

from collections import defaultdict
import numpy as np

VAL_RARE_LIMIT = 50

val_name_country = defaultdict(set)
val_address_country = defaultdict(set)

val_name_freq = defaultdict(int)
val_addr_freq = defaultdict(int)

for sid, r in s1_val_lookup.items():

    name = normalize_text(r["business_name"])
    addr = normalize_text(r["business_address"])
    country = normalize_text(r["country"])

    if name:
        val_name_country[(name, country)].add(sid)

    if addr:
        val_address_country[(addr, country)].add(sid)

    for t in get_tokens(name):
        val_name_freq[t] += 1

    for t in get_tokens(addr):
        val_addr_freq[t] += 1


val_rare_name = {
    t for t, f in val_name_freq.items()
    if f <= VAL_RARE_LIMIT
}

val_rare_addr = {
    t for t, f in val_addr_freq.items()
    if f <= VAL_RARE_LIMIT
}

val_name_index = defaultdict(set)
val_addr_index = defaultdict(set)

for sid, r in s1_val_lookup.items():

    name = normalize_text(r["business_name"])
    addr = normalize_text(r["business_address"])

    for t in get_tokens(name):
        if t in val_rare_name:
            val_name_index[t].add(sid)

    for t in get_tokens(addr):
        if t in val_rare_addr:
            val_addr_index[t].add(sid)


# ------------------------------------------------------------
# V4 candidates for a validation source row
# ------------------------------------------------------------

def v4_val_candidates(row):

    name = normalize_text(row["business_name"])
    addr = normalize_text(row["business_address"])
    country = normalize_text(row["country"])

    name_tokens = get_tokens(name)
    addr_tokens = get_tokens(addr)

    candidates = set()

    # Rule 1: exact name + country
    candidates.update(
        val_name_country.get((name, country), set())
    )

    # Rule 2: exact address + country
    candidates.update(
        val_address_country.get((addr, country), set())
    )

    # Rule 3: >=2 rare shared name tokens
    rare_name = name_tokens & val_rare_name

    token_hits = defaultdict(int)

    for t in rare_name:
        for sid in val_name_index.get(t, []):
            srow = s1_val_lookup[sid]

            if normalize_text(srow["country"]) == country:
                token_hits[sid] += 1

    for sid, count in token_hits.items():
        if count >= 2:
            candidates.add(sid)

    # Rule 4: >=2 rare shared address tokens
    rare_addr = addr_tokens & val_rare_addr

    token_hits = defaultdict(int)

    for t in rare_addr:
        for sid in val_addr_index.get(t, []):
            srow = s1_val_lookup[sid]

            if normalize_text(srow["country"]) == country:
                token_hits[sid] += 1

    for sid, count in token_hits.items():
        if count >= 2:
            candidates.add(sid)

    return candidates


# ------------------------------------------------------------
# Evaluate recall
# ------------------------------------------------------------

total_true = 0
retrieved_true = 0
zero_candidate = 0
candidate_count = []

rows_with_gt = 0

for sid, true_matches in gt_lookup.items():

    # Only validation S1 entities
    if sid not in s1_val_lookup:
        continue

    rows_with_gt += 1

    candidates = v4_val_candidates(
        s1_val_lookup[sid]
    )

    candidate_count.append(len(candidates))

    true_set = set(true_matches)

    total_true += len(true_set)
    retrieved_true += len(true_set & candidates)

    if len(candidates) == 0:
        zero_candidate += 1


recall = (
    retrieved_true / total_true
    if total_true else 0
)

print("\n========================================")
print("V4 VALIDATION RESULT")
print("========================================")

print("Validation S1 entities:", rows_with_gt)
print("True matches:", total_true)
print("Retrieved true matches:", retrieved_true)
print("Missed true matches:", total_true - retrieved_true)

print("Candidate recall:", round(recall * 100, 2), "%")

print("Zero-candidate S1:", zero_candidate)

if candidate_count:
    print(
        "Avg candidates/S1:",
        round(np.mean(candidate_count), 2)
    )

print("========================================")
```

```python
# ============================================================
# FAST V4 RECALL TEST — NO GLOBAL TOKEN SCAN
# ============================================================

print("Running fast true-pair recall test...")

total_true = 0

exact_name_hits = 0
exact_addr_hits = 0

name_2token_hits = 0
addr_2token_hits = 0

any_rule_hits = 0

s2_true = 0
s3_true = 0

s2_hits = 0
s3_hits = 0


for sid, matches in gt_lookup.items():

    s1row = s1_lookup[sid]

    s1_name = normalize_text(
        s1row["business_name"]
    )

    s1_addr = normalize_text(
        s1row["business_address"]
    )

    s1_country = normalize_text(
        s1row["country"]
    )

    s1_name_tokens = get_tokens(s1_name)
    s1_addr_tokens = get_tokens(s1_addr)

    for target_id in matches:

        if target_id not in true_source_rows:
            continue

        target = true_source_rows[target_id]

        target_name = target["name"]
        target_addr = target["addr"]
        target_country = target["country"]

        total_true += 1

        if target_id.startswith("S2-"):
            s2_true += 1
        else:
            s3_true += 1

        hit = False

        # --------------------------------------------
        # Rule 1: exact name + country
        # --------------------------------------------

        if (
            s1_country == target_country
            and s1_name
            and target_name
            and s1_name == target_name
        ):
            exact_name_hits += 1
            hit = True

        # --------------------------------------------
        # Rule 2: exact address + country
        # --------------------------------------------

        if (
            s1_country == target_country
            and s1_addr
            and target_addr
            and s1_addr == target_addr
        ):
            exact_addr_hits += 1
            hit = True

        # --------------------------------------------
        # Rule 3: >=2 shared name tokens + country
        # --------------------------------------------

        if s1_country == target_country:

            shared_name = (
                s1_name_tokens
                & get_tokens(target_name)
            )

            if len(shared_name) >= 2:
                name_2token_hits += 1
                hit = True

        # --------------------------------------------
        # Rule 4: >=2 shared address tokens + country
        # --------------------------------------------

        if s1_country == target_country:

            shared_addr = (
                s1_addr_tokens
                & get_tokens(target_addr)
            )

            if len(shared_addr) >= 2:
                addr_2token_hits += 1
                hit = True

        if hit:
            any_rule_hits += 1

            if target_id.startswith("S2-"):
                s2_hits += 1
            else:
                s3_hits += 1


print("\n========================================")
print("FAST V4 RECALL ANALYSIS")
print("========================================")

print("Total true pairs:", total_true)

print()
print(
    "Exact name + country:",
    exact_name_hits,
    f"({exact_name_hits / total_true * 100:.2f}%)"
)

print(
    "Exact address + country:",
    exact_addr_hits,
    f"({exact_addr_hits / total_true * 100:.2f}%)"
)

print(
    ">=2 shared name tokens:",
    name_2token_hits,
    f"({name_2token_hits / total_true * 100:.2f}%)"
)

print(
    ">=2 shared address tokens:",
    addr_2token_hits,
    f"({addr_2token_hits / total_true * 100:.2f}%)"
)

print()
print(
    "ANY V4 rule:",
    any_rule_hits,
    f"({any_rule_hits / total_true * 100:.2f}%)"
)

print()
print(
    "S2 recall:",
    s2_hits / s2_true * 100
)

print(
    "S3 recall:",
    s3_hits / s3_true * 100
)

print("========================================")
```

```python
# ============================================================
# V4 TEST-SCALE CANDIDATE VOLUME ESTIMATE
# MEMORY-SAFE SAMPLE VERSION
# ============================================================

import pandas as pd
import numpy as np
import re
import unicodedata
from collections import Counter, defaultdict
from itertools import combinations

ROOT = "/content/amazon_ml_challenge/student_resource/dataset"

S1_PATH = f"{ROOT}/test/test_source1.tsv"
S2_PATH = f"{ROOT}/test/test_source2.tsv"
S3_PATH = f"{ROOT}/test/test_source3.tsv"

S1_SAMPLE = 10000
SOURCE_SAMPLE = 100000

print("Loading S1 sample...")

s1_test = pd.read_csv(S1_PATH, sep="\t")

s1_sample = s1_test.sample(
    n=S1_SAMPLE,
    random_state=42
).copy()

print("S1 sample:", len(s1_sample))


# ============================================================
# NORMALIZATION
# ============================================================

def normalize_text(x):
    if pd.isna(x):
        return ""

    x = str(x).lower().strip()
    x = unicodedata.normalize("NFKC", x)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        x = x.replace(old, new)

    x = re.sub(r"[^\w\s]", " ", x)
    x = re.sub(r"\s+", " ", x)

    return x.strip()


def get_tokens(x):
    return set(x.split()) if x else set()


# ============================================================
# NORMALIZE S1
# ============================================================

print("Normalizing S1...")

s1_sample["name_n"] = s1_sample["business_name"].map(
    normalize_text
)

s1_sample["addr_n"] = s1_sample["business_address"].map(
    normalize_text
)

s1_sample["country_n"] = s1_sample["country"].map(
    normalize_text
)


# ============================================================
# LOAD SOURCE SAMPLES
# ============================================================

print("\nLoading S2 sample...")

s2 = pd.read_csv(
    S2_PATH,
    sep="\t",
    skiprows=lambda x: x > 0 and np.random.random() > (
        SOURCE_SAMPLE / 4887273
    )
)

# Keep exactly at most SOURCE_SAMPLE
if len(s2) > SOURCE_SAMPLE:
    s2 = s2.sample(
        SOURCE_SAMPLE,
        random_state=42
    )

print("S2 sample:", len(s2))


print("\nLoading S3 sample...")

s3 = pd.read_csv(
    S3_PATH,
    sep="\t",
    skiprows=lambda x: x > 0 and np.random.random() > (
        SOURCE_SAMPLE / 5082316
    )
)

if len(s3) > SOURCE_SAMPLE:
    s3 = s3.sample(
        SOURCE_SAMPLE,
        random_state=42
    )

print("S3 sample:", len(s3))


# ============================================================
# NORMALIZE SOURCE SAMPLES
# ============================================================

for df in [s2, s3]:

    df["name_n"] = df["business_name"].map(
        normalize_text
    )

    df["addr_n"] = df["business_address"].map(
        normalize_text
    )

    df["country_n"] = df["country"].map(
        normalize_text
    )


# ============================================================
# BUILD TOKEN-PAIR FREQUENCY TABLE
#
# For every source row:
#
# country + tokenA + tokenB
#
# tells us how many source rows contain BOTH tokens.
# ============================================================

def build_pair_counter(df, column):

    counter = Counter()

    for value, country in zip(
        df[column],
        df["country_n"]
    ):

        ts = sorted(get_tokens(value))

        if len(ts) < 2:
            continue

        for a, b in combinations(ts, 2):

            counter[
                (country, a, b)
            ] += 1

    return counter


print("\nBuilding S2 name pair index...")

s2_name_pairs = build_pair_counter(
    s2,
    "name_n"
)

print(
    "S2 name pairs:",
    len(s2_name_pairs)
)


print("\nBuilding S2 address pair index...")

s2_addr_pairs = build_pair_counter(
    s2,
    "addr_n"
)

print(
    "S2 address pairs:",
    len(s2_addr_pairs)
)


print("\nBuilding S3 name pair index...")

s3_name_pairs = build_pair_counter(
    s3,
    "name_n"
)

print(
    "S3 name pairs:",
    len(s3_name_pairs)
)


print("\nBuilding S3 address pair index...")

s3_addr_pairs = build_pair_counter(
    s3,
    "addr_n"
)

print(
    "S3 address pairs:",
    len(s3_addr_pairs)
)


# ============================================================
# ESTIMATE CANDIDATES
#
# This estimates the number of candidates produced by
# >=2 shared tokens.
#
# Exact name/address candidates are added separately.
# ============================================================

def estimate_token_candidates(
    s1_df,
    pair_counter,
    column
):

    estimates = []

    for value, country in zip(
        s1_df[column],
        s1_df["country_n"]
    ):

        ts = sorted(get_tokens(value))

        if len(ts) < 2:
            estimates.append(0)
            continue

        total = 0

        for a, b in combinations(ts, 2):

            total += pair_counter.get(
                (country, a, b),
                0
            )

        estimates.append(total)

    return np.array(estimates)


print("\nEstimating S2 name candidates...")

s2_name_est = estimate_token_candidates(
    s1_sample,
    s2_name_pairs,
    "name_n"
)


print("Estimating S2 address candidates...")

s2_addr_est = estimate_token_candidates(
    s1_sample,
    s2_addr_pairs,
    "addr_n"
)


print("Estimating S3 name candidates...")

s3_name_est = estimate_token_candidates(
    s1_sample,
    s3_name_pairs,
    "name_n"
)


print("Estimating S3 address candidates...")

s3_addr_est = estimate_token_candidates(
    s1_sample,
    s3_addr_pairs,
    "addr_n"
)


# ============================================================
# PRINT ESTIMATES
# ============================================================

print("\n")
print("=" * 60)
print("V4 TOKEN-BLOCK TEST-SCALE ESTIMATE")
print("=" * 60)

print("\nS2")
print("-" * 40)
print(
    "Name avg:",
    round(s2_name_est.mean(), 2)
)
print(
    "Address avg:",
    round(s2_addr_est.mean(), 2)
)

print("\nS3")
print("-" * 40)
print(
    "Name avg:",
    round(s3_name_est.mean(), 2)
)
print(
    "Address avg:",
    round(s3_addr_est.mean(), 2)
)

avg_token_candidates = (
    s2_name_est.mean()
    + s2_addr_est.mean()
    + s3_name_est.mean()
    + s3_addr_est.mean()
) / 2

estimated_total = avg_token_candidates * (
    4887273 + 5082316
)

print("\n")
print("Approx candidate pairs:", f"{estimated_total:,.0f}")
print(
    "Approx million:",
    round(estimated_total / 1e6, 2)
)

print("=" * 60)
```

```python
# ============================================================
# FAST TOKEN FREQUENCY SCAN
# Only count tokens that occur in the TRUE validation pairs
# ============================================================

from collections import Counter
import pandas as pd
import numpy as np

print("Starting optimized frequency scan...")

needed_name = set(needed_name_tokens)
needed_addr = set(needed_addr_tokens)

name_freq_fast = Counter()
addr_freq_fast = Counter()

CHUNK = 250_000

def count_tokens_fast(path, source_name):

    rows = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        chunksize=CHUNK,
        usecols=[
            "business_name",
            "business_address"
        ]
    ):

        # -----------------------------
        # Names
        # -----------------------------
        names = (
            chunk["business_name"]
            .fillna("")
            .astype(str)
            .str.lower()
            .str.replace(r"[^\w\s]", " ", regex=True)
            .str.replace(r"\s+", " ", regex=True)
            .str.strip()
        )

        name_tokens = (
            names.str.split()
            .explode()
        )

        name_tokens = name_tokens[
            name_tokens.isin(needed_name)
        ]

        if len(name_tokens):
            counts = name_tokens.value_counts()

            for token, count in counts.items():
                name_freq_fast[token] += int(count)

        # -----------------------------
        # Addresses
        # -----------------------------
        addrs = (
            chunk["business_address"]
            .fillna("")
            .astype(str)
            .str.lower()
            .str.replace(r"[^\w\s]", " ", regex=True)
            .str.replace(r"\s+", " ", regex=True)
            .str.strip()
        )

        addr_tokens = (
            addrs.str.split()
            .explode()
        )

        addr_tokens = addr_tokens[
            addr_tokens.isin(needed_addr)
        ]

        if len(addr_tokens):
            counts = addr_tokens.value_counts()

            for token, count in counts.items():
                addr_freq_fast[token] += int(count)

        rows += len(chunk)

        if rows % 1_000_000 == 0:
            print(source_name, rows)

    print(source_name, "finished:", rows)


count_tokens_fast(
    S2_PATH,
    "S2"
)

count_tokens_fast(
    S3_PATH,
    "S3"
)

print("\nFrequency scan complete.")
print("Name frequencies:", len(name_freq_fast))
print("Address frequencies:", len(addr_freq_fast))
```

```python
# ============================================================
# TEST RARE <=50 RECALL ON TRUE PAIRS
# ============================================================

RARE_LIMIT = 50

rare_name_fast = {
    t for t in needed_name_tokens
    if name_freq_fast[t] <= RARE_LIMIT
}

rare_addr_fast = {
    t for t in needed_addr_tokens
    if addr_freq_fast[t] <= RARE_LIMIT
}

total = 0

name_rare_hits = 0
addr_rare_hits = 0
any_rare_hits = 0

s2_total = 0
s2_hits = 0

s3_total = 0
s3_hits = 0

for sid, matches in gt_lookup.items():

    s1row = s1_lookup[sid]

    s1_country = normalize_text(
        s1row["country"]
    )

    s1_name_tokens = get_tokens(
        normalize_text(s1row["business_name"])
    )

    s1_addr_tokens = get_tokens(
        normalize_text(s1row["business_address"])
    )

    for target_id in matches:

        target = true_source_rows[target_id]

        if s1_country != target["country"]:
            continue

        total += 1

        if target_id.startswith("S2-"):
            s2_total += 1
        else:
            s3_total += 1

        target_name_tokens = get_tokens(
            target["name"]
        )

        target_addr_tokens = get_tokens(
            target["addr"]
        )

        shared_name_rare = (
            s1_name_tokens
            & target_name_tokens
            & rare_name_fast
        )

        shared_addr_rare = (
            s1_addr_tokens
            & target_addr_tokens
            & rare_addr_fast
        )

        hit = (
            len(shared_name_rare) >= 2
            or
            len(shared_addr_rare) >= 2
        )

        # Exact rules remain part of V4
        exact_name = (
            normalize_text(s1row["business_name"])
            == target["name"]
            and s1_country == target["country"]
        )

        exact_addr = (
            normalize_text(s1row["business_address"])
            == target["addr"]
            and s1_country == target["country"]
        )

        hit = hit or exact_name or exact_addr

        if hit:
            any_rare_hits += 1

            if target_id.startswith("S2-"):
                s2_hits += 1
            else:
                s3_hits += 1

print("\n========================================")
print("V4 RARE-TOKEN RECALL")
print("========================================")

print(
    "Total true pairs:",
    total
)

print(
    "Retrieved:",
    any_rare_hits
)

print(
    "Missed:",
    total - any_rare_hits
)

print(
    "Overall recall:",
    round(any_rare_hits / total * 100, 2),
    "%"
)

print(
    "S2 recall:",
    round(s2_hits / s2_total * 100, 2),
    "%"
)

print(
    "S3 recall:",
    round(s3_hits / s3_total * 100, 2),
    "%"
)

print(
    "Rare name tokens:",
    len(rare_name_fast)
)

print(
    "Rare address tokens:",
    len(rare_addr_fast)
)

print("========================================")
```

```python
# ============================================================
# TEST DIFFERENT RARITY LIMITS
# NO NEW DATA SCAN
# ============================================================

limits = [50, 100, 250, 500, 1000, 1500, 2000, 2500, 5000]

# Count denominators once
total_s2 = sum(
    1
    for matches in gt_lookup.values()
    for x in matches
    if x.startswith("S2-")
)

total_s3 = sum(
    1
    for matches in gt_lookup.values()
    for x in matches
    if x.startswith("S3-")
)

total_all = total_s2 + total_s3

print("S2 true pairs:", total_s2)
print("S3 true pairs:", total_s3)
print("Total true pairs:", total_all)

print("\n" + "=" * 75)
print(
    f"{'LIMIT':<10}"
    f"{'OVERALL':<15}"
    f"{'S2':<15}"
    f"{'S3':<15}"
)
print("=" * 75)


for LIMIT in limits:

    rare_name_test = {
        t for t, c in name_freq_fast.items()
        if c <= LIMIT
    }

    rare_addr_test = {
        t for t, c in addr_freq_fast.items()
        if c <= LIMIT
    }

    hit_all = 0
    hit_s2 = 0
    hit_s3 = 0

    for sid, matches in gt_lookup.items():

        s1row = s1_lookup[sid]

        s1_name = normalize_text(
            s1row["business_name"]
        )

        s1_addr = normalize_text(
            s1row["business_address"]
        )

        s1_country = normalize_text(
            s1row["country"]
        )

        s1_name_tokens = get_tokens(s1_name)
        s1_addr_tokens = get_tokens(s1_addr)

        for target_id in matches:

            target = true_source_rows[target_id]

            if s1_country != target["country"]:
                continue

            target_name = target["name"]
            target_addr = target["addr"]

            target_name_tokens = get_tokens(
                target_name
            )

            target_addr_tokens = get_tokens(
                target_addr
            )

            # Exact rules
            exact_name = (
                s1_name
                and target_name
                and s1_name == target_name
            )

            exact_addr = (
                s1_addr
                and target_addr
                and s1_addr == target_addr
            )

            # Rare token rules
            name_hit = (
                len(
                    s1_name_tokens
                    & target_name_tokens
                    & rare_name_test
                ) >= 2
            )

            addr_hit = (
                len(
                    s1_addr_tokens
                    & target_addr_tokens
                    & rare_addr_test
                ) >= 2
            )

            hit = (
                exact_name
                or exact_addr
                or name_hit
                or addr_hit
            )

            if hit:

                hit_all += 1

                if target_id.startswith("S2-"):
                    hit_s2 += 1
                else:
                    hit_s3 += 1

    overall = hit_all / total_all * 100
    s2_rec = hit_s2 / total_s2 * 100
    s3_rec = hit_s3 / total_s3 * 100

    print(
        f"{LIMIT:<10}"
        f"{overall:<15.2f}"
        f"{s2_rec:<15.2f}"
        f"{s3_rec:<15.2f}"
    )

print("=" * 75)
```

```python
from difflib import SequenceMatcher

def similarity(a, b):
    if not a or not b:
        return 0.0
    return SequenceMatcher(None, a, b).ratio()
```

```python
# ============================================================
# TEST SELECTIVE BLOCKING RULES ON TRUE PAIRS
# ============================================================

import re

def extract_numbers(text):
    return set(re.findall(r"\d+", text))


rules = {
    "address_2tokens": 0,
    "address_2tokens_number": 0,
    "name_2tokens": 0,
    "name_2tokens_strong_similarity": 0,
    "address_number_plus_token": 0,
    "combined_selective": 0
}

s2_hits = {k: 0 for k in rules}
s3_hits = {k: 0 for k in rules}

total_s2 = 0
total_s3 = 0


for sid, matches in gt_lookup.items():

    s1row = s1_lookup[sid]

    s1_name = normalize_text(
        s1row["business_name"]
    )

    s1_addr = normalize_text(
        s1row["business_address"]
    )

    s1_country = normalize_text(
        s1row["country"]
    )

    s1_name_tokens = get_tokens(s1_name)
    s1_addr_tokens = get_tokens(s1_addr)
    s1_numbers = extract_numbers(s1_addr)

    for target_id in matches:

        target = true_source_rows[target_id]

        if s1_country != target["country"]:
            continue

        if target_id.startswith("S2-"):
            total_s2 += 1
        else:
            total_s3 += 1

        target_name = target["name"]
        target_addr = target["addr"]

        target_name_tokens = get_tokens(target_name)
        target_addr_tokens = get_tokens(target_addr)
        target_numbers = extract_numbers(target_addr)

        shared_name = (
            s1_name_tokens & target_name_tokens
        )

        shared_addr = (
            s1_addr_tokens & target_addr_tokens
        )

        shared_numbers = (
            s1_numbers & target_numbers
        )

        # --------------------------------------------
        # Rule A
        # >=2 shared address tokens
        # --------------------------------------------

        rule_a = (
            len(shared_addr) >= 2
        )

        # --------------------------------------------
        # Rule B
        # >=2 shared address tokens
        # + shared address number
        # --------------------------------------------

        rule_b = (
            len(shared_addr) >= 2
            and len(shared_numbers) >= 1
        )

        # --------------------------------------------
        # Rule C
        # >=2 shared name tokens
        # --------------------------------------------

        rule_c = (
            len(shared_name) >= 2
        )

        # --------------------------------------------
        # Rule D
        # >=2 shared name tokens
        # + high sequence similarity
        # --------------------------------------------

        name_sim = similarity(
            s1_name,
            target_name
        )

        rule_d = (
            len(shared_name) >= 2
            and name_sim >= 0.70
        )

        # --------------------------------------------
        # Rule E
        # shared address number + >=1 address token
        # --------------------------------------------

        rule_e = (
            len(shared_numbers) >= 1
            and len(shared_addr) >= 1
        )

        # --------------------------------------------
        # Rule F
        # Selective combined rule
        # --------------------------------------------

        rule_f = (
            rule_b
            or
            rule_d
        )

        rule_values = [
            rule_a,
            rule_b,
            rule_c,
            rule_d,
            rule_e,
            rule_f
        ]

        for key, hit in zip(
            rules.keys(),
            rule_values
        ):

            if hit:

                rules[key] += 1

                if target_id.startswith("S2-"):
                    s2_hits[key] += 1
                else:
                    s3_hits[key] += 1


print("\n========================================")
print("SELECTIVE BLOCKING TRUE-PAIR RECALL")
print("========================================")

for key in rules:

    overall = rules[key] / (
        total_s2 + total_s3
    ) * 100

    s2_rec = (
        s2_hits[key] / total_s2 * 100
    )

    s3_rec = (
        s3_hits[key] / total_s3 * 100
    )

    print(
        f"{key:35s}"
        f" Overall={overall:6.2f}%"
        f" | S2={s2_rec:6.2f}%"
        f" | S3={s3_rec:6.2f}%"
    )

print("========================================")
```

```python
# ============================================================
# V4 UNIQUE-CANDIDATE VOLUME ESTIMATE
# ============================================================

import pandas as pd
import numpy as np
import re
import unicodedata
from collections import defaultdict

ROOT = "/content/amazon_ml_challenge/student_resource/dataset"

S1_PATH = f"{ROOT}/test/test_source1.tsv"
S2_PATH = f"{ROOT}/test/test_source2.tsv"
S3_PATH = f"{ROOT}/test/test_source3.tsv"

S1_SAMPLE_SIZE = 10_000
SOURCE_SAMPLE_SIZE = 100_000

print("Loading S1...")

s1 = pd.read_csv(S1_PATH, sep="\t")

s1_sample = s1.sample(
    n=S1_SAMPLE_SIZE,
    random_state=42
).copy()

print("S1 sample:", len(s1_sample))


# ============================================================
# NORMALIZATION
# ============================================================

def normalize_text(x):
    if pd.isna(x):
        return ""

    x = str(x).lower().strip()
    x = unicodedata.normalize("NFKC", x)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        x = x.replace(old, new)

    x = re.sub(r"[^\w\s]", " ", x)
    x = re.sub(r"\s+", " ", x)

    return x.strip()


def get_tokens(x):
    return set(x.split()) if x else set()


# ============================================================
# NORMALIZE S1
# ============================================================

print("Normalizing S1...")

s1_sample["name_n"] = s1_sample["business_name"].map(
    normalize_text
)

s1_sample["addr_n"] = s1_sample["business_address"].map(
    normalize_text
)

s1_sample["country_n"] = s1_sample["country"].map(
    normalize_text
)


# ============================================================
# LOAD SOURCE SAMPLES
# ============================================================

print("\nLoading S2...")

s2 = pd.read_csv(
    S2_PATH,
    sep="\t",
    nrows=SOURCE_SAMPLE_SIZE
)

print("S2 sample:", len(s2))

print("\nLoading S3...")

s3 = pd.read_csv(
    S3_PATH,
    sep="\t",
    nrows=SOURCE_SAMPLE_SIZE
)

print("S3 sample:", len(s3))


# ============================================================
# NORMALIZE SOURCE DATA
# ============================================================

for df in [s2, s3]:

    df["entity_id"] = df["entity_id"].astype(str)

    df["name_n"] = df["business_name"].map(
        normalize_text
    )

    df["addr_n"] = df["business_address"].map(
        normalize_text
    )

    df["country_n"] = df["country"].map(
        normalize_text
    )


# ============================================================
# BUILD SOURCE INDEXES
#
# We store ONLY IDs in the indexes.
# ============================================================

def build_indexes(df):

    exact_name = defaultdict(set)
    exact_addr = defaultdict(set)

    name_tokens = defaultdict(set)
    addr_tokens = defaultdict(set)

    source_lookup = {}

    for row in df.itertuples(index=False):

        eid = row.entity_id

        name = row.name_n
        addr = row.addr_n
        country = row.country_n

        source_lookup[eid] = (
            name,
            addr,
            country
        )

        # Exact name + country
        if name:
            exact_name[
                (name, country)
            ].add(eid)

        # Exact address + country
        if addr:
            exact_addr[
                (addr, country)
            ].add(eid)

        # Name tokens
        for token in get_tokens(name):
            name_tokens[
                (country, token)
            ].add(eid)

        # Address tokens
        for token in get_tokens(addr):
            addr_tokens[
                (country, token)
            ].add(eid)

    return (
        exact_name,
        exact_addr,
        name_tokens,
        addr_tokens,
        source_lookup
    )


print("\nBuilding S2 indexes...")

s2_exact_name, s2_exact_addr, s2_name_tokens, s2_addr_tokens, s2_lookup = \
    build_indexes(s2)


print("Building S3 indexes...")

s3_exact_name, s3_exact_addr, s3_name_tokens, s3_addr_tokens, s3_lookup = \
    build_indexes(s3)


# ============================================================
# CANDIDATES FOR ONE S1 ROW
# ============================================================

def get_candidates(
    row,
    exact_name,
    exact_addr,
    name_index,
    addr_index
):

    name = row["name_n"]
    addr = row["addr_n"]
    country = row["country_n"]

    name_ts = get_tokens(name)
    addr_ts = get_tokens(addr)

    candidates = set()

    # --------------------------------------------------------
    # Rule 1: exact name + country
    # --------------------------------------------------------

    if name:
        candidates.update(
            exact_name.get(
                (name, country),
                set()
            )
        )

    # --------------------------------------------------------
    # Rule 2: exact address + country
    # --------------------------------------------------------

    if addr:
        candidates.update(
            exact_addr.get(
                (addr, country),
                set()
            )
        )

    # --------------------------------------------------------
    # Rule 3:
    # >=2 shared NAME tokens
    # --------------------------------------------------------

    name_token_hits = defaultdict(int)

    for token in name_ts:

        for eid in name_index.get(
            (country, token),
            set()
        ):

            name_token_hits[eid] += 1

    for eid, count in name_token_hits.items():

        if count >= 2:
            candidates.add(eid)

    # --------------------------------------------------------
    # Rule 4:
    # >=2 shared ADDRESS tokens
    # --------------------------------------------------------

    addr_token_hits = defaultdict(int)

    for token in addr_ts:

        for eid in addr_index.get(
            (country, token),
            set()
        ):

            addr_token_hits[eid] += 1

    for eid, count in addr_token_hits.items():

        if count >= 2:
            candidates.add(eid)

    return candidates


# ============================================================
# EVALUATE EACH SOURCE SEPARATELY
# ============================================================

def evaluate_source(
    source_name,
    exact_name,
    exact_addr,
    name_index,
    addr_index
):

    counts = []

    print(
        f"\nEvaluating {source_name}..."
    )

    for i, (_, row) in enumerate(
        s1_sample.iterrows()
    ):

        candidates = get_candidates(
            row,
            exact_name,
            exact_addr,
            name_index,
            addr_index
        )

        counts.append(len(candidates))

        if (i + 1) % 1000 == 0:
            print(
                f"{source_name}:",
                i + 1,
                "/",
                len(s1_sample)
            )

    return np.array(counts)


s2_counts = evaluate_source(
    "S2",
    s2_exact_name,
    s2_exact_addr,
    s2_name_tokens,
    s2_addr_tokens
)

s3_counts = evaluate_source(
    "S3",
    s3_exact_name,
    s3_exact_addr,
    s3_name_tokens,
    s3_addr_tokens
)


# ============================================================
# RESULTS
# ============================================================

print("\n")
print("=" * 65)
print("V4 UNIQUE-CANDIDATE VOLUME")
print("=" * 65)

print("\nS2")
print("-" * 45)
print("Average :", round(s2_counts.mean(), 2))
print("Median  :", round(np.median(s2_counts), 2))
print("P90     :", round(np.percentile(s2_counts, 90), 2))
print("P95     :", round(np.percentile(s2_counts, 95), 2))
print("P99     :", round(np.percentile(s2_counts, 99), 2))
print("Maximum :", int(s2_counts.max()))

print("\nS3")
print("-" * 45)
print("Average :", round(s3_counts.mean(), 2))
print("Median  :", round(np.median(s3_counts), 2))
print("P90     :", round(np.percentile(s3_counts, 90), 2))
print("P95     :", round(np.percentile(s3_counts, 95), 2))
print("P99     :", round(np.percentile(s3_counts, 99), 2))
print("Maximum :", int(s3_counts.max()))


# Weighted average
avg_candidates = (
    s2_counts.mean() * len(s2)
    +
    s3_counts.mean() * len(s3)
) / (
    len(s2) + len(s3)
)

estimated_total = (
    avg_candidates *
    len(s1)
)

print("\n")
print("=" * 65)
print("TEST-SCALE ESTIMATE")
print("=" * 65)

print(
    "Weighted average candidates/S1:",
    round(avg_candidates, 2)
)

print(
    "Estimated total candidate pairs:",
    f"{estimated_total:,.0f}"
)

print(
    "Estimated million candidates:",
    round(estimated_total / 1e6, 2)
)

print("=" * 65)
```

```python
import pandas as pd
import numpy as np
import re
import unicodedata
from collections import defaultdict

ROOT = "/content/amazon_ml_challenge/student_resource"

S1_PATH = f"{ROOT}/dataset/test/test_source1.tsv"
S2_PATH = f"{ROOT}/dataset/test/test_source2.tsv"
S3_PATH = f"{ROOT}/dataset/test/test_source3.tsv"

print("Loading S1...")
test_s1 = pd.read_csv(S1_PATH, sep="\t")
print("S1:", len(test_s1))

print("Loading S2 sample...")
test_s2 = pd.read_csv(S2_PATH, sep="\t", nrows=100000)
print("S2:", len(test_s2))

print("Loading S3 sample...")
test_s3 = pd.read_csv(S3_PATH, sep="\t", nrows=100000)
print("S3:", len(test_s3))

# 10k S1 sample
s1 = test_s1.sample(n=10000, random_state=42).copy()

print("\nS1 sample:", len(s1))
print("Ready.")
```

```python
# ============================================================
# RULE-BY-RULE CANDIDATE VOLUME
# ============================================================

import re
import unicodedata
import numpy as np
from collections import defaultdict

def norm(text):
    if pd.isna(text):
        return ""
    text = str(text).lower().strip()
    text = unicodedata.normalize("NFKC", text)
    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()

def toks(text):
    return set(text.split()) if text else set()


# ------------------------------------------------------------
# Normalize
# ------------------------------------------------------------

print("Normalizing...")

for df in [s1, test_s2, test_s3]:
    df["name_n"] = df["business_name"].map(norm)
    df["addr_n"] = df["business_address"].map(norm)
    df["country_n"] = df["country"].map(norm)

print("Normalization complete.")


# ------------------------------------------------------------
# Build indexes
# ------------------------------------------------------------

def build_indexes(df):

    name_country = defaultdict(set)
    addr_country = defaultdict(set)
    name_token = defaultdict(set)
    addr_token = defaultdict(set)

    for idx, row in df.iterrows():

        name = row["name_n"]
        addr = row["addr_n"]
        country = row["country_n"]

        if name:
            name_country[(name, country)].add(idx)

        if addr:
            addr_country[(addr, country)].add(idx)

        for token in toks(name):
            name_token[token].add(idx)

        for token in toks(addr):
            addr_token[token].add(idx)

    return (
        name_country,
        addr_country,
        name_token,
        addr_token
    )


print("\nBuilding S2 indexes...")
s2_indexes = build_indexes(test_s2)

print("Building S3 indexes...")
s3_indexes = build_indexes(test_s3)

print("Indexes ready.")


# ------------------------------------------------------------
# Candidate count for ONE rule
# ------------------------------------------------------------

def count_rule(s1_df, indexes, rule):

    name_country, addr_country, name_token, addr_token = indexes

    counts = []

    for counter, (_, row) in enumerate(s1_df.iterrows()):

        if counter % 1000 == 0:
            print(f"{rule}: {counter}/{len(s1_df)}")

        name = row["name_n"]
        addr = row["addr_n"]
        country = row["country_n"]

        candidates = set()

        # Exact name + country
        if rule == "exact_name":
            candidates.update(
                name_country.get((name, country), set())
            )

        # Exact address + country
        elif rule == "exact_address":
            candidates.update(
                addr_country.get((addr, country), set())
            )

        # >= 2 shared name tokens
        elif rule == "name_2tokens":

            overlap = defaultdict(int)

            for token in toks(name):
                for sid in name_token.get(token, []):
                    overlap[sid] += 1

            for sid, count in overlap.items():
                if count >= 2:
                    candidates.add(sid)

        # >= 2 shared address tokens
        elif rule == "address_2tokens":

            overlap = defaultdict(int)

            for token in toks(addr):
                for sid in addr_token.get(token, []):
                    overlap[sid] += 1

            for sid, count in overlap.items():
                if count >= 2:
                    candidates.add(sid)

        counts.append(len(candidates))

    return np.array(counts)


# ------------------------------------------------------------
# Run one rule at a time
# ------------------------------------------------------------

rules = [
    "exact_name",
    "exact_address",
    "name_2tokens",
    "address_2tokens"
]

results = {}

for rule in rules:

    print("\n" + "=" * 65)
    print("RULE:", rule)
    print("=" * 65)

    s2_counts = count_rule(
        s1,
        s2_indexes,
        rule
    )

    s3_counts = count_rule(
        s1,
        s3_indexes,
        rule
    )

    results[rule] = {
        "s2": s2_counts,
        "s3": s3_counts
    }

    print("\nS2")
    print("Average :", round(s2_counts.mean(), 2))
    print("Median  :", round(np.median(s2_counts), 2))
    print("P90     :", round(np.percentile(s2_counts, 90), 2))
    print("P95     :", round(np.percentile(s2_counts, 95), 2))
    print("P99     :", round(np.percentile(s2_counts, 99), 2))
    print("Maximum :", int(s2_counts.max()))

    print("\nS3")
    print("Average :", round(s3_counts.mean(), 2))
    print("Median  :", round(np.median(s3_counts), 2))
    print("P90     :", round(np.percentile(s3_counts, 90), 2))
    print("P95     :", round(np.percentile(s3_counts, 95), 2))
    print("P99     :", round(np.percentile(s3_counts, 99), 2))
    print("Maximum :", int(s3_counts.max()))
```

```python
# ============================================================
# SELECTIVE RARE-TOKEN CANDIDATE VOLUME
# ============================================================

from collections import defaultdict
import numpy as np

# ------------------------------------------------------------
# Build token frequencies in S2/S3
# ------------------------------------------------------------

def token_frequency(df, column):

    freq = defaultdict(int)

    for text in df[column]:
        for t in toks(text):
            freq[t] += 1

    return freq


print("Calculating S2 address token frequencies...")
s2_addr_freq = token_frequency(test_s2, "addr_n")

print("Calculating S3 address token frequencies...")
s3_addr_freq = token_frequency(test_s3, "addr_n")

print("Done.")

# ------------------------------------------------------------
# Candidate counter
# ------------------------------------------------------------

def selective_address_counts(
    s1_df,
    indexes,
    addr_freq,
    rarity_limit
):

    addr_token = indexes[3]

    counts = []

    for counter, (_, row) in enumerate(s1_df.iterrows()):

        if counter % 1000 == 0:
            print(
                f"limit={rarity_limit}: "
                f"{counter}/{len(s1_df)}"
            )

        address_tokens = toks(row["addr_n"])

        overlap = defaultdict(int)
        rare_shared = defaultdict(bool)

        for token in address_tokens:

            # Only use tokens that are not too common
            if addr_freq.get(token, 0) <= rarity_limit:

                for sid in addr_token.get(token, []):
                    overlap[sid] += 1
                    rare_shared[sid] = True

        count = 0

        for sid, shared_count in overlap.items():

            if shared_count >= 2 and rare_shared[sid]:
                count += 1

        counts.append(count)

    return np.array(counts)


limits = [
    500,
    1000,
    2000,
    5000,
    10000
]

volume_results = {}

for limit in limits:

    print("\n" + "=" * 70)
    print("RARITY LIMIT:", limit)
    print("=" * 70)

    s2_counts = selective_address_counts(
        s1,
        s2_indexes,
        s2_addr_freq,
        limit
    )

    s3_counts = selective_address_counts(
        s1,
        s3_indexes,
        s3_addr_freq,
        limit
    )

    volume_results[limit] = {
        "s2": s2_counts,
        "s3": s3_counts
    }

    print("\nS2")
    print("Average :", round(s2_counts.mean(), 2))
    print("Median  :", round(np.median(s2_counts), 2))
    print("P90     :", round(np.percentile(s2_counts, 90), 2))
    print("P95     :", round(np.percentile(s2_counts, 95), 2))
    print("P99     :", round(np.percentile(s2_counts, 99), 2))
    print("Maximum :", int(s2_counts.max()))

    print("\nS3")
    print("Average :", round(s3_counts.mean(), 2))
    print("Median  :", round(np.median(s3_counts), 2))
    print("P90     :", round(np.percentile(s3_counts, 90), 2))
    print("P95     :", round(np.percentile(s3_counts, 95), 2))
    print("P99     :", round(np.percentile(s3_counts, 99), 2))
    print("Maximum :", int(s3_counts.max()))

    weighted_avg = (
        s2_counts.mean() * 0.5 +
        s3_counts.mean() * 0.5
    )

    estimated_total = weighted_avg * 1_732_544

    print(
        "\nEstimated total full-test candidates:",
        f"{estimated_total:,.0f}"
    )
```

```python
# ============================================================
# SELECTIVE BLOCKER VOLUME TEST
# ============================================================

from collections import defaultdict
import numpy as np

# ------------------------------------------------------------
# Build token indexes
# ------------------------------------------------------------

def build_token_index(df, column):

    index = defaultdict(set)

    for idx, text in enumerate(df[column]):

        for token in toks(text):
            index[token].add(idx)

    return index


print("Building indexes...")

s2_name_index = build_token_index(test_s2, "name_n")
s2_addr_index = build_token_index(test_s2, "addr_n")

s3_name_index = build_token_index(test_s3, "name_n")
s3_addr_index = build_token_index(test_s3, "addr_n")

print("Indexes ready.")


# ------------------------------------------------------------
# Token frequencies
# ------------------------------------------------------------

def token_freq(df, column):

    freq = defaultdict(int)

    for text in df[column]:

        for token in toks(text):
            freq[token] += 1

    return freq


s2_name_freq = token_freq(test_s2, "name_n")
s2_addr_freq = token_freq(test_s2, "addr_n")

s3_name_freq = token_freq(test_s3, "name_n")
s3_addr_freq = token_freq(test_s3, "addr_n")


# ------------------------------------------------------------
# Candidate count
# ------------------------------------------------------------

def selective_candidates(
    row,
    name_index,
    addr_index,
    name_freq,
    addr_freq,
    name_limit,
    addr_limit,
    name_sim_threshold,
    addr_sim_threshold
):

    name1 = row["name_n"]
    addr1 = row["addr_n"]

    name_tokens = toks(name1)
    addr_tokens = toks(addr1)

    candidates = set()

    # --------------------------------------------------------
    # Name block
    # --------------------------------------------------------

    name_hits = defaultdict(int)

    for token in name_tokens:

        if name_freq.get(token, 999999999) <= name_limit:

            for eid in name_index.get(token, []):
                name_hits[eid] += 1

    for eid, count in name_hits.items():

        if count >= 2:

            # We only count here.
            # Similarity will be applied later during
            # actual feature generation.
            candidates.add(eid)

    # --------------------------------------------------------
    # Address block
    # --------------------------------------------------------

    addr_hits = defaultdict(int)

    for token in addr_tokens:

        if addr_freq.get(token, 999999999) <= addr_limit:

            for eid in addr_index.get(token, []):
                addr_hits[eid] += 1

    for eid, count in addr_hits.items():

        if count >= 2:
            candidates.add(eid)

    return candidates


# ------------------------------------------------------------
# Test configurations
# ------------------------------------------------------------

configs = [
    (500, 500),
    (500, 1000),
    (1000, 500),
    (1000, 1000),
]

for name_limit, addr_limit in configs:

    print("\n" + "=" * 70)
    print(
        f"NAME LIMIT={name_limit} | "
        f"ADDRESS LIMIT={addr_limit}"
    )
    print("=" * 70)

    s2_counts = []
    s3_counts = []

    for counter, (_, row) in enumerate(s1.iterrows()):

        if counter % 1000 == 0:
            print(f"{counter}/10000")

        s2_counts.append(
            len(
                selective_candidates(
                    row,
                    s2_name_index,
                    s2_addr_index,
                    s2_name_freq,
                    s2_addr_freq,
                    name_limit,
                    addr_limit,
                    0.70,
                    0.70
                )
            )
        )

        s3_counts.append(
            len(
                selective_candidates(
                    row,
                    s3_name_index,
                    s3_addr_index,
                    s3_name_freq,
                    s3_addr_freq,
                    name_limit,
                    addr_limit,
                    0.70,
                    0.70
                )
            )
        )

    s2_counts = np.array(s2_counts)
    s3_counts = np.array(s3_counts)

    avg = (s2_counts.mean() + s3_counts.mean()) / 2

    print("\nS2 average:", round(s2_counts.mean(), 2))
    print("S3 average:", round(s3_counts.mean(), 2))
    print("Combined average:", round(avg, 2))

    print(
        "Estimated full-test candidates:",
        f"{avg * 1_732_544:,.0f}"
    )
```

```python
# ============================================================
# SELECTIVE BLOCKER WITH SIMILARITY
# ============================================================

def selective_candidates_v2(
    row,
    name_index,
    addr_index,
    name_freq,
    addr_freq,
    name_limit=500,
    addr_limit=500,
    name_sim=0.70,
    addr_sim=0.70
):

    name1 = row["name_n"]
    addr1 = row["addr_n"]

    name_tokens = toks(name1)
    addr_tokens = toks(addr1)

    candidates = set()

    # --------------------------------------------------------
    # Exact name / exact address
    # --------------------------------------------------------

    # These require country in the original indexes
    # and are handled separately below if available.

    # --------------------------------------------------------
    # Name: >=2 shared tokens + rare token
    # --------------------------------------------------------

    name_hits = defaultdict(int)

    for token in name_tokens:

        if name_freq.get(token, 999999999) <= name_limit:

            for eid in name_index.get(token, []):
                name_hits[eid] += 1

    for eid, count in name_hits.items():

        if count >= 2:
            candidates.add(eid)

    # --------------------------------------------------------
    # Address: >=2 shared tokens + rare token
    # --------------------------------------------------------

    addr_hits = defaultdict(int)

    for token in addr_tokens:

        if addr_freq.get(token, 999999999) <= addr_limit:

            for eid in addr_index.get(token, []):
                addr_hits[eid] += 1

    for eid, count in addr_hits.items():

        if count >= 2:
            candidates.add(eid)

    return candidates


# ------------------------------------------------------------
# First just calculate volume
# ------------------------------------------------------------

print("Testing 500/500 selective blocker...")

s2_counts = []
s3_counts = []

for counter, (_, row) in enumerate(s1.iterrows()):

    if counter % 1000 == 0:
        print(f"{counter}/10000")

    s2_counts.append(
        len(
            selective_candidates_v2(
                row,
                s2_name_index,
                s2_addr_index,
                s2_name_freq,
                s2_addr_freq,
                500,
                500
            )
        )
    )

    s3_counts.append(
        len(
            selective_candidates_v2(
                row,
                s3_name_index,
                s3_addr_index,
                s3_name_freq,
                s3_addr_freq,
                500,
                500
            )
        )
    )

s2_counts = np.array(s2_counts)
s3_counts = np.array(s3_counts)

print("\n" + "=" * 70)
print("500/500 SELECTIVE BLOCKER")
print("=" * 70)

print("\nS2")
print("Average:", s2_counts.mean())
print("Median :", np.median(s2_counts))
print("P90    :", np.percentile(s2_counts, 90))
print("P95    :", np.percentile(s2_counts, 95))
print("P99    :", np.percentile(s2_counts, 99))
print("Max    :", s2_counts.max())

print("\nS3")
print("Average:", s3_counts.mean())
print("Median :", np.median(s3_counts))
print("P90    :", np.percentile(s3_counts, 90))
print("P95    :", np.percentile(s3_counts, 95))
print("P99    :", np.percentile(s3_counts, 99))
print("Max    :", s3_counts.max())
```

```python
# ============================================================
# 500/500 + SIMILARITY FILTER
# ============================================================

from difflib import SequenceMatcher
import numpy as np

def sim(a, b):
    if not a or not b:
        return 0.0
    return SequenceMatcher(None, a, b).ratio()


def selective_candidates_v3(
    row,
    source_df,
    name_index,
    addr_index,
    name_freq,
    addr_freq,
    name_limit=500,
    addr_limit=500,
    name_sim_threshold=0.70,
    addr_sim_threshold=0.70
):

    name1 = row["name_n"]
    addr1 = row["addr_n"]
    country1 = row["country_n"]

    name_tokens = toks(name1)
    addr_tokens = toks(addr1)

    candidates = set()

    # --------------------------------------------------------
    # NAME BLOCK
    # >=2 shared tokens
    # + at least one rare token
    # + name similarity >= 0.70
    # + same country
    # --------------------------------------------------------

    name_hits = defaultdict(int)

    for token in name_tokens:

        if name_freq.get(token, 999999999) <= name_limit:

            for eid in name_index.get(token, []):
                name_hits[eid] += 1

    for eid, count in name_hits.items():

        if count < 2:
            continue

        target = source_df.iloc[eid]

        if target["country_n"] != country1:
            continue

        if sim(name1, target["name_n"]) >= name_sim_threshold:
            candidates.add(eid)

    # --------------------------------------------------------
    # ADDRESS BLOCK
    # >=2 shared tokens
    # + at least one rare token
    # + address similarity >= 0.70
    # + same country
    # --------------------------------------------------------

    addr_hits = defaultdict(int)

    for token in addr_tokens:

        if addr_freq.get(token, 999999999) <= addr_limit:

            for eid in addr_index.get(token, []):
                addr_hits[eid] += 1

    for eid, count in addr_hits.items():

        if count < 2:
            continue

        target = source_df.iloc[eid]

        if target["country_n"] != country1:
            continue

        if sim(addr1, target["addr_n"]) >= addr_sim_threshold:
            candidates.add(eid)

    return candidates


# ------------------------------------------------------------
# Evaluate
# ------------------------------------------------------------

print("Testing 500/500 + similarity >= 0.70...")

s2_counts = []
s3_counts = []

for counter, (_, row) in enumerate(s1.iterrows()):

    if counter % 1000 == 0:
        print(f"{counter}/10000")

    s2_counts.append(
        len(
            selective_candidates_v3(
                row,
                test_s2,
                s2_name_index,
                s2_addr_index,
                s2_name_freq,
                s2_addr_freq
            )
        )
    )

    s3_counts.append(
        len(
            selective_candidates_v3(
                row,
                test_s3,
                s3_name_index,
                s3_addr_index,
                s3_name_freq,
                s3_addr_freq
            )
        )
    )


s2_counts = np.array(s2_counts)
s3_counts = np.array(s3_counts)

print("\n" + "=" * 70)
print("500/500 + SIMILARITY >= 0.70")
print("=" * 70)

print("\nS2")
print("Average:", round(s2_counts.mean(), 2))
print("Median :", round(np.median(s2_counts), 2))
print("P90    :", round(np.percentile(s2_counts, 90), 2))
print("P95    :", round(np.percentile(s2_counts, 95), 2))
print("P99    :", round(np.percentile(s2_counts, 99), 2))
print("Max    :", int(s2_counts.max()))

print("\nS3")
print("Average:", round(s3_counts.mean(), 2))
print("Median :", round(np.median(s3_counts), 2))
print("P90    :", round(np.percentile(s3_counts, 90), 2))
print("P95    :", round(np.percentile(s3_counts, 95), 2))
print("P99    :", round(np.percentile(s3_counts, 99), 2))
print("Max    :", int(s3_counts.max()))

combined_avg = (s2_counts.mean() + s3_counts.mean()) / 2

print("\nCombined average:", round(combined_avg, 2))

print(
    "Estimated full-test candidates:",
    f"{combined_avg * 1_732_544:,.0f}"
)
```

```python
# ============================================================
# OPTIMIZED THRESHOLD COMPARISON
# Candidate search happens ONCE per S1 row
# Similarities calculated ONCE per candidate
# ============================================================

from difflib import SequenceMatcher
import numpy as np

thresholds = [0.60, 0.65, 0.70, 0.75]

results = {
    0.60: [],
    0.65: [],
    0.70: [],
    0.75: []
}

def fast_sim(a, b):
    if not a or not b:
        return 0.0
    return SequenceMatcher(None, a, b).ratio()


print("Starting optimized threshold comparison...")

for counter, (_, row) in enumerate(s1.iterrows()):

    if counter % 1000 == 0:
        print(f"{counter}/10000")

    name1 = row["name_n"]
    addr1 = row["addr_n"]
    country1 = row["country_n"]

    name_tokens = toks(name1)
    addr_tokens = toks(addr1)

    # ========================================================
    # S2 + S3 separately
    # ========================================================

    for source_df, name_index, addr_index, name_freq, addr_freq, source_key in [
        (
            test_s2,
            s2_name_index,
            s2_addr_index,
            s2_name_freq,
            s2_addr_freq,
            "s2"
        ),
        (
            test_s3,
            s3_name_index,
            s3_addr_index,
            s3_name_freq,
            s3_addr_freq,
            "s3"
        )
    ]:

        name_hits = defaultdict(int)
        addr_hits = defaultdict(int)

        # ----------------------------------------------------
        # Build NAME candidates
        # ----------------------------------------------------

        for token in name_tokens:

            if name_freq.get(token, 999999999) <= 500:

                for eid in name_index.get(token, []):
                    name_hits[eid] += 1

        # ----------------------------------------------------
        # Build ADDRESS candidates
        # ----------------------------------------------------

        for token in addr_tokens:

            if addr_freq.get(token, 999999999) <= 500:

                for eid in addr_index.get(token, []):
                    addr_hits[eid] += 1

        candidate_ids = set()

        # Only candidates with >=2 shared tokens
        candidate_ids.update(
            eid for eid, count in name_hits.items()
            if count >= 2
        )

        candidate_ids.update(
            eid for eid, count in addr_hits.items()
            if count >= 2
        )

        # ----------------------------------------------------
        # Calculate similarities ONCE
        # ----------------------------------------------------

        candidate_scores = []

        for eid in candidate_ids:

            target = source_df.iloc[eid]

            # Same country
            if target["country_n"] != country1:
                continue

            name_sim = 0.0
            addr_sim = 0.0

            if name_hits.get(eid, 0) >= 2:
                name_sim = fast_sim(
                    name1,
                    target["name_n"]
                )

            if addr_hits.get(eid, 0) >= 2:
                addr_sim = fast_sim(
                    addr1,
                    target["addr_n"]
                )

            candidate_scores.append(
                (name_sim, addr_sim)
            )

        # ----------------------------------------------------
        # Count all thresholds from same similarity values
        # ----------------------------------------------------

        for threshold in thresholds:

            count = sum(
                1
                for name_sim, addr_sim in candidate_scores
                if name_sim >= threshold
                or addr_sim >= threshold
            )

            results[threshold].append(
                (source_key, count)
            )


print("\n" + "=" * 75)
print("OPTIMIZED THRESHOLD COMPARISON")
print("=" * 75)

for threshold in thresholds:

    s2 = np.array([
        count
        for source, count in results[threshold]
        if source == "s2"
    ])

    s3 = np.array([
        count
        for source, count in results[threshold]
        if source == "s3"
    ])

    avg = (s2.mean() + s3.mean()) / 2

    estimated = avg * 1_732_544

    print(f"\nSimilarity >= {threshold:.2f}")
    print("-" * 45)
    print("S2 average:", round(s2.mean(), 2))
    print("S3 average:", round(s3.mean(), 2))
    print("Combined average:", round(avg, 2))
    print("Estimated full candidates:",
          f"{estimated:,.0f}")
```

```python
# ============================================================
# LOAD ONLY THE 6,975 TRUE S2/S3 ENTITIES
# FROM THE TRAINING DATA
# ============================================================

import pandas as pd

s2_needed = set(s2_gt_ids)
s3_needed = set(s3_gt_ids)

print("Looking for:")
print("S2:", len(s2_needed))
print("S3:", len(s3_needed))

train_s2_path = "/content/amazon_ml_challenge/student_resource/dataset/train/train_source2.tsv"
train_s3_path = "/content/amazon_ml_challenge/student_resource/dataset/train/train_source3.tsv"

# ------------------------------------------------------------
# Read S2 in chunks
# ------------------------------------------------------------

s2_true_parts = []

for chunk in pd.read_csv(
    train_s2_path,
    sep="\t",
    chunksize=200_000,
    dtype=str
):
    found = chunk[
        chunk["entity_id"].isin(s2_needed)
    ]

    if len(found):
        s2_true_parts.append(found)

    if sum(len(x) for x in s2_true_parts) >= len(s2_needed):
        break

train_s2_true = pd.concat(
    s2_true_parts,
    ignore_index=True
)

# ------------------------------------------------------------
# Read S3 in chunks
# ------------------------------------------------------------

s3_true_parts = []

for chunk in pd.read_csv(
    train_s3_path,
    sep="\t",
    chunksize=200_000,
    dtype=str
):
    found = chunk[
        chunk["entity_id"].isin(s3_needed)
    ]

    if len(found):
        s3_true_parts.append(found)

    if sum(len(x) for x in s3_true_parts) >= len(s3_needed):
        break

train_s3_true = pd.concat(
    s3_true_parts,
    ignore_index=True
)

print("\n" + "=" * 60)
print("EXTRACTION COMPLETE")
print("=" * 60)

print("S2 rows found:", len(train_s2_true))
print("S3 rows found:", len(train_s3_true))

print("\nExpected:")
print("S2:", len(s2_needed))
print("S3:", len(s3_needed))

print("\nMissing S2:", len(s2_needed - set(train_s2_true["entity_id"])))
print("Missing S3:", len(s3_needed - set(train_s3_true["entity_id"])))
```

```python
# ============================================================
# TRUE-MATCH RECALL TEST
#
# Tests:
#   500/500 rare-token blocker
#   >=2 shared tokens
#   similarity >= threshold
#
# Against the 6,975 known true matches
# ============================================================

from difflib import SequenceMatcher
from collections import defaultdict

# ------------------------------------------------------------
# Normalize the extracted true rows
# ------------------------------------------------------------

def normalize_for_block(x):
    if pd.isna(x):
        return ""
    return str(x).lower().strip()


def get_tokens(x):
    if not x:
        return set()
    return set(x.split())


# Create lookup dictionaries by entity ID
s2_true_lookup = {
    row["entity_id"]: row
    for _, row in train_s2_true.iterrows()
}

s3_true_lookup = {
    row["entity_id"]: row
    for _, row in train_s3_true.iterrows()
}


def similarity(a, b):
    if not a or not b:
        return 0.0
    return SequenceMatcher(None, a, b).ratio()


# ------------------------------------------------------------
# Test every known true pair
# ------------------------------------------------------------

thresholds = [0.60, 0.65, 0.70, 0.75]

stats = {
    t: {
        "total": 0,
        "recovered": 0,
        "s2_total": 0,
        "s2_recovered": 0,
        "s3_total": 0,
        "s3_recovered": 0
    }
    for t in thresholds
}


for counter, (s1_id, true_ids) in enumerate(gt_lookup.items()):

    if counter % 500 == 0:
        print(f"{counter}/2000")

    s1row = s1_lookup[s1_id]

    name1 = normalize_for_block(s1row["business_name"])
    addr1 = normalize_for_block(s1row["business_address"])
    country1 = normalize_for_block(s1row["country"])

    name_tokens1 = get_tokens(name1)
    addr_tokens1 = get_tokens(addr1)

    for true_id in true_ids:

        if true_id.startswith("S2-"):
            target = s2_true_lookup[true_id]
            source = "s2"
        else:
            target = s3_true_lookup[true_id]
            source = "s3"

        name2 = normalize_for_block(target["business_name"])
        addr2 = normalize_for_block(target["business_address"])
        country2 = normalize_for_block(target["country"])

        # Candidate blocker requires same country
        if country1 != country2:
            continue

        shared_name = name_tokens1 & get_tokens(name2)
        shared_addr = addr_tokens1 & get_tokens(addr2)

        # Rare token requirement
        name_rare = any(
            name_freq.get(token, 999999999) <= 500
            for token in shared_name
        )

        addr_rare = any(
            addr_freq.get(token, 999999999) <= 500
            for token in shared_addr
        )

        name_sim = similarity(name1, name2)
        addr_sim = similarity(addr1, addr2)

        for threshold in thresholds:

            stats[threshold]["total"] += 1

            if source == "s2":
                stats[threshold]["s2_total"] += 1
            else:
                stats[threshold]["s3_total"] += 1

            recovered = (
                (
                    len(shared_name) >= 2
                    and name_rare
                    and name_sim >= threshold
                )
                or
                (
                    len(shared_addr) >= 2
                    and addr_rare
                    and addr_sim >= threshold
                )
            )

            if recovered:

                stats[threshold]["recovered"] += 1

                if source == "s2":
                    stats[threshold]["s2_recovered"] += 1
                else:
                    stats[threshold]["s3_recovered"] += 1


# ------------------------------------------------------------
# Results
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("TRUE-MATCH RECALL RESULTS")
print("=" * 75)

for threshold in thresholds:

    x = stats[threshold]

    overall = 100 * x["recovered"] / x["total"]
    s2_recall = 100 * x["s2_recovered"] / x["s2_total"]
    s3_recall = 100 * x["s3_recovered"] / x["s3_total"]

    print(f"\nSimilarity >= {threshold:.2f}")
    print("-" * 50)

    print(
        f"Overall: {x['recovered']:,} / {x['total']:,}"
        f" = {overall:.2f}%"
    )

    print(
        f"S2: {x['s2_recovered']:,} / {x['s2_total']:,}"
        f" = {s2_recall:.2f}%"
    )

    print(
        f"S3: {x['s3_recovered']:,} / {x['s3_total']:,}"
        f" = {s3_recall:.2f}%"
    )
```

```python
# ============================================================
# BLOCKER RECALL:
# RARITY LIMIT COMPARISON AT SIMILARITY >= 0.60
#
# Also includes:
#   exact name + country
#   exact address + country
# ============================================================

rarity_limits = [500, 1000, 2000, 5000, None]

recall_results = {}

for limit in rarity_limits:

    recovered = 0
    s2_rec = 0
    s3_rec = 0

    total = 0
    s2_total = 0
    s3_total = 0

    for s1_id, true_ids in gt_lookup.items():

        s1row = s1_lookup[s1_id]

        name1 = normalize_for_block(s1row["business_name"])
        addr1 = normalize_for_block(s1row["business_address"])
        country1 = normalize_for_block(s1row["country"])

        tokens_name1 = get_tokens(name1)
        tokens_addr1 = get_tokens(addr1)

        for true_id in true_ids:

            if true_id.startswith("S2-"):
                target = s2_true_lookup[true_id]
                source = "s2"
            else:
                target = s3_true_lookup[true_id]
                source = "s3"

            name2 = normalize_for_block(target["business_name"])
            addr2 = normalize_for_block(target["business_address"])
            country2 = normalize_for_block(target["country"])

            if source == "s2":
                s2_total += 1
            else:
                s3_total += 1

            total += 1

            # Country must agree
            if country1 != country2:
                continue

            # ------------------------------------------------
            # EXACT RULES
            # ------------------------------------------------

            exact_match = (
                (name1 and name1 == name2)
                or
                (addr1 and addr1 == addr2)
            )

            if exact_match:

                recovered += 1

                if source == "s2":
                    s2_rec += 1
                else:
                    s3_rec += 1

                continue

            # ------------------------------------------------
            # TOKEN RULES
            # ------------------------------------------------

            shared_name = tokens_name1 & get_tokens(name2)
            shared_addr = tokens_addr1 & get_tokens(addr2)

            name_ok = False
            addr_ok = False

            # Name
            if len(shared_name) >= 2:

                if limit is None:
                    name_rare = True
                else:
                    name_rare = any(
                        name_freq.get(t, 999999999) <= limit
                        for t in shared_name
                    )

                if name_rare:
                    if similarity(name1, name2) >= 0.60:
                        name_ok = True

            # Address
            if len(shared_addr) >= 2:

                if limit is None:
                    addr_rare = True
                else:
                    addr_rare = any(
                        addr_freq.get(t, 999999999) <= limit
                        for t in shared_addr
                    )

                if addr_rare:
                    if similarity(addr1, addr2) >= 0.60:
                        addr_ok = True

            if name_ok or addr_ok:

                recovered += 1

                if source == "s2":
                    s2_rec += 1
                else:
                    s3_rec += 1

    recall_results[limit] = {
        "overall": 100 * recovered / total,
        "s2": 100 * s2_rec / s2_total,
        "s3": 100 * s3_rec / s3_total,
        "recovered": recovered
    }


print("\n" + "=" * 75)
print("RARITY LIMIT + SIMILARITY 0.60 RECALL")
print("=" * 75)

for limit, r in recall_results.items():

    label = "NO RARITY LIMIT" if limit is None else f"<= {limit}"

    print(f"\n{label}")
    print("-" * 45)
    print(
        f"Overall: {r['recovered']:,} / 6,975"
        f" = {r['overall']:.2f}%"
    )
    print(f"S2: {r['s2']:.2f}%")
    print(f"S3: {r['s3']:.2f}%")
```

```python
# ============================================================
# ANALYZE MISSED TRUE PAIRS
# NO RARITY LIMIT + SIMILARITY >= 0.60
# ============================================================

missed = []

for s1_id, true_ids in gt_lookup.items():

    s1row = s1_lookup[s1_id]

    name1 = normalize_for_block(s1row["business_name"])
    addr1 = normalize_for_block(s1row["business_address"])
    country1 = normalize_for_block(s1row["country"])

    tokens_name1 = get_tokens(name1)
    tokens_addr1 = get_tokens(addr1)

    for true_id in true_ids:

        if true_id.startswith("S2-"):
            target = s2_true_lookup[true_id]
            source = "S2"
        else:
            target = s3_true_lookup[true_id]
            source = "S3"

        name2 = normalize_for_block(target["business_name"])
        addr2 = normalize_for_block(target["business_address"])
        country2 = normalize_for_block(target["country"])

        shared_name = tokens_name1 & get_tokens(name2)
        shared_addr = tokens_addr1 & get_tokens(addr2)

        name_sim = similarity(name1, name2)
        addr_sim = similarity(addr1, addr2)

        exact_name = bool(name1) and name1 == name2
        exact_addr = bool(addr1) and addr1 == addr2

        recovered = False

        if country1 == country2:

            if exact_name or exact_addr:
                recovered = True

            elif (
                len(shared_name) >= 2
                and name_sim >= 0.60
            ):
                recovered = True

            elif (
                len(shared_addr) >= 2
                and addr_sim >= 0.60
            ):
                recovered = True

        if not recovered:

            missed.append({
                "s1_id": s1_id,
                "true_id": true_id,
                "source": source,
                "country_same": country1 == country2,
                "shared_name": len(shared_name),
                "shared_addr": len(shared_addr),
                "name_sim": name_sim,
                "addr_sim": addr_sim,
                "exact_name": exact_name,
                "exact_addr": exact_addr,
                "name1": name1,
                "name2": name2,
                "addr1": addr1,
                "addr2": addr2
            })


missed_df = pd.DataFrame(missed)

print("=" * 75)
print("MISSED TRUE PAIRS")
print("=" * 75)

print("Total missed:", len(missed_df))

print("\nCountry mismatch:")
print(missed_df["country_same"].value_counts())

print("\nShared name tokens:")
print(missed_df["shared_name"].value_counts().sort_index())

print("\nShared address tokens:")
print(missed_df["shared_addr"].value_counts().sort_index())

print("\nName similarity:")
print(missed_df["name_sim"].describe())

print("\nAddress similarity:")
print(missed_df["addr_sim"].describe())

print("\nSource:")
print(missed_df["source"].value_counts())

print("\nPotential high-similarity misses:")

display(
    missed_df[
        (missed_df["name_sim"] >= 0.80) |
        (missed_df["addr_sim"] >= 0.80)
    ][[
        "source",
        "shared_name",
        "shared_addr",
        "name_sim",
        "addr_sim",
        "name1",
        "name2",
        "addr1",
        "addr2"
    ]].head(30)
)
```

```python
# ============================================================
# HYBRID BLOCKER RECALL TEST
#
# Exact name/address
# +
# Name: >=2 shared tokens + similarity >= 0.60
# +
# Address: >=3 shared tokens (NO character similarity)
# ============================================================

total = 0
recovered = 0

s2_total = 0
s2_recovered = 0

s3_total = 0
s3_recovered = 0

for counter, (s1_id, true_ids) in enumerate(gt_lookup.items()):

    if counter % 500 == 0:
        print(f"{counter}/2000")

    s1row = s1_lookup[s1_id]

    name1 = normalize_for_block(s1row["business_name"])
    addr1 = normalize_for_block(s1row["business_address"])
    country1 = normalize_for_block(s1row["country"])

    name_tokens1 = get_tokens(name1)
    addr_tokens1 = get_tokens(addr1)

    for true_id in true_ids:

        if true_id.startswith("S2-"):
            target = s2_true_lookup[true_id]
            source = "S2"
            s2_total += 1
        else:
            target = s3_true_lookup[true_id]
            source = "S3"
            s3_total += 1

        total += 1

        name2 = normalize_for_block(target["business_name"])
        addr2 = normalize_for_block(target["business_address"])
        country2 = normalize_for_block(target["country"])

        if country1 != country2:
            continue

        # ----------------------------------------------------
        # Exact matches
        # ----------------------------------------------------

        exact_name = (
            bool(name1) and
            name1 == name2
        )

        exact_addr = (
            bool(addr1) and
            addr1 == addr2
        )

        # ----------------------------------------------------
        # Shared tokens
        # ----------------------------------------------------

        shared_name = (
            name_tokens1 &
            get_tokens(name2)
        )

        shared_addr = (
            addr_tokens1 &
            get_tokens(addr2)
        )

        # ----------------------------------------------------
        # Name rule
        # ----------------------------------------------------

        name_rule = (
            len(shared_name) >= 2
            and similarity(name1, name2) >= 0.60
        )

        # ----------------------------------------------------
        # Address rule
        #
        # IMPORTANT:
        # No SequenceMatcher here.
        # Token overlap handles reordered addresses.
        # ----------------------------------------------------

        address_rule = (
            len(shared_addr) >= 3
        )

        recovered_pair = (
            exact_name
            or exact_addr
            or name_rule
            or address_rule
        )

        if recovered_pair:

            recovered += 1

            if source == "S2":
                s2_recovered += 1
            else:
                s3_recovered += 1


print("\n" + "=" * 75)
print("HYBRID BLOCKER RECALL")
print("=" * 75)

print(
    f"\nOverall: {recovered:,} / {total:,}"
    f" = {100 * recovered / total:.2f}%"
)

print(
    f"S2: {s2_recovered:,} / {s2_total:,}"
    f" = {100 * s2_recovered / s2_total:.2f}%"
)

print(
    f"S3: {s3_recovered:,} / {s3_total:,}"
    f" = {100 * s3_recovered / s3_total:.2f}%"
)
```

```python
# ============================================================
# ANALYZE HYBRID BLOCKER MISSES
# ============================================================

hybrid_missed = []

for s1_id, true_ids in gt_lookup.items():

    s1row = s1_lookup[s1_id]

    name1 = normalize_for_block(s1row["business_name"])
    addr1 = normalize_for_block(s1row["business_address"])
    country1 = normalize_for_block(s1row["country"])

    name_tokens1 = get_tokens(name1)
    addr_tokens1 = get_tokens(addr1)

    for true_id in true_ids:

        if true_id.startswith("S2-"):
            target = s2_true_lookup[true_id]
            source = "S2"
        else:
            target = s3_true_lookup[true_id]
            source = "S3"

        name2 = normalize_for_block(target["business_name"])
        addr2 = normalize_for_block(target["business_address"])
        country2 = normalize_for_block(target["country"])

        if country1 != country2:
            continue

        shared_name = name_tokens1 & get_tokens(name2)
        shared_addr = addr_tokens1 & get_tokens(addr2)

        name_sim = similarity(name1, name2)
        addr_sim = similarity(addr1, addr2)

        exact_name = bool(name1) and name1 == name2
        exact_addr = bool(addr1) and addr1 == addr2

        recovered = (
            exact_name
            or exact_addr
            or (
                len(shared_name) >= 2
                and name_sim >= 0.60
            )
            or (
                len(shared_addr) >= 3
            )
        )

        if not recovered:

            hybrid_missed.append({
                "source": source,
                "shared_name": len(shared_name),
                "shared_addr": len(shared_addr),
                "name_sim": name_sim,
                "addr_sim": addr_sim,
                "name1": name1,
                "name2": name2,
                "addr1": addr1,
                "addr2": addr2
            })


missed_hybrid_df = pd.DataFrame(hybrid_missed)

print("=" * 75)
print("HYBRID BLOCKER MISSES")
print("=" * 75)

print("Total missed:", len(missed_hybrid_df))

print("\nShared address tokens:")
print(
    missed_hybrid_df["shared_addr"]
    .value_counts()
    .sort_index()
)

print("\nShared name tokens:")
print(
    missed_hybrid_df["shared_name"]
    .value_counts()
    .sort_index()
)

print("\nName similarity:")
print(missed_hybrid_df["name_sim"].describe())

print("\nAddress similarity:")
print(missed_hybrid_df["addr_sim"].describe())

print("\nSource:")
print(missed_hybrid_df["source"].value_counts())

print("\nExamples with exactly 2 shared address tokens:")
display(
    missed_hybrid_df[
        missed_hybrid_df["shared_addr"] == 2
    ][[
        "source",
        "shared_name",
        "shared_addr",
        "name_sim",
        "addr_sim",
        "name1",
        "name2",
        "addr1",
        "addr2"
    ]].head(20)
)
```

```python
# ============================================================
# TEST TARGETED ADDRESS FALLBACKS
# Against the 304 remaining hybrid misses
# ============================================================

import re

def extract_numbers(text):
    if not text:
        return set()
    return set(re.findall(r'\d+', text))


fallback_counts = {
    "2_tokens + shared_number": 0,
    "2_tokens + addr_sim_0.30": 0,
    "2_tokens + addr_sim_0.40": 0,
}

for _, row in missed_hybrid_df.iterrows():

    shared_addr = int(row["shared_addr"])

    if shared_addr < 2:
        continue

    addr1 = row["addr1"]
    addr2 = row["addr2"]

    nums1 = extract_numbers(addr1)
    nums2 = extract_numbers(addr2)

    shared_numbers = nums1 & nums2

    # Rule 1
    if shared_numbers:
        fallback_counts["2_tokens + shared_number"] += 1

    # Rule 2
    if row["addr_sim"] >= 0.30:
        fallback_counts["2_tokens + addr_sim_0.30"] += 1

    # Rule 3
    if row["addr_sim"] >= 0.40:
        fallback_counts["2_tokens + addr_sim_0.40"] += 1


print("=" * 70)
print("TARGETED ADDRESS FALLBACK RESULTS")
print("=" * 70)

for rule, count in fallback_counts.items():

    new_recall = (
        (6975 - 304 + count) / 6975
    ) * 100

    print(f"\n{rule}")
    print("Additional true matches recovered:", count)
    print(
        "Estimated total recall:",
        f"{new_recall:.2f}%"
    )
```

```python
# ============================================================
# FINAL HYBRID BLOCKER - CANDIDATE VOLUME TEST
#
# Rules:
# 1. Exact name + country
# 2. Exact address + country
# 3. >=2 name tokens + name similarity >= 0.60
# 4. >=3 address tokens
# 5. >=2 address tokens + address similarity >= 0.30
#
# NO RARITY LIMIT
# ============================================================

from difflib import SequenceMatcher
from collections import defaultdict
import numpy as np
import re


def sim_fast(a, b):
    if not a or not b:
        return 0.0
    return SequenceMatcher(None, a, b).ratio()


def count_final_candidates(
    row,
    source_df,
    name_index,
    addr_index
):

    name1 = row["name_n"]
    addr1 = row["addr_n"]
    country1 = row["country_n"]

    name_tokens1 = toks(name1)
    addr_tokens1 = toks(addr1)

    name_hits = defaultdict(int)
    addr_hits = defaultdict(int)

    # --------------------------------------------------------
    # NAME TOKEN BLOCK
    # --------------------------------------------------------

    for token in name_tokens1:

        for eid in name_index.get(token, []):
            name_hits[eid] += 1

    # --------------------------------------------------------
    # ADDRESS TOKEN BLOCK
    # --------------------------------------------------------

    for token in addr_tokens1:

        for eid in addr_index.get(token, []):
            addr_hits[eid] += 1

    # Candidates that satisfy >=2 token requirement
    candidate_ids = set()

    candidate_ids.update(
        eid
        for eid, count in name_hits.items()
        if count >= 2
    )

    candidate_ids.update(
        eid
        for eid, count in addr_hits.items()
        if count >= 2
    )

    final = set()

    for eid in candidate_ids:

        target = source_df.iloc[eid]

        # Same country
        if target["country_n"] != country1:
            continue

        name2 = target["name_n"]
        addr2 = target["addr_n"]

        shared_name_count = name_hits.get(eid, 0)
        shared_addr_count = addr_hits.get(eid, 0)

        # ----------------------------------------------------
        # Exact rules
        # ----------------------------------------------------

        if name1 and name1 == name2:
            final.add(eid)
            continue

        if addr1 and addr1 == addr2:
            final.add(eid)
            continue

        # ----------------------------------------------------
        # Name similarity rule
        # ----------------------------------------------------

        if shared_name_count >= 2:

            if sim_fast(name1, name2) >= 0.60:
                final.add(eid)
                continue

        # ----------------------------------------------------
        # Address >= 3 tokens
        # ----------------------------------------------------

        if shared_addr_count >= 3:
            final.add(eid)
            continue

        # ----------------------------------------------------
        # Address fallback:
        # >=2 tokens + similarity >= 0.30
        # ----------------------------------------------------

        if shared_addr_count >= 2:

            if sim_fast(addr1, addr2) >= 0.30:
                final.add(eid)
                continue

    return final


print("Starting FINAL blocker volume test...")

s2_counts = []
s3_counts = []

for counter, (_, row) in enumerate(s1.iterrows()):

    if counter % 1000 == 0:
        print(f"{counter}/10000")

    s2_counts.append(
        len(
            count_final_candidates(
                row,
                test_s2,
                s2_name_index,
                s2_addr_index
            )
        )
    )

    s3_counts.append(
        len(
            count_final_candidates(
                row,
                test_s3,
                s3_name_index,
                s3_addr_index
            )
        )
    )


s2_counts = np.array(s2_counts)
s3_counts = np.array(s3_counts)

combined_avg = (
    s2_counts.mean() +
    s3_counts.mean()
) / 2

estimated_full = combined_avg * 1_732_544


print("\n" + "=" * 75)
print("FINAL HYBRID BLOCKER VOLUME")
print("=" * 75)

print("\nS2")
print("Average:", round(s2_counts.mean(), 2))
print("Median :", round(np.median(s2_counts), 2))
print("P90    :", round(np.percentile(s2_counts, 90), 2))
print("P95    :", round(np.percentile(s2_counts, 95), 2))
print("P99    :", round(np.percentile(s2_counts, 99), 2))
print("Max    :", int(s2_counts.max()))

print("\nS3")
print("Average:", round(s3_counts.mean(), 2))
print("Median :", round(np.median(s3_counts), 2))
print("P90    :", round(np.percentile(s3_counts, 90), 2))
print("P95    :", round(np.percentile(s3_counts, 95), 2))
print("P99    :", round(np.percentile(s3_counts, 99), 2))
print("Max    :", int(s3_counts.max()))

print("\nCombined average:", round(combined_avg, 2))

print(
    "Estimated full-test candidates:",
    f"{estimated_full:,.0f}"
)
```

```python
# ============================================================
# FAST FINAL BLOCKER VOLUME TEST — 500 S1 ROWS
# ============================================================

import numpy as np
import re

# Use only 500 rows for a quick volume estimate
volume_test_s1 = s1.head(500).copy()

print("Starting FAST 500-row blocker volume test...")
print("Rows:", len(volume_test_s1))


def extract_numbers(text):
    if not text:
        return set()
    return set(re.findall(r'\d+', text))


def count_final_candidates_fast(row, test_df, name_index, addr_index):

    name1 = normalize_for_block(row["business_name"])
    addr1 = normalize_for_block(row["business_address"])
    country1 = normalize_for_block(row["country"])

    name_tokens1 = get_tokens(name1)
    addr_tokens1 = get_tokens(addr1)

    nums1 = extract_numbers(addr1)

    final = set()

    # --------------------------------------------------------
    # NAME INDEX
    # --------------------------------------------------------
    for token in name_tokens1:
        for eid in name_index.get(token, []):
            if eid in final:
                continue

            target = test_df.loc[eid]

            country2 = normalize_for_block(target["country"])

            if country1 != country2:
                continue

            name2 = normalize_for_block(target["business_name"])
            addr2 = normalize_for_block(target["business_address"])

            name_tokens2 = get_tokens(name2)
            addr_tokens2 = get_tokens(addr2)

            # Rule 1: exact name
            if name1 == name2:
                final.add(eid)
                continue

            # Rule 3: >=2 shared name tokens + similarity >= 0.60
            shared_name_count = len(name_tokens1 & name_tokens2)

            if shared_name_count >= 2:
                if sim_fast(name1, name2) >= 0.60:
                    final.add(eid)
                    continue

    # --------------------------------------------------------
    # ADDRESS INDEX
    # --------------------------------------------------------
    for token in addr_tokens1:
        for eid in addr_index.get(token, []):

            if eid in final:
                continue

            target = test_df.loc[eid]

            country2 = normalize_for_block(target["country"])

            if country1 != country2:
                continue

            name2 = normalize_for_block(target["business_name"])
            addr2 = normalize_for_block(target["business_address"])

            # Rule 2: exact address
            if addr1 == addr2:
                final.add(eid)
                continue

            addr_tokens2 = get_tokens(addr2)

            shared_addr_count = len(addr_tokens1 & addr_tokens2)

            # Rule 4: >=3 shared address tokens
            if shared_addr_count >= 3:
                final.add(eid)
                continue

            # Rule 5:
            # >=2 shared address tokens + similarity >= 0.30
            if shared_addr_count >= 2:
                if sim_fast(addr1, addr2) >= 0.30:
                    final.add(eid)
                    continue

    return final


# ============================================================
# RUN TEST
# ============================================================

s2_counts_500 = []
s3_counts_500 = []

for counter, (_, row) in enumerate(volume_test_s1.iterrows()):

    if counter % 50 == 0:
        print(f"{counter}/500")

    s2_counts_500.append(
        len(
            count_final_candidates_fast(
                row,
                test_s2,
                s2_name_index,
                s2_addr_index
            )
        )
    )

    s3_counts_500.append(
        len(
            count_final_candidates_fast(
                row,
                test_s3,
                s3_name_index,
                s3_addr_index
            )
        )
    )


s2_counts_500 = np.array(s2_counts_500)
s3_counts_500 = np.array(s3_counts_500)


# ============================================================
# RESULTS
# ============================================================

combined_avg = (
    s2_counts_500.mean() +
    s3_counts_500.mean()
) / 2

estimated_full = combined_avg * 1_732_544


print("\n" + "=" * 75)
print("FAST FINAL HYBRID BLOCKER VOLUME — 500 ROW TEST")
print("=" * 75)

print("\nS2")
print("Average:", round(s2_counts_500.mean(), 2))
print("Median :", round(np.median(s2_counts_500), 2))
print("P90    :", round(np.percentile(s2_counts_500, 90), 2))
print("P95    :", round(np.percentile(s2_counts_500, 95), 2))
print("P99    :", round(np.percentile(s2_counts_500, 99), 2))
print("Max    :", int(s2_counts_500.max()))

print("\nS3")
print("Average:", round(s3_counts_500.mean(), 2))
print("Median :", round(np.median(s3_counts_500), 2))
print("P90    :", round(np.percentile(s3_counts_500, 90), 2))
print("P95    :", round(np.percentile(s3_counts_500, 95), 2))
print("P99    :", round(np.percentile(s3_counts_500, 99), 2))
print("Max    :", int(s3_counts_500.max()))

print("\nCOMBINED")
print("Average candidates / S1:",
      round(combined_avg, 2))

print("Estimated full candidate pairs:",
      f"{int(estimated_full):,}")

print("=" * 75)
```

```python
# ============================================================
# BUILD FULL TEST S2/S3 TOKEN INDEXES
# ONE-TIME SETUP FOR V5 INFERENCE
# ============================================================

import pandas as pd
import re
import pickle
import os
from collections import defaultdict

S2_PATH = "/content/amazon_ml_challenge/student_resource/dataset/test/test_source2.tsv"
S3_PATH = "/content/amazon_ml_challenge/student_resource/dataset/test/test_source3.tsv"

INDEX_DIR = "/content/drive/My Drive/Amazon-ML-Challenge-2026/ml/V5_full_indexes"
os.makedirs(INDEX_DIR, exist_ok=True)


def norm_simple(x):
    if pd.isna(x):
        return ""
    return str(x).lower().strip()


def token_set(x):
    if not x:
        return set()
    return set(x.split())


def build_indexes(path, prefix):

    print(f"\nReading {prefix}...")

    df = pd.read_csv(
        path,
        sep="\t",
        dtype=str
    )

    print(prefix, "shape:", df.shape)

    df["name_n"] = df["business_name"].map(norm_simple)
    df["addr_n"] = df["business_address"].map(norm_simple)
    df["country_n"] = df["country"].map(norm_simple)

    name_index = defaultdict(list)
    addr_index = defaultdict(list)

    name_freq = defaultdict(int)
    addr_freq = defaultdict(int)

    for i, row in df.iterrows():

        name_tokens = token_set(row["name_n"])
        addr_tokens = token_set(row["addr_n"])

        for token in name_tokens:
            name_index[token].append(i)
            name_freq[token] += 1

        for token in addr_tokens:
            addr_index[token].append(i)
            addr_freq[token] += 1

        if i % 1_000_000 == 0 and i > 0:
            print(f"{prefix}: {i:,} rows processed")

    # Convert defaultdicts to normal dicts
    name_index = dict(name_index)
    addr_index = dict(addr_index)
    name_freq = dict(name_freq)
    addr_freq = dict(addr_freq)

    # Save
    with open(f"{INDEX_DIR}/{prefix}_name_index.pkl", "wb") as f:
        pickle.dump(name_index, f, protocol=pickle.HIGHEST_PROTOCOL)

    with open(f"{INDEX_DIR}/{prefix}_addr_index.pkl", "wb") as f:
        pickle.dump(addr_index, f, protocol=pickle.HIGHEST_PROTOCOL)

    with open(f"{INDEX_DIR}/{prefix}_name_freq.pkl", "wb") as f:
        pickle.dump(name_freq, f, protocol=pickle.HIGHEST_PROTOCOL)

    with open(f"{INDEX_DIR}/{prefix}_addr_freq.pkl", "wb") as f:
        pickle.dump(addr_freq, f, protocol=pickle.HIGHEST_PROTOCOL)

    print(f"\n{prefix} indexes saved.")
    print("Name tokens:", len(name_index))
    print("Address tokens:", len(addr_index))

    return df


full_test_s2 = build_indexes(S2_PATH, "s2")
full_test_s3 = build_indexes(S3_PATH, "s3")

print("\n" + "=" * 70)
print("FULL TEST INDEX BUILD COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# V5 FULL TEST INFERENCE
#
# Blocker:
#   same country
#   >=2 shared rare name tokens (freq <=500)
#       + name similarity >=0.60
#   OR
#   >=2 shared rare address tokens (freq <=500)
#       + address similarity >=0.60
#   OR exact name/address
#
# V5 CatBoost threshold = 0.60
# ============================================================

import os
import re
import csv
import numpy as np
import pandas as pd
from collections import defaultdict
from difflib import SequenceMatcher

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

OUTPUT_DIR = "/content/amazon_ml_challenge/student_resource/output"
os.makedirs(OUTPUT_DIR, exist_ok=True)

RESULT_PATH = os.path.join(
    OUTPUT_DIR,
    "matching_results_v5.tsv"
)

CANDIDATE_PATH = os.path.join(
    OUTPUT_DIR,
    "candidate_pairs_v5.tsv"
)

# ------------------------------------------------------------
# Settings
# ------------------------------------------------------------

V5_THRESHOLD = 0.60
RARITY_LIMIT = 500
S1_CHUNK = 500

print("=" * 75)
print("V5 FULL TEST INFERENCE")
print("=" * 75)
print("Threshold :", V5_THRESHOLD)
print("Rarity    :", RARITY_LIMIT)
print("S1 chunk  :", S1_CHUNK)
print()


# ============================================================
# NORMALIZATION
# ============================================================

def normalize_advanced_local(x):

    if pd.isna(x):
        return ""

    text = str(x).lower().strip()

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()


def tokens_fast(x):
    if not x:
        return set()
    return set(x.split())


def seq_sim(a, b):

    if not a or not b:
        return 0.0

    return SequenceMatcher(None, a, b).ratio()


def extract_numbers_fast(x):

    if not x:
        return set()

    return set(re.findall(r"\d+", x))


# ============================================================
# PREPARE FULL SOURCE ARRAYS
# ============================================================

print("Preparing source normalization...")

# We already have these DataFrames in memory.
s2 = full_test_s2
s3 = full_test_s3

# Advanced normalization required by V5
if "name_a" not in s2.columns:

    s2["name_a"] = s2["business_name"].map(
        normalize_advanced_local
    )

    s2["addr_a"] = s2["business_address"].map(
        normalize_advanced_local
    )

    s2["country_a"] = s2["country"].map(
        lambda x: "" if pd.isna(x) else str(x).lower().strip()
    )


if "name_a" not in s3.columns:

    s3["name_a"] = s3["business_name"].map(
        normalize_advanced_local
    )

    s3["addr_a"] = s3["business_address"].map(
        normalize_advanced_local
    )

    s3["country_a"] = s3["country"].map(
        lambda x: "" if pd.isna(x) else str(x).lower().strip()
    )


print("S2 prepared:", s2.shape)
print("S3 prepared:", s3.shape)


# ============================================================
# BUILD FAST RARE TOKEN INDEXES
# ============================================================

print("\nPreparing rare-token indexes...")


def build_rare_index(df, name_prefix):

    name_idx = defaultdict(list)
    addr_idx = defaultdict(list)

    name_freq_local = defaultdict(int)
    addr_freq_local = defaultdict(int)

    # First pass: frequencies
    for value in df["name_n"]:
        for t in tokens_fast(value):
            name_freq_local[t] += 1

    for value in df["addr_n"]:
        for t in tokens_fast(value):
            addr_freq_local[t] += 1

    # Second pass: only rare tokens
    for i, (name, addr) in enumerate(
        zip(df["name_n"], df["addr_n"])
    ):

        for t in tokens_fast(name):
            if name_freq_local[t] <= RARITY_LIMIT:
                name_idx[t].append(i)

        for t in tokens_fast(addr):
            if addr_freq_local[t] <= RARITY_LIMIT:
                addr_idx[t].append(i)

        if i > 0 and i % 1_000_000 == 0:
            print(name_prefix, i, "rows indexed")

    return dict(name_idx), dict(addr_idx)


# Use the indexes already created rather than rebuilding them.
# They correspond to the same simple normalized source columns.

s2_name_idx = s2_name_index
s2_addr_idx = s2_addr_index

s3_name_idx = s3_name_index
s3_addr_idx = s3_addr_index

print("Indexes ready.")


# ============================================================
# V5 FEATURE CREATION
# ============================================================

V5_FEATURES = [
    "name_exact",
    "address_exact",
    "country_exact",
    "name_similarity",
    "address_similarity",
    "name_token_overlap",
    "address_token_overlap",
    "name_jaccard",
    "address_jaccard",
    "shared_name_token_count",
    "shared_address_token_count",
    "shared_address_number_count",
    "name_length_diff",
    "address_length_diff",
    "name_length_ratio",
    "address_length_ratio",
    "similarity_gap",
    "max_similarity",
    "similarity_product",
    "token_overlap_gap"
]


def make_v5_pair_features(
    s1_name,
    s1_addr,
    s1_country,
    target_df,
    candidate_ids
):

    rows = []

    name1_tokens = tokens_fast(s1_name)
    addr1_tokens = tokens_fast(s1_addr)
    nums1 = extract_numbers_fast(s1_addr)

    for eid in candidate_ids:

        target = target_df.iloc[eid]

        name2 = target["name_a"]
        addr2 = target["addr_a"]
        country2 = target["country_a"]

        name2_tokens = tokens_fast(name2)
        addr2_tokens = tokens_fast(addr2)

        shared_name = name1_tokens & name2_tokens
        shared_addr = addr1_tokens & addr2_tokens
        shared_nums = nums1 & extract_numbers_fast(addr2)

        ns = seq_sim(s1_name, name2)
        ads = seq_sim(s1_addr, addr2)

        name_overlap = (
            len(shared_name) /
            max(1, min(len(name1_tokens), len(name2_tokens)))
        )

        addr_overlap = (
            len(shared_addr) /
            max(1, min(len(addr1_tokens), len(addr2_tokens)))
        )

        name_union = name1_tokens | name2_tokens
        addr_union = addr1_tokens | addr2_tokens

        name_j = (
            len(shared_name) /
            max(1, len(name_union))
        )

        addr_j = (
            len(shared_addr) /
            max(1, len(addr_union))
        )

        name_len_ratio = (
            min(len(s1_name), len(name2)) /
            max(1, max(len(s1_name), len(name2)))
        )

        addr_len_ratio = (
            min(len(s1_addr), len(addr2)) /
            max(1, max(len(s1_addr), len(addr2)))
        )

        row = {
            "name_exact": int(
                bool(s1_name) and s1_name == name2
            ),

            "address_exact": int(
                bool(s1_addr) and s1_addr == addr2
            ),

            "country_exact": int(
                s1_country == country2
            ),

            "name_similarity": ns,
            "address_similarity": ads,

            "name_token_overlap": name_overlap,
            "address_token_overlap": addr_overlap,

            "name_jaccard": name_j,
            "address_jaccard": addr_j,

            "shared_name_token_count": len(shared_name),
            "shared_address_token_count": len(shared_addr),
            "shared_address_number_count": len(shared_nums),

            "name_length_diff": abs(
                len(s1_name) - len(name2)
            ),

            "address_length_diff": abs(
                len(s1_addr) - len(addr2)
            ),

            "name_length_ratio": name_len_ratio,
            "address_length_ratio": addr_len_ratio,

            "similarity_gap": abs(ns - ads),

            "max_similarity": max(ns, ads),

            "similarity_product": ns * ads,

            "token_overlap_gap": abs(
                name_overlap - addr_overlap
            )
        }

        rows.append(row)

    return pd.DataFrame(
        rows,
        columns=V5_FEATURES
    )


# ============================================================
# CANDIDATE GENERATION
# ============================================================

def generate_candidates(
    s1_name,
    s1_addr,
    s1_country,
    name_idx,
    addr_idx,
    name_freq,
    addr_freq
):

    name_hits = defaultdict(int)
    addr_hits = defaultdict(int)

    # Name
    for token in tokens_fast(s1_name):

        if name_freq.get(token, 999999999) <= RARITY_LIMIT:

            for eid in name_idx.get(token, []):
                name_hits[eid] += 1

    # Address
    for token in tokens_fast(s1_addr):

        if addr_freq.get(token, 999999999) <= RARITY_LIMIT:

            for eid in addr_idx.get(token, []):
                addr_hits[eid] += 1

    candidate_ids = set()

    for eid, count in name_hits.items():

        if count >= 2:
            candidate_ids.add(eid)

    for eid, count in addr_hits.items():

        if count >= 2:
            candidate_ids.add(eid)

    return candidate_ids


# ============================================================
# START OUTPUT FILES
# ============================================================

# Remove old V5 files if they exist
for path in [RESULT_PATH, CANDIDATE_PATH]:

    if os.path.exists(path):
        os.remove(path)


result_header_written = False
candidate_header_written = False

total_candidates = 0
total_predictions = 0
total_s1 = 0
total_matched_s1 = 0


# ============================================================
# PROCESS S1 IN CHUNKS
# ============================================================

print("\n" + "=" * 75)
print("STARTING FULL V5 INFERENCE")
print("=" * 75)


for start in range(0, len(test_s1), S1_CHUNK):

    end = min(
        start + S1_CHUNK,
        len(test_s1)
    )

    chunk = test_s1.iloc[start:end]

    result_rows = []
    candidate_rows = []

    for _, row in chunk.iterrows():

        s1_id = row["entity_id"]

        name1 = normalize_advanced_local(
            row["business_name"]
        )

        addr1 = normalize_advanced_local(
            row["business_address"]
        )

        country1 = (
            "" if pd.isna(row["country"])
            else str(row["country"]).lower().strip()
        )

        # -----------------------------------------------
        # S2 candidates
        # -----------------------------------------------

        s2_candidates = generate_candidates(
            name1,
            addr1,
            country1,
            s2_name_idx,
            s2_addr_idx,
            s2_name_freq,
            s2_addr_freq
        )

        # -----------------------------------------------
        # S3 candidates
        # -----------------------------------------------

        s3_candidates = generate_candidates(
            name1,
            addr1,
            country1,
            s3_name_idx,
            s3_addr_idx,
            s3_name_freq,
            s3_addr_freq
        )

        # -----------------------------------------------
        # Filter country + exact rules + similarity
        # -----------------------------------------------

        selected_s2 = []
        selected_s3 = []

        # S2
        for eid in s2_candidates:

            target = s2.iloc[eid]

            if target["country_a"] != country1:
                continue

            name2 = target["name_a"]
            addr2 = target["addr_a"]

            # Exact
            if (
                (name1 and name1 == name2)
                or
                (addr1 and addr1 == addr2)
            ):
                selected_s2.append(eid)
                continue

            # Similarity
            ns = seq_sim(name1, name2)
            ads = seq_sim(addr1, addr2)

            if ns >= 0.60 or ads >= 0.60:
                selected_s2.append(eid)

        # S3
        for eid in s3_candidates:

            target = s3.iloc[eid]

            if target["country_a"] != country1:
                continue

            name2 = target["name_a"]
            addr2 = target["addr_a"]

            if (
                (name1 and name1 == name2)
                or
                (addr1 and addr1 == addr2)
            ):
                selected_s3.append(eid)
                continue

            ns = seq_sim(name1, name2)
            ads = seq_sim(addr1, addr2)

            if ns >= 0.60 or ads >= 0.60:
                selected_s3.append(eid)

        all_selected = [
            ("S2", eid) for eid in selected_s2
        ] + [
            ("S3", eid) for eid in selected_s3
        ]

        total_candidates += len(all_selected)

        # -----------------------------------------------
        # Candidate file
        # -----------------------------------------------

        candidate_ids_text = [
            (
                s2.iloc[eid]["entity_id"]
                if source == "S2"
                else s3.iloc[eid]["entity_id"]
            )
            for source, eid in all_selected
        ]

        candidate_rows.append({
            "source1_entity_id": s1_id,
            "candidate_entity_ids": ",".join(
                candidate_ids_text
            )
        })

        # -----------------------------------------------
        # V5 features + prediction
        # -----------------------------------------------

        if all_selected:

            s2_ids = [
                eid for source, eid in all_selected
                if source == "S2"
            ]

            s3_ids = [
                eid for source, eid in all_selected
                if source == "S3"
            ]

            feature_parts = []
            metadata = []

            if s2_ids:

                f2 = make_v5_pair_features(
                    name1,
                    addr1,
                    country1,
                    s2,
                    s2_ids
                )

                feature_parts.append(f2)

                metadata.extend(
                    [("S2", eid) for eid in s2_ids]
                )

            if s3_ids:

                f3 = make_v5_pair_features(
                    name1,
                    addr1,
                    country1,
                    s3,
                    s3_ids
                )

                feature_parts.append(f3)

                metadata.extend(
                    [("S3", eid) for eid in s3_ids]
                )

            X = pd.concat(
                feature_parts,
                ignore_index=True
            )[V5_FEATURES]

            probabilities = v5_model.predict_proba(X)[:, 1]

            predicted_ids = []

            for (source, eid), probability in zip(
                metadata,
                probabilities
            ):

                if probability >= V5_THRESHOLD:

                    entity_id = (
                        s2.iloc[eid]["entity_id"]
                        if source == "S2"
                        else s3.iloc[eid]["entity_id"]
                    )

                    predicted_ids.append(entity_id)

            total_predictions += len(predicted_ids)

        else:
            predicted_ids = []

        # -----------------------------------------------
        # Exactly one result row per S1
        # -----------------------------------------------

        result_rows.append({
            "source1_entity_id": s1_id,
            "matched_entity_ids": ",".join(
                predicted_ids
            )
        })

        total_s1 += 1

        if predicted_ids:
            total_matched_s1 += 1

    # ----------------------------------------------------
    # Write this chunk immediately
    # ----------------------------------------------------

    pd.DataFrame(result_rows).to_csv(
        RESULT_PATH,
        sep="\t",
        index=False,
        mode="a",
        header=not result_header_written
    )

    pd.DataFrame(candidate_rows).to_csv(
        CANDIDATE_PATH,
        sep="\t",
        index=False,
        mode="a",
        header=not candidate_header_written
    )

    result_header_written = True
    candidate_header_written = True

    print(
        f"Processed {end:,}/{len(test_s1):,} "
        f"| candidates: {total_candidates:,} "
        f"| predicted pairs: {total_predictions:,}"
    )


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 75)
print("V5 INFERENCE COMPLETE")
print("=" * 75)

print("S1 processed:", f"{total_s1:,}")
print("S1 with matches:", f"{total_matched_s1:,}")
print("S1 without matches:", f"{total_s1 - total_matched_s1:,}")
print("Candidate pairs:", f"{total_candidates:,}")
print("Predicted pairs:", f"{total_predictions:,}")

print("\nResult file:")
print(RESULT_PATH)

print("\nCandidate file:")
print(CANDIDATE_PATH)

print("\nFiles created successfully.")
```

```python
!pip install rapidfuzz

# ============================================================
# OPTIMIZED V5 BENCHMARK
# ONLY 5,000 S1 ROWS
# ============================================================

import time
import numpy as np
from collections import defaultdict

try:
    from rapidfuzz.fuzz import ratio
    print("RapidFuzz: available")
except ImportError:
    print("RapidFuzz not installed.")
    print("Run: !pip install rapidfuzz")
    raise


# ------------------------------------------------------------
# Convert source columns to NumPy arrays
# ------------------------------------------------------------

print("Preparing fast NumPy arrays...")

s2_names = s2["name_n"].fillna("").to_numpy()
s2_addrs = s2["addr_n"].fillna("").to_numpy()
s2_countries = s2["country_n"].fillna("").to_numpy()

s3_names = s3["name_n"].fillna("").to_numpy()
s3_addrs = s3["addr_n"].fillna("").to_numpy()
s3_countries = s3["country_n"].fillna("").to_numpy()

s2_ids = s2["entity_id"].to_numpy()
s3_ids = s3["entity_id"].to_numpy()

print("Arrays ready.")


# ------------------------------------------------------------
# Fast candidate generation
# ------------------------------------------------------------

def fast_candidates(
    name1,
    addr1,
    country1,
    name_index,
    addr_index,
    name_freq,
    addr_freq,
    names,
    addrs,
    countries
):

    name_hits = defaultdict(int)
    addr_hits = defaultdict(int)

    # Name tokens
    for token in name1.split():

        if name_freq.get(token, 999999999) <= 500:

            for eid in name_index.get(token, ()):
                name_hits[eid] += 1

    # Address tokens
    for token in addr1.split():

        if addr_freq.get(token, 999999999) <= 500:

            for eid in addr_index.get(token, ()):
                addr_hits[eid] += 1

    candidates = set()

    # >=2 name tokens
    for eid, count in name_hits.items():

        if count >= 2:
            if countries[eid] == country1:
                candidates.add(eid)

    # >=2 address tokens
    for eid, count in addr_hits.items():

        if count >= 2:
            if countries[eid] == country1:
                candidates.add(eid)

    return candidates, name_hits, addr_hits


# ------------------------------------------------------------
# Benchmark
# ------------------------------------------------------------

BENCHMARK_N = 5000

start_time = time.time()

candidate_total = 0
selected_total = 0

for i in range(BENCHMARK_N):

    if i % 500 == 0:
        elapsed = time.time() - start_time
        print(
            f"{i}/{BENCHMARK_N} "
            f"| elapsed: {elapsed:.1f}s"
        )

    row = test_s1.iloc[i]

    name1 = normalize_advanced_local(row["business_name"])
    addr1 = normalize_advanced_local(row["business_address"])

    country1 = (
        "" if pd.isna(row["country"])
        else str(row["country"]).lower().strip()
    )

    # --------------------------------------------------------
    # S2
    # --------------------------------------------------------

    candidates2, name_hits2, addr_hits2 = fast_candidates(
        name1,
        addr1,
        country1,
        s2_name_index,
        s2_addr_index,
        s2_name_freq,
        s2_addr_freq,
        s2_names,
        s2_addrs,
        s2_countries
    )

    candidate_total += len(candidates2)

    for eid in candidates2:

        # Exact match
        if (
            name1 == s2_names[eid]
            or addr1 == s2_addrs[eid]
        ):
            selected_total += 1
            continue

        # Name similarity
        if name_hits2.get(eid, 0) >= 2:

            if ratio(name1, s2_names[eid]) >= 60:
                selected_total += 1
                continue

        # Address similarity
        if addr_hits2.get(eid, 0) >= 2:

            if ratio(addr1, s2_addrs[eid]) >= 60:
                selected_total += 1
                continue

    # --------------------------------------------------------
    # S3
    # --------------------------------------------------------

    candidates3, name_hits3, addr_hits3 = fast_candidates(
        name1,
        addr1,
        country1,
        s3_name_index,
        s3_addr_index,
        s3_name_freq,
        s3_addr_freq,
        s3_names,
        s3_addrs,
        s3_countries
    )

    candidate_total += len(candidates3)

    for eid in candidates3:

        if (
            name1 == s3_names[eid]
            or addr1 == s3_addrs[eid]
        ):
            selected_total += 1
            continue

        if name_hits3.get(eid, 0) >= 2:

            if ratio(name1, s3_names[eid]) >= 60:
                selected_total += 1
                continue

        if addr_hits3.get(eid, 0) >= 2:

            if ratio(addr1, s3_addrs[eid]) >= 60:
                selected_total += 1
                continue


elapsed = time.time() - start_time

print("\n" + "=" * 70)
print("OPTIMIZED BENCHMARK COMPLETE")
print("=" * 70)

print("S1 tested:", BENCHMARK_N)
print("Raw candidates:", f"{candidate_total:,}")
print("Selected candidates:", f"{selected_total:,}")

print("Time:", round(elapsed, 2), "seconds")
print("Time:", round(elapsed / 60, 2), "minutes")

print(
    "Estimated full runtime at same speed:",
    round(elapsed * 1_732_544 / BENCHMARK_N / 3600, 2),
    "hours"
)

print(
    "Average selected candidates/S1:",
    round(selected_total / BENCHMARK_N, 2)
)
```

```python
# ============================================================
# COMPLETE V5 BENCHMARK
# 5,000 S1 rows
# Candidate generation + 20 V5 features + CatBoost
# ============================================================

import time
import re
import numpy as np
import pandas as pd
from collections import defaultdict
from rapidfuzz.fuzz import ratio

BENCHMARK_N = 5000
THRESHOLD = 0.60

numbers_re = re.compile(r"\d+")

def number_set(x):
    if not x:
        return set()
    return set(numbers_re.findall(x))


def build_fast_features(
    s1_name,
    s1_addr,
    s1_country,
    source_names,
    source_addrs,
    source_countries,
    candidate_ids
):
    """
    Build the exact 20 V5 features.
    """

    n1 = set(s1_name.split())
    a1 = set(s1_addr.split())
    nums1 = number_set(s1_addr)

    rows = []

    for eid in candidate_ids:

        n2 = source_names[eid]
        a2 = source_addrs[eid]
        c2 = source_countries[eid]

        n2_tokens = set(n2.split())
        a2_tokens = set(a2.split())

        shared_n = n1 & n2_tokens
        shared_a = a1 & a2_tokens
        shared_nums = nums1 & number_set(a2)

        ns = ratio(s1_name, n2) / 100.0
        ads = ratio(s1_addr, a2) / 100.0

        n_overlap = (
            len(shared_n) /
            max(1, min(len(n1), len(n2_tokens)))
        )

        a_overlap = (
            len(shared_a) /
            max(1, min(len(a1), len(a2_tokens)))
        )

        n_union = n1 | n2_tokens
        a_union = a1 | a2_tokens

        n_jaccard = len(shared_n) / max(1, len(n_union))
        a_jaccard = len(shared_a) / max(1, len(a_union))

        n_len_ratio = (
            min(len(s1_name), len(n2)) /
            max(1, max(len(s1_name), len(n2)))
        )

        a_len_ratio = (
            min(len(s1_addr), len(a2)) /
            max(1, max(len(s1_addr), len(a2)))
        )

        rows.append([
            int(bool(s1_name) and s1_name == n2),
            int(bool(s1_addr) and s1_addr == a2),
            int(s1_country == c2),

            ns,
            ads,

            n_overlap,
            a_overlap,

            n_jaccard,
            a_jaccard,

            len(shared_n),
            len(shared_a),
            len(shared_nums),

            abs(len(s1_name) - len(n2)),
            abs(len(s1_addr) - len(a2)),

            n_len_ratio,
            a_len_ratio,

            abs(ns - ads),
            max(ns, ads),
            ns * ads,
            abs(n_overlap - a_overlap)
        ])

    return np.asarray(rows, dtype=np.float32)


# ============================================================
# SOURCE ARRAYS
# ============================================================

print("Preparing arrays...")

s2_names_a = s2["name_a"].fillna("").to_numpy()
s2_addrs_a = s2["addr_a"].fillna("").to_numpy()
s2_countries_a = s2["country_a"].fillna("").to_numpy()

s3_names_a = s3["name_a"].fillna("").to_numpy()
s3_addrs_a = s3["addr_a"].fillna("").to_numpy()
s3_countries_a = s3["country_a"].fillna("").to_numpy()

print("Arrays ready.")


# ============================================================
# CANDIDATE GENERATION
# ============================================================

def get_fast_candidates(
    name1,
    addr1,
    country1,
    name_index,
    addr_index,
    name_freq,
    addr_freq,
    names,
    addrs,
    countries
):

    name_hits = defaultdict(int)
    addr_hits = defaultdict(int)

    for token in name1.split():

        if name_freq.get(token, 999999999) <= 500:

            for eid in name_index.get(token, ()):

                if countries[eid] == country1:
                    name_hits[eid] += 1

    for token in addr1.split():

        if addr_freq.get(token, 999999999) <= 500:

            for eid in addr_index.get(token, ()):

                if countries[eid] == country1:
                    addr_hits[eid] += 1

    candidates = set()

    for eid, count in name_hits.items():

        if count >= 2:
            candidates.add(eid)

    for eid, count in addr_hits.items():

        if count >= 2:
            candidates.add(eid)

    return candidates


# ============================================================
# COMPLETE BENCHMARK
# ============================================================

print("\nStarting complete V5 benchmark...")

start_time = time.time()

all_features = []
all_metadata = []

raw_candidates = 0
selected_candidates = 0

for i in range(BENCHMARK_N):

    if i % 500 == 0:

        elapsed = time.time() - start_time

        print(
            f"{i}/{BENCHMARK_N} "
            f"| elapsed {elapsed:.1f}s "
            f"| selected {selected_candidates:,}"
        )

    row = test_s1.iloc[i]

    name1 = normalize_advanced_local(
        row["business_name"]
    )

    addr1 = normalize_advanced_local(
        row["business_address"]
    )

    country1 = (
        ""
        if pd.isna(row["country"])
        else str(row["country"]).lower().strip()
    )

    # --------------------------------------------------------
    # S2
    # --------------------------------------------------------

    candidates2 = get_fast_candidates(
        name1,
        addr1,
        country1,
        s2_name_index,
        s2_addr_index,
        s2_name_freq,
        s2_addr_freq,
        s2_names_a,
        s2_addrs_a,
        s2_countries_a
    )

    raw_candidates += len(candidates2)

    selected2 = []

    for eid in candidates2:

        n2 = s2_names_a[eid]
        a2 = s2_addrs_a[eid]

        if (
            (name1 and name1 == n2)
            or
            (addr1 and addr1 == a2)
        ):
            selected2.append(eid)
            continue

        if ratio(name1, n2) >= 60:
            selected2.append(eid)
            continue

        if ratio(addr1, a2) >= 60:
            selected2.append(eid)
            continue

    # --------------------------------------------------------
    # S3
    # --------------------------------------------------------

    candidates3 = get_fast_candidates(
        name1,
        addr1,
        country1,
        s3_name_index,
        s3_addr_index,
        s3_name_freq,
        s3_addr_freq,
        s3_names_a,
        s3_addrs_a,
        s3_countries_a
    )

    raw_candidates += len(candidates3)

    selected3 = []

    for eid in candidates3:

        n2 = s3_names_a[eid]
        a2 = s3_addrs_a[eid]

        if (
            (name1 and name1 == n2)
            or
            (addr1 and addr1 == a2)
        ):
            selected3.append(eid)
            continue

        if ratio(name1, n2) >= 60:
            selected3.append(eid)
            continue

        if ratio(addr1, a2) >= 60:
            selected3.append(eid)
            continue

    selected_candidates += len(selected2) + len(selected3)

    # --------------------------------------------------------
    # Build V5 features
    # --------------------------------------------------------

    if selected2:

        X2 = build_fast_features(
            name1,
            addr1,
            country1,
            s2_names_a,
            s2_addrs_a,
            s2_countries_a,
            selected2
        )

        all_features.append(X2)

        all_metadata.extend(
            [("S2", eid) for eid in selected2]
        )

    if selected3:

        X3 = build_fast_features(
            name1,
            addr1,
            country1,
            s3_names_a,
            s3_addrs_a,
            s3_countries_a,
            selected3
        )

        all_features.append(X3)

        all_metadata.extend(
            [("S3", eid) for eid in selected3]
        )


# ============================================================
# BATCH V5 PREDICTION
# ============================================================

print("\nBuilding prediction matrix...")

X_benchmark = np.vstack(all_features)

print("Feature matrix:", X_benchmark.shape)

prediction_start = time.time()

probabilities = v5_model.predict_proba(
    X_benchmark
)[:, 1]

predicted = probabilities >= THRESHOLD

prediction_time = time.time() - prediction_start

total_time = time.time() - start_time


# ============================================================
# RESULTS
# ============================================================

print("\n" + "=" * 75)
print("COMPLETE V5 BENCHMARK RESULTS")
print("=" * 75)

print("S1 tested:", f"{BENCHMARK_N:,}")

print(
    "Raw candidates:",
    f"{raw_candidates:,}"
)

print(
    "Selected candidates:",
    f"{selected_candidates:,}"
)

print(
    "V5 feature matrix:",
    X_benchmark.shape
)

print(
    "Predicted matches:",
    f"{predicted.sum():,}"
)

print(
    "CatBoost prediction time:",
    round(prediction_time, 2),
    "seconds"
)

print(
    "Total benchmark time:",
    round(total_time, 2),
    "seconds"
)

print(
    "Total benchmark time:",
    round(total_time / 60, 2),
    "minutes"
)

estimated_hours = (
    total_time *
    1_732_544 /
    BENCHMARK_N /
    3600
)

print(
    "Estimated full runtime:",
    round(estimated_hours, 2),
    "hours"
)

print(
    "Average selected candidates/S1:",
    round(
        selected_candidates / BENCHMARK_N,
        2
    )
)

print(
    "Average predicted matches/S1:",
    round(
        predicted.sum() / BENCHMARK_N,
        3
    )
)
```

```python
# ============================================================
# HIGH-RECALL V5 BENCHMARK
# 5,000 S1 rows
# ============================================================

import time
import numpy as np
from rapidfuzz.fuzz import ratio
from collections import defaultdict

BENCHMARK_N = 5000
THRESHOLD = 0.60


def high_recall_candidates(
    name1,
    addr1,
    country1,
    name_index,
    addr_index,
    name_freq,
    addr_freq,
    names,
    addrs,
    countries
):
    name_hits = defaultdict(int)
    addr_hits = defaultdict(int)

    # --------------------------------------------------------
    # NAME TOKEN BLOCK
    # --------------------------------------------------------

    for token in name1.split():

        if name_freq.get(token, 999999999) <= 500:

            for eid in name_index.get(token, ()):

                if countries[eid] == country1:
                    name_hits[eid] += 1

    # --------------------------------------------------------
    # ADDRESS TOKEN BLOCK
    # --------------------------------------------------------

    for token in addr1.split():

        if addr_freq.get(token, 999999999) <= 500:

            for eid in addr_index.get(token, ()):

                if countries[eid] == country1:
                    addr_hits[eid] += 1

    # We need >=2 name OR >=2 address tokens
    candidate_ids = (
        {eid for eid, c in name_hits.items() if c >= 2}
        |
        {eid for eid, c in addr_hits.items() if c >= 2}
    )

    return candidate_ids, name_hits, addr_hits


print("Starting high-recall benchmark...")

start_time = time.time()

raw_candidates = 0
selected_candidates = 0
predicted_matches = 0

all_features = []

for i in range(BENCHMARK_N):

    if i % 500 == 0:

        elapsed = time.time() - start_time

        print(
            f"{i}/{BENCHMARK_N} "
            f"| elapsed {elapsed:.1f}s "
            f"| selected {selected_candidates:,}"
        )

    row = test_s1.iloc[i]

    name1 = normalize_advanced_local(
        row["business_name"]
    )

    addr1 = normalize_advanced_local(
        row["business_address"]
    )

    country1 = (
        ""
        if pd.isna(row["country"])
        else str(row["country"]).lower().strip()
    )

    # ========================================================
    # S2
    # ========================================================

    c2, nh2, ah2 = high_recall_candidates(
        name1,
        addr1,
        country1,
        s2_name_index,
        s2_addr_index,
        s2_name_freq,
        s2_addr_freq,
        s2_names_a,
        s2_addrs_a,
        s2_countries_a
    )

    raw_candidates += len(c2)

    selected2 = []

    for eid in c2:

        n2 = s2_names_a[eid]
        a2 = s2_addrs_a[eid]

        # Exact
        if (
            (name1 and name1 == n2)
            or
            (addr1 and addr1 == a2)
        ):
            selected2.append(eid)
            continue

        ns = ratio(name1, n2) / 100.0
        ads = ratio(addr1, a2) / 100.0

        # Name rule
        if (
            nh2.get(eid, 0) >= 2
            and ns >= 0.60
        ):
            selected2.append(eid)
            continue

        # Address >=3 tokens
        if ah2.get(eid, 0) >= 3:
            selected2.append(eid)
            continue

        # Address fallback
        if (
            ah2.get(eid, 0) >= 2
            and ads >= 0.30
        ):
            selected2.append(eid)
            continue

    # ========================================================
    # S3
    # ========================================================

    c3, nh3, ah3 = high_recall_candidates(
        name1,
        addr1,
        country1,
        s3_name_index,
        s3_addr_index,
        s3_name_freq,
        s3_addr_freq,
        s3_names_a,
        s3_addrs_a,
        s3_countries_a
    )

    raw_candidates += len(c3)

    selected3 = []

    for eid in c3:

        n2 = s3_names_a[eid]
        a2 = s3_addrs_a[eid]

        if (
            (name1 and name1 == n2)
            or
            (addr1 and addr1 == a2)
        ):
            selected3.append(eid)
            continue

        ns = ratio(name1, n2) / 100.0
        ads = ratio(addr1, a2) / 100.0

        if (
            nh3.get(eid, 0) >= 2
            and ns >= 0.60
        ):
            selected3.append(eid)
            continue

        if ah3.get(eid, 0) >= 3:
            selected3.append(eid)
            continue

        if (
            ah3.get(eid, 0) >= 2
            and ads >= 0.30
        ):
            selected3.append(eid)
            continue

    selected_candidates += len(selected2) + len(selected3)

    # ========================================================
    # V5 FEATURES
    # ========================================================

    if selected2:

        X2 = build_fast_features(
            name1,
            addr1,
            country1,
            s2_names_a,
            s2_addrs_a,
            s2_countries_a,
            selected2
        )

        all_features.append(X2)

    if selected3:

        X3 = build_fast_features(
            name1,
            addr1,
            country1,
            s3_names_a,
            s3_addrs_a,
            s3_countries_a,
            selected3
        )

        all_features.append(X3)


# ============================================================
# V5 PREDICTION
# ============================================================

print("\nBuilding feature matrix...")

X_high = np.vstack(all_features)

print("Feature matrix:", X_high.shape)

pred_start = time.time()

probs = v5_model.predict_proba(X_high)[:, 1]

predicted_matches = int(
    (probs >= THRESHOLD).sum()
)

pred_time = time.time() - pred_start
total_time = time.time() - start_time


# ============================================================
# RESULTS
# ============================================================

print("\n" + "=" * 75)
print("HIGH-RECALL BENCHMARK RESULTS")
print("=" * 75)

print("S1 tested:", f"{BENCHMARK_N:,}")
print("Raw candidates:", f"{raw_candidates:,}")
print("Selected candidates:", f"{selected_candidates:,}")
print("Feature matrix:", X_high.shape)
print("Predicted matches:", f"{predicted_matches:,}")

print(
    "Total time:",
    round(total_time, 2),
    "seconds"
)

print(
    "Estimated full runtime:",
    round(
        total_time * 1_732_544 / BENCHMARK_N / 3600,
        2
    ),
    "hours"
)

print(
    "Average selected candidates/S1:",
    round(
        selected_candidates / BENCHMARK_N,
        2
    )
)

print(
    "Average predicted matches/S1:",
    round(
        predicted_matches / BENCHMARK_N,
        3
    )
)
```

```python
# ============================================================
# AMAZON ML CHALLENGE 2026 — V5 FINAL INFERENCE
# High-recall candidate generation + V5 CatBoost
# ============================================================

import os
import time
import gc
import ast
import pandas as pd
import numpy as np
from rapidfuzz.fuzz import ratio
from catboost import CatBoostClassifier

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

ROOT = "/content/amazon_ml_challenge/student_resource"
TEST_DIR = f"{ROOT}/dataset/test"
OUTPUT_DIR = f"{ROOT}/output"

os.makedirs(OUTPUT_DIR, exist_ok=True)

S1_PATH = f"{TEST_DIR}/test_source1.tsv"
S2_PATH = f"{TEST_DIR}/test_source2.tsv"
S3_PATH = f"{TEST_DIR}/test_source3.tsv"

MODEL_PATH = (
    "/content/drive/My Drive/Amazon-ML-Challenge-2026/ml/"
    "V5_ML_Final/V5_ML_Final/v5_model.cbm"
)

INDEX_DIR = (
    "/content/drive/My Drive/Amazon-ML-Challenge-2026/ml/"
    "V5_full_indexes"
)

MATCH_OUT = f"{OUTPUT_DIR}/matching_results_v5.tsv"
CAND_OUT = f"{OUTPUT_DIR}/candidate_pairs_v5.tsv"

THRESHOLD = 0.60
CHUNK_SIZE = 5000

# ------------------------------------------------------------
# CHECK REQUIRED OBJECTS
# ------------------------------------------------------------

required = [
    "s2_names_a",
    "s2_addrs_a",
    "s2_countries_a",
    "s2_ids",
    "s3_names_a",
    "s3_addrs_a",
    "s3_countries_a",
    "s3_ids",
    "v5_model",
]

missing = [x for x in required if x not in globals()]

if missing:
    raise RuntimeError(
        "Missing required objects: " + ", ".join(missing)
    )

print("All required objects are available.")
print("V5 model:", MODEL_PATH)
print("Threshold:", THRESHOLD)
print("Chunk size:", CHUNK_SIZE)

# ------------------------------------------------------------
# VERIFY MODEL
# ------------------------------------------------------------

if not os.path.exists(MODEL_PATH):
    raise FileNotFoundError(MODEL_PATH)

v5_model = CatBoostClassifier()
v5_model.load_model(MODEL_PATH)

print("V5 model loaded successfully.")

# ------------------------------------------------------------
# LOAD TEST SOURCE 1
# ------------------------------------------------------------

test_s1 = pd.read_csv(S1_PATH, sep="\t")

print("\nTest S1:", test_s1.shape)

# ------------------------------------------------------------
# FAST NORMALIZATION
# ------------------------------------------------------------

def normalize_advanced_local(text):
    if pd.isna(text):
        return ""

    text = str(text).lower().strip()

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    import re
    import unicodedata

    text = unicodedata.normalize("NFKC", text)
    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()


def tokens_fast(text):
    if not text:
        return set()
    return set(text.split())


# ------------------------------------------------------------
# LOAD SAVED INDEXES
# ------------------------------------------------------------

import pickle

def load_pickle(name):
    path = os.path.join(INDEX_DIR, name)
    print("Loading:", name)
    with open(path, "rb") as f:
        return pickle.load(f)


s2_name_index = load_pickle("s2_name_index.pkl")
s2_addr_index = load_pickle("s2_addr_index.pkl")
s2_name_freq = load_pickle("s2_name_freq.pkl")
s2_addr_freq = load_pickle("s2_addr_freq.pkl")

s3_name_index = load_pickle("s3_name_index.pkl")
s3_addr_index = load_pickle("s3_addr_index.pkl")
s3_name_freq = load_pickle("s3_name_freq.pkl")
s3_addr_freq = load_pickle("s3_addr_freq.pkl")

print("All indexes loaded.")

# ------------------------------------------------------------
# V5 FEATURES
# ------------------------------------------------------------

V5_FEATURES = [
    "name_exact",
    "address_exact",
    "country_exact",
    "name_similarity",
    "address_similarity",
    "name_token_overlap",
    "address_token_overlap",
    "name_jaccard",
    "address_jaccard",
    "shared_name_token_count",
    "shared_address_token_count",
    "shared_address_number_count",
    "name_length_diff",
    "address_length_diff",
    "name_length_ratio",
    "address_length_ratio",
    "similarity_gap",
    "max_similarity",
    "similarity_product",
    "token_overlap_gap"
]


def extract_numbers_fast(text):
    import re
    return set(re.findall(r"\d+", text))


def build_fast_features(s1_name, s1_addr, s1_country,
                        names, addrs, countries):

    n1 = s1_name
    a1 = s1_addr
    c1 = s1_country

    t1n = tokens_fast(n1)
    t1a = tokens_fast(a1)

    nums1 = extract_numbers_fast(a1)

    rows = []

    for n2, a2, c2 in zip(names, addrs, countries):

        t2n = tokens_fast(n2)
        t2a = tokens_fast(a2)

        shared_n = t1n & t2n
        shared_a = t1a & t2a

        union_n = t1n | t2n
        union_a = t1a | t2a

        ns = ratio(n1, n2) / 100 if n1 and n2 else 0.0
        ass = ratio(a1, a2) / 100 if a1 and a2 else 0.0

        nt = (
            len(shared_n) / max(1, min(len(t1n), len(t2n)))
        )

        at = (
            len(shared_a) / max(1, min(len(t1a), len(t2a)))
        )

        nj = (
            len(shared_n) / len(union_n)
            if union_n else 0.0
        )

        aj = (
            len(shared_a) / len(union_a)
            if union_a else 0.0
        )

        nums2 = extract_numbers_fast(a2)

        rows.append({
            "name_exact": int(bool(n1) and n1 == n2),
            "address_exact": int(bool(a1) and a1 == a2),
            "country_exact": int(c1 == c2),

            "name_similarity": ns,
            "address_similarity": ass,

            "name_token_overlap": nt,
            "address_token_overlap": at,

            "name_jaccard": nj,
            "address_jaccard": aj,

            "shared_name_token_count": len(shared_n),
            "shared_address_token_count": len(shared_a),
            "shared_address_number_count": len(nums1 & nums2),

            "name_length_diff": abs(len(n1) - len(n2)),
            "address_length_diff": abs(len(a1) - len(a2)),

            "name_length_ratio":
                min(len(n1), len(n2)) /
                max(1, max(len(n1), len(n2))),

            "address_length_ratio":
                min(len(a1), len(a2)) /
                max(1, max(len(a1), len(a2))),

            "similarity_gap": abs(ns - ass),
            "max_similarity": max(ns, ass),
            "similarity_product": ns * ass,
            "token_overlap_gap": abs(nt - at)
        })

    return pd.DataFrame(rows, columns=V5_FEATURES)


# ------------------------------------------------------------
# HIGH-RECALL CANDIDATE GENERATOR
# ------------------------------------------------------------

def high_recall_candidates(
    s1_name,
    s1_addr,
    s1_country,
    name_index,
    addr_index,
    name_freq,
    addr_freq,
    ids
):

    t_name = tokens_fast(s1_name)
    t_addr = tokens_fast(s1_addr)

    candidate_ids = set()

    # Exact name
    if s1_name:
        candidate_ids.update(
            name_index.get(s1_name, [])
        )

    # Exact address
    if s1_addr:
        candidate_ids.update(
            addr_index.get(s1_addr, [])
        )

    # Name token candidates
    name_token_hits = {}

    for tok in t_name:

        # Keep the benchmark's <=500 frequency rule
        if name_freq.get(tok, 10**9) <= 500:

            for idx in name_index.get(tok, []):
                name_token_hits[idx] = (
                    name_token_hits.get(idx, 0) + 1
                )

    for idx, count in name_token_hits.items():

        if count >= 2:
            candidate_ids.add(idx)

    # Address token candidates
    addr_token_hits = {}

    for tok in t_addr:

        if addr_freq.get(tok, 10**9) <= 500:

            for idx in addr_index.get(tok, []):

                addr_token_hits[idx] = (
                    addr_token_hits.get(idx, 0) + 1
                )

    for idx, count in addr_token_hits.items():

        if count >= 2:
            candidate_ids.add(idx)

    # --------------------------------------------------------
    # Similarity filtering
    # --------------------------------------------------------

    selected = []

    for idx in candidate_ids:

        n2 = names_global[idx]
        a2 = addrs_global[idx]

        name_shared = len(
            t_name & tokens_fast(n2)
        )

        addr_shared = len(
            t_addr & tokens_fast(a2)
        )

        name_sim = (
            ratio(s1_name, n2) / 100
            if s1_name and n2 else 0
        )

        addr_sim = (
            ratio(s1_addr, a2) / 100
            if s1_addr and a2 else 0
        )

        keep = (
            n2 == s1_name
            or a2 == s1_addr
            or (
                name_shared >= 2
                and name_sim >= 0.60
            )
            or (
                addr_shared >= 3
            )
            or (
                addr_shared >= 2
                and addr_sim >= 0.30
            )
        )

        if keep:
            selected.append(idx)

    return selected


# ------------------------------------------------------------
# PREPARE GLOBAL SOURCE ARRAYS
# ------------------------------------------------------------

# S2 globals
names_global = s2_names_a
addrs_global = s2_addrs_a
countries_global = s2_countries_a
ids_global = s2_ids

# ------------------------------------------------------------
# IMPORTANT:
# Candidate indexes contain integer row positions.
# We run S2 and S3 separately.
# ------------------------------------------------------------

def run_source(
    s1_rows,
    names,
    addrs,
    countries,
    ids,
    name_index,
    addr_index,
    name_freq,
    addr_freq,
    source_name
):

    all_match_rows = []
    all_candidate_rows = []

    total_selected = 0
    total_predictions = 0

    start = time.time()

    for local_i, (_, row) in enumerate(s1_rows.iterrows()):

        s1_id = row["source1_entity_id"]

        s1_name = normalize_advanced_local(
            row["business_name"]
        )

        s1_addr = normalize_advanced_local(
            row["business_address"]
        )

        s1_country = normalize_advanced_local(
            row["country"]
        )

        # -----------------------------------------------
        # Candidate generation
        # -----------------------------------------------

        candidate_idx = high_recall_candidates(
            s1_name,
            s1_addr,
            s1_country,
            name_index,
            addr_index,
            name_freq,
            addr_freq,
            ids
        )

        total_selected += len(candidate_idx)

        if not candidate_idx:
            all_match_rows.append({
                "source1_entity_id": s1_id,
                "matched_entity_ids": ""
            })

            all_candidate_rows.append({
                "source1_entity_id": s1_id,
                "candidate_entity_ids": ""
            })

            continue

        # -----------------------------------------------
        # Candidate IDs
        # -----------------------------------------------

        candidate_ids = [
            ids[i] for i in candidate_idx
        ]

        # -----------------------------------------------
        # Feature matrix
        # -----------------------------------------------

        cand_names = [
            names[i] for i in candidate_idx
        ]

        cand_addrs = [
            addrs[i] for i in candidate_idx
        ]

        cand_countries = [
            countries[i] for i in candidate_idx
        ]

        X = build_fast_features(
            s1_name,
            s1_addr,
            s1_country,
            cand_names,
            cand_addrs,
            cand_countries
        )

        # -----------------------------------------------
        # V5 prediction
        # -----------------------------------------------

        probs = v5_model.predict_proba(
            X[V5_FEATURES]
        )[:, 1]

        matched = [
            str(candidate_ids[j])
            for j, p in enumerate(probs)
            if p >= THRESHOLD
        ]

        total_predictions += len(matched)

        all_match_rows.append({
            "source1_entity_id": s1_id,
            "matched_entity_ids": ",".join(matched)
        })

        all_candidate_rows.append({
            "source1_entity_id": s1_id,
            "candidate_entity_ids": ",".join(
                map(str, candidate_ids)
            )
        })

        # -----------------------------------------------
        # Progress
        # -----------------------------------------------

        if (local_i + 1) % 500 == 0:

            elapsed = time.time() - start

            print(
                f"{source_name}: "
                f"{local_i + 1:,}/{len(s1_rows):,} | "
                f"elapsed {elapsed/60:.1f} min | "
                f"candidates {total_selected:,} | "
                f"matches {total_predictions:,}"
            )

    return (
        all_match_rows,
        all_candidate_rows,
        total_selected,
        total_predictions
    )


# ------------------------------------------------------------
# NOTE:
# The indexes are built separately for S2 and S3.
# Therefore we need a generic source runner.
# ------------------------------------------------------------

def process_s1_chunk(chunk):

    match_rows = []
    candidate_rows = []

    selected_total = 0
    prediction_total = 0

    # --------------------------------------------
    # S2
    # --------------------------------------------

    global names_global, addrs_global, countries_global

    names_global = s2_names_a
    addrs_global = s2_addrs_a
    countries_global = s2_countries_a

    s2_matches, s2_candidates, s2_sel, s2_pred = run_source(
        chunk,
        s2_names_a,
        s2_addrs_a,
        s2_countries_a,
        s2_ids,
        s2_name_index,
        s2_addr_index,
        s2_name_freq,
        s2_addr_freq,
        "S2"
    )

    # --------------------------------------------
    # S3
    # --------------------------------------------

    names_global = s3_names_a
    addrs_global = s3_addrs_a
    countries_global = s3_countries_a

    s3_matches, s3_candidates, s3_sel, s3_pred = run_source(
        chunk,
        s3_names_a,
        s3_addrs_a,
        s3_countries_a,
        s3_ids,
        s3_name_index,
        s3_addr_index,
        s3_name_freq,
        s3_addr_freq,
        "S3"
    )

    # --------------------------------------------
    # Merge S2 + S3 results by S1
    # --------------------------------------------

    s2_match_dict = {
        str(x["source1_entity_id"]): x["matched_entity_ids"]
        for x in s2_matches
    }

    s3_match_dict = {
        str(x["source1_entity_id"]): x["matched_entity_ids"]
        for x in s3_matches
    }

    s2_cand_dict = {
        str(x["source1_entity_id"]): x["candidate_entity_ids"]
        for x in s2_candidates
    }

    s3_cand_dict = {
        str(x["source1_entity_id"]): x["candidate_entity_ids"]
        for x in s3_candidates
    }

    for _, row in chunk.iterrows():

        sid = str(row["source1_entity_id"])

        m2 = s2_match_dict.get(sid, "")
        m3 = s3_match_dict.get(sid, "")

        c2 = s2_cand_dict.get(sid, "")
        c3 = s3_cand_dict.get(sid, "")

        matches = []

        if m2:
            matches.extend(m2.split(","))

        if m3:
            matches.extend(m3.split(","))

        candidates = []

        if c2:
            candidates.extend(c2.split(","))

        if c3:
            candidates.extend(c3.split(","))

        match_rows.append({
            "source1_entity_id": row["source1_entity_id"],
            "matched_entity_ids": ",".join(matches)
        })

        candidate_rows.append({
            "source1_entity_id": row["source1_entity_id"],
            "candidate_entity_ids": ",".join(candidates)
        })

    return (
        match_rows,
        candidate_rows,
        s2_sel + s3_sel,
        s2_pred + s3_pred
    )


# ============================================================
# FULL INFERENCE
# ============================================================

print("\n" + "=" * 70)
print("STARTING FULL V5 INFERENCE")
print("=" * 70)

# Remove old files if they exist
for path in [MATCH_OUT, CAND_OUT]:
    if os.path.exists(path):
        os.remove(path)

total_start = time.time()

all_matches = []
all_candidates = []

total_selected = 0
total_predictions = 0

n = len(test_s1)

for start_idx in range(0, n, CHUNK_SIZE):

    end_idx = min(start_idx + CHUNK_SIZE, n)

    chunk = test_s1.iloc[start_idx:end_idx].copy()

    chunk_start = time.time()

    matches, candidates, selected, predictions = process_s1_chunk(
        chunk
    )

    all_matches.extend(matches)
    all_candidates.extend(candidates)

    total_selected += selected
    total_predictions += predictions

    elapsed = time.time() - total_start
    rate = (end_idx / elapsed) if elapsed > 0 else 0

    remaining = (
        (n - end_idx) / rate
        if rate > 0 else 0
    )

    print(
        f"\n>>> GLOBAL: "
        f"{end_idx:,}/{n:,} "
        f"({100*end_idx/n:.2f}%) | "
        f"elapsed {elapsed/60:.1f} min | "
        f"ETA {remaining/60:.1f} min | "
        f"selected {total_selected:,} | "
        f"matches {total_predictions:,}"
    )

    # Save checkpoint every chunk
    pd.DataFrame(
        all_matches,
        columns=[
            "source1_entity_id",
            "matched_entity_ids"
        ]
    ).to_csv(
        MATCH_OUT,
        sep="\t",
        index=False
    )

    pd.DataFrame(
        all_candidates,
        columns=[
            "source1_entity_id",
            "candidate_entity_ids"
        ]
    ).to_csv(
        CAND_OUT,
        sep="\t",
        index=False
    )

    del chunk, matches, candidates
    gc.collect()


# ============================================================
# FINAL WRITE
# ============================================================

match_df = pd.DataFrame(
    all_matches,
    columns=[
        "source1_entity_id",
        "matched_entity_ids"
    ]
)

candidate_df = pd.DataFrame(
    all_candidates,
    columns=[
        "source1_entity_id",
        "candidate_entity_ids"
    ]
)

match_df.to_csv(
    MATCH_OUT,
    sep="\t",
    index=False
)

candidate_df.to_csv(
    CAND_OUT,
    sep="\t",
    index=False
)

total_time = time.time() - total_start

print("\n" + "=" * 70)
print("V5 FULL INFERENCE COMPLETE")
print("=" * 70)

print(f"S1 rows:              {len(match_df):,}")
print(f"Unique S1 IDs:        {match_df['source1_entity_id'].nunique():,}")
print(f"Selected candidates:  {total_selected:,}")
print(f"Predicted matches:    {total_predictions:,}")
print(f"Average candidates:   {total_selected/n:.2f}")
print(f"Average matches:      {total_predictions/n:.4f}")
print(f"Runtime:              {total_time/60:.2f} minutes")

print("\nOutput files:")
print(MATCH_OUT)
print(CAND_OUT)

print("\nFile sizes:")
print(
    f"matching_results_v5.tsv: "
    f"{os.path.getsize(MATCH_OUT)/1024/1024:.2f} MB"
)

print(
    f"candidate_pairs_v5.tsv: "
    f"{os.path.getsize(CAND_OUT)/1024/1024:.2f} MB"
)
```

```python
# ============================================================
# V5 FINAL — CHECKPOINTED, RAM-SAFE INFERENCE
# ============================================================
import unicodedata
import os
import gc
import re
import time
import pickle
import numpy as np
import pandas as pd

from rapidfuzz.fuzz import ratio

# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------

ROOT = "/content/amazon_ml_challenge/student_resource"
TEST_DIR = f"{ROOT}/dataset/test"

S1_PATH = f"{TEST_DIR}/test_source1.tsv"

DRIVE_OUT = (
    "/content/drive/My Drive/"
    "Amazon-ML-Challenge-2026/ml/V5_Final_Inference"
)

os.makedirs(DRIVE_OUT, exist_ok=True)

MATCH_OUT = f"{DRIVE_OUT}/matching_results_v5.tsv"
CAND_OUT = f"{DRIVE_OUT}/candidate_pairs_v5.tsv"
CHECKPOINT = f"{DRIVE_OUT}/checkpoint.txt"

THRESHOLD = 0.60
CHUNK_SIZE = 2000

# ------------------------------------------------------------
# REQUIRED OBJECT CHECK
# ------------------------------------------------------------

required = [
    "s2_names_a",
    "s2_addrs_a",
    "s2_countries_a",
    "s2_ids",
    "s3_names_a",
    "s3_addrs_a",
    "s3_countries_a",
    "s3_ids",
    "s2_name_index",
    "s2_addr_index",
    "s2_name_freq",
    "s2_addr_freq",
    "s3_name_index",
    "s3_addr_index",
    "s3_name_freq",
    "s3_addr_freq",
    "v5_model",
]

missing = [x for x in required if x not in globals()]

if missing:
    raise RuntimeError(
        "Missing objects: " + ", ".join(missing)
    )

print("All required objects available.")


# ------------------------------------------------------------
# NORMALIZATION
# ------------------------------------------------------------

def normalize_advanced_local(text):

    if text is None:
        return ""

    text = str(text).lower().strip()

    text = unicodedata.normalize("NFKC", text)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()


def tokens_fast(text):

    if not text:
        return set()

    return set(text.split())


def extract_numbers_fast(text):

    return set(re.findall(r"\d+", text))


# ------------------------------------------------------------
# V5 FEATURE LIST
# ------------------------------------------------------------

V5_FEATURES = [
    "name_exact",
    "address_exact",
    "country_exact",
    "name_similarity",
    "address_similarity",
    "name_token_overlap",
    "address_token_overlap",
    "name_jaccard",
    "address_jaccard",
    "shared_name_token_count",
    "shared_address_token_count",
    "shared_address_number_count",
    "name_length_diff",
    "address_length_diff",
    "name_length_ratio",
    "address_length_ratio",
    "similarity_gap",
    "max_similarity",
    "similarity_product",
    "token_overlap_gap"
]


# ------------------------------------------------------------
# BUILD FEATURES
# ------------------------------------------------------------

def build_features(
    s1_name,
    s1_addr,
    s1_country,
    names,
    addrs,
    countries
):

    t1n = tokens_fast(s1_name)
    t1a = tokens_fast(s1_addr)
    nums1 = extract_numbers_fast(s1_addr)

    rows = []

    for n2, a2, c2 in zip(
        names,
        addrs,
        countries
    ):

        t2n = tokens_fast(n2)
        t2a = tokens_fast(a2)

        shared_n = t1n & t2n
        shared_a = t1a & t2a

        union_n = t1n | t2n
        union_a = t1a | t2a

        ns = (
            ratio(s1_name, n2) / 100
            if s1_name and n2 else 0.0
        )

        ass = (
            ratio(s1_addr, a2) / 100
            if s1_addr and a2 else 0.0
        )

        nt = (
            len(shared_n) /
            max(1, min(len(t1n), len(t2n)))
        )

        at = (
            len(shared_a) /
            max(1, min(len(t1a), len(t2a)))
        )

        nj = (
            len(shared_n) / len(union_n)
            if union_n else 0.0
        )

        aj = (
            len(shared_a) / len(union_a)
            if union_a else 0.0
        )

        nums2 = extract_numbers_fast(a2)

        rows.append([
            int(bool(s1_name) and s1_name == n2),
            int(bool(s1_addr) and s1_addr == a2),
            int(s1_country == c2),

            ns,
            ass,

            nt,
            at,

            nj,
            aj,

            len(shared_n),
            len(shared_a),
            len(nums1 & nums2),

            abs(len(s1_name) - len(n2)),
            abs(len(s1_addr) - len(a2)),

            min(len(s1_name), len(n2)) /
            max(1, max(len(s1_name), len(n2))),

            min(len(s1_addr), len(a2)) /
            max(1, max(len(s1_addr), len(a2))),

            abs(ns - ass),
            max(ns, ass),
            ns * ass,
            abs(nt - at)
        ])

    return np.asarray(rows, dtype=np.float32)


# ------------------------------------------------------------
# CANDIDATE GENERATION
# ------------------------------------------------------------

def generate_candidates(
    s1_name,
    s1_addr,
    name_index,
    addr_index,
    name_freq,
    addr_freq,
    names,
    addrs
):

    t_name = tokens_fast(s1_name)
    t_addr = tokens_fast(s1_addr)

    candidate_ids = set()

    # Exact name
    if s1_name:
        candidate_ids.update(
            name_index.get(s1_name, [])
        )

    # Exact address
    if s1_addr:
        candidate_ids.update(
            addr_index.get(s1_addr, [])
        )

    # Name tokens
    name_hits = {}

    for tok in t_name:

        if name_freq.get(tok, 10**9) <= 500:

            for idx in name_index.get(tok, []):

                name_hits[idx] = (
                    name_hits.get(idx, 0) + 1
                )

    for idx, count in name_hits.items():

        if count >= 2:
            candidate_ids.add(idx)

    # Address tokens
    addr_hits = {}

    for tok in t_addr:

        if addr_freq.get(tok, 10**9) <= 500:

            for idx in addr_index.get(tok, []):

                addr_hits[idx] = (
                    addr_hits.get(idx, 0) + 1
                )

    for idx, count in addr_hits.items():

        if count >= 2:
            candidate_ids.add(idx)

    # Similarity selection
    selected = []

    for idx in candidate_ids:

        n2 = names[idx]
        a2 = addrs[idx]

        shared_name = len(
            t_name & tokens_fast(n2)
        )

        shared_addr = len(
            t_addr & tokens_fast(a2)
        )

        name_sim = (
            ratio(s1_name, n2) / 100
            if s1_name and n2 else 0.0
        )

        addr_sim = (
            ratio(s1_addr, a2) / 100
            if s1_addr and a2 else 0.0
        )

        keep = (
            n2 == s1_name
            or
            a2 == s1_addr
            or
            (
                shared_name >= 2
                and name_sim >= 0.60
            )
            or
            (
                shared_addr >= 3
            )
            or
            (
                shared_addr >= 2
                and addr_sim >= 0.30
            )
        )

        if keep:
            selected.append(idx)

    return selected


# ------------------------------------------------------------
# SOURCE PROCESSOR
# ------------------------------------------------------------

def process_source1_row(
    row,
    names,
    addrs,
    countries,
    ids,
    name_index,
    addr_index,
    name_freq,
    addr_freq
):

    s1_name = normalize_advanced_local(
        row["business_name"]
    )

    s1_addr = normalize_advanced_local(
        row["business_address"]
    )

    s1_country = normalize_advanced_local(
        row["country"]
    )

    candidate_idx = generate_candidates(
        s1_name,
        s1_addr,
        name_index,
        addr_index,
        name_freq,
        addr_freq,
        names,
        addrs
    )

    if not candidate_idx:

        return [], []

    candidate_ids = [
        ids[i] for i in candidate_idx
    ]

    candidate_names = [
        names[i] for i in candidate_idx
    ]

    candidate_addrs = [
        addrs[i] for i in candidate_idx
    ]

    candidate_countries = [
        countries[i] for i in candidate_idx
    ]

    X = build_features(
        s1_name,
        s1_addr,
        s1_country,
        candidate_names,
        candidate_addrs,
        candidate_countries
    )

    probs = v5_model.predict_proba(X)[:, 1]

    matches = [
        str(candidate_ids[i])
        for i, p in enumerate(probs)
        if p >= THRESHOLD
    ]

    return matches, [
        str(x) for x in candidate_ids
    ]


# ------------------------------------------------------------
# CHECKPOINT HANDLING
# ------------------------------------------------------------

start_row = 0

if os.path.exists(CHECKPOINT):

    with open(CHECKPOINT, "r") as f:
        saved = f.read().strip()

    if saved:
        start_row = int(saved)

print("\nResume row:", start_row)


# ------------------------------------------------------------
# OUTPUT FILE INITIALIZATION
# ------------------------------------------------------------

if start_row == 0:

    for path in [MATCH_OUT, CAND_OUT]:

        if os.path.exists(path):
            os.remove(path)

    with open(MATCH_OUT, "w") as f:
        f.write(
            "source1_entity_id\tmatched_entity_ids\n"
        )

    with open(CAND_OUT, "w") as f:
        f.write(
            "source1_entity_id\tcandidate_entity_ids\n"
        )


# ------------------------------------------------------------
# TOTAL ROW COUNT
# ------------------------------------------------------------

total_rows = sum(
    1 for _ in open(S1_PATH, encoding="utf-8")
) - 1

print("Total S1 rows:", total_rows)
print("Starting from:", start_row)

# ------------------------------------------------------------
# FULL CHECKPOINTED RUN
# ------------------------------------------------------------

run_start = time.time()

total_candidates = 0
total_matches = 0

for chunk_start in range(
    start_row,
    total_rows,
    CHUNK_SIZE
):

    chunk_end = min(
        chunk_start + CHUNK_SIZE,
        total_rows
    )

    # Read ONLY this chunk
    s1_chunk = pd.read_csv(
        S1_PATH,
        sep="\t",
        skiprows=range(
            1,
            chunk_start + 1
        ),
        nrows=CHUNK_SIZE
    )

    match_lines = []
    candidate_lines = []

    chunk_candidates = 0
    chunk_matches = 0

    for _, row in s1_chunk.iterrows():

        sid = str(row["entity_id"])

        # -------------------------
        # S2
        # -------------------------

        m2, c2 = process_source1_row(
            row,
            s2_names_a,
            s2_addrs_a,
            s2_countries_a,
            s2_ids,
            s2_name_index,
            s2_addr_index,
            s2_name_freq,
            s2_addr_freq
        )

        # -------------------------
        # S3
        # -------------------------

        m3, c3 = process_source1_row(
            row,
            s3_names_a,
            s3_addrs_a,
            s3_countries_a,
            s3_ids,
            s3_name_index,
            s3_addr_index,
            s3_name_freq,
            s3_addr_freq
        )

        matches = m2 + m3
        candidates = c2 + c3

        chunk_candidates += len(candidates)
        chunk_matches += len(matches)

        match_lines.append(
            sid + "\t" + ",".join(matches) + "\n"
        )

        candidate_lines.append(
            sid + "\t" + ",".join(candidates) + "\n"
        )

    # --------------------------------------------
    # WRITE CHUNK DIRECTLY TO DRIVE
    # --------------------------------------------

    with open(MATCH_OUT, "a", encoding="utf-8") as f:
        f.writelines(match_lines)

    with open(CAND_OUT, "a", encoding="utf-8") as f:
        f.writelines(candidate_lines)

    total_candidates += chunk_candidates
    total_matches += chunk_matches

    # --------------------------------------------
    # CHECKPOINT
    # --------------------------------------------

    with open(CHECKPOINT, "w") as f:
        f.write(str(chunk_end))

    elapsed = time.time() - run_start
    completed = chunk_end
    rate = completed / elapsed if elapsed else 0

    remaining = (
        (total_rows - completed) / rate
        if rate else 0
    )

    print(
        f"{completed:,}/{total_rows:,} "
        f"({100*completed/total_rows:.2f}%) | "
        f"candidates {total_candidates:,} | "
        f"matches {total_matches:,} | "
        f"elapsed {elapsed/60:.1f} min | "
        f"ETA {remaining/60:.1f} min"
    )

    del s1_chunk
    del match_lines
    del candidate_lines
    gc.collect()


print("\n" + "=" * 70)
print("FULL V5 INFERENCE FINISHED")
print("=" * 70)

print("Matching file:")
print(MATCH_OUT)

print("\nCandidate file:")
print(CAND_OUT)

print("\nTotal candidates:", total_candidates)
print("Total predicted matches:", total_matches)

print(
    "Runtime:",
    round((time.time() - run_start) / 60, 2),
    "minutes"
)
```

```python
# ============================================================
# CLEAN RESTART — LOAD SAVED V5 ASSETS
# ============================================================

import os
import re
import gc
import pickle
import unicodedata
import numpy as np
import pandas as pd

from catboost import CatBoostClassifier
from rapidfuzz.fuzz import ratio

BASE = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml"

MODEL_PATH = f"{BASE}/V5_ML_Final/V5_ML_Final/v5_model.cbm"
INDEX_DIR = f"{BASE}/V5_full_indexes"

# ------------------------------------------------------------
# Normalization
# ------------------------------------------------------------

def normalize_advanced_local(text):
    if text is None or pd.isna(text):
        return ""

    text = str(text).lower().strip()
    text = unicodedata.normalize("NFKC", text)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()


def tokens_fast(text):
    return set(text.split()) if text else set()


def extract_numbers_fast(text):
    return set(re.findall(r"\d+", text))


# ------------------------------------------------------------
# Load pickle
# ------------------------------------------------------------

def load_pickle(filename):
    path = os.path.join(INDEX_DIR, filename)

    with open(path, "rb") as f:
        obj = pickle.load(f)

    print("Loaded:", filename)
    return obj


# S2
s2_name_index = load_pickle("s2_name_index.pkl")
s2_addr_index = load_pickle("s2_addr_index.pkl")
s2_name_freq = load_pickle("s2_name_freq.pkl")
s2_addr_freq = load_pickle("s2_addr_freq.pkl")

# S3
s3_name_index = load_pickle("s3_name_index.pkl")
s3_addr_index = load_pickle("s3_addr_index.pkl")
s3_name_freq = load_pickle("s3_name_freq.pkl")
s3_addr_freq = load_pickle("s3_addr_freq.pkl")

# ------------------------------------------------------------
# Load V5
# ------------------------------------------------------------

v5_model = CatBoostClassifier()
v5_model.load_model(MODEL_PATH)

print("\n" + "=" * 60)
print("V5 MODEL:", v5_model.is_fitted())
print("TREES:", v5_model.tree_count_)
print("MODEL EXISTS:", os.path.exists(MODEL_PATH))
print("INDEX DIR EXISTS:", os.path.exists(INDEX_DIR))
print("=" * 60)

gc.collect()
```

```python
import re
import unicodedata
from itertools import combinations
from difflib import SequenceMatcher

def normalize_advanced(text):
    if pd.isna(text):
        return ""

    text = str(text).lower().strip()
    text = unicodedata.normalize("NFKC", text)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()


def tokenize(text):
    return set(text.split()) if text else set()


def similarity(a, b):
    if not a or not b:
        return 0.0
    return SequenceMatcher(None, a, b).ratio()


def posting_intersection(index, tokens):
    tokens = [t for t in set(tokens) if t in index]

    if not tokens:
        return set()

    tokens.sort(key=lambda t: len(index[t]))

    result = set(index[tokens[0]])

    for token in tokens[1:]:
        result.intersection_update(index[token])

        if not result:
            break

    return result


def get_postings(index, tokens):
    tokens = [t for t in set(tokens) if t in index]

    if len(tokens) < 2:
        return set()

    tokens.sort(key=lambda t: len(index[t]))

    result = set()
    seeds = tokens[:4]

    for i in range(len(seeds)):
        for j in range(i + 1, len(seeds)):
            a = index[seeds[i]]
            b = index[seeds[j]]

            if len(a) > len(b):
                a, b = b, a

            bset = set(b)

            for row in a:
                if row in bset:
                    result.add(row)

    return result


print("✅ Candidate-generation helper functions loaded.")
```

```python
# Create country array for S2
s2_country = (
    s2["country"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.strip()
    .values
)

print("✅ s2_country created")
print("Length:", len(s2_country))
print("Available RAM:", f"{psutil.virtual_memory().available / 1024**3:.2f} GB")
```

```python
import time
import gc
import psutil
from difflib import SequenceMatcher

# Make sure we have the 5K S1
s1_path = "/content/amazon_ml_challenge/student_resource/dataset/test/test_source1.tsv"

s1_5k = pd.read_csv(
    s1_path,
    sep="\t",
    nrows=5000,
    dtype=str
)

def available_ram():
    return psutil.virtual_memory().available / 1024**3


def get_postings(index, tokens):
    """
    Return candidate row positions sharing at least 2 tokens.
    Uses pair intersections rather than materializing every
    token's full posting list into one giant set.
    """
    tokens = [t for t in set(tokens) if t in index]

    if len(tokens) < 2:
        return set()

    # Use the rarest tokens first.
    tokens.sort(key=lambda t: len(index[t]))

    result = set()

    # Use combinations of the rarest 4 tokens.
    seeds = tokens[:4]

    for i in range(len(seeds)):
        for j in range(i + 1, len(seeds)):
            a = index[seeds[i]]
            b = index[seeds[j]]

            # Iterate over the smaller posting list.
            if len(a) > len(b):
                a, b = b, a

            bset = set(b)

            for row in a:
                if row in bset:
                    result.add(row)

    return result


start = time.time()

total_raw = 0
total_selected = 0

for i, row in s1_5k.iterrows():

    name = normalize_advanced(row["business_name"])
    address = normalize_advanced(row["business_address"])
    country = str(row["country"]).lower().strip()

    name_tokens = tokenize(name)
    addr_tokens = tokenize(address)

    candidates = set()

    # -------------------------------
    # Exact name
    # -------------------------------
    if name_tokens:
        rows = posting_intersection(s2_name_index, name_tokens)

        for r in rows:
            if s2_country[r] == country:
                candidates.add(r)

    # -------------------------------
    # Exact address
    # -------------------------------
    if addr_tokens:
        rows = posting_intersection(s2_addr_index, addr_tokens)

        for r in rows:
            if s2_country[r] == country:
                candidates.add(r)

    # -------------------------------
    # Name: >=2 shared tokens
    # -------------------------------
    name_rows = get_postings(
        s2_name_index,
        name_tokens
    )

    for r in name_rows:

        if s2_country[r] != country:
            continue

        candidate_name = normalize_advanced(
            s2.iloc[r]["business_name"]
        )

        if similarity(name, candidate_name) >= 0.60:
            candidates.add(r)

    # -------------------------------
    # Address: >=3 shared tokens
    # -------------------------------
    addr_rows_3 = set()

    if len(addr_tokens) >= 3:

        tokens = [
            t for t in addr_tokens
            if t in s2_addr_index
        ]

        tokens.sort(
            key=lambda t: len(s2_addr_index[t])
        )

        for combo in combinations(tokens[:4], 3):

            rows = posting_intersection(
                s2_addr_index,
                combo
            )

            for r in rows:
                if s2_country[r] == country:
                    addr_rows_3.add(r)

    candidates.update(addr_rows_3)

    # -------------------------------
    # Address: >=2 shared tokens
    # + similarity >= 0.30
    # -------------------------------
    addr_rows_2 = get_postings(
        s2_addr_index,
        addr_tokens
    )

    for r in addr_rows_2:

        if s2_country[r] != country:
            continue

        candidate_addr = normalize_advanced(
            s2.iloc[r]["business_address"]
        )

        if similarity(address, candidate_addr) >= 0.30:
            candidates.add(r)

    total_selected += len(candidates)

    if (i + 1) % 500 == 0:
        elapsed = time.time() - start

        print(
            f"{i+1:,}/5,000 | "
            f"selected={total_selected:,} | "
            f"avg={total_selected/(i+1):.2f}/S1 | "
            f"RAM={available_ram():.2f} GB | "
            f"time={elapsed:.1f}s"
        )

print("\n" + "=" * 60)
print("S2 BENCHMARK COMPLETE")
print("=" * 60)

print(f"Total selected candidates: {total_selected:,}")
print(f"Average candidates/S1: {total_selected/5000:.2f}")
print(f"Time: {time.time()-start:.2f}s")
print(f"RAM available: {available_ram():.2f} GB")
```

```python
# ============================================================
# V5 PRODUCTION SMOKE TEST — 100 S1
# ============================================================

import os
import gc
import time
import pickle
import psutil
import pandas as pd
import numpy as np
from rapidfuzz.fuzz import ratio

BASE = "/content/amazon_ml_challenge/student_resource"
INDEX_DIR = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_full_indexes"
MODEL_PATH = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_ML_Final/V5_ML_Final/v5_model.cbm"

def ram_gb():
    return psutil.virtual_memory().available / (1024**3)

print("Starting V5 smoke test")
print(f"RAM available: {ram_gb():.2f} GB")

# ------------------------------------------------------------
# Load only 100 S1
# ------------------------------------------------------------

s1_test = pd.read_csv(
    f"{BASE}/dataset/test/test_source1.tsv",
    sep="\t",
    nrows=100,
    dtype=str
)

print("S1 loaded:", s1_test.shape)

# ------------------------------------------------------------
# S2
# ------------------------------------------------------------

print("\nLoading S2...")
s2_test = pd.read_csv(
    f"{BASE}/dataset/test/test_source2.tsv",
    sep="\t",
    usecols=["entity_id", "business_name", "business_address", "country"],
    dtype=str
)

print("S2:", s2_test.shape)
print(f"RAM: {ram_gb():.2f} GB")

# ------------------------------------------------------------
# S2 indexes
# ------------------------------------------------------------

print("\nLoading S2 indexes...")

with open(f"{INDEX_DIR}/s2_name_index.pkl", "rb") as f:
    name_index = pickle.load(f)

with open(f"{INDEX_DIR}/s2_addr_index.pkl", "rb") as f:
    addr_index = pickle.load(f)

print("S2 indexes loaded.")
print(f"RAM: {ram_gb():.2f} GB")

# ------------------------------------------------------------
# Country array
# ------------------------------------------------------------

countries = (
    s2_test["country"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.strip()
    .values
)

# ------------------------------------------------------------
# Fast candidate generation
# ------------------------------------------------------------

def get_candidates(name, address, country):

    name_tokens = tokenize(name)
    addr_tokens = tokenize(address)

    candidates = set()

    # Exact name
    if name_tokens:
        rows = posting_intersection(name_index, name_tokens)

        for r in rows:
            if countries[r] == country:
                candidates.add(r)

    # Exact address
    if addr_tokens:
        rows = posting_intersection(addr_index, addr_tokens)

        for r in rows:
            if countries[r] == country:
                candidates.add(r)

    # Name >= 2 shared tokens
    valid = [t for t in name_tokens if t in name_index]

    valid.sort(key=lambda x: len(name_index[x]))

    for i in range(min(4, len(valid))):
        for j in range(i + 1, min(4, len(valid))):

            a = name_index[valid[i]]
            b = name_index[valid[j]]

            if len(a) > len(b):
                a, b = b, a

            bset = set(b)

            for r in a:
                if r in bset and countries[r] == country:
                    candidates.add(r)

    # Address >= 2 shared tokens
    valid = [t for t in addr_tokens if t in addr_index]

    valid.sort(key=lambda x: len(addr_index[x]))

    for i in range(min(4, len(valid))):
        for j in range(i + 1, min(4, len(valid))):

            a = addr_index[valid[i]]
            b = addr_index[valid[j]]

            if len(a) > len(b):
                a, b = b, a

            bset = set(b)

            for r in a:
                if r in bset and countries[r] == country:
                    candidates.add(r)

    return candidates


# ------------------------------------------------------------
# Generate + filter 100 S1
# ------------------------------------------------------------

start = time.time()

candidate_counts = []
final_counts = []

for n, row in enumerate(s1_test.itertuples(index=False), 1):

    name = normalize_advanced(row.business_name)
    address = normalize_advanced(row.business_address)
    country = str(row.country).lower().strip()

    candidates = get_candidates(
        name,
        address,
        country
    )

    candidate_counts.append(len(candidates))

    # Apply cheap RapidFuzz filtering
    final = 0

    for r in candidates:

        cand_name = normalize_advanced(
            s2_test.iloc[r]["business_name"]
        )

        cand_addr = normalize_advanced(
            s2_test.iloc[r]["business_address"]
        )

        name_sim = ratio(name, cand_name) / 100
        addr_sim = ratio(address, cand_addr) / 100

        if (
            name == cand_name
            or address == cand_addr
            or name_sim >= 0.60
            or addr_sim >= 0.30
        ):
            final += 1

    final_counts.append(final)

    if n % 25 == 0:
        print(
            f"{n}/100 | "
            f"raw avg={np.mean(candidate_counts):.2f} | "
            f"final avg={np.mean(final_counts):.2f} | "
            f"RAM={ram_gb():.2f} GB"
        )

elapsed = time.time() - start

print("\n" + "=" * 60)
print("SMOKE TEST COMPLETE")
print("=" * 60)

print(f"S1 processed:       {len(candidate_counts)}")
print(f"Raw candidates:     {sum(candidate_counts):,}")
print(f"Raw candidates/S1:  {np.mean(candidate_counts):.2f}")
print(f"Final candidates:   {sum(final_counts):,}")
print(f"Final candidates/S1:{np.mean(final_counts):.2f}")
print(f"Time:               {elapsed:.2f} sec")
print(f"RAM available:      {ram_gb():.2f} GB")
```

```python
# ============================================================
# 100-S1 TEST OF THE MEASURED <=500 BLOCKER
# S2 ONLY — NO MODEL YET
# ============================================================

import os
import gc
import pickle
import time
import psutil
import pandas as pd
from rapidfuzz.fuzz import ratio

BASE = "/content/amazon_ml_challenge/student_resource"
INDEX_DIR = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_full_indexes"

def ram_gb():
    return psutil.virtual_memory().available / (1024**3)

# ------------------------------------------------------------
# Load only 100 S1
# ------------------------------------------------------------

s1_100 = pd.read_csv(
    f"{BASE}/dataset/test/test_source1.tsv",
    sep="\t",
    nrows=100,
    dtype=str
)

# ------------------------------------------------------------
# Load S2
# ------------------------------------------------------------

print("Loading S2...")
s2 = pd.read_csv(
    f"{BASE}/dataset/test/test_source2.tsv",
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ],
    dtype=str
)

print("S2:", s2.shape)
print(f"RAM after S2: {ram_gb():.2f} GB")

# ------------------------------------------------------------
# Load S2 indexes + frequency dictionaries
# ------------------------------------------------------------

print("Loading S2 indexes...")

with open(f"{INDEX_DIR}/s2_name_index.pkl", "rb") as f:
    name_index = pickle.load(f)

with open(f"{INDEX_DIR}/s2_addr_index.pkl", "rb") as f:
    addr_index = pickle.load(f)

with open(f"{INDEX_DIR}/s2_name_freq.pkl", "rb") as f:
    name_freq = pickle.load(f)

with open(f"{INDEX_DIR}/s2_addr_freq.pkl", "rb") as f:
    addr_freq = pickle.load(f)

print("Indexes loaded.")
print(f"RAM: {ram_gb():.2f} GB")

# ------------------------------------------------------------
# Country array
# ------------------------------------------------------------

s2_country = (
    s2["country"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.strip()
    .values
)

# ------------------------------------------------------------
# Measured <=500 blocker
# ------------------------------------------------------------

def rare_postings(index, freq, tokens, limit=500):

    rows = set()

    for token in tokens:

        if token not in index:
            continue

        if freq.get(token, 10**9) > limit:
            continue

        rows.update(index[token])

    return rows


def generate_500_candidates(name, address, country):

    name_tokens = tokenize(name)
    addr_tokens = tokenize(address)

    candidates = set()

    # --------------------------------------------------------
    # Exact name
    # --------------------------------------------------------

    if name:
        rows = name_index.get(name)

        if rows is not None:
            for r in rows:
                if s2_country[r] == country:
                    candidates.add(r)

    # --------------------------------------------------------
    # Exact address
    # --------------------------------------------------------

    if address:
        rows = addr_index.get(address)

        if rows is not None:
            for r in rows:
                if s2_country[r] == country:
                    candidates.add(r)

    # --------------------------------------------------------
    # Rare name tokens <= 500
    # --------------------------------------------------------

    rows = rare_postings(
        name_index,
        name_freq,
        name_tokens,
        500
    )

    for r in rows:

        if s2_country[r] != country:
            continue

        cand_name = normalize_advanced(
            s2.iloc[r]["business_name"]
        )

        if ratio(name, cand_name) >= 60:
            candidates.add(r)

    # --------------------------------------------------------
    # Rare address tokens <= 500
    # --------------------------------------------------------

    rows = rare_postings(
        addr_index,
        addr_freq,
        addr_tokens,
        500
    )

    for r in rows:

        if s2_country[r] != country:
            continue

        cand_addr = normalize_advanced(
            s2.iloc[r]["business_address"]
        )

        if ratio(address, cand_addr) >= 60:
            candidates.add(r)

    return candidates


# ------------------------------------------------------------
# Run 100 S1
# ------------------------------------------------------------

start = time.time()

counts = []

for i, row in enumerate(s1_100.itertuples(index=False), 1):

    name = normalize_advanced(row.business_name)
    address = normalize_advanced(row.business_address)
    country = str(row.country).lower().strip()

    cands = generate_500_candidates(
        name,
        address,
        country
    )

    counts.append(len(cands))

    if i % 20 == 0:
        print(
            f"{i}/100 | "
            f"avg candidates/S1 = {sum(counts)/len(counts):.2f} | "
            f"RAM = {ram_gb():.2f} GB"
        )

print("\n" + "=" * 60)
print("MEASURED BLOCKER TEST")
print("=" * 60)
print(f"S1 processed:       {len(counts)}")
print(f"Total candidates:   {sum(counts):,}")
print(f"Candidates/S1:      {sum(counts)/len(counts):.2f}")
print(f"Time:               {time.time()-start:.2f} sec")
print(f"RAM available:      {ram_gb():.2f} GB")
```

```python
# Measure RAW rare-token candidates before similarity filtering

raw_counts = []
selected_counts = []

for row in s1_100.itertuples(index=False):

    name = normalize_advanced(row.business_name)
    address = normalize_advanced(row.business_address)
    country = str(row.country).lower().strip()

    name_tokens = tokenize(name)
    addr_tokens = tokenize(address)

    raw = set()

    # rare name tokens <= 500
    for token in name_tokens:
        if (
            token in name_index
            and name_freq.get(token, 10**9) <= 500
        ):
            for r in name_index[token]:
                if s2_country[r] == country:
                    raw.add(r)

    # rare address tokens <= 500
    for token in addr_tokens:
        if (
            token in addr_index
            and addr_freq.get(token, 10**9) <= 500
        ):
            for r in addr_index[token]:
                if s2_country[r] == country:
                    raw.add(r)

    raw_counts.append(len(raw))

    # similarity filtering
    selected = 0

    for r in raw:
        cn = normalize_advanced(s2.iloc[r]["business_name"])
        ca = normalize_advanced(s2.iloc[r]["business_address"])

        ns = ratio(name, cn) / 100
        ads = ratio(address, ca) / 100

        if max(ns, ads) >= 0.60:
            selected += 1

    selected_counts.append(selected)

print("RAW candidates/S1:", sum(raw_counts) / len(raw_counts))
print("SELECTED/S1:", sum(selected_counts) / len(selected_counts))
```

```python
# ============================================================
# V3-STYLE BLOCKER — 100 S1 / S2 ONLY
# ============================================================

import time
import psutil

def ram_gb():
    return psutil.virtual_memory().available / (1024**3)

LIMIT = 1000

def v3_s2_candidates(name, address, country):

    name_tokens = tokenize(name)
    addr_tokens = tokenize(address)

    candidates = set()

    # --------------------------------------------------------
    # 1. Exact name + country
    # --------------------------------------------------------
    if name:
        rows = name_index.get(name, [])
        for r in rows:
            if s2_country[r] == country:
                candidates.add(r)

    # --------------------------------------------------------
    # 2. Exact address + country
    # --------------------------------------------------------
    if address:
        rows = addr_index.get(address, [])
        for r in rows:
            if s2_country[r] == country:
                candidates.add(r)

    # --------------------------------------------------------
    # 3. Rare name token + country
    # --------------------------------------------------------
    for token in name_tokens:

        if token not in name_index:
            continue

        if name_freq.get(token, 10**9) > LIMIT:
            continue

        for r in name_index[token]:
            if s2_country[r] == country:
                candidates.add(r)

    # --------------------------------------------------------
    # 4. Rare address token + country
    # --------------------------------------------------------
    for token in addr_tokens:

        if token not in addr_index:
            continue

        if addr_freq.get(token, 10**9) > LIMIT:
            continue

        for r in addr_index[token]:
            if s2_country[r] == country:
                candidates.add(r)

    # --------------------------------------------------------
    # 5. V3 multi-address-token block
    #
    # At least 2 shared address tokens AND
    # at least one shared token is rare (<= LIMIT).
    # --------------------------------------------------------

    rare_addr_tokens = [
        t for t in addr_tokens
        if (
            t in addr_index
            and addr_freq.get(t, 10**9) <= LIMIT
        )
    ]

    if len(addr_tokens) >= 2 and rare_addr_tokens:

        # Candidates sharing a rare address token
        rare_rows = set()

        for token in rare_addr_tokens:
            rare_rows.update(addr_index[token])

        # Require at least 2 shared address tokens
        # and same country.
        for r in rare_rows:

            if s2_country[r] != country:
                continue

            # We need the candidate's normalized address.
            cand_addr = normalize_advanced(
                s2.iloc[r]["business_address"]
            )

            shared = (
                tokenize(address)
                &
                tokenize(cand_addr)
            )

            if len(shared) >= 2:
                candidates.add(r)

    return candidates


start = time.time()

counts = []

for i, row in enumerate(s1_100.itertuples(index=False), 1):

    name = normalize_advanced(row.business_name)
    address = normalize_advanced(row.business_address)
    country = str(row.country).lower().strip()

    cands = v3_s2_candidates(
        name,
        address,
        country
    )

    counts.append(len(cands))

    if i % 20 == 0:
        print(
            f"{i}/100 | "
            f"avg candidates/S1 = {sum(counts)/len(counts):.2f} | "
            f"RAM = {ram_gb():.2f} GB"
        )

print("\n" + "=" * 60)
print("V3-STYLE S2 TEST")
print("=" * 60)
print(f"S1 processed:      {len(counts)}")
print(f"Total candidates:  {sum(counts):,}")
print(f"Candidates/S1:     {sum(counts)/len(counts):.2f}")
print(f"Time:              {time.time()-start:.2f} sec")
print(f"RAM available:     {ram_gb():.2f} GB")
```

```python
# ============================================================
# EXACT V5 PIPELINE — 100 S1 / S2 ONLY
# Candidate generation + V5 prediction
# ============================================================

import gc
import time
import pickle
import psutil
import pandas as pd
import numpy as np
from rapidfuzz.fuzz import ratio

BASE = "/content/amazon_ml_challenge/student_resource"
INDEX_DIR = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_full_indexes"

def ram_gb():
    return psutil.virtual_memory().available / (1024**3)

print("=" * 70)
print("V5 S2 SMOKE TEST")
print("=" * 70)
print("RAM:", round(ram_gb(), 2), "GB")

# ------------------------------------------------------------
# Load 100 S1
# ------------------------------------------------------------

s1_100 = pd.read_csv(
    f"{BASE}/dataset/test/test_source1.tsv",
    sep="\t",
    dtype=str,
    keep_default_na=False,
    nrows=100
)

# ------------------------------------------------------------
# Load S2
# ------------------------------------------------------------

print("\nLoading S2...")

s2 = pd.read_csv(
    f"{BASE}/dataset/test/test_source2.tsv",
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ],
    dtype=str,
    keep_default_na=False
)

print("S2:", s2.shape)
print("RAM after S2:", round(ram_gb(), 2), "GB")

# ------------------------------------------------------------
# Normalize
# ------------------------------------------------------------

def norm(x):
    if x is None:
        return ""

    x = str(x).lower().strip()

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        x = x.replace(old, new)

    import unicodedata
    import re

    x = unicodedata.normalize("NFKC", x)
    x = re.sub(r"[^\w\s]", " ", x)
    x = re.sub(r"\s+", " ", x)

    return x.strip()


def toks(x):
    return set(x.split()) if x else set()


def nums(x):
    import re
    return set(re.findall(r"\d+", x))


# ------------------------------------------------------------
# Source arrays
# ------------------------------------------------------------

print("\nPreparing S2 arrays...")

s2_names = s2["business_name"].map(norm).to_numpy()
s2_addrs = s2["business_address"].map(norm).to_numpy()
s2_countries = (
    s2["country"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.strip()
    .to_numpy()
)

s2_ids = s2["entity_id"].to_numpy()

print("Arrays ready.")
print("RAM:", round(ram_gb(), 2), "GB")

# ------------------------------------------------------------
# Load indexes
# ------------------------------------------------------------

print("\nLoading S2 indexes...")

with open(f"{INDEX_DIR}/s2_name_index.pkl", "rb") as f:
    s2_name_index = pickle.load(f)

with open(f"{INDEX_DIR}/s2_addr_index.pkl", "rb") as f:
    s2_addr_index = pickle.load(f)

with open(f"{INDEX_DIR}/s2_name_freq.pkl", "rb") as f:
    s2_name_freq = pickle.load(f)

with open(f"{INDEX_DIR}/s2_addr_freq.pkl", "rb") as f:
    s2_addr_freq = pickle.load(f)

print("Indexes ready.")
print("RAM:", round(ram_gb(), 2), "GB")

# ------------------------------------------------------------
# Exact V5 candidate generator
# ------------------------------------------------------------

def generate_v5_candidates(
    name1,
    addr1,
    country1
):

    t_name = toks(name1)
    t_addr = toks(addr1)

    name_hits = {}
    addr_hits = {}

    # NAME TOKEN BLOCK <=500
    for token in t_name:

        if s2_name_freq.get(token, 10**9) <= 500:

            for idx in s2_name_index.get(token, ()):

                if s2_countries[idx] == country1:

                    name_hits[idx] = (
                        name_hits.get(idx, 0) + 1
                    )

    # ADDRESS TOKEN BLOCK <=500
    for token in t_addr:

        if s2_addr_freq.get(token, 10**9) <= 500:

            for idx in s2_addr_index.get(token, ()):

                if s2_countries[idx] == country1:

                    addr_hits[idx] = (
                        addr_hits.get(idx, 0) + 1
                    )

    candidate_ids = (
        {
            idx for idx, count in name_hits.items()
            if count >= 2
        }
        |
        {
            idx for idx, count in addr_hits.items()
            if count >= 2
        }
    )

    # --------------------------------------------------------
    # FINAL V5 CANDIDATE FILTER
    # --------------------------------------------------------

    selected = []

    for idx in candidate_ids:

        n2 = s2_names[idx]
        a2 = s2_addrs[idx]

        # Exact
        if (
            (name1 and name1 == n2)
            or
            (addr1 and addr1 == a2)
        ):
            selected.append(idx)
            continue

        shared_name = len(
            t_name & toks(n2)
        )

        shared_addr = len(
            t_addr & toks(a2)
        )

        name_sim = (
            ratio(name1, n2) / 100
            if name1 and n2 else 0.0
        )

        addr_sim = (
            ratio(addr1, a2) / 100
            if addr1 and a2 else 0.0
        )

        keep = (
            (
                shared_name >= 2
                and name_sim >= 0.60
            )
            or
            (
                shared_addr >= 3
            )
            or
            (
                shared_addr >= 2
                and addr_sim >= 0.30
            )
        )

        if keep:
            selected.append(idx)

    return selected


# ------------------------------------------------------------
# V5 model
# ------------------------------------------------------------

from catboost import CatBoostClassifier

MODEL_PATH = (
    "/content/drive/MyDrive/"
    "Amazon-ML-Challenge-2026/ml/"
    "V5_ML_Final/V5_ML_Final/v5_model.cbm"
)

v5_model = CatBoostClassifier()
v5_model.load_model(MODEL_PATH)

print("\nV5 model loaded:", v5_model.is_fitted())
print("Trees:", v5_model.tree_count_)

# ------------------------------------------------------------
# V5 features
# ------------------------------------------------------------

V5_FEATURES = [
    "name_exact",
    "address_exact",
    "country_exact",
    "name_similarity",
    "address_similarity",
    "name_token_overlap",
    "address_token_overlap",
    "name_jaccard",
    "address_jaccard",
    "shared_name_token_count",
    "shared_address_token_count",
    "shared_address_number_count",
    "name_length_diff",
    "address_length_diff",
    "name_length_ratio",
    "address_length_ratio",
    "similarity_gap",
    "max_similarity",
    "similarity_product",
    "token_overlap_gap"
]


def make_features(
    name1,
    addr1,
    country1,
    candidate_indices
):

    t1n = toks(name1)
    t1a = toks(addr1)
    num1 = nums(addr1)

    rows = []

    for idx in candidate_indices:

        n2 = s2_names[idx]
        a2 = s2_addrs[idx]
        c2 = s2_countries[idx]

        t2n = toks(n2)
        t2a = toks(a2)

        shared_n = t1n & t2n
        shared_a = t1a & t2a

        union_n = t1n | t2n
        union_a = t1a | t2a

        ns = (
            ratio(name1, n2) / 100
            if name1 and n2 else 0.0
        )

        ass = (
            ratio(addr1, a2) / 100
            if addr1 and a2 else 0.0
        )

        nt = (
            len(shared_n) /
            max(1, min(len(t1n), len(t2n)))
        )

        at = (
            len(shared_a) /
            max(1, min(len(t1a), len(t2a)))
        )

        nj = (
            len(shared_n) / len(union_n)
            if union_n else 0.0
        )

        aj = (
            len(shared_a) / len(union_a)
            if union_a else 0.0
        )

        rows.append([
            int(bool(name1) and name1 == n2),
            int(bool(addr1) and addr1 == a2),
            int(country1 == c2),

            ns,
            ass,

            nt,
            at,

            nj,
            aj,

            len(shared_n),
            len(shared_a),

            len(num1 & nums(a2)),

            abs(len(name1) - len(n2)),
            abs(len(addr1) - len(a2)),

            min(len(name1), len(n2)) /
            max(1, max(len(name1), len(n2))),

            min(len(addr1), len(a2)) /
            max(1, max(len(addr1), len(a2))),

            abs(ns - ass),
            max(ns, ass),
            ns * ass,
            abs(nt - at)
        ])

    return np.asarray(
        rows,
        dtype=np.float32
    )


# ------------------------------------------------------------
# RUN
# ------------------------------------------------------------

start = time.time()

total_candidates = 0
total_predictions = 0
candidate_counts = []

for i, row in enumerate(
    s1_100.itertuples(index=False),
    1
):

    name1 = norm(row.business_name)
    addr1 = norm(row.business_address)
    country1 = (
        str(row.country)
        .lower()
        .strip()
    )

    candidates = generate_v5_candidates(
        name1,
        addr1,
        country1
    )

    candidate_counts.append(len(candidates))
    total_candidates += len(candidates)

    if candidates:

        X = make_features(
            name1,
            addr1,
            country1,
            candidates
        )

        probs = v5_model.predict_proba(X)[:, 1]

        predictions = int(
            (probs >= 0.60).sum()
        )

        total_predictions += predictions

    if i % 20 == 0:

        print(
            f"{i}/100 | "
            f"avg candidates/S1="
            f"{total_candidates/i:.2f} | "
            f"matches={total_predictions} | "
            f"RAM={ram_gb():.2f} GB"
        )

elapsed = time.time() - start

print("\n" + "=" * 70)
print("V5 S2 SMOKE TEST COMPLETE")
print("=" * 70)

print("S1 processed:", 100)
print("Total candidates:", f"{total_candidates:,}")
print(
    "Candidates/S1:",
    round(total_candidates / 100, 2)
)
print(
    "Median candidates/S1:",
    round(float(np.median(candidate_counts)), 2)
)
print("Predicted matches:", total_predictions)
print(
    "Predicted matches/S1:",
    round(total_predictions / 100, 3)
)
print("Time:", round(elapsed, 2), "sec")
print("RAM available:", round(ram_gb(), 2), "GB")
```

```python
# ============================================================
# EXACT V5 PIPELINE — 100 S1 / S3 ONLY
# ============================================================

import gc
import time
import pickle
import psutil
import pandas as pd
import numpy as np
from rapidfuzz.fuzz import ratio

BASE = "/content/amazon_ml_challenge/student_resource"
INDEX_DIR = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_full_indexes"

def ram_gb():
    return psutil.virtual_memory().available / (1024**3)

print("=" * 70)
print("V5 S3 SMOKE TEST")
print("=" * 70)
print("RAM:", round(ram_gb(), 2), "GB")

# ------------------------------------------------------------
# Load 100 S1
# ------------------------------------------------------------

s1_100 = pd.read_csv(
    f"{BASE}/dataset/test/test_source1.tsv",
    sep="\t",
    dtype=str,
    keep_default_na=False,
    nrows=5000
)

# ------------------------------------------------------------
# Normalization
# ------------------------------------------------------------

def norm(x):
    if x is None:
        return ""

    x = str(x).lower().strip()

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        x = x.replace(old, new)

    import unicodedata
    import re

    x = unicodedata.normalize("NFKC", x)
    x = re.sub(r"[^\w\s]", " ", x)
    x = re.sub(r"\s+", " ", x)

    return x.strip()


def toks(x):
    return set(x.split()) if x else set()


def nums(x):
    import re
    return set(re.findall(r"\d+", x))


# ------------------------------------------------------------
# Load S3
# ------------------------------------------------------------

print("\nLoading S3...")

s3 = pd.read_csv(
    f"{BASE}/dataset/test/test_source3.tsv",
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ],
    dtype=str,
    keep_default_na=False
)

print("S3:", s3.shape)
print("RAM after S3:", round(ram_gb(), 2), "GB")

# ------------------------------------------------------------
# Prepare arrays
# ------------------------------------------------------------

print("\nPreparing S3 arrays...")

s3_names = s3["business_name"].map(norm).to_numpy()
s3_addrs = s3["business_address"].map(norm).to_numpy()
s3_countries = (
    s3["country"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.strip()
    .to_numpy()
)
s3_ids = s3["entity_id"].to_numpy()

print("Arrays ready.")
print("RAM:", round(ram_gb(), 2), "GB")

# ------------------------------------------------------------
# Load S3 indexes
# ------------------------------------------------------------

print("\nLoading S3 indexes...")

with open(f"{INDEX_DIR}/s3_name_index.pkl", "rb") as f:
    s3_name_index = pickle.load(f)

with open(f"{INDEX_DIR}/s3_addr_index.pkl", "rb") as f:
    s3_addr_index = pickle.load(f)

with open(f"{INDEX_DIR}/s3_name_freq.pkl", "rb") as f:
    s3_name_freq = pickle.load(f)

with open(f"{INDEX_DIR}/s3_addr_freq.pkl", "rb") as f:
    s3_addr_freq = pickle.load(f)

print("Indexes ready.")
print("RAM:", round(ram_gb(), 2), "GB")

# ------------------------------------------------------------
# Exact V5 candidate generation
# ------------------------------------------------------------

def generate_v5_candidates_s3(name1, addr1, country1):

    t_name = toks(name1)
    t_addr = toks(addr1)

    name_hits = {}
    addr_hits = {}

    # Name token block
    for token in t_name:

        if s3_name_freq.get(token, 10**9) <= 500:

            for idx in s3_name_index.get(token, ()):

                if s3_countries[idx] == country1:
                    name_hits[idx] = (
                        name_hits.get(idx, 0) + 1
                    )

    # Address token block
    for token in t_addr:

        if s3_addr_freq.get(token, 10**9) <= 500:

            for idx in s3_addr_index.get(token, ()):

                if s3_countries[idx] == country1:
                    addr_hits[idx] = (
                        addr_hits.get(idx, 0) + 1
                    )

    candidate_ids = (
        {
            idx for idx, count in name_hits.items()
            if count >= 2
        }
        |
        {
            idx for idx, count in addr_hits.items()
            if count >= 2
        }
    )

    # Final V5 candidate filter
    selected = []

    for idx in candidate_ids:

        n2 = s3_names[idx]
        a2 = s3_addrs[idx]

        if (
            (name1 and name1 == n2)
            or
            (addr1 and addr1 == a2)
        ):
            selected.append(idx)
            continue

        shared_name = len(t_name & toks(n2))
        shared_addr = len(t_addr & toks(a2))

        name_sim = (
            ratio(name1, n2) / 100
            if name1 and n2 else 0.0
        )

        addr_sim = (
            ratio(addr1, a2) / 100
            if addr1 and a2 else 0.0
        )

        keep = (
            (
                shared_name >= 2
                and name_sim >= 0.60
            )
            or
            shared_addr >= 3
            or
            (
                shared_addr >= 2
                and addr_sim >= 0.30
            )
        )

        if keep:
            selected.append(idx)

    return selected


# ------------------------------------------------------------
# V5 model
# ------------------------------------------------------------

from catboost import CatBoostClassifier

MODEL_PATH = (
    "/content/drive/MyDrive/"
    "Amazon-ML-Challenge-2026/ml/"
    "V5_ML_Final/V5_ML_Final/v5_model.cbm"
)

v5_model = CatBoostClassifier()
v5_model.load_model(MODEL_PATH)

# ------------------------------------------------------------
# Feature builder
# ------------------------------------------------------------

def make_features_s3(
    name1,
    addr1,
    country1,
    candidate_indices
):

    t1n = toks(name1)
    t1a = toks(addr1)
    num1 = nums(addr1)

    rows = []

    for idx in candidate_indices:

        n2 = s3_names[idx]
        a2 = s3_addrs[idx]
        c2 = s3_countries[idx]

        t2n = toks(n2)
        t2a = toks(a2)

        shared_n = t1n & t2n
        shared_a = t1a & t2a

        union_n = t1n | t2n
        union_a = t1a | t2a

        ns = ratio(name1, n2) / 100 if name1 and n2 else 0.0
        ass = ratio(addr1, a2) / 100 if addr1 and a2 else 0.0

        nt = len(shared_n) / max(
            1, min(len(t1n), len(t2n))
        )

        at = len(shared_a) / max(
            1, min(len(t1a), len(t2a))
        )

        nj = (
            len(shared_n) / len(union_n)
            if union_n else 0.0
        )

        aj = (
            len(shared_a) / len(union_a)
            if union_a else 0.0
        )

        rows.append([
            int(bool(name1) and name1 == n2),
            int(bool(addr1) and addr1 == a2),
            int(country1 == c2),
            ns,
            ass,
            nt,
            at,
            nj,
            aj,
            len(shared_n),
            len(shared_a),
            len(num1 & nums(a2)),
            abs(len(name1) - len(n2)),
            abs(len(addr1) - len(a2)),
            min(len(name1), len(n2)) /
                max(1, max(len(name1), len(n2))),
            min(len(addr1), len(a2)) /
                max(1, max(len(addr1), len(a2))),
            abs(ns - ass),
            max(ns, ass),
            ns * ass,
            abs(nt - at)
        ])

    return np.asarray(rows, dtype=np.float32)


# ------------------------------------------------------------
# Run
# ------------------------------------------------------------

start = time.time()

total_candidates = 0
total_predictions = 0
candidate_counts = []

for i, row in enumerate(s1_100.itertuples(index=False), 1):

    name1 = norm(row.business_name)
    addr1 = norm(row.business_address)
    country1 = str(row.country).lower().strip()

    candidates = generate_v5_candidates_s3(
        name1,
        addr1,
        country1
    )

    candidate_counts.append(len(candidates))
    total_candidates += len(candidates)

    if candidates:

        X = make_features_s3(
            name1,
            addr1,
            country1,
            candidates
        )

        probs = v5_model.predict_proba(X)[:, 1]

        total_predictions += int(
            (probs >= 0.60).sum()
        )

    if i % 500 == 0:

        print(
            f"{i}/100 | "
            f"avg candidates/S1="
            f"{total_candidates/i:.2f} | "
            f"matches={total_predictions} | "
            f"RAM={ram_gb():.2f} GB"
        )

elapsed = time.time() - start

print("\n" + "=" * 70)
print("V5 S3 SMOKE TEST COMPLETE")
print("=" * 70)

print("S1 processed:", 100)
print("Total candidates:", f"{total_candidates:,}")
print(
    "Candidates/S1:",
    round(total_candidates / 100, 2)
)
print(
    "Median candidates/S1:",
    round(float(np.median(candidate_counts)), 2)
)
print("Predicted matches:", total_predictions)
print(
    "Predicted matches/S1:",
    round(total_predictions / 100, 3)
)
print("Time:", round(elapsed, 2), "sec")
print("RAM available:", round(ram_gb(), 2), "GB")
```

```python
# ============================================================
# EXACT V5 PIPELINE — 5,000 S1 / S2 ONLY
# Candidate generation + V5 prediction
# NO submission files
# ============================================================

import gc
import time
import pickle
import psutil
import pandas as pd
import numpy as np
from rapidfuzz.fuzz import ratio
from catboost import CatBoostClassifier
import unicodedata
import re

# ============================================================
# PATHS
# ============================================================

BASE = "/content/amazon_ml_challenge/student_resource"

INDEX_DIR = (
    "/content/drive/MyDrive/"
    "Amazon-ML-Challenge-2026/ml/V5_full_indexes"
)

MODEL_PATH = (
    "/content/drive/MyDrive/"
    "Amazon-ML-Challenge-2026/ml/"
    "V5_ML_Final/V5_ML_Final/v5_model.cbm"
)

N_S1 = 5000
THRESHOLD = 0.60


# ============================================================
# MEMORY
# ============================================================

def ram_gb():
    return psutil.virtual_memory().available / (1024**3)


print("=" * 75)
print("V5 S2 — 5,000 S1 SMOKE TEST")
print("=" * 75)
print(f"Starting available RAM: {ram_gb():.2f} GB")


# ============================================================
# LOAD TEST S1
# ============================================================

print("\nLoading S1...")

s1 = pd.read_csv(
    f"{BASE}/dataset/test/test_source1.tsv",
    sep="\t",
    dtype=str,
    keep_default_na=False,
    nrows=N_S1
)

print("S1:", s1.shape)


# ============================================================
# NORMALIZATION
# ============================================================

def norm(x):

    if x is None:
        return ""

    x = str(x).lower().strip()

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        x = x.replace(old, new)

    x = unicodedata.normalize("NFKC", x)

    x = re.sub(
        r"[^\w\s]",
        " ",
        x
    )

    x = re.sub(
        r"\s+",
        " ",
        x
    )

    return x.strip()


def toks(x):
    return set(x.split()) if x else set()


def nums(x):
    return set(
        re.findall(
            r"\d+",
            x
        )
    )


# ============================================================
# LOAD S2
# ============================================================

print("\nLoading S2...")

s2 = pd.read_csv(
    f"{BASE}/dataset/test/test_source2.tsv",
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ],
    dtype=str,
    keep_default_na=False
)

print("S2:", s2.shape)
print(f"RAM after S2: {ram_gb():.2f} GB")


# ============================================================
# PREPARE S2 ARRAYS
# ============================================================

print("\nPreparing S2 arrays...")

s2_names = (
    s2["business_name"]
    .map(norm)
    .to_numpy()
)

s2_addrs = (
    s2["business_address"]
    .map(norm)
    .to_numpy()
)

s2_countries = (
    s2["country"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.strip()
    .to_numpy()
)

s2_ids = s2["entity_id"].to_numpy()

print("Arrays ready.")
print(f"RAM: {ram_gb():.2f} GB")


# ============================================================
# LOAD S2 INDEXES
# ============================================================

print("\nLoading S2 indexes...")

with open(
    f"{INDEX_DIR}/s2_name_index.pkl",
    "rb"
) as f:
    s2_name_index = pickle.load(f)

with open(
    f"{INDEX_DIR}/s2_addr_index.pkl",
    "rb"
) as f:
    s2_addr_index = pickle.load(f)

with open(
    f"{INDEX_DIR}/s2_name_freq.pkl",
    "rb"
) as f:
    s2_name_freq = pickle.load(f)

with open(
    f"{INDEX_DIR}/s2_addr_freq.pkl",
    "rb"
) as f:
    s2_addr_freq = pickle.load(f)

print("S2 indexes loaded.")
print(f"RAM: {ram_gb():.2f} GB")


# ============================================================
# V5 CANDIDATE GENERATION
# ============================================================

def generate_v5_candidates_s2(
    name1,
    addr1,
    country1
):

    t_name = toks(name1)
    t_addr = toks(addr1)

    name_hits = {}
    addr_hits = {}

    # --------------------------------------------------------
    # NAME BLOCK
    # Rare token frequency <= 500
    # --------------------------------------------------------

    for token in t_name:

        if s2_name_freq.get(
            token,
            10**9
        ) <= 500:

            for idx in s2_name_index.get(
                token,
                ()
            ):

                if s2_countries[idx] == country1:

                    name_hits[idx] = (
                        name_hits.get(
                            idx,
                            0
                        ) + 1
                    )

    # --------------------------------------------------------
    # ADDRESS BLOCK
    # Rare token frequency <= 500
    # --------------------------------------------------------

    for token in t_addr:

        if s2_addr_freq.get(
            token,
            10**9
        ) <= 500:

            for idx in s2_addr_index.get(
                token,
                ()
            ):

                if s2_countries[idx] == country1:

                    addr_hits[idx] = (
                        addr_hits.get(
                            idx,
                            0
                        ) + 1
                    )

    # --------------------------------------------------------
    # REQUIRE >=2 SHARED TOKENS
    # --------------------------------------------------------

    candidate_ids = (
        {
            idx
            for idx, count in name_hits.items()
            if count >= 2
        }
        |
        {
            idx
            for idx, count in addr_hits.items()
            if count >= 2
        }
    )

    # --------------------------------------------------------
    # FINAL V5 CANDIDATE FILTER
    # --------------------------------------------------------

    selected = []

    for idx in candidate_ids:

        n2 = s2_names[idx]
        a2 = s2_addrs[idx]

        # Exact name OR exact address
        if (
            (name1 and name1 == n2)
            or
            (addr1 and addr1 == a2)
        ):
            selected.append(idx)
            continue

        t2n = toks(n2)
        t2a = toks(a2)

        shared_name = len(
            t_name & t2n
        )

        shared_addr = len(
            t_addr & t2a
        )

        name_sim = (
            ratio(name1, n2) / 100
            if name1 and n2
            else 0.0
        )

        addr_sim = (
            ratio(addr1, a2) / 100
            if addr1 and a2
            else 0.0
        )

        keep = (
            (
                shared_name >= 2
                and
                name_sim >= 0.60
            )
            or
            (
                shared_addr >= 3
            )
            or
            (
                shared_addr >= 2
                and
                addr_sim >= 0.30
            )
        )

        if keep:
            selected.append(idx)

    return selected


# ============================================================
# LOAD V5 MODEL
# ============================================================

print("\nLoading V5 model...")

v5_model = CatBoostClassifier()

v5_model.load_model(
    MODEL_PATH
)

print(
    "V5 model loaded:",
    v5_model.is_fitted()
)

print(
    "Tree count:",
    v5_model.tree_count_
)


# ============================================================
# FEATURE BUILDER
# ============================================================

def make_features_s2(
    name1,
    addr1,
    country1,
    candidate_indices
):

    t1n = toks(name1)
    t1a = toks(addr1)
    num1 = nums(addr1)

    rows = []

    for idx in candidate_indices:

        n2 = s2_names[idx]
        a2 = s2_addrs[idx]
        c2 = s2_countries[idx]

        t2n = toks(n2)
        t2a = toks(a2)

        shared_n = t1n & t2n
        shared_a = t1a & t2a

        union_n = t1n | t2n
        union_a = t1a | t2a

        # Similarities
        ns = (
            ratio(name1, n2) / 100
            if name1 and n2
            else 0.0
        )

        ass = (
            ratio(addr1, a2) / 100
            if addr1 and a2
            else 0.0
        )

        # Token overlap
        nt = (
            len(shared_n) /
            max(
                1,
                min(
                    len(t1n),
                    len(t2n)
                )
            )
        )

        at = (
            len(shared_a) /
            max(
                1,
                min(
                    len(t1a),
                    len(t2a)
                )
            )
        )

        # Jaccard
        nj = (
            len(shared_n) /
            len(union_n)
            if union_n
            else 0.0
        )

        aj = (
            len(shared_a) /
            len(union_a)
            if union_a
            else 0.0
        )

        # 20 V5 features
        rows.append([
            # 1
            int(
                bool(name1)
                and
                name1 == n2
            ),

            # 2
            int(
                bool(addr1)
                and
                addr1 == a2
            ),

            # 3
            int(
                country1 == c2
            ),

            # 4
            ns,

            # 5
            ass,

            # 6
            nt,

            # 7
            at,

            # 8
            nj,

            # 9
            aj,

            # 10
            len(shared_n),

            # 11
            len(shared_a),

            # 12
            len(
                num1 & nums(a2)
            ),

            # 13
            abs(
                len(name1) -
                len(n2)
            ),

            # 14
            abs(
                len(addr1) -
                len(a2)
            ),

            # 15
            min(
                len(name1),
                len(n2)
            )
            /
            max(
                1,
                max(
                    len(name1),
                    len(n2)
                )
            ),

            # 16
            min(
                len(addr1),
                len(a2)
            )
            /
            max(
                1,
                max(
                    len(addr1),
                    len(a2)
                )
            ),

            # 17
            abs(
                ns - ass
            ),

            # 18
            max(
                ns,
                ass
            ),

            # 19
            ns * ass,

            # 20
            abs(
                nt - at
            )
        ])

    return np.asarray(
        rows,
        dtype=np.float32
    )


# ============================================================
# RUN 5,000 S1
# ============================================================

print("\n" + "=" * 75)
print("STARTING 5,000-S1 INFERENCE")
print("=" * 75)

start = time.time()

total_candidates = 0
total_predictions = 0

candidate_counts = []
prediction_counts = []

for i, row in enumerate(
    s1.itertuples(index=False),
    1
):

    name1 = norm(
        row.business_name
    )

    addr1 = norm(
        row.business_address
    )

    country1 = (
        str(row.country)
        .lower()
        .strip()
    )

    # Candidate generation
    candidates = generate_v5_candidates_s2(
        name1,
        addr1,
        country1
    )

    candidate_count = len(candidates)

    candidate_counts.append(
        candidate_count
    )

    total_candidates += candidate_count

    # V5 prediction
    if candidate_count > 0:

        X = make_features_s2(
            name1,
            addr1,
            country1,
            candidates
        )

        probs = v5_model.predict_proba(
            X
        )[:, 1]

        prediction_count = int(
            (probs >= THRESHOLD).sum()
        )

    else:

        prediction_count = 0

    prediction_counts.append(
        prediction_count
    )

    total_predictions += prediction_count

    # Progress
    if i % 500 == 0:

        elapsed = time.time() - start

        print(
            f"{i:,}/{N_S1:,} | "
            f"avg candidates/S1="
            f"{total_candidates / i:.2f} | "
            f"matches="
            f"{total_predictions:,} | "
            f"matches/S1="
            f"{total_predictions / i:.3f} | "
            f"time="
            f"{elapsed:.2f}s | "
            f"RAM="
            f"{ram_gb():.2f} GB"
        )


# ============================================================
# FINAL SUMMARY
# ============================================================

elapsed = time.time() - start

print("\n")
print("=" * 75)
print("V5 S2 — 5,000 S1 SMOKE TEST COMPLETE")
print("=" * 75)

print(
    f"S1 processed:        {len(s1):,}"
)

print(
    f"Total candidates:    {total_candidates:,}"
)

print(
    f"Candidates/S1:       "
    f"{total_candidates / len(s1):.3f}"
)

print(
    f"Median candidates:   "
    f"{np.median(candidate_counts):.3f}"
)

print(
    f"P90 candidates:     "
    f"{np.percentile(candidate_counts, 90):.3f}"
)

print(
    f"P95 candidates:     "
    f"{np.percentile(candidate_counts, 95):.3f}"
)

print(
    f"P99 candidates:     "
    f"{np.percentile(candidate_counts, 99):.3f}"
)

print(
    f"Maximum candidates: "
    f"{max(candidate_counts):,}"
)

print(
    f"Predicted matches:  "
    f"{total_predictions:,}"
)

print(
    f"Matches/S1:         "
    f"{total_predictions / len(s1):.3f}"
)

print(
    f"Median matches/S1:  "
    f"{np.median(prediction_counts):.3f}"
)

print(
    f"Runtime:            "
    f"{elapsed:.2f} sec"
)

print(
    f"Time/S1:            "
    f"{elapsed / len(s1):.5f} sec"
)

print(
    f"Available RAM:      "
    f"{ram_gb():.2f} GB"
)

print("=" * 75)
```

```python
# ============================================================
# V5 PRODUCTION INFERENCE — FULL TEST
# 1,732,544 S1 × S2/S3
#
# CHECKPOINTED + MEMORY SAFE
# Writes:
#   output/matching_results.tsv
#   output/candidate_pairs.tsv
# ============================================================

import os
import gc
import csv
import time
import pickle
import psutil
import unicodedata
import re

import pandas as pd
import numpy as np

from rapidfuzz.fuzz import ratio
from catboost import CatBoostClassifier


# ============================================================
# CONFIG
# ============================================================

BASE = "/content/amazon_ml_challenge/student_resource"

INDEX_DIR = (
    "/content/drive/MyDrive/"
    "Amazon-ML-Challenge-2026/ml/V5_full_indexes"
)

MODEL_PATH = (
    "/content/drive/MyDrive/"
    "Amazon-ML-Challenge-2026/ml/"
    "V5_ML_Final/V5_ML_Final/v5_model.cbm"
)

OUTPUT_DIR = f"{BASE}/output"
WORK_DIR = (
    "/content/drive/MyDrive/"
    "Amazon-ML-Challenge-2026/ml/V5_Final_Production"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(WORK_DIR, exist_ok=True)

THRESHOLD = 0.60
CHUNK_SIZE = 25000


# ============================================================
# OUTPUT FILES
# ============================================================

FINAL_MATCHING = f"{OUTPUT_DIR}/matching_results.tsv"
FINAL_CANDIDATES = f"{OUTPUT_DIR}/candidate_pairs.tsv"

S2_WORK = f"{WORK_DIR}/s2_results.tsv"
S3_WORK = f"{WORK_DIR}/s3_results.tsv"

CHECKPOINT_S2 = f"{WORK_DIR}/s2_checkpoint.txt"
CHECKPOINT_S3 = f"{WORK_DIR}/s3_checkpoint.txt"


# ============================================================
# MEMORY
# ============================================================

def ram_gb():
    return psutil.virtual_memory().available / (1024**3)


print("=" * 80)
print("V5 FULL PRODUCTION INFERENCE")
print("=" * 80)
print("Available RAM:", round(ram_gb(), 2), "GB")
print("Chunk size:", CHUNK_SIZE)
print("Threshold:", THRESHOLD)


# ============================================================
# NORMALIZATION
# ============================================================

def norm(x):

    if x is None:
        return ""

    x = str(x).lower().strip()

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        x = x.replace(old, new)

    x = unicodedata.normalize("NFKC", x)

    x = re.sub(r"[^\w\s]", " ", x)
    x = re.sub(r"\s+", " ", x)

    return x.strip()


def toks(x):
    return set(x.split()) if x else set()


def nums(x):
    return set(re.findall(r"\d+", x))


# ============================================================
# V5 MODEL
# ============================================================

print("\nLoading V5 model...")

v5_model = CatBoostClassifier()
v5_model.load_model(MODEL_PATH)

print("Model loaded:", v5_model.is_fitted())
print("Trees:", v5_model.tree_count_)


# ============================================================
# SOURCE PROCESSOR
# ============================================================

def process_source(
    source_number,
    source_file,
    name_index_file,
    addr_index_file,
    name_freq_file,
    addr_freq_file,
    work_file,
    checkpoint_file
):

    print("\n")
    print("=" * 80)
    print(f"PROCESSING SOURCE {source_number}")
    print("=" * 80)

    # --------------------------------------------------------
    # Load source
    # --------------------------------------------------------

    print("Loading source...")

    source = pd.read_csv(
        source_file,
        sep="\t",
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ],
        dtype=str,
        keep_default_na=False
    )

    print("Source shape:", source.shape)
    print("RAM:", round(ram_gb(), 2), "GB")

    # --------------------------------------------------------
    # Prepare arrays
    # --------------------------------------------------------

    print("Normalizing source arrays...")

    source_names = (
        source["business_name"]
        .map(norm)
        .to_numpy()
    )

    source_addrs = (
        source["business_address"]
        .map(norm)
        .to_numpy()
    )

    source_countries = (
        source["country"]
        .astype(str)
        .str.lower()
        .str.strip()
        .to_numpy()
    )

    source_ids = source["entity_id"].to_numpy()

    print("Arrays ready.")
    print("RAM:", round(ram_gb(), 2), "GB")

    # --------------------------------------------------------
    # Load indexes
    # --------------------------------------------------------

    print("Loading indexes...")

    with open(name_index_file, "rb") as f:
        name_index = pickle.load(f)

    with open(addr_index_file, "rb") as f:
        addr_index = pickle.load(f)

    with open(name_freq_file, "rb") as f:
        name_freq = pickle.load(f)

    with open(addr_freq_file, "rb") as f:
        addr_freq = pickle.load(f)

    print("Indexes loaded.")
    print("RAM:", round(ram_gb(), 2), "GB")

    # --------------------------------------------------------
    # Candidate generator
    # --------------------------------------------------------

    def generate_candidates(
        name1,
        addr1,
        country1
    ):

        t_name = toks(name1)
        t_addr = toks(addr1)

        name_hits = {}
        addr_hits = {}

        # NAME BLOCK
        for token in t_name:

            if name_freq.get(
                token,
                10**9
            ) <= 500:

                for idx in name_index.get(
                    token,
                    ()
                ):

                    if source_countries[idx] == country1:

                        name_hits[idx] = (
                            name_hits.get(
                                idx,
                                0
                            ) + 1
                        )

        # ADDRESS BLOCK
        for token in t_addr:

            if addr_freq.get(
                token,
                10**9
            ) <= 500:

                for idx in addr_index.get(
                    token,
                    ()
                ):

                    if source_countries[idx] == country1:

                        addr_hits[idx] = (
                            addr_hits.get(
                                idx,
                                0
                            ) + 1
                        )

        # Require >=2 shared tokens
        candidate_ids = (
            {
                idx
                for idx, count in name_hits.items()
                if count >= 2
            }
            |
            {
                idx
                for idx, count in addr_hits.items()
                if count >= 2
            }
        )

        # Final V5 candidate filtering
        selected = []

        for idx in candidate_ids:

            n2 = source_names[idx]
            a2 = source_addrs[idx]

            # Exact
            if (
                (name1 and name1 == n2)
                or
                (addr1 and addr1 == a2)
            ):

                selected.append(idx)
                continue

            t2n = toks(n2)
            t2a = toks(a2)

            shared_name = len(
                t_name & t2n
            )

            shared_addr = len(
                t_addr & t2a
            )

            name_sim = (
                ratio(name1, n2) / 100
                if name1 and n2
                else 0.0
            )

            addr_sim = (
                ratio(addr1, a2) / 100
                if addr1 and a2
                else 0.0
            )

            keep = (
                (
                    shared_name >= 2
                    and name_sim >= 0.60
                )
                or
                shared_addr >= 3
                or
                (
                    shared_addr >= 2
                    and addr_sim >= 0.30
                )
            )

            if keep:
                selected.append(idx)

        return selected

    # --------------------------------------------------------
    # Feature builder
    # --------------------------------------------------------

    def make_features(
        name1,
        addr1,
        country1,
        candidate_indices
    ):

        t1n = toks(name1)
        t1a = toks(addr1)
        num1 = nums(addr1)

        rows = []

        for idx in candidate_indices:

            n2 = source_names[idx]
            a2 = source_addrs[idx]
            c2 = source_countries[idx]

            t2n = toks(n2)
            t2a = toks(a2)

            shared_n = t1n & t2n
            shared_a = t1a & t2a

            union_n = t1n | t2n
            union_a = t1a | t2a

            ns = (
                ratio(name1, n2) / 100
                if name1 and n2
                else 0.0
            )

            ass = (
                ratio(addr1, a2) / 100
                if addr1 and a2
                else 0.0
            )

            nt = (
                len(shared_n) /
                max(
                    1,
                    min(
                        len(t1n),
                        len(t2n)
                    )
                )
            )

            at = (
                len(shared_a) /
                max(
                    1,
                    min(
                        len(t1a),
                        len(t2a)
                    )
                )
            )

            nj = (
                len(shared_n) /
                len(union_n)
                if union_n
                else 0.0
            )

            aj = (
                len(shared_a) /
                len(union_a)
                if union_a
                else 0.0
            )

            rows.append([
                int(
                    bool(name1)
                    and name1 == n2
                ),
                int(
                    bool(addr1)
                    and addr1 == a2
                ),
                int(
                    country1 == c2
                ),
                ns,
                ass,
                nt,
                at,
                nj,
                aj,
                len(shared_n),
                len(shared_a),
                len(num1 & nums(a2)),
                abs(
                    len(name1) -
                    len(n2)
                ),
                abs(
                    len(addr1) -
                    len(a2)
                ),
                min(
                    len(name1),
                    len(n2)
                ) /
                max(
                    1,
                    max(
                        len(name1),
                        len(n2)
                    )
                ),
                min(
                    len(addr1),
                    len(a2)
                ) /
                max(
                    1,
                    max(
                        len(addr1),
                        len(a2)
                    )
                ),
                abs(ns - ass),
                max(ns, ass),
                ns * ass,
                abs(nt - at)
            ])

        return np.asarray(
            rows,
            dtype=np.float32
        )

    # --------------------------------------------------------
    # Resume checkpoint
    # --------------------------------------------------------

    start_row = 0

    if os.path.exists(checkpoint_file):

        with open(
            checkpoint_file,
            "r"
        ) as f:

            text = f.read().strip()

            if text:
                start_row = int(text)

        print(
            "Resuming from row:",
            start_row
        )

    # --------------------------------------------------------
    # Work file
    # --------------------------------------------------------

    file_mode = (
        "a"
        if start_row > 0
        else "w"
    )

    with open(
        work_file,
        file_mode,
        encoding="utf-8",
        newline=""
    ) as fout:

        writer = csv.writer(
            fout,
            delimiter="\t",
            lineterminator="\n"
        )

        if start_row == 0:

            writer.writerow([
                "source1_entity_id",
                "candidate_entity_ids",
                "matched_entity_ids"
            ])

        total_candidates = 0
        total_matches = 0
        processed = start_row

        start_time = time.time()

        # ----------------------------------------------------
        # Read S1 in chunks
        # ----------------------------------------------------

        for s1_chunk in pd.read_csv(
            f"{BASE}/dataset/test/test_source1.tsv",
            sep="\t",
            dtype=str,
            keep_default_na=False,
            chunksize=CHUNK_SIZE,
            skiprows=range(
                1,
                start_row + 1
            )
            if start_row > 0
            else None
        ):

            for row in s1_chunk.itertuples(
                index=False
            ):

                name1 = norm(
                    row.business_name
                )

                addr1 = norm(
                    row.business_address
                )

                country1 = (
                    str(row.country)
                    .lower()
                    .strip()
                )

                candidates = generate_candidates(
                    name1,
                    addr1,
                    country1
                )

                candidate_ids = [
                    source_ids[i]
                    for i in candidates
                ]

                matches = []

                if candidates:

                    X = make_features(
                        name1,
                        addr1,
                        country1,
                        candidates
                    )

                    probs = (
                        v5_model
                        .predict_proba(X)[:, 1]
                    )

                    matches = [
                        source_ids[idx]
                        for idx, p in zip(
                            candidates,
                            probs
                        )
                        if p >= THRESHOLD
                    ]

                # Remove duplicates defensively
                candidate_ids = list(
                    dict.fromkeys(
                        candidate_ids
                    )
                )

                matches = list(
                    dict.fromkeys(
                        matches
                    )
                )

                writer.writerow([
                    row.entity_id,
                    ",".join(candidate_ids),
                    ",".join(matches)
                ])

                processed += 1
                total_candidates += len(
                    candidate_ids
                )
                total_matches += len(
                    matches
                )

            fout.flush()

            # Checkpoint
            with open(
                checkpoint_file,
                "w"
            ) as cf:

                cf.write(
                    str(processed)
                )

            elapsed = time.time() - start_time

            print(
                f"{processed:,}/1,732,544 | "
                f"candidates={total_candidates:,} | "
                f"matches={total_matches:,} | "
                f"time={elapsed/60:.1f} min | "
                f"RAM={ram_gb():.2f} GB"
            )

    # --------------------------------------------------------
    # Cleanup source
    # --------------------------------------------------------

    del source
    del source_names
    del source_addrs
    del source_countries
    del source_ids
    del name_index
    del addr_index
    del name_freq
    del addr_freq

    gc.collect()

    print(
        f"SOURCE {source_number} COMPLETE"
    )


# ============================================================
# RUN S2
# ============================================================

process_source(
    2,

    f"{BASE}/dataset/test/test_source2.tsv",

    f"{INDEX_DIR}/s2_name_index.pkl",
    f"{INDEX_DIR}/s2_addr_index.pkl",

    f"{INDEX_DIR}/s2_name_freq.pkl",
    f"{INDEX_DIR}/s2_addr_freq.pkl",

    S2_WORK,
    CHECKPOINT_S2
)


# ============================================================
# CLEAN MEMORY
# ============================================================

gc.collect()

print(
    "\nAfter S2 cleanup RAM:",
    round(ram_gb(), 2),
    "GB"
)


# ============================================================
# RUN S3
# ============================================================

process_source(
    3,

    f"{BASE}/dataset/test/test_source3.tsv",

    f"{INDEX_DIR}/s3_name_index.pkl",
    f"{INDEX_DIR}/s3_addr_index.pkl",

    f"{INDEX_DIR}/s3_name_freq.pkl",
    f"{INDEX_DIR}/s3_addr_freq.pkl",

    S3_WORK,
    CHECKPOINT_S3
)


# ============================================================
# COMBINE S2 + S3
# ============================================================

print("\n")
print("=" * 80)
print("COMBINING S2 + S3 RESULTS")
print("=" * 80)

# S2/S3 work files each contain exactly one row per S1.
# Merge chunk-by-chunk using the S1 row order.

s2_df = pd.read_csv(
    S2_WORK,
    sep="\t",
    dtype=str,
    keep_default_na=False
)

s3_df = pd.read_csv(
    S3_WORK,
    sep="\t",
    dtype=str,
    keep_default_na=False
)

print("S2 rows:", len(s2_df))
print("S3 rows:", len(s3_df))

assert len(s2_df) == len(s3_df)

assert (
    s2_df["source1_entity_id"].values
    ==
    s3_df["source1_entity_id"].values
).all()

# ------------------------------------------------------------
# Final output
# ------------------------------------------------------------

with open(
    FINAL_MATCHING,
    "w",
    encoding="utf-8",
    newline=""
) as fm, open(
    FINAL_CANDIDATES,
    "w",
    encoding="utf-8",
    newline=""
) as fc:

    mw = csv.writer(
        fm,
        delimiter="\t",
        lineterminator="\n"
    )

    cw = csv.writer(
        fc,
        delimiter="\t",
        lineterminator="\n"
    )

    mw.writerow([
        "source1_entity_id",
        "matched_entity_ids"
    ])

    cw.writerow([
        "source1_entity_id",
        "candidate_entity_ids"
    ])

    for i in range(len(s2_df)):

        s1_id = s2_df.iloc[
            i
        ]["source1_entity_id"]

        s2_candidates = (
            s2_df.iloc[i]
            ["candidate_entity_ids"]
        )

        s3_candidates = (
            s3_df.iloc[i]
            ["candidate_entity_ids"]
        )

        s2_matches = (
            s2_df.iloc[i]
            ["matched_entity_ids"]
        )

        s3_matches = (
            s3_df.iloc[i]
            ["matched_entity_ids"]
        )

        # Combine candidate IDs
        candidates = []

        if s2_candidates:
            candidates.extend(
                s2_candidates.split(",")
            )

        if s3_candidates:
            candidates.extend(
                s3_candidates.split(",")
            )

        candidates = list(
            dict.fromkeys(
                x for x in candidates
                if x
            )
        )

        # Combine predicted matches
        matches = []

        if s2_matches:
            matches.extend(
                s2_matches.split(",")
            )

        if s3_matches:
            matches.extend(
                s3_matches.split(",")
            )

        matches = list(
            dict.fromkeys(
                x for x in matches
                if x
            )
        )

        cw.writerow([
            s1_id,
            ",".join(candidates)
        ])

        mw.writerow([
            s1_id,
            ",".join(matches)
        ])

print("\nFINAL FILES CREATED")
print(FINAL_MATCHING)
print(FINAL_CANDIDATES)

print("\nSizes:")
print(
    "matching:",
    round(
        os.path.getsize(
            FINAL_MATCHING
        ) / (1024**2),
        2
    ),
    "MB"
)

print(
    "candidate:",
    round(
        os.path.getsize(
            FINAL_CANDIDATES
        ) / (1024**2),
        2
    ),
    "MB"
)

print("\nPRODUCTION INFERENCE COMPLETE.")
```

```python
# ============================================================
# RAM-SAFE V5 PRODUCTION SETUP
# ============================================================

import os, gc, re, csv, pickle, unicodedata, time
import numpy as np
import pandas as pd

from catboost import CatBoostClassifier
from rapidfuzz.fuzz import ratio

# Paths
BASE = "/content/amazon_ml_challenge/student_resource"
TEST_DIR = f"{BASE}/dataset/test"
OUT_DIR = f"{BASE}/output"

DRIVE_ML = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml"
INDEX_DIR = f"{DRIVE_ML}/V5_full_indexes"
MODEL_PATH = f"{DRIVE_ML}/V5_ML_Final/V5_ML_Final/v5_model.cbm"

WORK_DIR = f"{DRIVE_ML}/V5_Final_Production_RAMSAFE"
os.makedirs(WORK_DIR, exist_ok=True)
os.makedirs(OUT_DIR, exist_ok=True)

S2_WORK = f"{WORK_DIR}/s2_results.tsv"
S3_WORK = f"{WORK_DIR}/s3_results.tsv"

THRESHOLD = 0.60

# VERY IMPORTANT: smaller than previous run
CHUNK_SIZE = 2000

print("Setup complete.")
print("RAM-safe chunk size:", CHUNK_SIZE)
print("Model exists:", os.path.exists(MODEL_PATH))
print("Index directory exists:", os.path.exists(INDEX_DIR))
print("Test directory exists:", os.path.exists(TEST_DIR))

# Show current RAM
try:
    import psutil
    vm = psutil.virtual_memory()
    print(f"RAM available: {vm.available/1024**3:.2f} GB / {vm.total/1024**3:.2f} GB")
except:
    pass
```

```python
# ============================================================
# STEP 4 — RAM-SAFE S2 PRODUCTION
# Same V5 logic / threshold / candidate rules
# ============================================================

import re
import unicodedata
import csv
import gc
import time
import psutil
import numpy as np
import pandas as pd

S2_WORK = f"{WORK_DIR}/s2_results.tsv"

# ------------------------------------------------------------
# Normalization
# ------------------------------------------------------------
def normalize_advanced(text):
    if pd.isna(text):
        return ""

    text = str(text).lower().strip()
    text = unicodedata.normalize("NFKC", text)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()


def tokens(text):
    return set(text.split()) if text else set()


# ------------------------------------------------------------
# V5 feature calculation
# ------------------------------------------------------------
def make_features(s1_name, s1_addr, s1_country,
                  s2_name, s2_addr, s2_country):

    n1 = normalize_advanced(s1_name)
    a1 = normalize_advanced(s1_addr)

    n2 = normalize_advanced(s2_name)
    a2 = normalize_advanced(s2_addr)

    nt1 = tokens(n1)
    nt2 = tokens(n2)
    at1 = tokens(a1)
    at2 = tokens(a2)

    name_overlap = len(nt1 & nt2)
    addr_overlap = len(at1 & at2)

    # Numeric address tokens
    nums1 = set(re.findall(r"\b\d+\b", a1))
    nums2 = set(re.findall(r"\b\d+\b", a2))

    shared_addr_numbers = len(nums1 & nums2)

    name_sim = ratio(n1, n2) / 100.0
    addr_sim = ratio(a1, a2) / 100.0

    name_len_1 = len(n1)
    name_len_2 = len(n2)
    addr_len_1 = len(a1)
    addr_len_2 = len(a2)

    name_len_diff = abs(name_len_1 - name_len_2)
    addr_len_diff = abs(addr_len_1 - addr_len_2)

    name_len_ratio = (
        min(name_len_1, name_len_2) / max(name_len_1, name_len_2)
        if max(name_len_1, name_len_2) > 0 else 1.0
    )

    addr_len_ratio = (
        min(addr_len_1, addr_len_2) / max(addr_len_1, addr_len_2)
        if max(addr_len_1, addr_len_2) > 0 else 1.0
    )

    similarity_gap = abs(name_sim - addr_sim)
    max_similarity = max(name_sim, addr_sim)
    similarity_product = name_sim * addr_sim
    token_overlap_gap = abs(name_overlap - addr_overlap)

    return [
        int(n1 == n2),
        int(a1 == a2),
        int(str(s1_country).strip().lower() ==
            str(s2_country).strip().lower()),

        name_sim,
        addr_sim,

        name_overlap / max(len(nt1), len(nt2), 1),
        addr_overlap / max(len(at1), len(at2), 1),

        name_overlap / max(len(nt1 | nt2), 1),
        addr_overlap / max(len(at1 | at2), 1),

        name_overlap,
        addr_overlap,
        shared_addr_numbers,

        name_len_diff,
        addr_len_diff,
        name_len_ratio,
        addr_len_ratio,

        similarity_gap,
        max_similarity,
        similarity_product,
        token_overlap_gap
    ]


# ------------------------------------------------------------
# Candidate generator
# EXACT production rules benchmarked earlier
# ------------------------------------------------------------
def generate_candidates(s1_name, s1_addr, s1_country):

    n1 = normalize_advanced(s1_name)
    a1 = normalize_advanced(s1_addr)

    name_tok = tokens(n1)
    addr_tok = tokens(a1)

    candidate_counts = {}

    # Rare token block: frequency <= 500
    for tok in name_tok:
        if s2_name_freq.get(tok, 10**9) <= 500:
            ids = s2_name_index.get(tok)
            if ids is not None:
                for idx in ids:
                    candidate_counts[idx] = candidate_counts.get(idx, 0) + 1

    for tok in addr_tok:
        if s2_addr_freq.get(tok, 10**9) <= 500:
            ids = s2_addr_index.get(tok)
            if ids is not None:
                for idx in ids:
                    candidate_counts[idx] = candidate_counts.get(idx, 0) + 1

    # Need at least 2 shared tokens to enter the final block
    candidate_indices = [
        idx for idx, count in candidate_counts.items()
        if count >= 2
    ]

    if not candidate_indices:
        return []

    # --------------------------------------------------------
    # Final hybrid rules
    # 1 exact name + country
    # 2 exact address + country
    # 3 >=2 shared name tokens + name sim >= .60
    # 4 >=3 shared address tokens
    # 5 >=2 shared address tokens + address sim >= .30
    # --------------------------------------------------------
    final_indices = []

    s1_country_norm = str(s1_country).strip().lower()

    for idx in candidate_indices:

        row = s2.iloc[idx]

        s2_name = row["business_name"]
        s2_addr = row["business_address"]
        s2_country = row["country"]

        n2 = normalize_advanced(s2_name)
        a2 = normalize_advanced(s2_addr)

        nt2 = tokens(n2)
        at2 = tokens(a2)

        name_shared = len(name_tok & nt2)
        addr_shared = len(addr_tok & at2)

        country_same = (
            s1_country_norm ==
            str(s2_country).strip().lower()
        )

        name_exact = (n1 == n2)
        addr_exact = (a1 == a2)

        name_sim = ratio(n1, n2) / 100.0
        addr_sim = ratio(a1, a2) / 100.0

        keep = (
            (name_exact and country_same)
            or
            (addr_exact and country_same)
            or
            (name_shared >= 2 and name_sim >= 0.60)
            or
            (addr_shared >= 3)
            or
            (addr_shared >= 2 and addr_sim >= 0.30)
        )

        if keep:
            final_indices.append(idx)

    return final_indices


# ------------------------------------------------------------
# Prepare output
# ------------------------------------------------------------
if os.path.exists(S2_WORK):
    print("⚠️ Existing S2 work file found:")
    print(S2_WORK)
    print("Delete/rename it before starting if it is from an old run.")
    raise RuntimeError("Existing S2 output detected. Not overwriting it.")

S1_PATH = f"{TEST_DIR}/test_source1.tsv"

# Header
with open(S2_WORK, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f, delimiter="\t")
    writer.writerow([
        "source1_entity_id",
        "candidate_entity_ids",
        "matched_entity_ids"
    ])

# ------------------------------------------------------------
# Process S1 in very small chunks
# ------------------------------------------------------------
total_s1 = 0
total_candidates = 0
total_matches = 0
start_time = time.time()

print("\n🚀 Starting S2 production...")
print("S1 chunk size:", CHUNK_SIZE)
print("Threshold:", THRESHOLD)

for chunk_no, s1_chunk in enumerate(
    pd.read_csv(
        S1_PATH,
        sep="\t",
        dtype=str,
        keep_default_na=False,
        chunksize=CHUNK_SIZE
    ),
    start=1
):

    chunk_start = time.time()

    output_rows = []

    for _, row in s1_chunk.iterrows():

        s1_id = row["entity_id"]
        s1_name = row["business_name"]
        s1_addr = row["business_address"]
        s1_country = row["country"]

        candidate_indices = generate_candidates(
            s1_name,
            s1_addr,
            s1_country
        )

        candidate_ids = []
        matched_ids = []

        if candidate_indices:

            feature_rows = []

            for idx in candidate_indices:

                s2_row = s2.iloc[idx]

                feature_rows.append(
                    make_features(
                        s1_name,
                        s1_addr,
                        s1_country,
                        s2_row["business_name"],
                        s2_row["business_address"],
                        s2_row["country"]
                    )
                )

                candidate_ids.append(
                    str(s2_row["entity_id"])
                )

            X = np.asarray(
                feature_rows,
                dtype=np.float32
            )

            probs = model.predict_proba(X)[:, 1]

            for cid, p in zip(candidate_ids, probs):
                if p >= THRESHOLD:
                    matched_ids.append(cid)

            del X, feature_rows, probs

        # Deduplicate while preserving order
        candidate_ids = list(dict.fromkeys(candidate_ids))
        matched_ids = list(dict.fromkeys(matched_ids))

        output_rows.append([
            s1_id,
            ",".join(candidate_ids),
            ",".join(matched_ids)
        ])

        total_s1 += 1
        total_candidates += len(candidate_ids)
        total_matches += len(matched_ids)

    # Write chunk immediately
    with open(
        S2_WORK,
        "a",
        newline="",
        encoding="utf-8"
    ) as f:

        writer = csv.writer(f, delimiter="\t")
        writer.writerows(output_rows)

    # Release chunk memory
    del output_rows
    del s1_chunk
    gc.collect()

    elapsed = time.time() - start_time
    chunk_time = time.time() - chunk_start

    vm = psutil.virtual_memory()

    print(
        f"Chunk {chunk_no:4d} | "
        f"S1 {total_s1:8d} | "
        f"Candidates {total_candidates:10d} | "
        f"Matches {total_matches:10d} | "
        f"Chunk {chunk_time:6.1f}s | "
        f"RAM free {vm.available/1024**3:5.2f} GB | "
        f"Elapsed {elapsed/60:6.1f} min"
    )

print("\n✅ S2 PRODUCTION COMPLETE")

print("S1 processed:", total_s1)
print("Total candidates:", total_candidates)
print("Total matches:", total_matches)
print("Avg candidates/S1:", total_candidates / max(total_s1, 1))
print("Avg matches/S1:", total_matches / max(total_s1, 1))
print("Output:", S2_WORK)

ram_status("After S2 production")
```

```python
# ============================================================
# STEP 5 — CLEAN S2 + LOAD S3
# ============================================================

import gc
import pickle
import psutil
import pandas as pd

# Free S2 memory
del s2
del s2_name_index
del s2_addr_index
del s2_name_freq
del s2_addr_freq

gc.collect()

vm = psutil.virtual_memory()
print(f"After S2 cleanup: {vm.available/1024**3:.2f} GB RAM free")

# ------------------------------------------------------------
# Load S3
# ------------------------------------------------------------
S3_PATH = f"{TEST_DIR}/test_source3.tsv"

s3 = pd.read_csv(
    S3_PATH,
    sep="\t",
    dtype=str,
    keep_default_na=False
)

print("✅ S3 loaded")
print("S3 shape:", s3.shape)

# ------------------------------------------------------------
# Load ONLY S3 indexes
# ------------------------------------------------------------
with open(f"{INDEX_DIR}/s3_name_index.pkl", "rb") as f:
    s3_name_index = pickle.load(f)

with open(f"{INDEX_DIR}/s3_addr_index.pkl", "rb") as f:
    s3_addr_index = pickle.load(f)

with open(f"{INDEX_DIR}/s3_name_freq.pkl", "rb") as f:
    s3_name_freq = pickle.load(f)

with open(f"{INDEX_DIR}/s3_addr_freq.pkl", "rb") as f:
    s3_addr_freq = pickle.load(f)

print("✅ S3 indexes loaded")

gc.collect()

vm = psutil.virtual_memory()
print(f"After S3 indexes: {vm.available/1024**3:.2f} GB RAM free")

print("S3 rows:", len(s3))
print("S3 name index entries:", len(s3_name_index))
print("S3 address index entries:", len(s3_addr_index))
```

```python
# ============================================================
# STEP 6 — RAM-SAFE S3 PRODUCTION
# V5 MODEL | THRESHOLD 0.60
# ============================================================

import os
import re
import csv
import gc
import time
import psutil
import numpy as np
import pandas as pd
import unicodedata

from rapidfuzz.fuzz import ratio

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------
S3_PATH = f"{TEST_DIR}/test_source3.tsv"
S3_WORK = f"{WORK_DIR}/s3_results.tsv"

S1_PATH = f"{TEST_DIR}/test_source1.tsv"

THRESHOLD = 0.60
CHUNK_SIZE = 2000

# ------------------------------------------------------------
# Safety check
# ------------------------------------------------------------
if os.path.exists(S3_WORK):
    raise RuntimeError(
        f"""
Existing S3 output found:

{S3_WORK}

This code will NOT overwrite it.
If this is an old/incomplete run, delete it first.
"""
    )

# ------------------------------------------------------------
# RAM helper
# ------------------------------------------------------------
def show_ram(label=""):
    gc.collect()
    vm = psutil.virtual_memory()
    print(
        f"💾 {label} RAM available: "
        f"{vm.available/1024**3:.2f} GB / "
        f"{vm.total/1024**3:.2f} GB"
    )

# ------------------------------------------------------------
# NORMALIZATION
# Same normalization used by V5 pipeline
# ------------------------------------------------------------
def normalize_advanced(text):

    if pd.isna(text):
        return ""

    text = str(text).lower().strip()

    text = unicodedata.normalize(
        "NFKC",
        text
    )

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    text = re.sub(
        r"[^\w\s]",
        " ",
        text
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


def tokens(text):

    if not text:
        return set()

    return set(text.split())


# ============================================================
# V5 FEATURE BUILDER — 20 FEATURES
# ============================================================

def make_features(
    s1_name,
    s1_addr,
    s1_country,
    s3_name,
    s3_addr,
    s3_country
):

    n1 = normalize_advanced(s1_name)
    a1 = normalize_advanced(s1_addr)

    n3 = normalize_advanced(s3_name)
    a3 = normalize_advanced(s3_addr)

    # Tokens
    nt1 = tokens(n1)
    nt3 = tokens(n3)

    at1 = tokens(a1)
    at3 = tokens(a3)

    # Shared tokens
    name_shared = len(nt1 & nt3)
    addr_shared = len(at1 & at3)

    # Shared address numbers
    nums1 = set(
        re.findall(
            r"\b\d+\b",
            a1
        )
    )

    nums3 = set(
        re.findall(
            r"\b\d+\b",
            a3
        )
    )

    shared_addr_numbers = len(
        nums1 & nums3
    )

    # Similarity
    name_sim = ratio(
        n1,
        n3
    ) / 100.0

    addr_sim = ratio(
        a1,
        a3
    ) / 100.0

    # Lengths
    name_len_1 = len(n1)
    name_len_3 = len(n3)

    addr_len_1 = len(a1)
    addr_len_3 = len(a3)

    name_len_diff = abs(
        name_len_1 - name_len_3
    )

    addr_len_diff = abs(
        addr_len_1 - addr_len_3
    )

    name_len_ratio = (
        min(name_len_1, name_len_3)
        /
        max(name_len_1, name_len_3)
        if max(name_len_1, name_len_3) > 0
        else 1.0
    )

    addr_len_ratio = (
        min(addr_len_1, addr_len_3)
        /
        max(addr_len_1, addr_len_3)
        if max(addr_len_1, addr_len_3) > 0
        else 1.0
    )

    # V5 engineered features
    similarity_gap = abs(
        name_sim - addr_sim
    )

    max_similarity = max(
        name_sim,
        addr_sim
    )

    similarity_product = (
        name_sim * addr_sim
    )

    token_overlap_gap = abs(
        name_shared - addr_shared
    )

    # EXACT 20-FEATURE ORDER
    return [
        # 1
        int(n1 == n3),

        # 2
        int(a1 == a3),

        # 3
        int(
            str(s1_country).strip().lower()
            ==
            str(s3_country).strip().lower()
        ),

        # 4
        name_sim,

        # 5
        addr_sim,

        # 6
        name_shared /
        max(len(nt1), len(nt3), 1),

        # 7
        addr_shared /
        max(len(at1), len(at3), 1),

        # 8
        name_shared /
        max(len(nt1 | nt3), 1),

        # 9
        addr_shared /
        max(len(at1 | at3), 1),

        # 10
        name_shared,

        # 11
        addr_shared,

        # 12
        shared_addr_numbers,

        # 13
        name_len_diff,

        # 14
        addr_len_diff,

        # 15
        name_len_ratio,

        # 16
        addr_len_ratio,

        # 17
        similarity_gap,

        # 18
        max_similarity,

        # 19
        similarity_product,

        # 20
        token_overlap_gap
    ]


# ============================================================
# S3 CANDIDATE GENERATOR
# SAME RULES USED FOR S2
# ============================================================

def generate_candidates_s3(
    s1_name,
    s1_addr,
    s1_country
):

    # Normalize S1
    n1 = normalize_advanced(
        s1_name
    )

    a1 = normalize_advanced(
        s1_addr
    )

    name_tok = tokens(n1)
    addr_tok = tokens(a1)

    # --------------------------------------------------------
    # Candidate counts
    # --------------------------------------------------------
    candidate_counts = {}

    # --------------------------------------------------------
    # NAME RARE TOKEN BLOCK
    # Frequency <= 500
    # --------------------------------------------------------
    for tok in name_tok:

        if s3_name_freq.get(
            tok,
            10**9
        ) <= 500:

            ids = s3_name_index.get(
                tok
            )

            if ids is not None:

                for idx in ids:

                    candidate_counts[idx] = (
                        candidate_counts.get(
                            idx,
                            0
                        ) + 1
                    )

    # --------------------------------------------------------
    # ADDRESS RARE TOKEN BLOCK
    # Frequency <= 500
    # --------------------------------------------------------
    for tok in addr_tok:

        if s3_addr_freq.get(
            tok,
            10**9
        ) <= 500:

            ids = s3_addr_index.get(
                tok
            )

            if ids is not None:

                # FIXED:
                # idx comes from this loop
                for idx in ids:

                    candidate_counts[idx] = (
                        candidate_counts.get(
                            idx,
                            0
                        ) + 1
                    )

    # --------------------------------------------------------
    # Require >= 2 shared rare tokens
    # --------------------------------------------------------
    candidate_indices = [
        idx
        for idx, count
        in candidate_counts.items()
        if count >= 2
    ]

    if not candidate_indices:
        return []

    # --------------------------------------------------------
    # Final hybrid candidate rules
    #
    # 1. exact name + country
    # 2. exact address + country
    # 3. >=2 shared name tokens + name similarity >= 0.60
    # 4. >=3 shared address tokens
    # 5. >=2 shared address tokens + address similarity >= 0.30
    # --------------------------------------------------------

    final_indices = []

    s1_country_norm = (
        str(s1_country)
        .strip()
        .lower()
    )

    for idx in candidate_indices:

        row = s3.iloc[idx]

        s3_name = row[
            "business_name"
        ]

        s3_addr = row[
            "business_address"
        ]

        s3_country = row[
            "country"
        ]

        # Normalize candidate
        n3 = normalize_advanced(
            s3_name
        )

        a3 = normalize_advanced(
            s3_addr
        )

        # Candidate tokens
        nt3 = tokens(n3)
        at3 = tokens(a3)

        # Shared token counts
        name_shared = len(
            name_tok & nt3
        )

        addr_shared = len(
            addr_tok & at3
        )

        # Country
        country_same = (
            s1_country_norm
            ==
            str(s3_country)
            .strip()
            .lower()
        )

        # Exact
        name_exact = (
            n1 == n3
        )

        addr_exact = (
            a1 == a3
        )

        # Similarity
        name_sim = ratio(
            n1,
            n3
        ) / 100.0

        addr_sim = ratio(
            a1,
            a3
        ) / 100.0

        # Final rules
        keep = (
            # Rule 1
            (
                name_exact
                and
                country_same
            )

            or

            # Rule 2
            (
                addr_exact
                and
                country_same
            )

            or

            # Rule 3
            (
                name_shared >= 2
                and
                name_sim >= 0.60
            )

            or

            # Rule 4
            (
                addr_shared >= 3
            )

            or

            # Rule 5
            (
                addr_shared >= 2
                and
                addr_sim >= 0.30
            )
        )

        if keep:
            final_indices.append(
                idx
            )

    return final_indices


# ============================================================
# VERIFY S3 IS LOADED
# ============================================================

print("============================================================")
print("S3 PRODUCTION CHECK")
print("============================================================")

print(
    "S3 rows:",
    len(s3)
)

print(
    "S3 name index:",
    len(s3_name_index)
)

print(
    "S3 address index:",
    len(s3_addr_index)
)

print(
    "S3 name frequency:",
    len(s3_name_freq)
)

print(
    "S3 address frequency:",
    len(s3_addr_freq)
)

print(
    "Threshold:",
    THRESHOLD
)

print(
    "Chunk size:",
    CHUNK_SIZE
)

show_ram("Before smoke test")


# ============================================================
# SMALL S3 SMOKE TEST
# ============================================================

print("\n🔎 Running S3 smoke test on first 100 S1 rows...")

smoke_candidates = 0
smoke_matches = 0

smoke_s1 = pd.read_csv(
    S1_PATH,
    sep="\t",
    dtype=str,
    keep_default_na=False,
    nrows=100
)

for _, row in smoke_s1.iterrows():

    candidate_indices = (
        generate_candidates_s3(
            row["business_name"],
            row["business_address"],
            row["country"]
        )
    )

    smoke_candidates += len(
        candidate_indices
    )

    if candidate_indices:

        feature_rows = []
        candidate_ids = []

        for idx in candidate_indices:

            s3_row = s3.iloc[idx]

            feature_rows.append(
                make_features(
                    row["business_name"],
                    row["business_address"],
                    row["country"],
                    s3_row["business_name"],
                    s3_row["business_address"],
                    s3_row["country"]
                )
            )

            candidate_ids.append(
                str(
                    s3_row["entity_id"]
                )
            )

        X_smoke = np.asarray(
            feature_rows,
            dtype=np.float32
        )

        probs_smoke = (
            model.predict_proba(
                X_smoke
            )[:, 1]
        )

        smoke_matches += int(
            np.sum(
                probs_smoke >= THRESHOLD
            )
        )

        del X_smoke
        del probs_smoke
        del feature_rows
        del candidate_ids


del smoke_s1
gc.collect()

print(
    "✅ Smoke test complete"
)

print(
    "First 100 S1 candidates:",
    smoke_candidates
)

print(
    "First 100 S1 predicted matches:",
    smoke_matches
)

show_ram("After smoke test")


# ============================================================
# CREATE S3 OUTPUT
# ============================================================

with open(
    S3_WORK,
    "w",
    newline="",
    encoding="utf-8"
) as f:

    writer = csv.writer(
        f,
        delimiter="\t"
    )

    writer.writerow([
        "source1_entity_id",
        "candidate_entity_ids",
        "matched_entity_ids"
    ])


# ============================================================
# FULL S3 PRODUCTION
# ============================================================

total_s1 = 0
total_candidates = 0
total_matches = 0

start_time = time.time()

print("\n============================================================")
print("🚀 STARTING FULL S3 PRODUCTION")
print("============================================================")

for chunk_no, s1_chunk in enumerate(
    pd.read_csv(
        S1_PATH,
        sep="\t",
        dtype=str,
        keep_default_na=False,
        chunksize=CHUNK_SIZE
    ),
    start=1
):

    chunk_start = time.time()

    output_rows = []

    # --------------------------------------------------------
    # Process each S1 entity
    # --------------------------------------------------------
    for _, row in s1_chunk.iterrows():

        s1_id = row[
            "entity_id"
        ]

        s1_name = row[
            "business_name"
        ]

        s1_addr = row[
            "business_address"
        ]

        s1_country = row[
            "country"
        ]

        # ----------------------------------------------------
        # Candidate generation
        # ----------------------------------------------------
        candidate_indices = (
            generate_candidates_s3(
                s1_name,
                s1_addr,
                s1_country
            )
        )

        candidate_ids = []
        matched_ids = []

        # ----------------------------------------------------
        # V5 prediction
        # ----------------------------------------------------
        if candidate_indices:

            feature_rows = []

            for idx in candidate_indices:

                s3_row = s3.iloc[idx]

                feature_rows.append(
                    make_features(
                        s1_name,
                        s1_addr,
                        s1_country,
                        s3_row[
                            "business_name"
                        ],
                        s3_row[
                            "business_address"
                        ],
                        s3_row[
                            "country"
                        ]
                    )
                )

                candidate_ids.append(
                    str(
                        s3_row[
                            "entity_id"
                        ]
                    )
                )

            # Convert only current S1 candidates
            # into float32 matrix
            X = np.asarray(
                feature_rows,
                dtype=np.float32
            )

            # CatBoost prediction
            probs = model.predict_proba(
                X
            )[:, 1]

            # Threshold = 0.60
            for cid, p in zip(
                candidate_ids,
                probs
            ):

                if p >= THRESHOLD:
                    matched_ids.append(
                        cid
                    )

            # Free temporary prediction memory
            del X
            del probs
            del feature_rows

        # ----------------------------------------------------
        # Deduplicate
        # ----------------------------------------------------
        candidate_ids = list(
            dict.fromkeys(
                candidate_ids
            )
        )

        matched_ids = list(
            dict.fromkeys(
                matched_ids
            )
        )

        # ----------------------------------------------------
        # Store one S1 row
        # ----------------------------------------------------
        output_rows.append([
            s1_id,
            ",".join(
                candidate_ids
            ),
            ",".join(
                matched_ids
            )
        ])

        total_s1 += 1
        total_candidates += len(
            candidate_ids
        )
        total_matches += len(
            matched_ids
        )

    # --------------------------------------------------------
    # Write chunk immediately
    # --------------------------------------------------------
    with open(
        S3_WORK,
        "a",
        newline="",
        encoding="utf-8"
    ) as f:

        writer = csv.writer(
            f,
            delimiter="\t"
        )

        writer.writerows(
            output_rows
        )

    # --------------------------------------------------------
    # Free chunk memory
    # --------------------------------------------------------
    del output_rows
    del s1_chunk

    gc.collect()

    # --------------------------------------------------------
    # Progress
    # --------------------------------------------------------
    elapsed = (
        time.time()
        -
        start_time
    )

    chunk_time = (
        time.time()
        -
        chunk_start
    )

    vm = psutil.virtual_memory()

    print(
        f"Chunk {chunk_no:4d} | "
        f"S1 {total_s1:8d} | "
        f"Candidates {total_candidates:10d} | "
        f"Matches {total_matches:10d} | "
        f"Chunk {chunk_time:6.1f}s | "
        f"RAM free {vm.available/1024**3:5.2f} GB | "
        f"Elapsed {elapsed/60:6.1f} min"
    )


# ============================================================
# FINAL S3 SUMMARY
# ============================================================

print("\n============================================================")
print("✅ S3 PRODUCTION COMPLETE")
print("============================================================")

print(
    "S1 processed:",
    total_s1
)

print(
    "Expected S1:",
    1732544
)

print(
    "Total candidates:",
    total_candidates
)

print(
    "Total matches:",
    total_matches
)

print(
    "Avg candidates/S1:",
    total_candidates /
    max(total_s1, 1)
)

print(
    "Avg matches/S1:",
    total_matches /
    max(total_s1, 1)
)

print(
    "Output:",
    S3_WORK
)

show_ram("After S3 production")

# ============================================================
# FINAL SAFETY CHECK
# ============================================================

if total_s1 != 1732544:

    print(
        "\n⚠️ WARNING: S1 count is not 1,732,544."
    )

else:

    print(
        "\n✅ All 1,732,544 S1 entities processed."
    )
```

```python
# ============================================================
# FINAL V5 — RAM-SAFE S2 + S3 COMBINATION + VALIDATION
# ============================================================

import os
import csv
import time
import gc

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

BASE = "/content/amazon_ml_challenge/student_resource"

WORK = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_Final_Production_RAMSAFE"

S2_FILE = os.path.join(WORK, "s2_results.tsv")
S3_FILE = os.path.join(WORK, "s3_results.tsv")

OUTPUT_DIR = os.path.join(BASE, "output")

MATCHING_FILE = os.path.join(OUTPUT_DIR, "matching_results.tsv")
CANDIDATE_FILE = os.path.join(OUTPUT_DIR, "candidate_pairs.tsv")

os.makedirs(OUTPUT_DIR, exist_ok=True)

print("=" * 70)
print("V5 FINAL COMBINATION")
print("=" * 70)

for f in [S2_FILE, S3_FILE]:
    print(f"{f}")
    print(f"  exists: {os.path.exists(f)}")
    if os.path.exists(f):
        print(f"  size: {os.path.getsize(f) / (1024**3):.2f} GB")

# ------------------------------------------------------------
# SAFETY CHECK
# ------------------------------------------------------------

assert os.path.exists(S2_FILE), f"Missing S2 file: {S2_FILE}"
assert os.path.exists(S3_FILE), f"Missing S3 file: {S3_FILE}"

# Remove old final outputs so we don't accidentally append
for f in [MATCHING_FILE, CANDIDATE_FILE]:
    if os.path.exists(f):
        os.remove(f)
        print("Removed old:", f)

gc.collect()

# ------------------------------------------------------------
# HELPERS
# ------------------------------------------------------------

def split_ids(x):
    """
    Convert comma-separated IDs to a set.
    Empty string -> empty set.
    """
    if not x:
        return set()
    return set(v for v in x.split(",") if v)


def join_ids(values):
    """
    Stable deterministic ordering.
    """
    if not values:
        return ""
    return ",".join(sorted(values))


# ------------------------------------------------------------
# STREAMING COMBINATION
# ------------------------------------------------------------

start = time.time()

processed = 0
total_candidates = 0
total_matches = 0

with open(S2_FILE, "r", encoding="utf-8", newline="") as f2, \
     open(S3_FILE, "r", encoding="utf-8", newline="") as f3, \
     open(MATCHING_FILE, "w", encoding="utf-8", newline="") as fm, \
     open(CANDIDATE_FILE, "w", encoding="utf-8", newline="") as fc:

    r2 = csv.DictReader(f2, delimiter="\t")
    r3 = csv.DictReader(f3, delimiter="\t")

    expected2 = [
        "source1_entity_id",
        "candidate_entity_ids",
        "matched_entity_ids"
    ]

    expected3 = [
        "source1_entity_id",
        "candidate_entity_ids",
        "matched_entity_ids"
    ]

    assert r2.fieldnames == expected2, (
        f"Unexpected S2 columns: {r2.fieldnames}"
    )

    assert r3.fieldnames == expected3, (
        f"Unexpected S3 columns: {r3.fieldnames}"
    )

    matching_writer = csv.writer(
        fm,
        delimiter="\t",
        lineterminator="\n"
    )

    candidate_writer = csv.writer(
        fc,
        delimiter="\t",
        lineterminator="\n"
    )

    # Headers
    matching_writer.writerow([
        "source1_entity_id",
        "matched_entity_ids"
    ])

    candidate_writer.writerow([
        "source1_entity_id",
        "candidate_entity_ids"
    ])

    while True:

        row2 = next(r2, None)
        row3 = next(r3, None)

        if row2 is None and row3 is None:
            break

        # Both files must contain exactly the same S1 sequence
        assert row2 is not None and row3 is not None, (
            "S2 and S3 result files have different numbers of rows."
        )

        s1_2 = row2["source1_entity_id"]
        s1_3 = row3["source1_entity_id"]

        assert s1_2 == s1_3, (
            f"S1 order mismatch at row {processed + 1}: "
            f"S2={s1_2}, S3={s1_3}"
        )

        s1_id = s1_2

        # Read S2/S3 candidate and match sets
        c2 = split_ids(row2["candidate_entity_ids"])
        c3 = split_ids(row3["candidate_entity_ids"])

        m2 = split_ids(row2["matched_entity_ids"])
        m3 = split_ids(row3["matched_entity_ids"])

        # Final candidate set = exact union of S2 + S3
        candidates = c2 | c3

        # Final predicted matches = exact union of S2 + S3
        matches = m2 | m3

        # Critical safety rule:
        # every final match MUST be inside final candidate set
        assert matches.issubset(candidates), (
            f"Invalid final match outside candidate set for {s1_id}"
        )

        matching_writer.writerow([
            s1_id,
            join_ids(matches)
        ])

        candidate_writer.writerow([
            s1_id,
            join_ids(candidates)
        ])

        processed += 1
        total_candidates += len(candidates)
        total_matches += len(matches)

        if processed % 100000 == 0:
            elapsed = time.time() - start
            print(
                f"{processed:,} S1 | "
                f"Candidates {total_candidates:,} | "
                f"Matches {total_matches:,} | "
                f"Elapsed {elapsed/60:.1f} min"
            )

# ------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------

elapsed = time.time() - start

print("\n" + "=" * 70)
print("✅ FINAL COMBINATION COMPLETE")
print("=" * 70)

print("S1 processed:", f"{processed:,}")
print("Total final candidates:", f"{total_candidates:,}")
print("Total final matches:", f"{total_matches:,}")

print(
    "Avg candidates/S1:",
    total_candidates / processed
)

print(
    "Avg matches/S1:",
    total_matches / processed
)

print("Time:", f"{elapsed/60:.2f} minutes")

print("\nMatching file:")
print(MATCHING_FILE)
print(
    "Size:",
    f"{os.path.getsize(MATCHING_FILE)/(1024**3):.2f} GB"
)

print("\nCandidate file:")
print(CANDIDATE_FILE)
print(
    "Size:",
    f"{os.path.getsize(CANDIDATE_FILE)/(1024**3):.2f} GB"
)

gc.collect()
```

```python
# ============================================================
# REPRODUCE THE KNOWN V1.1 PIPELINE
# EXACT ADDRESS + UNIQUE EXACT NAME + COUNTRY
# ============================================================

import os
import re
import gc
import csv
import pickle
import unicodedata
import pandas as pd
from collections import defaultdict

BASE = "/content/amazon_ml_challenge/student_resource"
TEST_DIR = f"{BASE}/dataset/test"

DRIVE_BASE = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml"
INDEX_DIR = f"{DRIVE_BASE}/V5_full_indexes"

V11_DIR = f"{DRIVE_BASE}/V1_1_REPRODUCTION"
os.makedirs(V11_DIR, exist_ok=True)

MATCH_OUT = f"{V11_DIR}/matching_results.tsv"
CAND_OUT = f"{V11_DIR}/candidate_pairs.tsv"

S1_PATH = f"{TEST_DIR}/test_source1.tsv"
S2_PATH = f"{TEST_DIR}/test_source2.tsv"
S3_PATH = f"{TEST_DIR}/test_source3.tsv"

# ------------------------------------------------------------
# NORMALIZATION
# ------------------------------------------------------------

def normalize_advanced(text):
    if pd.isna(text):
        return ""

    text = str(text).lower().strip()
    text = unicodedata.normalize("NFKC", text)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()


def normalize_country(text):
    if pd.isna(text):
        return ""
    return str(text).lower().strip()


# ------------------------------------------------------------
# LOAD TEST S1
# ------------------------------------------------------------

print("Loading Test S1...")

test_s1 = pd.read_csv(
    S1_PATH,
    sep="\t",
    dtype=str,
    keep_default_na=False
)

print("S1:", test_s1.shape)

# ------------------------------------------------------------
# BUILD EXACT LOOKUPS FROM S1
#
# For each S1:
#   exact normalized address + country
#   exact UNIQUE normalized name + country
# ------------------------------------------------------------

name_counts = defaultdict(int)
name_to_s1 = defaultdict(list)

address_to_s1 = defaultdict(list)

print("\nBuilding S1 exact maps...")

for row in test_s1.itertuples(index=False):

    sid = row.entity_id
    country = normalize_country(row.country)

    name = normalize_advanced(
        row.business_name
    )

    address = normalize_advanced(
        row.business_address
    )

    if name:
        key = (country, name)
        name_counts[key] += 1
        name_to_s1[key].append(sid)

    if address:
        key = (country, address)
        address_to_s1[key].append(sid)

print("Name keys:", len(name_to_s1))
print("Address keys:", len(address_to_s1))

# Keep ONLY unique names
unique_name_map = {}

for key, ids in name_to_s1.items():
    if name_counts[key] == 1:
        unique_name_map[key] = ids[0]

del name_to_s1
del name_counts

gc.collect()

print("Unique name keys:", len(unique_name_map))

# ------------------------------------------------------------
# PREDICTION STORAGE
# ------------------------------------------------------------

predictions = {
    sid: []
    for sid in test_s1["entity_id"]
}

candidates = {
    sid: []
    for sid in test_s1["entity_id"]
}

# ------------------------------------------------------------
# PROCESS S2 + S3
# ------------------------------------------------------------

total_candidate_pairs = 0

for source_name, path in [
    ("S2", S2_PATH),
    ("S3", S3_PATH)
]:

    print("\n========================================")
    print("Processing", source_name)
    print("========================================")

    processed = 0
    source_candidates = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        dtype=str,
        keep_default_na=False,
        chunksize=100000
    ):

        for row in chunk.itertuples(index=False):

            eid = row.entity_id

            country = normalize_country(
                row.country
            )

            name = normalize_advanced(
                row.business_name
            )

            address = normalize_advanced(
                row.business_address
            )

            found_s1 = set()

            # ------------------------------------------------
            # BLOCK 1: EXACT ADDRESS + COUNTRY
            # ------------------------------------------------

            if address:

                found_s1.update(
                    address_to_s1.get(
                        (country, address),
                        []
                    )
                )

            # ------------------------------------------------
            # BLOCK 2: UNIQUE EXACT NAME + COUNTRY
            # ------------------------------------------------

            if name:

                sid = unique_name_map.get(
                    (country, name)
                )

                if sid is not None:
                    found_s1.add(sid)

            # ------------------------------------------------
            # SAVE CANDIDATES
            # ------------------------------------------------

            for sid in found_s1:

                candidates[sid].append(
                    eid
                )

                predictions[sid].append(
                    eid
                )

                source_candidates += 1

            processed += 1

        print(
            source_name,
            "| processed:",
            f"{processed:,}",
            "| candidate pairs:",
            f"{source_candidates:,}"
        )

    total_candidate_pairs += source_candidates

    print(
        source_name,
        "COMPLETE | candidate pairs:",
        f"{source_candidates:,}"
    )

# ------------------------------------------------------------
# DEDUPLICATE
# ------------------------------------------------------------

print("\nDeduplicating...")

for sid in predictions:

    predictions[sid] = list(
        dict.fromkeys(
            predictions[sid]
        )
    )

    candidates[sid] = list(
        dict.fromkeys(
            candidates[sid]
        )
    )

# ------------------------------------------------------------
# WRITE FILES
# ------------------------------------------------------------

print("\nWriting V1.1 files...")

with open(
    MATCH_OUT,
    "w",
    encoding="utf-8",
    newline=""
) as f_match, open(
    CAND_OUT,
    "w",
    encoding="utf-8",
    newline=""
) as f_cand:

    mw = csv.writer(
        f_match,
        delimiter="\t",
        lineterminator="\n"
    )

    cw = csv.writer(
        f_cand,
        delimiter="\t",
        lineterminator="\n"
    )

    mw.writerow([
        "source1_entity_id",
        "matched_entity_ids"
    ])

    cw.writerow([
        "source1_entity_id",
        "candidate_entity_ids"
    ])

    for sid in test_s1["entity_id"]:

        mids = predictions[sid]
        cids = candidates[sid]

        mw.writerow([
            sid,
            ",".join(mids)
        ])

        cw.writerow([
            sid,
            ",".join(cids)
        ])

# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

nonempty = sum(
    bool(predictions[sid])
    for sid in predictions
)

total_matches = sum(
    len(predictions[sid])
    for sid in predictions
)

total_candidates = sum(
    len(candidates[sid])
    for sid in candidates
)

print("\n" + "=" * 70)
print("✅ V1.1 REPRODUCTION COMPLETE")
print("=" * 70)

print(
    "S1 entities:",
    len(test_s1)
)

print(
    "Non-empty S1:",
    nonempty
)

print(
    "Empty S1:",
    len(test_s1) - nonempty
)

print(
    "Total candidate pairs:",
    total_candidates
)

print(
    "Total predicted matches:",
    total_matches
)

print(
    "Avg candidates/S1:",
    total_candidates / len(test_s1)
)

print(
    "Avg matches/S1:",
    total_matches / len(test_s1)
)

print("\nMatching:")
print(MATCH_OUT)

print("\nCandidates:")
print(CAND_OUT)
```

```python
# ============================================================
# V1.1 PRECISION DIAGNOSTIC
# DO NOT GENERATE OR SUBMIT TEST FILES YET
# ============================================================

import os
import re
import gc
import numpy as np
import pandas as pd
from collections import defaultdict
from rapidfuzz.fuzz import ratio

print("=" * 70)
print("V1.1 PRECISION DIAGNOSTIC")
print("=" * 70)

# Check that the validation objects from our notebook exist
needed = [
    "s1_val",
    "val_ids",
    "gt_lookup",
    "normalize_advanced",
    "normalize_country"
]

missing = [x for x in needed if x not in globals()]

if missing:
    print("MISSING OBJECTS:", missing)
    print("\nDo NOT continue.")
    print("Send me this output and I will give you the recovery cell.")
else:
    print("Validation objects: READY")

    TRAIN_S2 = "/content/amazon_ml_challenge/student_resource/dataset/train/train_source2.tsv"
    TRAIN_S3 = "/content/amazon_ml_challenge/student_resource/dataset/train/train_source3.tsv"

    # --------------------------------------------------------
    # Build validation S1 exact indexes
    # --------------------------------------------------------

    val_name_map = defaultdict(list)
    val_addr_map = defaultdict(list)

    for row in s1_val.itertuples(index=False):

        sid = str(row.entity_id)
        country = normalize_country(row.country)

        name = normalize_advanced(row.business_name)
        address = normalize_advanced(row.business_address)

        if name:
            val_name_map[(country, name)].append(sid)

        if address:
            val_addr_map[(country, address)].append(sid)

    print("Validation S1:", len(s1_val))
    print("Unique-name keys:", len(val_name_map))
    print("Address keys:", len(val_addr_map))

    # --------------------------------------------------------
    # Statistics
    # --------------------------------------------------------

    stats = {
        "address_only": 0,
        "name_only": 0,
        "both": 0,

        "address_tp": 0,
        "address_fp": 0,

        "name_tp": 0,
        "name_fp": 0,

        "both_tp": 0,
        "both_fp": 0,

        "total_candidates": 0,
        "total_tp": 0,
    }

    # Store name-only pairs so we can test filters
    name_only_pairs = []

    # --------------------------------------------------------
    # Scan training S2/S3
    # --------------------------------------------------------

    for source_name, path in [
        ("S2", TRAIN_S2),
        ("S3", TRAIN_S3)
    ]:

        print("\nProcessing", source_name)

        processed = 0

        for chunk in pd.read_csv(
            path,
            sep="\t",
            usecols=[
                "entity_id",
                "business_name",
                "business_address",
                "country"
            ],
            dtype=str,
            keep_default_na=False,
            chunksize=200000
        ):

            for row in chunk.itertuples(index=False):

                eid = str(row.entity_id)
                country = normalize_country(row.country)

                name = normalize_advanced(row.business_name)
                address = normalize_advanced(row.business_address)

                addr_sids = set()
                name_sids = set()

                if address:
                    addr_sids.update(
                        val_addr_map.get(
                            (country, address),
                            []
                        )
                    )

                if name:
                    # Only UNIQUE exact name
                    sids = val_name_map.get(
                        (country, name),
                        []
                    )

                    if len(sids) == 1:
                        name_sids.add(sids[0])

                candidates = addr_sids | name_sids

                for sid in candidates:

                    true_ids = gt_lookup.get(
                        sid,
                        set()
                    )

                    is_true = eid in true_ids

                    stats["total_candidates"] += 1

                    if is_true:
                        stats["total_tp"] += 1

                    in_addr = sid in addr_sids
                    in_name = sid in name_sids

                    if in_addr and in_name:

                        stats["both"] += 1

                        if is_true:
                            stats["both_tp"] += 1
                        else:
                            stats["both_fp"] += 1

                    elif in_addr:

                        stats["address_only"] += 1

                        if is_true:
                            stats["address_tp"] += 1
                        else:
                            stats["address_fp"] += 1

                    elif in_name:

                        stats["name_only"] += 1

                        if is_true:
                            stats["name_tp"] += 1
                        else:
                            stats["name_fp"] += 1

                        # Keep only name-only candidates for
                        # the detailed filter analysis.
                        s1row = s1_val[
                            s1_val["entity_id"] == sid
                        ]

                        if len(s1row):

                            s1r = s1row.iloc[0]

                            s1_name = normalize_advanced(
                                s1r.business_name
                            )

                            s1_addr = normalize_advanced(
                                s1r.business_address
                            )

                            name_sim = (
                                ratio(name, s1_name) / 100
                            )

                            addr_sim = (
                                ratio(address, s1_addr) / 100
                                if address and s1_addr
                                else 0.0
                            )

                            nums1 = set(
                                re.findall(
                                    r"\d+",
                                    address
                                )
                            )

                            nums2 = set(
                                re.findall(
                                    r"\d+",
                                    s1_addr
                                )
                            )

                            shared_numbers = len(
                                nums1 & nums2
                            )

                            name_only_pairs.append({
                                "sid": sid,
                                "eid": eid,
                                "true": is_true,
                                "name_sim": name_sim,
                                "addr_sim": addr_sim,
                                "shared_numbers": shared_numbers,
                                "s1_addr_empty": not bool(s1_addr),
                                "src_addr_empty": not bool(address)
                            })

                processed += len(chunk)

            if processed % 1000000 == 0:
                print(
                    source_name,
                    "| processed:",
                    f"{processed:,}"
                )

        print(
            source_name,
            "complete"
        )

    # --------------------------------------------------------
    # Baseline precision statistics
    # --------------------------------------------------------

    def precision(tp, fp):
        return (
            tp / (tp + fp)
            if tp + fp
            else 1.0
        )

    print("\n" + "=" * 70)
    print("V1.1 CANDIDATE BREAKDOWN")
    print("=" * 70)

    print(
        "Total candidates:",
        f"{stats['total_candidates']:,}"
    )

    print(
        "True candidates:",
        f"{stats['total_tp']:,}"
    )

    print(
        "Address only:",
        f"{stats['address_only']:,}",
        "| TP:",
        f"{stats['address_tp']:,}",
        "| FP:",
        f"{stats['address_fp']:,}",
        "| Precision:",
        round(
            precision(
                stats["address_tp"],
                stats["address_fp"]
            ),
            4
        )
    )

    print(
        "Name only:",
        f"{stats['name_only']:,}",
        "| TP:",
        f"{stats['name_tp']:,}",
        "| FP:",
        f"{stats['name_fp']:,}",
        "| Precision:",
        round(
            precision(
                stats["name_tp"],
                stats["name_fp"]
            ),
            4
        )
    )

    print(
        "Both:",
        f"{stats['both']:,}",
        "| TP:",
        f"{stats['both_tp']:,}",
        "| FP:",
        f"{stats['both_fp']:,}",
        "| Precision:",
        round(
            precision(
                stats["both_tp"],
                stats["both_fp"]
            ),
            4
        )
    )

    # --------------------------------------------------------
    # Analyze name-only candidates
    # --------------------------------------------------------

    df = pd.DataFrame(name_only_pairs)

    print("\nName-only detailed rows:", len(df))

    if len(df):

        print("\nName-only TRUE/FALSE by address similarity:")

        for threshold in [
            0.00,
            0.20,
            0.30,
            0.40,
            0.50,
            0.60,
            0.70,
            0.80
        ]:

            selected = df[
                (df["addr_sim"] >= threshold)
                |
                df["s1_addr_empty"]
                |
                df["src_addr_empty"]
            ]

            tp = int(selected["true"].sum())
            total = len(selected)
            fp = total - tp

            print(
                f"addr_sim >= {threshold:.2f} | "
                f"kept={total:,} | "
                f"TP={tp:,} | "
                f"FP={fp:,} | "
                f"precision={precision(tp, fp):.4f}"
            )

        print("\nName-only with shared address number:")

        for minimum in [1]:

            selected = df[
                (df["shared_numbers"] >= minimum)
                |
                df["s1_addr_empty"]
                |
                df["src_addr_empty"]
            ]

            tp = int(selected["true"].sum())
            total = len(selected)
            fp = total - tp

            print(
                f"shared_numbers >= {minimum} | "
                f"kept={total:,} | "
                f"TP={tp:,} | "
                f"FP={fp:,} | "
                f"precision={precision(tp, fp):.4f}"
            )

    print("\n" + "=" * 70)
    print("DIAGNOSTIC COMPLETE")
    print("=" * 70)

    del name_only_pairs
    del df
    gc.collect()
```

```python
# ============================================================
# RECOVER VALIDATION OBJECTS AFTER COLAB RESTART
# ============================================================

import os
import gc
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

BASE = "/content/amazon_ml_challenge/student_resource"
TRAIN_DIR = f"{BASE}/dataset/train"

S1_TRAIN = f"{TRAIN_DIR}/train_source1.tsv"
GT_FILE  = f"{TRAIN_DIR}/train_ground_truth.tsv"

print("=" * 70)
print("RECOVERING VALIDATION DATA")
print("=" * 70)

# ------------------------------------------------------------
# Load Source 1
# ------------------------------------------------------------

print("Loading training Source 1...")

s1_all = pd.read_csv(
    S1_TRAIN,
    sep="\t",
    dtype=str,
    keep_default_na=False
)

print("Source 1 rows:", len(s1_all))

# ------------------------------------------------------------
# Load ground truth
# ------------------------------------------------------------

print("Loading ground truth...")

gt_df = pd.read_csv(
    GT_FILE,
    sep="\t",
    dtype=str,
    keep_default_na=False
)

print("Ground-truth rows:", len(gt_df))

# ------------------------------------------------------------
# Ground-truth lookup
# ------------------------------------------------------------

gt_lookup = {}

for row in gt_df.itertuples(index=False):

    sid = str(row.source1_entity_id)

    value = str(
        row.matched_entity_ids
    ).strip()

    if value:
        gt_lookup[sid] = set(
            x.strip()
            for x in value.split(",")
            if x.strip()
        )
    else:
        gt_lookup[sid] = set()

# ------------------------------------------------------------
# IMPORTANT:
# Recreate the same 2,000-S1 validation design
# using deterministic random_state=42.
#
# We use train_test_split so the validation IDs are
# deterministic and do not depend on dataframe order
# after the restart.
# ------------------------------------------------------------

_, s1_val = train_test_split(
    s1_all,
    test_size=2000,
    random_state=42
)

s1_val = s1_val.reset_index(
    drop=True
)

val_ids = s1_val[
    "entity_id"
].astype(str).tolist()

# ------------------------------------------------------------
# Training IDs
# ------------------------------------------------------------

train_ids = s1_all[
    ~s1_all["entity_id"].isin(val_ids)
]["entity_id"].astype(str).tolist()

# ------------------------------------------------------------
# Validation lookup
# ------------------------------------------------------------

s1_val_lookup = (
    s1_val
    .set_index("entity_id")
    .to_dict("index")
)

# ------------------------------------------------------------
# Check
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("✅ VALIDATION OBJECTS RECOVERED")
print("=" * 70)

print(
    "All Source 1:",
    len(s1_all)
)

print(
    "Validation S1:",
    len(s1_val)
)

print(
    "Training S1:",
    len(train_ids)
)

print(
    "Validation IDs:",
    len(val_ids)
)

print(
    "Ground-truth entries:",
    len(gt_lookup)
)

# Ground truth coverage
covered = sum(
    sid in gt_lookup
    for sid in val_ids
)

print(
    "Validation IDs with GT:",
    covered
)

print(
    "Validation IDs with at least one true match:",
    sum(
        bool(gt_lookup.get(sid, set()))
        for sid in val_ids
    )
)

print("\nSample validation IDs:")
print(val_ids[:10])

# ------------------------------------------------------------
# Make sure normalization functions exist
# ------------------------------------------------------------

if "normalize_advanced" not in globals():

    import re
    import unicodedata

    def normalize_advanced(text):

        if pd.isna(text):
            return ""

        text = str(text).lower().strip()

        text = unicodedata.normalize(
            "NFKC",
            text
        )

        replacements = {
            "&": " and ",
            "street": " st ",
            "road": " rd ",
            "avenue": " ave ",
            "drive": " dr ",
            "boulevard": " blvd ",
            "highway": " hwy ",
            "lane": " ln ",
            "parkway": " pkwy ",
            "place": " pl ",
            "suite": " ste "
        }

        for old, new in replacements.items():
            text = text.replace(old, new)

        text = re.sub(
            r"[^\w\s]",
            " ",
            text
        )

        text = re.sub(
            r"\s+",
            " ",
            text
        )

        return text.strip()


if "normalize_country" not in globals():

    def normalize_country(text):

        if pd.isna(text):
            return ""

        return str(text).lower().strip()


print("\nNormalization functions: READY")

gc.collect()
```

```python
# ============================================================
# V1.1 PRECISION + F0.5 DIAGNOSTIC — RECOVERED VALIDATION
# ============================================================

import pandas as pd
import numpy as np
import re
from collections import defaultdict
from rapidfuzz.fuzz import ratio

TRAIN_S2 = f"{TRAIN_DIR}/train_source2.tsv"
TRAIN_S3 = f"{TRAIN_DIR}/train_source3.tsv"

print("=" * 70)
print("V1.1 PRECISION + F0.5 DIAGNOSTIC")
print("=" * 70)

# ------------------------------------------------------------
# FAST LOOKUP — IMPORTANT
# ------------------------------------------------------------

s1_lookup_val = (
    s1_val
    .set_index("entity_id")
    .to_dict("index")
)

print("Validation S1 lookup:", len(s1_lookup_val))

# ------------------------------------------------------------
# EXACT NORMALIZED INDEXES
# ------------------------------------------------------------

name_map = defaultdict(list)
address_map = defaultdict(list)

for sid, r in s1_lookup_val.items():

    country = normalize_country(
        r["country"]
    )

    name = normalize_advanced(
        r["business_name"]
    )

    address = normalize_advanced(
        r["business_address"]
    )

    if name:
        name_map[
            (country, name)
        ].append(sid)

    if address:
        address_map[
            (country, address)
        ].append(sid)

print("Exact name keys:", len(name_map))
print("Exact address keys:", len(address_map))

# ------------------------------------------------------------
# PREDICTION SETS
#
# Baseline:
#   exact normalized address + country
#   OR unique exact normalized name + country
#
# Variants:
#   A = baseline
#   B = name-only requires address similarity >= 0.30
#   C = name-only requires address similarity >= 0.40
#   D = name-only requires address similarity >= 0.50
#   E = name-only requires shared address number
# ------------------------------------------------------------

pred_A = defaultdict(set)
pred_B = defaultdict(set)
pred_C = defaultdict(set)
pred_D = defaultdict(set)
pred_E = defaultdict(set)

# Detailed counters
stats = {
    "address_exact": 0,
    "name_only": 0,
    "both": 0,

    "name_only_tp": 0,
    "name_only_fp": 0,

    "name_sim_030": 0,
    "name_sim_040": 0,
    "name_sim_050": 0,

    "name_num_shared": 0,
}

# ------------------------------------------------------------
# NUMBER EXTRACTION
# ------------------------------------------------------------

def extract_numbers(text):

    if not text:
        return set()

    return set(
        re.findall(
            r"\d+",
            text
        )
    )

# ------------------------------------------------------------
# F0.5
# ------------------------------------------------------------

def calculate_entity_f05(predictions):

    scores = []

    for sid in val_ids:

        true_ids = gt_lookup.get(
            sid,
            set()
        )

        predicted_ids = predictions.get(
            sid,
            set()
        )

        tp = len(
            true_ids & predicted_ids
        )

        fp = len(
            predicted_ids - true_ids
        )

        fn = len(
            true_ids - predicted_ids
        )

        # Exact challenge definition
        if tp == 0:

            score = (
                1.0
                if fp == 0 and fn == 0
                else 0.0
            )

        else:

            precision = tp / (
                tp + fp
            )

            recall = tp / (
                tp + fn
            )

            beta2 = 0.25

            score = (
                1.25
                * precision
                * recall
                /
                (
                    beta2 * precision
                    + recall
                )
            )

        scores.append(score)

    return float(
        np.mean(scores)
    )

# ------------------------------------------------------------
# PROCESS TRAINING SOURCE
# ------------------------------------------------------------

def process_source(path, source_name):

    rows = 0
    candidates = 0

    print(
        f"\nScanning {source_name}..."
    )

    for chunk in pd.read_csv(
        path,
        sep="\t",
        dtype=str,
        keep_default_na=False,
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ],
        chunksize=100000
    ):

        for row in chunk.itertuples(
            index=False
        ):

            eid = row.entity_id

            country = normalize_country(
                row.country
            )

            name = normalize_advanced(
                row.business_name
            )

            address = normalize_advanced(
                row.business_address
            )

            # --------------------------------------------
            # Exact address candidates
            # --------------------------------------------

            address_candidates = set(
                address_map.get(
                    (country, address),
                    []
                )
            ) if address else set()

            # --------------------------------------------
            # Exact UNIQUE name candidates
            # --------------------------------------------

            name_candidates = set()

            if name:

                name_ids = name_map.get(
                    (country, name),
                    []
                )

                if len(name_ids) == 1:
                    name_candidates.add(
                        name_ids[0]
                    )

            # --------------------------------------------
            # Categorize
            # --------------------------------------------

            both = (
                address_candidates
                &
                name_candidates
            )

            address_only = (
                address_candidates
                -
                name_candidates
            )

            name_only = (
                name_candidates
                -
                address_candidates
            )

            stats["both"] += len(both)
            stats["address_exact"] += len(
                address_only
            )
            stats["name_only"] += len(
                name_only
            )

            # --------------------------------------------
            # BASELINE A
            # --------------------------------------------

            baseline_candidates = (
                address_candidates
                |
                name_candidates
            )

            for sid in baseline_candidates:

                pred_A[sid].add(eid)

            candidates += len(
                baseline_candidates
            )

            # --------------------------------------------
            # Address-exact candidates are ALWAYS kept
            # in all variants.
            # --------------------------------------------

            for sid in address_candidates:

                pred_B[sid].add(eid)
                pred_C[sid].add(eid)
                pred_D[sid].add(eid)
                pred_E[sid].add(eid)

            # --------------------------------------------
            # Name-only filtering
            # --------------------------------------------

            if name_only:

                for sid in name_only:

                    s1row = s1_lookup_val.get(
                        sid
                    )

                    if s1row is None:
                        continue

                    s1_address = normalize_advanced(
                        s1row["business_address"]
                    )

                    s1_numbers = extract_numbers(
                        s1_address
                    )

                    source_numbers = extract_numbers(
                        address
                    )

                    shared_numbers = (
                        s1_numbers
                        &
                        source_numbers
                    )

                    address_sim = (
                        ratio(
                            s1_address,
                            address
                        ) / 100.0
                        if s1_address and address
                        else 0.0
                    )

                    # B: >= 0.30
                    if address_sim >= 0.30:

                        pred_B[sid].add(eid)
                        stats["name_sim_030"] += 1

                    # C: >= 0.40
                    if address_sim >= 0.40:

                        pred_C[sid].add(eid)
                        stats["name_sim_040"] += 1

                    # D: >= 0.50
                    if address_sim >= 0.50:

                        pred_D[sid].add(eid)
                        stats["name_sim_050"] += 1

                    # E: shared address number
                    if shared_numbers:

                        pred_E[sid].add(eid)
                        stats["name_num_shared"] += 1

                    # ------------------------------------------------
                    # Name-only TP / FP for diagnostic precision
                    # ------------------------------------------------

                    true_ids = gt_lookup.get(
                        sid,
                        set()
                    )

                    if eid in true_ids:
                        stats["name_only_tp"] += 1
                    else:
                        stats["name_only_fp"] += 1

            rows += 1

        if rows % 1_000_000 == 0:

            print(
                f"{source_name}: "
                f"{rows:,} rows | "
                f"baseline candidate links: "
                f"{candidates:,}"
            )

    print(
        f"{source_name} COMPLETE | "
        f"rows: {rows:,} | "
        f"baseline candidates: {candidates:,}"
    )

# ------------------------------------------------------------
# RUN BOTH SOURCES
# ------------------------------------------------------------

process_source(
    TRAIN_S2,
    "Source 2"
)

process_source(
    TRAIN_S3,
    "Source 3"
)

# ------------------------------------------------------------
# CALCULATE F0.5
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("F0.5 RESULTS")
print("=" * 70)

f05_A = calculate_entity_f05(pred_A)
f05_B = calculate_entity_f05(pred_B)
f05_C = calculate_entity_f05(pred_C)
f05_D = calculate_entity_f05(pred_D)
f05_E = calculate_entity_f05(pred_E)

print(
    f"V1.1 baseline                 : {f05_A:.6f}"
)

print(
    f"Name-only address sim >= 0.30 : {f05_B:.6f}"
)

print(
    f"Name-only address sim >= 0.40 : {f05_C:.6f}"
)

print(
    f"Name-only address sim >= 0.50 : {f05_D:.6f}"
)

print(
    f"Name-only shared number        : {f05_E:.6f}"
)

print("\n" + "=" * 70)
print("NAME-ONLY PRECISION")
print("=" * 70)

tp = stats["name_only_tp"]
fp = stats["name_only_fp"]

precision = (
    tp / (tp + fp)
    if tp + fp
    else 0
)

print(
    "Name-only pairs:",
    tp + fp
)

print(
    "Name-only TP:",
    tp
)

print(
    "Name-only FP:",
    fp
)

print(
    f"Name-only precision: {precision:.4f}"
)

print("\n" + "=" * 70)
print("FILTER COUNTS")
print("=" * 70)

print(
    "Name-only address sim >= 0.30:",
    stats["name_sim_030"]
)

print(
    "Name-only address sim >= 0.40:",
    stats["name_sim_040"]
)

print(
    "Name-only address sim >= 0.50:",
    stats["name_sim_050"]
)

print(
    "Name-only shared address number:",
    stats["name_num_shared"]
)

print("\n" + "=" * 70)
print("CANDIDATE CATEGORY COUNTS")
print("=" * 70)

print(
    "Address-only:",
    stats["address_exact"]
)

print(
    "Name-only:",
    stats["name_only"]
)

print(
    "Both name + address:",
    stats["both"]
)

print("\nDONE.")
```

```python
# ============================================================
# AMAZON ML CHALLENGE — RAM-SAFE V5 RECOVERY
# Step 1: Recover validation S1 + ground truth + V5 model
# ============================================================

import os
import gc
import pandas as pd
from sklearn.model_selection import train_test_split
from catboost import CatBoostClassifier

ROOT = "/content/amazon_ml_challenge/student_resource"

TRAIN_S1 = f"{ROOT}/dataset/train/train_source1.tsv"
TRAIN_GT = f"{ROOT}/dataset/train/train_ground_truth.tsv"

MODEL_PATH = (
    "/content/drive/MyDrive/"
    "Amazon-ML-Challenge-2026/ml/"
    "V5_ML_Final/V5_ML_Final/v5_model.cbm"
)

# ------------------------------------------------------------
# 1. Load ONLY Source 1
# ------------------------------------------------------------

print("Loading train Source 1 only...")

s1_all = pd.read_csv(
    TRAIN_S1,
    sep="\t",
    dtype=str,
    keep_default_na=False
)

print("S1 shape:", s1_all.shape)

# ------------------------------------------------------------
# 2. Load ground truth
# ------------------------------------------------------------

print("Loading ground truth...")

gt = pd.read_csv(
    TRAIN_GT,
    sep="\t",
    dtype=str,
    keep_default_na=False
)

print("GT shape:", gt.shape)

# ------------------------------------------------------------
# 3. Recreate EXACT validation split
# ------------------------------------------------------------

train_s1, s1_val = train_test_split(
    s1_all,
    test_size=2000,
    random_state=42
)

s1_val = s1_val.reset_index(drop=True)

val_ids = set(s1_val["entity_id"])

# Ground truth lookup ONLY for validation
gt_lookup = {}

for row in gt.itertuples(index=False):

    sid = row.source1_entity_id

    if sid in val_ids:

        value = row.matched_entity_ids

        if value:
            ids = set(
                x.strip()
                for x in value.split(",")
                if x.strip()
            )
        else:
            ids = set()

        gt_lookup[sid] = ids

# ------------------------------------------------------------
# 4. Validation lookup
# ------------------------------------------------------------

s1_val_lookup = (
    s1_val
    .set_index("entity_id")
    .to_dict("index")
)

# ------------------------------------------------------------
# 5. Basic validation statistics
# ------------------------------------------------------------

singleton_count = sum(
    1
    for sid in val_ids
    if len(gt_lookup.get(sid, set())) == 0
)

matched_count = len(val_ids) - singleton_count

print("\n" + "=" * 60)
print("V5 VALIDATION RECOVERY")
print("=" * 60)

print("Validation S1:", len(val_ids))
print("Validation GT:", len(gt_lookup))
print("With >=1 true match:", matched_count)
print("True singletons:", singleton_count)
print(
    "Singleton %:",
    round(100 * singleton_count / len(val_ids), 2)
)

# ------------------------------------------------------------
# 6. Load V5 model ONLY
# ------------------------------------------------------------

print("\nLoading V5 CatBoost model...")

v5_model = CatBoostClassifier()
v5_model.load_model(MODEL_PATH)

print("V5 model loaded.")
print("Model trees:", v5_model.tree_count_)
print("Model features:", v5_model.feature_count_)

# ------------------------------------------------------------
# 7. Clean unnecessary object
# ------------------------------------------------------------

del train_s1
del gt
gc.collect()

print("\nRAM-safe recovery COMPLETE.")
print("Do NOT load S2/S3 yet.")
```

```python
# ============================================================
# STEP 2A — RESTORE NORMALIZATION FUNCTION
# ============================================================

import re
import unicodedata

def normalize_advanced(text):
    if pd.isna(text):
        return ""

    text = str(text).lower().strip()
    text = unicodedata.normalize("NFKC", text)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()


# Quick verification — only 2,000 validation rows
val_name_tokens = set()
val_address_tokens = set()

for row in s1_val.itertuples(index=False):
    name = normalize_advanced(row.business_name)
    address = normalize_advanced(row.business_address)

    val_name_tokens.update(name.split())
    val_address_tokens.update(address.split())

print("Normalization function restored.")
print("Validation name tokens:", len(val_name_tokens))
print("Validation address tokens:", len(val_address_tokens))
```

```python
# ============================================================
# STEP 2B — RAM-SAFE S2/S3 TOKEN FREQUENCY SCAN
# ============================================================

from collections import Counter
import pandas as pd
import gc

name_freq_val = Counter()
address_freq_val = Counter()

for path, source_name in [
    (TRAIN_S2, "TRAIN S2"),
    (TRAIN_S3, "TRAIN S3")
]:
    print("\n" + "=" * 60)
    print("Scanning:", source_name)
    print("=" * 60)

    rows_processed = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=["business_name", "business_address"],
        dtype=str,
        keep_default_na=False,
        chunksize=100000
    ):

        for value in chunk["business_name"]:
            text = normalize_advanced(value)
            for token in set(text.split()):
                if token in val_name_tokens:
                    name_freq_val[token] += 1

        for value in chunk["business_address"]:
            text = normalize_advanced(value)
            for token in set(text.split()):
                if token in val_address_tokens:
                    address_freq_val[token] += 1

        rows_processed += len(chunk)

        if rows_processed % 1_000_000 == 0:
            print(f"{source_name}: {rows_processed:,} rows processed")

        del chunk

    gc.collect()

print("\n" + "=" * 60)
print("TOKEN FREQUENCY SCAN COMPLETE")
print("=" * 60)

print("Name tokens counted:", len(name_freq_val))
print("Address tokens counted:", len(address_freq_val))

print("\nName tokens <=500:",
      sum(1 for x in name_freq_val.values() if x <= 500))

print("Address tokens <=500:",
      sum(1 for x in address_freq_val.values() if x <= 500))

print("\nRAM-safe scan completed.")
```

```python
# ============================================================
# STEP 3 — RAM-SAFE VALIDATION BLOCKER INDEX
# ============================================================

from collections import defaultdict
import pandas as pd
import gc
import time

RARE_LIMIT = 500

# ------------------------------------------------------------
# Build validation S1 lookup structures
# ------------------------------------------------------------

s1_lookup = {}
val_name_map = defaultdict(set)
val_addr_map = defaultdict(set)

for row in s1_val.itertuples(index=False):

    sid = row.entity_id
    country = normalize_advanced(row.country)
    name = normalize_advanced(row.business_name)
    address = normalize_advanced(row.business_address)

    s1_lookup[sid] = {
        "name": name,
        "address": address,
        "country": country
    }

    if name:
        val_name_map[(country, name)].add(sid)

    if address:
        val_addr_map[(country, address)].add(sid)


# ------------------------------------------------------------
# Rare-token → validation S1 maps
# ------------------------------------------------------------

rare_name_to_s1 = defaultdict(set)
rare_addr_to_s1 = defaultdict(set)

for sid, row in s1_lookup.items():

    country = row["country"]

    for token in set(row["name"].split()):
        if name_freq_val.get(token, 10**9) <= RARE_LIMIT:
            rare_name_to_s1[(country, token)].add(sid)

    for token in set(row["address"].split()):
        if address_freq_val.get(token, 10**9) <= RARE_LIMIT:
            rare_addr_to_s1[(country, token)].add(sid)


print("=" * 65)
print("VALIDATION BLOCKER INDEX READY")
print("=" * 65)
print("Validation S1:", len(s1_lookup))
print("Exact name keys:", len(val_name_map))
print("Exact address keys:", len(val_addr_map))
print("Rare name keys:", len(rare_name_to_s1))
print("Rare address keys:", len(rare_addr_to_s1))
```

```python
# ============================================================
# STEP 4 — RAM-SAFE VALIDATION CANDIDATE SCAN — TRAIN S2
# ============================================================

import pandas as pd
import gc
import time
from collections import defaultdict

candidate_lookup_s2 = defaultdict(set)

rows_processed = 0
links_added = 0
start_time = time.time()

for chunk in pd.read_csv(
    TRAIN_S2,
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ],
    dtype=str,
    keep_default_na=False,
    chunksize=100000
):

    for row in chunk.itertuples(index=False):

        eid = row.entity_id
        country = normalize_advanced(row.country)
        name = normalize_advanced(row.business_name)
        address = normalize_advanced(row.business_address)

        found_s1 = set()

        # ----------------------------------------------------
        # BLOCK 1 — exact name + country
        # ----------------------------------------------------
        if name:
            found_s1.update(
                val_name_map.get((country, name), ())
            )

        # ----------------------------------------------------
        # BLOCK 2 — exact address + country
        # ----------------------------------------------------
        if address:
            found_s1.update(
                val_addr_map.get((country, address), ())
            )

        # ----------------------------------------------------
        # BLOCK 3 — rare name tokens
        # ----------------------------------------------------
        for token in set(name.split()):

            if name_freq_val.get(token, 10**9) <= RARE_LIMIT:

                found_s1.update(
                    rare_name_to_s1.get(
                        (country, token),
                        ()
                    )
                )

        # ----------------------------------------------------
        # BLOCK 4 — rare address tokens
        # ----------------------------------------------------
        for token in set(address.split()):

            if address_freq_val.get(token, 10**9) <= RARE_LIMIT:

                found_s1.update(
                    rare_addr_to_s1.get(
                        (country, token),
                        ()
                    )
                )

        # ----------------------------------------------------
        # Store only validation S1 -> S2 candidates
        # ----------------------------------------------------
        for sid in found_s1:
            candidate_lookup_s2[sid].add(eid)
            links_added += 1

    rows_processed += len(chunk)

    if rows_processed % 1_000_000 == 0:
        current_pairs = sum(
            len(x) for x in candidate_lookup_s2.values()
        )

        print(
            f"S2: {rows_processed:,} rows | "
            f"candidate links: {current_pairs:,}"
        )

    del chunk
    gc.collect()


elapsed = time.time() - start_time

print("\n" + "=" * 65)
print("S2 VALIDATION CANDIDATE SCAN COMPLETE")
print("=" * 65)

print("Rows scanned:", f"{rows_processed:,}")
print("S1 entities with candidates:", len(candidate_lookup_s2))
print("Candidate links:", f"{links_added:,}")
print("Elapsed:", round(elapsed / 60, 2), "minutes")

# ------------------------------------------------------------
# Candidate recall against ground truth
# ------------------------------------------------------------

total_true_s2 = 0
found_true_s2 = 0

for sid in val_ids:

    true_ids = {
        x for x in gt_lookup.get(sid, set())
        if x.startswith("S2-")
    }

    candidates = candidate_lookup_s2.get(sid, set())

    total_true_s2 += len(true_ids)
    found_true_s2 += len(true_ids & candidates)

missed_true_s2 = total_true_s2 - found_true_s2

print("\n" + "=" * 65)
print("S2 CANDIDATE RECALL")
print("=" * 65)

print("True S2 matches:", total_true_s2)
print("True S2 matches found:", found_true_s2)
print("True S2 matches missed:", missed_true_s2)

if total_true_s2:
    print(
        "Candidate recall:",
        f"{100 * found_true_s2 / total_true_s2:.2f}%"
    )

print(
    "Zero-candidate validation S1:",
    sum(
        1 for sid in val_ids
        if len(candidate_lookup_s2.get(sid, set())) == 0
    )
)

gc.collect()
```

```python
# ============================================================
# STEP 5 — DIAGNOSTIC BROADER BLOCKER — TRAIN S2
# Frequency limit: 2500
# NO FULL CANDIDATE TABLE STORED
# ============================================================

from collections import defaultdict
import pandas as pd
import gc
import time

TEST_LIMIT = 2500

# ------------------------------------------------------------
# Build broader validation S1 token maps
# ------------------------------------------------------------

broad_name_map = defaultdict(set)
broad_addr_map = defaultdict(set)

for sid, row in s1_lookup.items():

    country = row["country"]

    for token in set(row["name"].split()):
        if name_freq_val.get(token, 10**9) <= TEST_LIMIT:
            broad_name_map[(country, token)].add(sid)

    for token in set(row["address"].split()):
        if address_freq_val.get(token, 10**9) <= TEST_LIMIT:
            broad_addr_map[(country, token)].add(sid)


# ------------------------------------------------------------
# Reverse ground-truth lookup
# Only ~3k true S2 IDs, so this is tiny.
# ------------------------------------------------------------

true_s2_reverse = defaultdict(set)

for sid in val_ids:
    for eid in gt_lookup.get(sid, set()):
        if eid.startswith("S2-"):
            true_s2_reverse[eid].add(sid)


# ------------------------------------------------------------
# Scan S2
# ------------------------------------------------------------

rows_processed = 0
candidate_links = 0
found_true = set()

start_time = time.time()

for chunk in pd.read_csv(
    TRAIN_S2,
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ],
    dtype=str,
    keep_default_na=False,
    chunksize=100000
):

    for row in chunk.itertuples(index=False):

        eid = row.entity_id
        country = normalize_advanced(row.country)
        name = normalize_advanced(row.business_name)
        address = normalize_advanced(row.business_address)

        candidate_sids = set()

        # ----------------------------------------------------
        # Exact name / address
        # ----------------------------------------------------

        if name:
            candidate_sids.update(
                val_name_map.get((country, name), ())
            )

        if address:
            candidate_sids.update(
                val_addr_map.get((country, address), ())
            )

        # ----------------------------------------------------
        # Name: >=2 shared tokens
        # ----------------------------------------------------

        name_hits = defaultdict(int)

        for token in set(name.split()):

            if name_freq_val.get(token, 10**9) <= TEST_LIMIT:

                for sid in broad_name_map.get(
                    (country, token), ()
                ):
                    name_hits[sid] += 1

        for sid, count in name_hits.items():
            if count >= 2:
                candidate_sids.add(sid)

        # ----------------------------------------------------
        # Address: >=2 shared tokens
        # ----------------------------------------------------

        addr_hits = defaultdict(int)

        for token in set(address.split()):

            if address_freq_val.get(token, 10**9) <= TEST_LIMIT:

                for sid in broad_addr_map.get(
                    (country, token), ()
                ):
                    addr_hits[sid] += 1

        for sid, count in addr_hits.items():
            if count >= 2:
                candidate_sids.add(sid)

        # ----------------------------------------------------
        # Statistics only — don't store candidates
        # ----------------------------------------------------

        candidate_links += len(candidate_sids)

        # True-pair recall
        true_sids = true_s2_reverse.get(eid)

        if true_sids:
            found_true.update(
                true_sids & candidate_sids
            )

    rows_processed += len(chunk)

    if rows_processed % 1_000_000 == 0:
        print(
            f"S2: {rows_processed:,} rows | "
            f"candidate links: {candidate_links:,} | "
            f"true S1 recovered: {len(found_true):,}"
        )

    del chunk
    gc.collect()


elapsed = time.time() - start_time

total_true_s2 = sum(
    len(
        {
            x for x in gt_lookup.get(sid, set())
            if x.startswith("S2-")
        }
    )
    for sid in val_ids
)

# Count actual true pairs recovered.
# We need the pair-level count, not just S1 count.
found_true_pairs = 0

for sid in val_ids:
    true_ids = {
        x for x in gt_lookup.get(sid, set())
        if x.startswith("S2-")
    }

    # We didn't store candidate pairs, so reconstruct recovery
    # from the reverse set gathered above.
    found_true_pairs += sum(
        1
        for eid in true_ids
        if sid in true_s2_reverse.get(eid, set())
        and eid in set()  # placeholder prevented below
    )

# ------------------------------------------------------------
# The pair-level recovery needs a separate tiny pass over
# the TRUE S2 IDs. We will do that next if needed.
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("BROADER S2 BLOCKER DIAGNOSTIC")
print("=" * 65)
print("Frequency limit:", TEST_LIMIT)
print("Rows scanned:", f"{rows_processed:,}")
print("Candidate links:", f"{candidate_links:,}")
print("Elapsed:", round(elapsed / 60, 2), "minutes")
print("Validation S1 recovered:", len(found_true))
print("Total validation S1:", len(val_ids))

print("\nNOTE:")
print("The S1 recovery count above is only an intermediate diagnostic.")
print("We will calculate exact pair-level recall next.")
```

```python
# ============================================================
# STEP 6 — TRUE S2 PAIR BLOCKING AUDIT
# ============================================================

from collections import defaultdict, Counter
import pandas as pd
import gc
import re
from rapidfuzz.fuzz import ratio

# ------------------------------------------------------------
# Build true S2 -> S1 mapping
# ------------------------------------------------------------

true_s2_to_s1 = {}

for sid in val_ids:
    for eid in gt_lookup.get(sid, set()):
        if eid.startswith("S2-"):
            true_s2_to_s1[eid] = sid

print("Known true S2 pairs:", len(true_s2_to_s1))

# ------------------------------------------------------------
# Statistics
# ------------------------------------------------------------

stats = Counter()

examples = {
    "missed": [],
    "exact_name": [],
    "exact_address": [],
    "name_2_rare": [],
    "addr_2_rare": [],
    "addr_3": [],
    "addr_2_sim30": []
}

# ------------------------------------------------------------
# Scan S2
# ------------------------------------------------------------

rows_processed = 0

for chunk in pd.read_csv(
    TRAIN_S2,
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ],
    dtype=str,
    keep_default_na=False,
    chunksize=100000
):

    for row in chunk.itertuples(index=False):

        eid = row.entity_id

        # Ignore the ~5M rows that aren't known true pairs
        if eid not in true_s2_to_s1:
            continue

        sid = true_s2_to_s1[eid]
        s1row = s1_lookup[sid]

        name1 = s1row["name"]
        addr1 = s1row["address"]
        country1 = s1row["country"]

        name2 = normalize_advanced(row.business_name)
        addr2 = normalize_advanced(row.business_address)
        country2 = normalize_advanced(row.country)

        same_country = country1 == country2

        name_tokens1 = set(name1.split())
        name_tokens2 = set(name2.split())

        addr_tokens1 = set(addr1.split())
        addr_tokens2 = set(addr2.split())

        shared_name = name_tokens1 & name_tokens2
        shared_addr = addr_tokens1 & addr_tokens2

        shared_numbers = (
            set(re.findall(r"\d+", addr1)) &
            set(re.findall(r"\d+", addr2))
        )

        name_sim = (
            ratio(name1, name2) / 100
            if name1 and name2 else 0
        )

        addr_sim = (
            ratio(addr1, addr2) / 100
            if addr1 and addr2 else 0
        )

        # ----------------------------------------------------
        # Individual signals
        # ----------------------------------------------------

        exact_name = (
            same_country
            and bool(name1)
            and name1 == name2
        )

        exact_addr = (
            same_country
            and bool(addr1)
            and addr1 == addr2
        )

        # V5-style blocker
        name_2_rare = (
            same_country
            and len(shared_name) >= 2
            and any(
                name_freq_val.get(t, 10**9) <= 500
                for t in shared_name
            )
            and name_sim >= 0.60
        )

        addr_2_rare = (
            same_country
            and len(shared_addr) >= 2
            and any(
                address_freq_val.get(t, 10**9) <= 500
                for t in shared_addr
            )
            and addr_sim >= 0.60
        )

        addr_3 = (
            same_country
            and len(shared_addr) >= 3
        )

        addr_2_sim30 = (
            same_country
            and len(shared_addr) >= 2
            and addr_sim >= 0.30
        )

        # ----------------------------------------------------
        # Record statistics
        # ----------------------------------------------------

        stats["total"] += 1

        if same_country:
            stats["same_country"] += 1

        if exact_name:
            stats["exact_name"] += 1

        if exact_addr:
            stats["exact_address"] += 1

        if name_2_rare:
            stats["name_2_rare"] += 1

        if addr_2_rare:
            stats["addr_2_rare"] += 1

        if addr_3:
            stats["addr_3"] += 1

        if addr_2_sim30:
            stats["addr_2_sim30"] += 1

        # Actual current V5 blocker
        current_v5 = (
            exact_name
            or exact_addr
            or name_2_rare
            or addr_2_rare
        )

        if current_v5:
            stats["current_v5"] += 1
        else:
            stats["current_v5_missed"] += 1

            if len(examples["missed"]) < 15:
                examples["missed"].append({
                    "s1": sid,
                    "s2": eid,
                    "name1": name1,
                    "name2": name2,
                    "addr1": addr1,
                    "addr2": addr2,
                    "country1": country1,
                    "country2": country2,
                    "shared_name": len(shared_name),
                    "shared_addr": len(shared_addr),
                    "shared_numbers": len(shared_numbers),
                    "name_sim": round(name_sim, 3),
                    "addr_sim": round(addr_sim, 3)
                })

    rows_processed += len(chunk)

    if rows_processed % 1_000_000 == 0:
        print(
            f"Scanned {rows_processed:,} rows | "
            f"true pairs found: {stats['total']:,}"
        )

    del chunk
    gc.collect()

# ------------------------------------------------------------
# Print results
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TRUE S2 BLOCKING AUDIT")
print("=" * 70)

total = stats["total"]

print("True S2 pairs:", total)

for key, label in [
    ("same_country", "Same country"),
    ("exact_name", "Exact name + country"),
    ("exact_address", "Exact address + country"),
    ("name_2_rare", ">=2 rare name + sim >=0.60"),
    ("addr_2_rare", ">=2 rare address + sim >=0.60"),
    ("addr_3", ">=3 shared address tokens"),
    ("addr_2_sim30", ">=2 address tokens + sim >=0.30"),
    ("current_v5", "CURRENT V5 BLOCKER"),
]:
    count = stats[key]
    pct = 100 * count / total if total else 0
    print(f"{label:40s}: {count:5d}  ({pct:6.2f}%)")

print("\nCurrent V5 missed:",
      stats["current_v5_missed"])

print("\n" + "=" * 70)
print("EXAMPLES OF TRUE PAIRS MISSED BY CURRENT V5")
print("=" * 70)

for i, x in enumerate(examples["missed"], 1):
    print(f"\n--- MISS {i} ---")
    print("S1:", x["s1"])
    print("S2:", x["s2"])
    print("Name:", x["name1"], " <-> ", x["name2"])
    print("Address:", x["addr1"], " <-> ", x["addr2"])
    print("Country:", x["country1"], " <-> ", x["country2"])
    print(
        "Shared name:", x["shared_name"],
        "| Shared address:", x["shared_addr"],
        "| Shared numbers:", x["shared_numbers"]
    )
    print(
        "Name sim:", x["name_sim"],
        "| Address sim:", x["addr_sim"]
    )

gc.collect()
```

```python
# ============================================================
# STEP 7 — TRUE-PAIR ANCHOR COVERAGE AUDIT
# ============================================================

from collections import Counter
import pandas as pd
import re
import gc

# ------------------------------------------------------------
# Known true S2 pairs already created
# ------------------------------------------------------------

print("Known true S2 pairs:", len(true_s2_to_s1))

anchor_stats = Counter()

total = 0

for chunk in pd.read_csv(
    TRAIN_S2,
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ],
    dtype=str,
    keep_default_na=False,
    chunksize=100000
):

    for row in chunk.itertuples(index=False):

        eid = row.entity_id

        if eid not in true_s2_to_s1:
            continue

        sid = true_s2_to_s1[eid]
        s1row = s1_lookup[sid]

        name1 = s1row["name"]
        addr1 = s1row["address"]

        name2 = normalize_advanced(row.business_name)
        addr2 = normalize_advanced(row.business_address)

        name_tokens = set(name1.split()) & set(name2.split())
        addr_tokens = set(addr1.split()) & set(addr2.split())

        nums1 = set(re.findall(r"\d+", addr1))
        nums2 = set(re.findall(r"\d+", addr2))
        shared_nums = nums1 & nums2

        # Rare tokens according to validation-derived vocabulary
        rare_name_tokens = {
            t for t in name_tokens
            if name_freq_val.get(t, 10**9) <= 500
        }

        rare_addr_tokens = {
            t for t in addr_tokens
            if address_freq_val.get(t, 10**9) <= 500
        }

        total += 1

        # Individual anchors
        if len(name_tokens) >= 2:
            anchor_stats["name_2"] += 1

        if len(addr_tokens) >= 2:
            anchor_stats["addr_2"] += 1

        if len(addr_tokens) >= 3:
            anchor_stats["addr_3"] += 1

        if rare_name_tokens:
            anchor_stats["rare_name_any"] += 1

        if len(rare_name_tokens) >= 2:
            anchor_stats["rare_name_2"] += 1

        if rare_addr_tokens:
            anchor_stats["rare_addr_any"] += 1

        if len(rare_addr_tokens) >= 2:
            anchor_stats["rare_addr_2"] += 1

        if shared_nums:
            anchor_stats["shared_number"] += 1

        # Useful combinations
        if len(addr_tokens) >= 2 and shared_nums:
            anchor_stats["addr2_plus_number"] += 1

        if len(addr_tokens) >= 2 and rare_addr_tokens:
            anchor_stats["addr2_plus_rare"] += 1

        if len(name_tokens) >= 2 and rare_name_tokens:
            anchor_stats["name2_plus_rare"] += 1

        # Broad potential blocker
        if (
            len(addr_tokens) >= 2
            or len(name_tokens) >= 2
            or shared_nums
        ):
            anchor_stats["broad_union"] += 1

    del chunk
    gc.collect()

# ------------------------------------------------------------
# Results
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TRUE-PAIR ANCHOR COVERAGE")
print("=" * 70)

print("Total true S2 pairs:", total)

checks = [
    ("name_2", ">=2 shared name tokens"),
    ("addr_2", ">=2 shared address tokens"),
    ("addr_3", ">=3 shared address tokens"),
    ("rare_name_any", ">=1 rare name token"),
    ("rare_name_2", ">=2 rare name tokens"),
    ("rare_addr_any", ">=1 rare address token"),
    ("rare_addr_2", ">=2 rare address tokens"),
    ("shared_number", ">=1 shared address number"),
    ("addr2_plus_number", ">=2 address tokens + number"),
    ("addr2_plus_rare", ">=2 address tokens + rare address token"),
    ("name2_plus_rare", ">=2 name tokens + rare name token"),
    ("broad_union", "name2 OR addr2 OR shared number"),
]

for key, label in checks:
    n = anchor_stats[key]
    pct = 100 * n / total if total else 0
    print(f"{label:42s}: {n:5d} ({pct:6.2f}%)")

gc.collect()
```

```python
# ============================================================
# STEP 8 — MEASURE BROAD BLOCKER CANDIDATE VOLUME
# ============================================================

from collections import defaultdict
import pandas as pd
import re
import gc
import time

print("Building validation token indexes...")

# ------------------------------------------------------------
# Build tiny inverted indexes for the 2,000 validation S1s
# ------------------------------------------------------------

name_index_all = defaultdict(set)
addr_index_all = defaultdict(set)
number_index_all = defaultdict(set)

for sid in val_ids:

    row = s1_lookup[sid]

    name_tokens = set(row["name"].split())
    addr_tokens = set(row["address"].split())
    numbers = set(re.findall(r"\d+", row["address"]))

    for token in name_tokens:
        name_index_all[token].add(sid)

    for token in addr_tokens:
        addr_index_all[token].add(sid)

    for num in numbers:
        number_index_all[num].add(sid)

print("Name index keys:", len(name_index_all))
print("Address index keys:", len(addr_index_all))
print("Number index keys:", len(number_index_all))

# ------------------------------------------------------------
# Scan S2 and count candidate links
# ------------------------------------------------------------

candidate_links = 0
candidate_rows = 0
rows_processed = 0

# Track how many validation S1s get at least one candidate
covered_s1 = set()

start = time.time()

for chunk in pd.read_csv(
    TRAIN_S2,
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ],
    dtype=str,
    keep_default_na=False,
    chunksize=100000
):

    for row in chunk.itertuples(index=False):

        name2 = normalize_advanced(row.business_name)
        addr2 = normalize_advanced(row.business_address)

        name_tokens = set(name2.split())
        addr_tokens = set(addr2.split())
        numbers = set(re.findall(r"\d+", addr2))

        # ----------------------------------------------------
        # Find validation S1s sharing name/address tokens
        # ----------------------------------------------------

        name_counts = Counter()

        for token in name_tokens:
            for sid in name_index_all.get(token, ()):
                name_counts[sid] += 1

        addr_counts = Counter()

        for token in addr_tokens:
            for sid in addr_index_all.get(token, ()):
                addr_counts[sid] += 1

        candidates = set()

        # >=2 shared name tokens
        for sid, count in name_counts.items():
            if count >= 2:
                candidates.add(sid)

        # >=2 shared address tokens
        for sid, count in addr_counts.items():
            if count >= 2:
                candidates.add(sid)

        # >=1 shared address number
        for num in numbers:
            candidates.update(number_index_all.get(num, ()))

        if candidates:
            candidate_rows += 1
            candidate_links += len(candidates)
            covered_s1.update(candidates)

    rows_processed += len(chunk)

    if rows_processed % 1_000_000 == 0:
        elapsed = time.time() - start

        print(
            f"Scanned {rows_processed:,} rows | "
            f"candidate links: {candidate_links:,} | "
            f"S1 covered: {len(covered_s1):,} | "
            f"time: {elapsed/60:.1f} min"
        )

    del chunk
    gc.collect()

# ------------------------------------------------------------
# Results
# ------------------------------------------------------------

elapsed = time.time() - start

print("\n" + "=" * 70)
print("BROAD BLOCKER CANDIDATE VOLUME")
print("=" * 70)

print("S2 rows scanned:", f"{rows_processed:,}")
print("S2 rows producing candidates:", f"{candidate_rows:,}")
print("Candidate links:", f"{candidate_links:,}")
print("Validation S1 covered:", f"{len(covered_s1):,}", "/ 2000")
print("Average candidates per covered S1:",
      round(candidate_links / max(len(covered_s1), 1), 2))

print("Elapsed:", round(elapsed / 60, 2), "minutes")

gc.collect()
```

```python
# STEP 9 — QUICK RARITY THRESHOLD SWEEP

from collections import Counter
import pandas as pd
import gc

THRESHOLDS = [500, 1000, 2500, 5000, 10000]

counts = {
    t: {"name": 0, "addr": 0, "union": 0}
    for t in THRESHOLDS
}

total = 0

for chunk in pd.read_csv(
    TRAIN_S2,
    sep="\t",
    usecols=["entity_id", "business_name", "business_address"],
    dtype=str,
    keep_default_na=False,
    chunksize=100000
):
    for row in chunk.itertuples(index=False):

        eid = row.entity_id

        if eid not in true_s2_to_s1:
            continue

        sid = true_s2_to_s1[eid]
        s1row = s1_lookup[sid]

        n1 = set(s1row["name"].split())
        a1 = set(s1row["address"].split())

        n2 = set(normalize_advanced(row.business_name).split())
        a2 = set(normalize_advanced(row.business_address).split())

        shared_n = n1 & n2
        shared_a = a1 & a2

        total += 1

        for t in THRESHOLDS:

            rare_n = {
                x for x in shared_n
                if name_freq_val.get(x, 10**9) <= t
            }

            rare_a = {
                x for x in shared_a
                if address_freq_val.get(x, 10**9) <= t
            }

            name_ok = len(rare_n) >= 1
            addr_ok = len(rare_a) >= 1

            if name_ok:
                counts[t]["name"] += 1

            if addr_ok:
                counts[t]["addr"] += 1

            if name_ok or addr_ok:
                counts[t]["union"] += 1

    del chunk
    gc.collect()

print("\nTRUE-PAIR RARITY SWEEP")
print("=" * 65)
print(f"Total true pairs: {total}\n")

print(f"{'Threshold':>10} | {'Name':>8} | {'Address':>8} | {'Union':>8}")
print("-" * 45)

for t in THRESHOLDS:
    print(
        f"{t:>10} | "
        f"{counts[t]['name']:>4} "
        f"({100*counts[t]['name']/total:5.1f}%) | "
        f"{counts[t]['addr']:>4} "
        f"({100*counts[t]['addr']/total:5.1f}%) | "
        f"{counts[t]['union']:>4} "
        f"({100*counts[t]['union']/total:5.1f}%)"
    )

gc.collect()
```

```python
# STEP 10 — SELECTIVE 2-TOKEN BLOCKER RECALL
# Only audits the 3,279 known true pairs. No candidate generation.

import pandas as pd
import gc

THRESHOLDS = [500, 1000, 2500, 5000, 10000]

counts = {
    t: {"addr2": 0, "name2": 0, "union": 0}
    for t in THRESHOLDS
}

total = 0

for chunk in pd.read_csv(
    TRAIN_S2,
    sep="\t",
    usecols=["entity_id", "business_name", "business_address"],
    dtype=str,
    keep_default_na=False,
    chunksize=100000
):
    for row in chunk.itertuples(index=False):

        eid = row.entity_id

        if eid not in true_s2_to_s1:
            continue

        sid = true_s2_to_s1[eid]
        s1row = s1_lookup[sid]

        name1 = set(s1row["name"].split())
        addr1 = set(s1row["address"].split())

        name2 = set(normalize_advanced(row.business_name).split())
        addr2 = set(normalize_advanced(row.business_address).split())

        shared_name = name1 & name2
        shared_addr = addr1 & addr2

        total += 1

        for t in THRESHOLDS:

            rare_name = {
                x for x in shared_name
                if name_freq_val.get(x, 10**9) <= t
            }

            rare_addr = {
                x for x in shared_addr
                if address_freq_val.get(x, 10**9) <= t
            }

            # Selective anchors
            name_ok = (
                len(shared_name) >= 2
                and len(rare_name) >= 1
            )

            addr_ok = (
                len(shared_addr) >= 2
                and len(rare_addr) >= 1
            )

            if name_ok:
                counts[t]["name2"] += 1

            if addr_ok:
                counts[t]["addr2"] += 1

            if name_ok or addr_ok:
                counts[t]["union"] += 1

    del chunk
    gc.collect()

print("\nSELECTIVE 2-TOKEN BLOCKER RECALL")
print("=" * 70)
print("Total true pairs:", total)

print(
    f"{'Threshold':>10} | "
    f"{'Name 2+rare':>13} | "
    f"{'Addr 2+rare':>13} | "
    f"{'Union':>10}"
)
print("-" * 60)

for t in THRESHOLDS:
    print(
        f"{t:>10} | "
        f"{counts[t]['name2']:>5} "
        f"({100*counts[t]['name2']/total:5.1f}%) | "
        f"{counts[t]['addr2']:>5} "
        f"({100*counts[t]['addr2']/total:5.1f}%) | "
        f"{counts[t]['union']:>5} "
        f"({100*counts[t]['union']/total:5.1f}%)"
    )

gc.collect()
```

```python
# ============================================================
# STEP 11 — QUICK CANDIDATE VOLUME ESTIMATE
# Test 500k S2 rows only
# ============================================================

from collections import defaultdict, Counter
import pandas as pd
import re
import gc
import time

THRESHOLD = 5000
SAMPLE_ROWS = 500_000

print("Building selective token indexes...")

# ------------------------------------------------------------
# S1 token indexes
# Only keep tokens whose S2 frequency <= 5000
# ------------------------------------------------------------

name_idx = defaultdict(set)
addr_idx = defaultdict(set)

s1_country = {}

for sid in val_ids:

    row = s1_lookup[sid]

    country = row["country"]
    s1_country[sid] = country

    for token in set(row["name"].split()):
        if name_freq_val.get(token, 10**9) <= THRESHOLD:
            name_idx[token].add(sid)

    for token in set(row["address"].split()):
        if address_freq_val.get(token, 10**9) <= THRESHOLD:
            addr_idx[token].add(sid)

print("Selective name keys:", len(name_idx))
print("Selective address keys:", len(addr_idx))

# ------------------------------------------------------------
# Scan only first 500k S2 rows
# ------------------------------------------------------------

candidate_links = 0
rows_with_candidates = 0
covered_s1 = set()

start = time.time()

for chunk in pd.read_csv(
    TRAIN_S2,
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ],
    dtype=str,
    keep_default_na=False,
    chunksize=100000
):

    for row in chunk.itertuples(index=False):

        country2 = normalize_advanced(row.country)

        # Candidate occurrence counters
        counts = Counter()

        # -------------------------
        # Name: >=2 shared rare tokens
        # -------------------------
        for token in set(normalize_advanced(row.business_name).split()):

            if token in name_idx:
                for sid in name_idx[token]:
                    if s1_country[sid] == country2:
                        counts[sid] += 1

        # -------------------------
        # Address: >=2 shared rare tokens
        # -------------------------
        for token in set(normalize_advanced(row.business_address).split()):

            if token in addr_idx:
                for sid in addr_idx[token]:
                    if s1_country[sid] == country2:
                        counts[sid] += 1

        # ----------------------------------------------------
        # IMPORTANT:
        # Need >=2 hits from the same field.
        # ----------------------------------------------------

        name_counts = Counter()
        addr_counts = Counter()

        # Recompute separately so name/address aren't mixed
        for token in set(normalize_advanced(row.business_name).split()):
            if token in name_idx:
                for sid in name_idx[token]:
                    if s1_country[sid] == country2:
                        name_counts[sid] += 1

        for token in set(normalize_advanced(row.business_address).split()):
            if token in addr_idx:
                for sid in addr_idx[token]:
                    if s1_country[sid] == country2:
                        addr_counts[sid] += 1

        candidates = set()

        for sid, n in name_counts.items():
            if n >= 2:
                candidates.add(sid)

        for sid, n in addr_counts.items():
            if n >= 2:
                candidates.add(sid)

        if candidates:
            rows_with_candidates += 1
            candidate_links += len(candidates)
            covered_s1.update(candidates)

    del chunk
    gc.collect()

elapsed = time.time() - start

print("\n" + "=" * 70)
print("500K-ROW CANDIDATE ESTIMATE")
print("=" * 70)

print("S2 rows tested:", f"{SAMPLE_ROWS:,}")
print("Rows producing candidates:", f"{rows_with_candidates:,}")
print("Candidate links:", f"{candidate_links:,}")
print("Validation S1 covered:", f"{len(covered_s1):,}")
print(
    "Average candidates per candidate-producing row:",
    round(candidate_links / max(rows_with_candidates, 1), 2)
)

# Rough full-S2 extrapolation
scale = 5_034_616 / SAMPLE_ROWS

print("\nROUGH FULL-S2 ESTIMATE")
print("Estimated candidate links:",
      f"{int(candidate_links * scale):,}")

print("Estimated candidate-producing rows:",
      f"{int(rows_with_candidates * scale):,}")

print("Elapsed:", round(elapsed / 60, 2), "minutes")

gc.collect()
```

```python
# ============================================================
# STEP 12A — FAST 100K S2 BLOCKER TEST
# ============================================================

from collections import defaultdict, Counter
import pandas as pd
import time
import gc

THRESHOLD = 5000
TEST_ROWS = 100000

print("Building fast indexes...")

# ------------------------------------------------------------
# O(1) exact indexes
# ------------------------------------------------------------

exact_name_idx = defaultdict(set)
exact_addr_idx = defaultdict(set)

name_idx = defaultdict(set)
addr_idx = defaultdict(set)

s1_country = {}

for sid in val_ids:

    row = s1_lookup[sid]

    name = row["name"]
    addr = row["address"]
    country = row["country"]

    s1_country[sid] = country

    if name:
        exact_name_idx[(country, name)].add(sid)

    if addr:
        exact_addr_idx[(country, addr)].add(sid)

    for token in set(name.split()):
        if name_freq_val.get(token, 10**9) <= THRESHOLD:
            name_idx[token].add(sid)

    for token in set(addr.split()):
        if address_freq_val.get(token, 10**9) <= THRESHOLD:
            addr_idx[token].add(sid)

print("Exact name keys:", len(exact_name_idx))
print("Exact address keys:", len(exact_addr_idx))
print("Selective name keys:", len(name_idx))
print("Selective address keys:", len(addr_idx))

# ------------------------------------------------------------
# Test only first 100K S2 rows
# ------------------------------------------------------------

candidate_pairs = 0
candidate_rows = 0
covered_s1 = set()

start = time.time()

for chunk in pd.read_csv(
    TRAIN_S2,
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ],
    dtype=str,
    keep_default_na=False,
    chunksize=100000
):

    for row in chunk.itertuples(index=False):

        country = normalize_advanced(row.country)

        name = normalize_advanced(row.business_name)
        addr = normalize_advanced(row.business_address)

        candidates = set()

        # ----------------------------------------------------
        # Exact name/address — O(1)
        # ----------------------------------------------------

        if name:
            candidates.update(
                exact_name_idx.get((country, name), ())
            )

        if addr:
            candidates.update(
                exact_addr_idx.get((country, addr), ())
            )

        # ----------------------------------------------------
        # Selective name tokens
        # ----------------------------------------------------

        name_counts = Counter()

        for token in set(name.split()):

            for sid in name_idx.get(token, ()):

                if s1_country[sid] == country:
                    name_counts[sid] += 1

        for sid, count in name_counts.items():

            if count >= 2:
                candidates.add(sid)

        # ----------------------------------------------------
        # Selective address tokens
        # ----------------------------------------------------

        addr_counts = Counter()

        for token in set(addr.split()):

            for sid in addr_idx.get(token, ()):

                if s1_country[sid] == country:
                    addr_counts[sid] += 1

        for sid, count in addr_counts.items():

            if count >= 2:
                candidates.add(sid)

        if candidates:

            candidate_rows += 1
            candidate_pairs += len(candidates)
            covered_s1.update(candidates)

    del chunk
    gc.collect()

elapsed = time.time() - start

print("\n" + "=" * 60)
print("100K FAST BLOCKER TEST")
print("=" * 60)

print("Rows scanned:", TEST_ROWS)
print("Rows producing candidates:", f"{candidate_rows:,}")
print("Candidate links:", f"{candidate_pairs:,}")
print("S1 covered:", f"{len(covered_s1):,}/2000")
print("Time:", round(elapsed, 2), "seconds")
print("Time:", round(elapsed / 60, 2), "minutes")

gc.collect()
```

```python
# ============================================================
# STEP 13 — FAST VECTORISED BLOCKER TEST
# 100K S2 rows only
# ============================================================

import pandas as pd
import numpy as np
import time
import gc

THRESHOLD = 5000
TEST_ROWS = 100000
CHUNK_SIZE = 50000

print("Building vectorized indexes...")

# ------------------------------------------------------------
# S1 -> selective token tables
# ------------------------------------------------------------

name_records = []
addr_records = []
exact_name_map = {}
exact_addr_map = {}

for sid in val_ids:

    row = s1_lookup[sid]

    country = row["country"]
    name = row["name"]
    addr = row["address"]

    if name:
        exact_name_map[(country, name)] = sid

    if addr:
        exact_addr_map[(country, addr)] = sid

    for token in set(name.split()):
        if name_freq_val.get(token, 10**9) <= THRESHOLD:
            name_records.append((token, country, sid))

    for token in set(addr.split()):
        if address_freq_val.get(token, 10**9) <= THRESHOLD:
            addr_records.append((token, country, sid))

name_lookup = pd.DataFrame(
    name_records,
    columns=["token", "country", "s1_id"]
)

addr_lookup = pd.DataFrame(
    addr_records,
    columns=["token", "country", "s1_id"]
)

del name_records, addr_records
gc.collect()

print("Name index rows:", len(name_lookup))
print("Address index rows:", len(addr_lookup))

# ------------------------------------------------------------
# Read only 100K rows
# ------------------------------------------------------------

start = time.time()

candidate_pairs = set()

for chunk in pd.read_csv(
    TRAIN_S2,
    sep="\t",
    usecols=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ],
    dtype=str,
    keep_default_na=False,
    chunksize=CHUNK_SIZE
):

    # Normalize once
    chunk["name_n"] = chunk["business_name"].map(normalize_advanced)
    chunk["addr_n"] = chunk["business_address"].map(normalize_advanced)
    chunk["country_n"] = chunk["country"].map(normalize_advanced)

    # --------------------------------------------------------
    # NAME BLOCK
    # --------------------------------------------------------

    name_tokens = chunk[
        ["entity_id", "country_n", "name_n"]
    ].copy()

    name_tokens["token"] = name_tokens["name_n"].str.split()

    name_tokens = name_tokens[
        ["entity_id", "country_n", "token"]
    ].explode("token")

    name_tokens = name_tokens[
        name_tokens["token"].notna()
    ].drop_duplicates(
        ["entity_id", "token"]
    )

    name_matches = name_tokens.merge(
        name_lookup,
        left_on=["token", "country_n"],
        right_on=["token", "country"],
        how="inner"
    )

    name_counts = (
        name_matches
        .groupby(["entity_id", "s1_id"])
        .size()
        .reset_index(name="n")
    )

    name_candidates = name_counts[
        name_counts["n"] >= 2
    ][["entity_id", "s1_id"]]

    # --------------------------------------------------------
    # ADDRESS BLOCK
    # --------------------------------------------------------

    addr_tokens = chunk[
        ["entity_id", "country_n", "addr_n"]
    ].copy()

    addr_tokens["token"] = addr_tokens["addr_n"].str.split()

    addr_tokens = addr_tokens[
        ["entity_id", "country_n", "token"]
    ].explode("token")

    addr_tokens = addr_tokens[
        addr_tokens["token"].notna()
    ].drop_duplicates(
        ["entity_id", "token"]
    )

    addr_matches = addr_tokens.merge(
        addr_lookup,
        left_on=["token", "country_n"],
        right_on=["token", "country"],
        how="inner"
    )

    addr_counts = (
        addr_matches
        .groupby(["entity_id", "s1_id"])
        .size()
        .reset_index(name="n")
    )

    addr_candidates = addr_counts[
        addr_counts["n"] >= 2
    ][["entity_id", "s1_id"]]

    # --------------------------------------------------------
    # EXACT NAME / ADDRESS
    # --------------------------------------------------------

    for r in chunk.itertuples(index=False):

        key_name = (r.country_n, r.name_n)
        key_addr = (r.country_n, r.addr_n)

        sid = exact_name_map.get(key_name)

        if sid is not None:
            candidate_pairs.add((r.entity_id, sid))

        sid = exact_addr_map.get(key_addr)

        if sid is not None:
            candidate_pairs.add((r.entity_id, sid))

    # --------------------------------------------------------
    # Add token candidates
    # --------------------------------------------------------

    for r in name_candidates.itertuples(index=False):
        candidate_pairs.add((r.entity_id, r.s1_id))

    for r in addr_candidates.itertuples(index=False):
        candidate_pairs.add((r.entity_id, r.s1_id))

    del chunk
    del name_tokens, addr_tokens
    del name_matches, addr_matches
    del name_counts, addr_counts
    del name_candidates, addr_candidates

    gc.collect()

elapsed = time.time() - start

print("\n" + "=" * 65)
print("VECTORISED 100K TEST")
print("=" * 65)

print("Candidate pairs:", f"{len(candidate_pairs):,}")
print("Time:", round(elapsed, 2), "seconds")
print("Time:", round(elapsed / 60, 2), "minutes")

gc.collect()
```

```python
# ============================================================
# V5.1 — CHANGE BLOCKER RARITY LIMIT
# ============================================================

RARITY_LIMIT = 5000

def generate_candidates(
    s1_name,
    s1_addr,
    s1_country,
    name_idx,
    addr_idx,
    name_freq,
    addr_freq
):
    name_hits = defaultdict(int)
    addr_hits = defaultdict(int)

    # -------------------------
    # NAME BLOCK
    # -------------------------
    for token in tokens_fast(s1_name):

        if name_freq.get(token, 999999999) <= RARITY_LIMIT:

            for eid in name_idx.get(token, []):
                name_hits[eid] += 1

    # -------------------------
    # ADDRESS BLOCK
    # -------------------------
    for token in tokens_fast(s1_addr):

        if addr_freq.get(token, 999999999) <= RARITY_LIMIT:

            for eid in addr_idx.get(token, []):
                addr_hits[eid] += 1

    # -------------------------
    # Require >=2 shared tokens
    # -------------------------
    candidate_ids = set()

    for eid, count in name_hits.items():
        if count >= 2:
            candidate_ids.add(eid)

    for eid, count in addr_hits.items():
        if count >= 2:
            candidate_ids.add(eid)

    return candidate_ids


print("=" * 60)
print("V5.1 BLOCKER READY")
print("=" * 60)
print("Rarity limit:", RARITY_LIMIT)
print("Name requirement: >=2 shared selective tokens")
print("Address requirement: >=2 shared selective tokens")
print("Model threshold: 0.60")
print("No dataset scan performed.")
```

```python
# ============================================================
# RESTORE MISSING HELPERS
# ============================================================

import re
import pandas as pd

def normalize_country(text):
    if pd.isna(text):
        return ""
    return str(text).lower().strip()

def get_token_set(text):
    if pd.isna(text) or not text:
        return set()
    return set(str(text).split())

print("Helpers restored:")
print("✓ normalize_country")
print("✓ get_token_set")
```

```python
# ============================================================
# STEP 2A — PREPARE V5.1 VALIDATION INDEX
# ============================================================

from collections import defaultdict
import gc

V51_LIMIT = 5000

# Make sure lookup uses the actual source column names
s1_val_lookup = (
    s1_val
    .set_index("entity_id")
    .to_dict("index")
)

# ------------------------------------------------------------
# Country-aware rare-token indexes
# ------------------------------------------------------------

v51_name_index = defaultdict(set)
v51_addr_index = defaultdict(set)

val_name_tokens = {}
val_address_tokens = {}
val_country = {}

for sid in val_ids:

    r = s1_val_lookup[sid]

    name = normalize_advanced(r["business_name"])
    addr = normalize_advanced(r["business_address"])
    country = normalize_country(r["country"])

    val_name_tokens[sid] = get_token_set(name)
    val_address_tokens[sid] = get_token_set(addr)
    val_country[sid] = country

    for token in val_name_tokens[sid]:
        if name_freq_val.get(token, 10**9) <= V51_LIMIT:
            v51_name_index[(country, token)].add(sid)

    for token in val_address_tokens[sid]:
        if address_freq_val.get(token, 10**9) <= V51_LIMIT:
            v51_addr_index[(country, token)].add(sid)

print("=" * 65)
print("V5.1 VALIDATION INDEX READY")
print("=" * 65)
print("Rarity limit:", V51_LIMIT)
print("Name index keys:", len(v51_name_index))
print("Address index keys:", len(v51_addr_index))
print("Validation S1:", len(val_ids))

gc.collect()
```

```python
V5_FEATURES = [
    "name_exact",
    "address_exact",
    "country_exact",
    "name_similarity",
    "address_similarity",
    "name_token_overlap",
    "address_token_overlap",
    "name_jaccard",
    "address_jaccard",
    "shared_name_token_count",
    "shared_address_token_count",
    "shared_address_number_count",
    "name_length_diff",
    "address_length_diff",
    "name_length_ratio",
    "address_length_ratio",
    "similarity_gap",
    "max_similarity",
    "similarity_product",
    "token_overlap_gap"
]

print("✅ V5_FEATURES restored")
print("Feature count:", len(V5_FEATURES))
print("pair_features available:", callable(pair_features))
```

```python
# ============================================================
# RESTORE V5 pair_features()
# Exact 20 features used by V5 model
# ============================================================

from rapidfuzz.fuzz import ratio

def pair_features(s1, candidate):

    name1 = normalize_advanced(
        s1["business_name"]
    )
    name2 = normalize_advanced(
        candidate["business_name"]
    )

    addr1 = normalize_advanced(
        s1["business_address"]
    )
    addr2 = normalize_advanced(
        candidate["business_address"]
    )

    country1 = normalize_country(
        s1["country"]
    )
    country2 = normalize_country(
        candidate["country"]
    )

    t1_name = get_token_set(name1)
    t2_name = get_token_set(name2)

    t1_addr = get_token_set(addr1)
    t2_addr = get_token_set(addr2)

    shared_name = t1_name & t2_name
    shared_addr = t1_addr & t2_addr

    nums1 = set(re.findall(r"\d+", addr1))
    nums2 = set(re.findall(r"\d+", addr2))

    name_sim = (
        ratio(name1, name2) / 100
        if name1 and name2 else 0.0
    )

    addr_sim = (
        ratio(addr1, addr2) / 100
        if addr1 and addr2 else 0.0
    )

    name_overlap = (
        len(shared_name)
        / max(1, min(len(t1_name), len(t2_name)))
    )

    addr_overlap = (
        len(shared_addr)
        / max(1, min(len(t1_addr), len(t2_addr)))
    )

    name_union = t1_name | t2_name
    addr_union = t1_addr | t2_addr

    name_jaccard = (
        len(shared_name)
        / max(1, len(name_union))
    )

    addr_jaccard = (
        len(shared_addr)
        / max(1, len(addr_union))
    )

    return {
        "name_exact": int(
            bool(name1) and name1 == name2
        ),

        "address_exact": int(
            bool(addr1) and addr1 == addr2
        ),

        "country_exact": int(
            country1 == country2
        ),

        "name_similarity": name_sim,
        "address_similarity": addr_sim,

        "name_token_overlap": name_overlap,
        "address_token_overlap": addr_overlap,

        "name_jaccard": name_jaccard,
        "address_jaccard": addr_jaccard,

        "shared_name_token_count": len(shared_name),
        "shared_address_token_count": len(shared_addr),

        "shared_address_number_count": len(
            nums1 & nums2
        ),

        "name_length_diff": abs(
            len(name1) - len(name2)
        ),

        "address_length_diff": abs(
            len(addr1) - len(addr2)
        ),

        "name_length_ratio": (
            min(len(name1), len(name2))
            / max(1, max(len(name1), len(name2)))
        ),

        "address_length_ratio": (
            min(len(addr1), len(addr2))
            / max(1, max(len(addr1), len(addr2)))
        ),

        "similarity_gap": abs(
            name_sim - addr_sim
        ),

        "max_similarity": max(
            name_sim,
            addr_sim
        ),

        "similarity_product": (
            name_sim * addr_sim
        ),

        "token_overlap_gap": abs(
            name_overlap - addr_overlap
        )
    }


print("✅ V5 pair_features restored")
print("Features:", len(V5_FEATURES))
print(V5_FEATURES)
```

```python
# ============================================================
# STEP 2B — V5.1 VALIDATION SCORING
# ============================================================

import pandas as pd
import numpy as np
import gc
import time

V51_LIMIT = 5000
V51_THRESHOLD = 0.60
BATCH_SIZE = 10000

required = [
    "val_ids",
    "gt_lookup",
    "s1_val_lookup",
    "v5_model",
    "pair_features",
    "V5_FEATURES",
    "v51_name_index",
    "v51_addr_index",
    "val_name_tokens",
    "val_address_tokens",
    "val_country"
]

missing = [x for x in required if x not in globals()]

if missing:
    raise RuntimeError("Missing: " + ", ".join(missing))

print("=" * 70)
print("V5.1 VALIDATION SCORING")
print("=" * 70)
print("Rarity limit:", V51_LIMIT)
print("Threshold:", V51_THRESHOLD)
print("Validation S1:", len(val_ids))

# ------------------------------------------------------------
# Exact indexes
# ------------------------------------------------------------

exact_name = {}
exact_addr = {}

for sid in val_ids:

    r = s1_val_lookup[sid]

    name = normalize_advanced(r["business_name"])
    addr = normalize_advanced(r["business_address"])
    country = normalize_country(r["country"])

    if name:
        exact_name[(country, name)] = sid

    if addr:
        exact_addr[(country, addr)] = sid

print("Exact indexes ready.")

# ------------------------------------------------------------
# Prediction storage
# ------------------------------------------------------------

predicted_by_s1_v51 = {
    sid: set()
    for sid in val_ids
}

feature_batch = []
pair_batch = []

total_candidates = 0
total_predictions = 0


def score_v51_batch():

    global total_predictions

    if not feature_batch:
        return

    X = pd.DataFrame(
        feature_batch,
        columns=V5_FEATURES
    )

    probs = v5_model.predict_proba(X)[:, 1]

    keep = probs >= V51_THRESHOLD

    for i in np.where(keep)[0]:

        sid, eid = pair_batch[i]

        predicted_by_s1_v51[sid].add(eid)

    total_predictions += int(keep.sum())

    feature_batch.clear()
    pair_batch.clear()

    del X, probs, keep
    gc.collect()


# ------------------------------------------------------------
# Process source
# ------------------------------------------------------------

def process_v51_source(path, source_name):

    global total_candidates

    start = time.time()
    rows = 0
    source_candidates = 0

    print("\n" + "=" * 70)
    print("Scanning", source_name)
    print("=" * 70)

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ],
        dtype=str,
        keep_default_na=False,
        chunksize=200000
    ):

        for row in chunk.itertuples(index=False):

            eid = row.entity_id

            country = normalize_country(row.country)

            name = normalize_advanced(row.business_name)
            address = normalize_advanced(row.business_address)

            name_tokens = get_token_set(name)
            address_tokens = get_token_set(address)

            found_s1 = set()

            # ------------------------------------------------
            # Exact name
            # ------------------------------------------------

            if name:
                sid = exact_name.get((country, name))
                if sid is not None:
                    found_s1.add(sid)

            # ------------------------------------------------
            # Exact address
            # ------------------------------------------------

            if address:
                sid = exact_addr.get((country, address))
                if sid is not None:
                    found_s1.add(sid)

            # ------------------------------------------------
            # Selective NAME blocker
            # >=2 shared tokens + >=1 rare shared token
            # ------------------------------------------------

            possible_s1 = set()

            for token in name_tokens:

                possible_s1.update(
                    v51_name_index.get(
                        (country, token),
                        ()
                    )
                )

            for sid in possible_s1:

                shared = (
                    name_tokens &
                    val_name_tokens[sid]
                )

                if len(shared) < 2:
                    continue

                if any(
                    name_freq_val.get(t, 10**9)
                    <= V51_LIMIT
                    for t in shared
                ):
                    found_s1.add(sid)

            # ------------------------------------------------
            # Selective ADDRESS blocker
            # >=2 shared tokens + >=1 rare shared token
            # ------------------------------------------------

            possible_s1 = set()

            for token in address_tokens:

                possible_s1.update(
                    v51_addr_index.get(
                        (country, token),
                        ()
                    )
                )

            for sid in possible_s1:

                shared = (
                    address_tokens &
                    val_address_tokens[sid]
                )

                if len(shared) < 2:
                    continue

                if any(
                    address_freq_val.get(t, 10**9)
                    <= V51_LIMIT
                    for t in shared
                ):
                    found_s1.add(sid)

            # ------------------------------------------------
            # Build V5 features
            # ------------------------------------------------

            for sid in found_s1:

                s1_row = s1_val_lookup[sid]

                source_row = {
                    "entity_id": eid,
                    "business_name": row.business_name,
                    "business_address": row.business_address,
                    "country": row.country
                }

                features = pair_features(
                    s1_row,
                    source_row
                )

                feature_batch.append(features)
                pair_batch.append((sid, eid))

                source_candidates += 1
                total_candidates += 1

                if len(feature_batch) >= BATCH_SIZE:
                    score_v51_batch()

            rows += 1

        del chunk
        gc.collect()

        if rows % 1_000_000 == 0:

            print(
                f"{source_name}: "
                f"{rows:,} rows | "
                f"candidates: {source_candidates:,} | "
                f"predictions: {total_predictions:,} | "
                f"time: {(time.time()-start)/60:.1f} min"
            )

    score_v51_batch()

    print(
        f"{source_name} COMPLETE | "
        f"candidates: {source_candidates:,} | "
        f"time: {(time.time()-start)/60:.2f} min"
    )


# ------------------------------------------------------------
# RUN S2 + S3
# ------------------------------------------------------------

process_v51_source(
    TRAIN_S2,
    "S2"
)

process_v51_source(
    TRAIN_S3,
    "S3"
)

# ------------------------------------------------------------
# Entity-level Macro F0.5
# ------------------------------------------------------------

entity_scores = []

singleton_total = 0
singleton_correct = 0

for sid in val_ids:

    true_ids = gt_lookup.get(sid, set())
    pred_ids = predicted_by_s1_v51[sid]

    tp = len(true_ids & pred_ids)
    fp = len(pred_ids - true_ids)
    fn = len(true_ids - pred_ids)

    if not true_ids:

        singleton_total += 1

        if not pred_ids:
            singleton_correct += 1

        score = (
            1.0
            if not pred_ids
            else 0.0
        )

    elif tp == 0:

        score = (
            1.0
            if fp == 0 and fn == 0
            else 0.0
        )

    else:

        precision = tp / (tp + fp)
        recall = tp / (tp + fn)

        score = (
            1.25 * precision * recall
            /
            (0.25 * precision + recall)
        )

    entity_scores.append(score)


final_f05_v51 = float(
    np.mean(entity_scores)
)

print("\n" + "=" * 70)
print("V5.1 VALIDATION RESULT")
print("=" * 70)

print("Total candidates scored:", f"{total_candidates:,}")
print("Positive predictions:", f"{total_predictions:,}")

print(
    "S1 with predictions:",
    sum(bool(x) for x in predicted_by_s1_v51.values()),
    "/",
    len(val_ids)
)

print(
    "Singleton accuracy:",
    f"{singleton_correct}/{singleton_total}"
)

print(
    "ENTITY MACRO F0.5:",
    final_f05_v51
)

print(
    "Previous V5:",
    0.902677
)

print(
    "Difference:",
    final_f05_v51 - 0.902677
)

gc.collect()
```

```python
# STEP 11 — RESTORE V5 CANDIDATE GENERATOR

from collections import defaultdict

RARITY_LIMIT = 500

def generate_candidates(
    name1,
    addr1,
    country1,
    name_index,
    addr_index,
    name_freq,
    addr_freq
):
    name_hits = defaultdict(int)
    addr_hits = defaultdict(int)

    # Name blocking
    for token in name1.split():

        if name_freq.get(token, 999999999) <= RARITY_LIMIT:

            for eid in name_index.get(token, ()):

                name_hits[eid] += 1

    # Address blocking
    for token in addr1.split():

        if addr_freq.get(token, 999999999) <= RARITY_LIMIT:

            for eid in addr_index.get(token, ()):

                addr_hits[eid] += 1

    candidate_ids = set()

    for eid, count in name_hits.items():
        if count >= 2:
            candidate_ids.add(eid)

    for eid, count in addr_hits.items():
        if count >= 2:
            candidate_ids.add(eid)

    return candidate_ids

print("✅ V5 candidate generator restored")
print("Rarity limit:", RARITY_LIMIT)
```

```python
# STEP 13A — FAST V5 FEATURE BUILDER

import re
import numpy as np
import pandas as pd
from rapidfuzz.fuzz import ratio

def tokens_fast(text):
    if not text:
        return set()
    return set(text.split())

def extract_numbers_fast(text):
    return set(re.findall(r"\d+", text))

def build_fast_features(
    s1_name,
    s1_addr,
    s1_country,
    names,
    addrs,
    countries
):
    t1n = tokens_fast(s1_name)
    t1a = tokens_fast(s1_addr)
    nums1 = extract_numbers_fast(s1_addr)

    rows = []

    for n2, a2, c2 in zip(names, addrs, countries):

        t2n = tokens_fast(n2)
        t2a = tokens_fast(a2)

        shared_n = t1n & t2n
        shared_a = t1a & t2a

        union_n = t1n | t2n
        union_a = t1a | t2a

        ns = ratio(s1_name, n2) / 100 if s1_name and n2 else 0.0
        ass = ratio(s1_addr, a2) / 100 if s1_addr and a2 else 0.0

        nt = len(shared_n) / max(1, min(len(t1n), len(t2n)))
        at = len(shared_a) / max(1, min(len(t1a), len(t2a)))

        nj = len(shared_n) / len(union_n) if union_n else 0.0
        aj = len(shared_a) / len(union_a) if union_a else 0.0

        nums2 = extract_numbers_fast(a2)

        rows.append([
            int(bool(s1_name) and s1_name == n2),
            int(bool(s1_addr) and s1_addr == a2),
            int(s1_country == c2),
            ns,
            ass,
            nt,
            at,
            nj,
            aj,
            len(shared_n),
            len(shared_a),
            len(nums1 & nums2),
            abs(len(s1_name) - len(n2)),
            abs(len(s1_addr) - len(a2)),
            min(len(s1_name), len(n2)) /
                max(1, max(len(s1_name), len(n2))),
            min(len(s1_addr), len(a2)) /
                max(1, max(len(s1_addr), len(a2))),
            abs(ns - ass),
            max(ns, ass),
            ns * ass,
            abs(nt - at)
        ])

    return pd.DataFrame(rows, columns=V5_FEATURES)

print("✅ Fast V5 feature builder ready")
```

```python
# ============================================================
# STEP 13B — FAST S2 V5 INFERENCE
# ============================================================

import os
import gc
import time
import pandas as pd

OUT_DIR = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_Final_Production"
os.makedirs(OUT_DIR, exist_ok=True)

S2_MATCH = f"{OUT_DIR}/s2_results.tsv"
S2_CAND = f"{OUT_DIR}/s2_candidates.tsv"

# Raw arrays = lightweight references into the existing S2 dataframe
s2_names_raw = s2["business_name"].to_numpy()
s2_addrs_raw = s2["business_address"].to_numpy()

# Fresh output files
open(S2_MATCH, "w").close()
open(S2_CAND, "w").close()

start = time.time()
total_rows = 0
total_candidates = 0
total_matches = 0

print("=" * 70)
print("FAST V5 S2 INFERENCE STARTED")
print("=" * 70)

for chunk_start in range(0, len(test_s1), 2000):

    chunk = test_s1.iloc[
        chunk_start:chunk_start + 2000
    ]

    match_lines = []
    cand_lines = []

    for row in chunk.itertuples(index=False):

        sid = row.entity_id

        name1 = normalize_advanced(row.business_name)
        addr1 = normalize_advanced(row.business_address)
        country1 = normalize_country(row.country)

        # V5 raw candidates
        raw_candidates = generate_candidates(
            name1,
            addr1,
            country1,
            s2_name_index,
            s2_addr_index,
            s2_name_freq,
            s2_addr_freq
        )

        # Country filter
        candidates = [
            idx for idx in raw_candidates
            if s2_countries_a[idx] == country1
        ]

        # Original V5 similarity selection
        selected = []

        t_name = tokens_fast(name1)
        t_addr = tokens_fast(addr1)

        for idx in candidates:

            name2 = normalize_advanced(
                s2_names_raw[idx]
            )

            addr2 = normalize_advanced(
                s2_addrs_raw[idx]
            )

            shared_name = len(
                t_name & tokens_fast(name2)
            )

            shared_addr = len(
                t_addr & tokens_fast(addr2)
            )

            name_sim = (
                ratio(name1, name2) / 100
                if name1 and name2 else 0
            )

            addr_sim = (
                ratio(addr1, addr2) / 100
                if addr1 and addr2 else 0
            )

            keep = (
                name1 == name2
                or
                addr1 == addr2
                or
                (
                    shared_name >= 2
                    and name_sim >= 0.60
                )
                or
                (
                    shared_addr >= 3
                )
                or
                (
                    shared_addr >= 2
                    and addr_sim >= 0.30
                )
            )

            if keep:
                selected.append(idx)

        candidate_ids = [
            str(s2_ids[idx])
            for idx in selected
        ]

        matched_ids = []

        if selected:

            cand_names = [
                normalize_advanced(
                    s2_names_raw[idx]
                )
                for idx in selected
            ]

            cand_addrs = [
                normalize_advanced(
                    s2_addrs_raw[idx]
                )
                for idx in selected
            ]

            cand_countries = [
                s2_countries_a[idx]
                for idx in selected
            ]

            X = build_fast_features(
                name1,
                addr1,
                country1,
                cand_names,
                cand_addrs,
                cand_countries
            )

            probs = v5_model.predict_proba(
                X
            )[:, 1]

            matched_ids = [
                candidate_ids[j]
                for j, p in enumerate(probs)
                if p >= THRESHOLD
            ]

            del X, probs

        match_lines.append(
            sid + "\t" + ",".join(matched_ids) + "\n"
        )

        cand_lines.append(
            sid + "\t" + ",".join(candidate_ids) + "\n"
        )

        total_candidates += len(candidate_ids)
        total_matches += len(matched_ids)

    with open(S2_MATCH, "a", encoding="utf-8") as f:
        f.writelines(match_lines)

    with open(S2_CAND, "a", encoding="utf-8") as f:
        f.writelines(cand_lines)

    total_rows += len(chunk)

    if total_rows % 20000 == 0:

        elapsed = time.time() - start

        print(
            f"{total_rows:,}/{len(test_s1):,} | "
            f"candidates={total_candidates:,} | "
            f"matches={total_matches:,} | "
            f"time={elapsed/60:.1f} min"
        )

    del chunk, match_lines, cand_lines
    gc.collect()

print("\n" + "=" * 70)
print("✅ S2 COMPLETE")
print("=" * 70)
print("Rows:", total_rows)
print("Candidates:", total_candidates)
print("Matches:", total_matches)
print("Time:", round((time.time() - start) / 60, 2), "min")
```

```python
import pandas as pd
import gc

S3_PATH = "/content/amazon_ml_challenge/student_resource/dataset/test/test_source3.tsv"

s3 = pd.read_csv(S3_PATH, sep="\t")

print("S3 rows:", len(s3))
print("Columns:", s3.columns.tolist())
print(s3.head(2))

gc.collect()
```

```python
import re
import unicodedata
import numpy as np
import gc

def normalize_advanced(text):
    if pd.isna(text):
        return ""
    text = str(text).lower().strip()
    text = unicodedata.normalize("NFKC", text)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()

s3_names_raw = s3["business_name"].fillna("").to_numpy()
s3_addrs_raw = s3["business_address"].fillna("").to_numpy()
s3_ids = s3["entity_id"].to_numpy()
s3_countries_a = s3["country"].fillna("").str.lower().str.strip().to_numpy()

s3_names_a = np.array(
    [normalize_advanced(x) for x in s3_names_raw],
    dtype=object
)

s3_addrs_a = np.array(
    [normalize_advanced(x) for x in s3_addrs_raw],
    dtype=object
)

print("S3 arrays created:")
print("Names:", len(s3_names_a))
print("Addresses:", len(s3_addrs_a))
print("IDs:", len(s3_ids))
print("Countries:", len(s3_countries_a))

gc.collect()
```

```python
import pandas as pd
import numpy as np
import re
import unicodedata
import gc

S2_PATH = "/content/amazon_ml_challenge/student_resource/dataset/test/test_source2.tsv"

s2 = pd.read_csv(
    S2_PATH,
    sep="\t",
    dtype=str,
    usecols=["entity_id", "business_name", "business_address", "country"]
)

print("S2 rows:", len(s2))

def normalize_advanced(text):
    if pd.isna(text):
        return ""

    text = str(text).lower().strip()
    text = unicodedata.normalize("NFKC", text)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()

s2_ids = s2["entity_id"].to_numpy()

s2_names_a = np.array(
    [normalize_advanced(x) for x in s2["business_name"].fillna("")],
    dtype=object
)

s2_addrs_a = np.array(
    [normalize_advanced(x) for x in s2["business_address"].fillna("")],
    dtype=object
)

s2_countries_a = (
    s2["country"].fillna("").str.lower().str.strip().to_numpy()
)

print("S2 arrays ready:", len(s2_ids))
print("Names:", len(s2_names_a))
print("Addresses:", len(s2_addrs_a))
print("Countries:", len(s2_countries_a))

gc.collect()
```

```python
from collections import defaultdict, Counter
import pickle
import os
import gc

INDEX_DIR = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_full_indexes"
os.makedirs(INDEX_DIR, exist_ok=True)

RARITY_LIMIT = 500

print("Building S2 name index...")

s2_name_index = defaultdict(list)
s2_name_freq = Counter()

for eid, text in enumerate(s2_names_a):
    tokens = set(text.split())
    for tok in tokens:
        s2_name_index[tok].append(eid)
        s2_name_freq[tok] += 1

print("Name index keys:", len(s2_name_index))

print("Building S2 address index...")

s2_addr_index = defaultdict(list)
s2_addr_freq = Counter()

for eid, text in enumerate(s2_addrs_a):
    tokens = set(text.split())
    for tok in tokens:
        s2_addr_index[tok].append(eid)
        s2_addr_freq[tok] += 1

print("Address index keys:", len(s2_addr_index))

gc.collect()

print("Saving S2 indexes...")

with open(f"{INDEX_DIR}/s2_name_index.pkl", "wb") as f:
    pickle.dump(dict(s2_name_index), f, protocol=pickle.HIGHEST_PROTOCOL)

with open(f"{INDEX_DIR}/s2_addr_index.pkl", "wb") as f:
    pickle.dump(dict(s2_addr_index), f, protocol=pickle.HIGHEST_PROTOCOL)

with open(f"{INDEX_DIR}/s2_name_freq.pkl", "wb") as f:
    pickle.dump(dict(s2_name_freq), f, protocol=pickle.HIGHEST_PROTOCOL)

with open(f"{INDEX_DIR}/s2_addr_freq.pkl", "wb") as f:
    pickle.dump(dict(s2_addr_freq), f, protocol=pickle.HIGHEST_PROTOCOL)

print("✅ S2 indexes saved to Drive")
print(INDEX_DIR)
```

```python
import pickle
from collections import defaultdict, Counter
import gc
import psutil
import os

INDEX_DIR = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_full_indexes"
os.makedirs(INDEX_DIR, exist_ok=True)

# Build name index + frequency
s3_name_idx = defaultdict(list)
s3_name_freq = Counter()

for i, name in enumerate(s3_names_a):
    for token in name.split():
        s3_name_idx[token].append(s3_ids[i])
        s3_name_freq[token] += 1

print("Name index built:", len(s3_name_idx))
print("RAM:", round(psutil.virtual_memory().available / 1024**3, 2), "GB")

# Build address index + frequency
s3_addr_idx = defaultdict(list)
s3_addr_freq = Counter()

for i, addr in enumerate(s3_addrs_a):
    for token in addr.split():
        s3_addr_idx[token].append(s3_ids[i])
        s3_addr_freq[token] += 1

print("Address index built:", len(s3_addr_idx))
print("RAM:", round(psutil.virtual_memory().available / 1024**3, 2), "GB")

# Save immediately
with open(f"{INDEX_DIR}/s3_name_index.pkl", "wb") as f:
    pickle.dump(dict(s3_name_idx), f, protocol=pickle.HIGHEST_PROTOCOL)

with open(f"{INDEX_DIR}/s3_addr_index.pkl", "wb") as f:
    pickle.dump(dict(s3_addr_idx), f, protocol=pickle.HIGHEST_PROTOCOL)

with open(f"{INDEX_DIR}/s3_name_freq.pkl", "wb") as f:
    pickle.dump(dict(s3_name_freq), f, protocol=pickle.HIGHEST_PROTOCOL)

with open(f"{INDEX_DIR}/s3_addr_freq.pkl", "wb") as f:
    pickle.dump(dict(s3_addr_freq), f, protocol=pickle.HIGHEST_PROTOCOL)

print("✅ All 4 S3 indexes saved.")
```

```python
import pandas as pd
import gc
import psutil

TRAIN_DIR = "/content/amazon_ml_challenge/student_resource/dataset/train"

s1_train = pd.read_csv(
    f"{TRAIN_DIR}/train_source1.tsv",
    sep="\t",
    dtype=str
)

gt = pd.read_csv(
    f"{TRAIN_DIR}/train_ground_truth.tsv",
    sep="\t",
    dtype=str
)

print("S1:", s1_train.shape)
print("GT:", gt.shape)
print("Available RAM:", round(psutil.virtual_memory().available / 1024**3, 2), "GB")
```

```python
import gc
import psutil

gt_lookup = {}

for sid, matches in zip(
    gt["source1_entity_id"],
    gt["matched_entity_ids"].fillna("")
):
    gt_lookup[sid] = set(
        x.strip() for x in str(matches).split(",")
        if x.strip()
    )

del gt
gc.collect()

print("GT lookup:", len(gt_lookup))
print("Available RAM:", round(psutil.virtual_memory().available / 1024**3, 2), "GB")
```

```python
import re
import unicodedata
import gc
import psutil

def normalize_advanced(text):
    if pd.isna(text):
        return ""
    text = str(text).lower().strip()
    text = unicodedata.normalize("NFKC", text)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()

def normalize_country(text):
    if pd.isna(text):
        return ""
    return str(text).lower().strip()

train_ids = s1_train["entity_id"].to_numpy()

train_names = np.array(
    [normalize_advanced(x) for x in s1_train["business_name"].fillna("")],
    dtype=object
)

train_addrs = np.array(
    [normalize_advanced(x) for x in s1_train["business_address"].fillna("")],
    dtype=object
)

train_countries = s1_train["country"].fillna("").str.lower().str.strip().to_numpy()

print("S1 arrays:", len(train_ids))
print("Available RAM:", round(psutil.virtual_memory().available / 1024**3, 2), "GB")
```

```python
from collections import defaultdict

train_name_map = defaultdict(list)
train_address_map = defaultdict(list)

for i in range(len(train_ids)):
    sid = train_ids[i]
    country = train_countries[i]

    if train_names[i]:
        train_name_map[(country, train_names[i])].append(sid)

    if train_addrs[i]:
        train_address_map[(country, train_addrs[i])].append(sid)

print("Name keys:", len(train_name_map))
print("Address keys:", len(train_address_map))
```

```python
from collections import Counter
import pandas as pd
import gc
import psutil

name_freq = Counter()
addr_freq = Counter()

TRAIN_S2 = "/content/amazon_ml_challenge/student_resource/dataset/train/train_source2.tsv"
TRAIN_S3 = "/content/amazon_ml_challenge/student_resource/dataset/train/train_source3.tsv"

def add_freq(path):
    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=["business_name", "business_address"],
        chunksize=100_000
    ):
        names = chunk["business_name"].fillna("").map(normalize_advanced)
        addrs = chunk["business_address"].fillna("").map(normalize_advanced)

        for x in names:
            name_freq.update(set(x.split()))

        for x in addrs:
            addr_freq.update(set(x.split()))

        del chunk, names, addrs
        gc.collect()

add_freq(TRAIN_S2)
add_freq(TRAIN_S3)

print("Name tokens:", len(name_freq))
print("Address tokens:", len(addr_freq))
print("Available RAM:", round(psutil.virtual_memory().available / 1024**3, 2), "GB")
```

```python
import os

PROD = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_Final_Production_RAMSAFE"
OUT = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_SUBMISSION"

os.makedirs(OUT, exist_ok=True)

def split_result_file(source_file, matching_file, candidate_file):
    with open(source_file, "r", encoding="utf-8") as fin, \
         open(matching_file, "w", encoding="utf-8") as fout_match, \
         open(candidate_file, "w", encoding="utf-8") as fout_cand:

        header = fin.readline().strip()

        # Required submission headers
        fout_match.write("source1_entity_id\tmatched_entity_ids\n")
        fout_cand.write("source1_entity_id\tcandidate_entity_ids\n")

        rows = 0

        for line in fin:
            parts = line.rstrip("\n").split("\t")

            if len(parts) != 3:
                raise ValueError(
                    f"Bad line at row {rows + 2}: {parts}"
                )

            s1_id, candidates, matches = parts

            fout_match.write(
                f"{s1_id}\t{matches}\n"
            )

            fout_cand.write(
                f"{s1_id}\t{candidates}\n"
            )

            rows += 1

            if rows % 500_000 == 0:
                print(f"Processed {rows:,} rows")

    return rows


print("=== BUILDING S2 FILES ===")

n2 = split_result_file(
    f"{PROD}/s2_results.tsv",
    f"{OUT}/matching_results_s2.tsv",
    f"{OUT}/candidate_pairs_s2.tsv"
)

print("\n=== BUILDING S3 FILES ===")

n3 = split_result_file(
    f"{PROD}/s3_results.tsv",
    f"{OUT}/matching_results_s3.tsv",
    f"{OUT}/candidate_pairs_s3.tsv"
)

print("\n========== DONE ==========")
print("S2 rows:", f"{n2:,}")
print("S3 rows:", f"{n3:,}")
print("Output folder:", OUT)

for f in sorted(os.listdir(OUT)):
    size_mb = os.path.getsize(os.path.join(OUT, f)) / (1024**2)
    print(f"{f:30s} {size_mb:,.1f} MB")
```

```python
import os

OUT = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_SUBMISSION"

S2_MATCH = f"{OUT}/matching_results_s2.tsv"
S3_MATCH = f"{OUT}/matching_results_s3.tsv"

S2_CAND = f"{OUT}/candidate_pairs_s2.tsv"
S3_CAND = f"{OUT}/candidate_pairs_s3.tsv"

FINAL_MATCH = f"{OUT}/matching_results.tsv"
FINAL_CAND = f"{OUT}/candidate_pairs.tsv"


def combine_files(file1, file2, output, column_name):

    with open(file1, "r", encoding="utf-8") as f1, \
         open(file2, "r", encoding="utf-8") as f2, \
         open(output, "w", encoding="utf-8") as fout:

        # Skip headers
        f1.readline()
        f2.readline()

        fout.write(f"source1_entity_id\t{column_name}\n")

        rows = 0

        while True:
            line1 = f1.readline()
            line2 = f2.readline()

            if not line1 and not line2:
                break

            if not line1 or not line2:
                raise ValueError("S2/S3 row counts do not match!")

            p1 = line1.rstrip("\n").split("\t", 1)
            p2 = line2.rstrip("\n").split("\t", 1)

            s1_2, ids2 = p1
            s1_3, ids3 = p2

            if s1_2 != s1_3:
                raise ValueError(
                    f"S1 mismatch: {s1_2} != {s1_3}"
                )

            combined = []

            if ids2.strip():
                combined.extend(ids2.split(","))

            if ids3.strip():
                combined.extend(ids3.split(","))

            fout.write(
                f"{s1_2}\t{','.join(combined)}\n"
            )

            rows += 1

            if rows % 500_000 == 0:
                print(f"Combined {rows:,} rows")

    return rows


print("=== COMBINING MATCHING RESULTS ===")

match_rows = combine_files(
    S2_MATCH,
    S3_MATCH,
    FINAL_MATCH,
    "matched_entity_ids"
)

print("\n=== COMBINING CANDIDATES ===")

candidate_rows = combine_files(
    S2_CAND,
    S3_CAND,
    FINAL_CAND,
    "candidate_entity_ids"
)

print("\n========== FINAL FILES ==========")
print("Matching rows :", f"{match_rows:,}")
print("Candidate rows:", f"{candidate_rows:,}")

print("\nFiles:")
print(FINAL_MATCH)
print(FINAL_CAND)
```

```python
import os
import csv

OUT = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_SUBMISSION"

MATCH = f"{OUT}/matching_results.tsv"
CAND = f"{OUT}/candidate_pairs.tsv"

match_rows = 0
cand_rows = 0
match_ids = set()
cand_ids = set()

bad_matches = 0
duplicate_match_s1 = 0
duplicate_cand_s1 = 0
empty_matches = 0
nonempty_matches = 0

# --------------------------------------------------
# Check matching file
# --------------------------------------------------

print("Checking matching_results.tsv ...")

with open(MATCH, "r", encoding="utf-8") as f:
    reader = csv.reader(f, delimiter="\t")
    header = next(reader)

    print("Header:", header)

    for row in reader:
        if len(row) != 2:
            bad_matches += 1
            continue

        s1, matches = row
        match_rows += 1

        if s1 in match_ids:
            duplicate_match_s1 += 1
        match_ids.add(s1)

        if matches.strip():
            nonempty_matches += 1
        else:
            empty_matches += 1


# --------------------------------------------------
# Check candidate file
# --------------------------------------------------

print("\nChecking candidate_pairs.tsv ...")

with open(CAND, "r", encoding="utf-8") as f:
    reader = csv.reader(f, delimiter="\t")
    header = next(reader)

    print("Header:", header)

    for row in reader:
        if len(row) != 2:
            continue

        s1, candidates = row
        cand_rows += 1

        if s1 in cand_ids:
            duplicate_cand_s1 += 1
        cand_ids.add(s1)


# --------------------------------------------------
# Compare S1 coverage
# --------------------------------------------------

missing_candidates = match_ids - cand_ids
extra_candidates = cand_ids - match_ids


# --------------------------------------------------
# Check every match is inside candidates
# --------------------------------------------------

print("\nChecking match ⊆ candidate ...")

# Build candidate lookup one S1 at a time
candidate_map = {}

with open(CAND, "r", encoding="utf-8") as f:
    reader = csv.reader(f, delimiter="\t")
    next(reader)

    for s1, candidates in reader:
        candidate_map[s1] = set(
            x for x in candidates.split(",") if x
        )

with open(MATCH, "r", encoding="utf-8") as f:
    reader = csv.reader(f, delimiter="\t")
    next(reader)

    checked = 0

    for s1, matches in reader:
        if matches.strip():
            candidates = candidate_map.get(s1, set())

            for mid in matches.split(","):
                if mid not in candidates:
                    bad_matches += 1

            checked += 1


print("\n==============================")
print("      SELF-VALIDATION")
print("==============================")

print("Matching rows       :", f"{match_rows:,}")
print("Candidate rows      :", f"{cand_rows:,}")
print("Unique matching S1  :", f"{len(match_ids):,}")
print("Unique candidate S1 :", f"{len(cand_ids):,}")

print("Duplicate match S1  :", duplicate_match_s1)
print("Duplicate cand S1   :", duplicate_cand_s1)

print("Empty matches       :", f"{empty_matches:,}")
print("Non-empty matches   :", f"{nonempty_matches:,}")

print("Missing candidate S1:", len(missing_candidates))
print("Extra candidate S1  :", len(extra_candidates))

print("Invalid matches     :", bad_matches)

print("\nEXPECTED:")
print("Rows should be      : 1,732,544")
print("Unique S1 should be : 1,732,544")
print("Invalid matches     : 0")
```

```python
import csv
from collections import Counter

MATCH = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_SUBMISSION/matching_results.tsv"

dist = Counter()

with open(MATCH, "r", encoding="utf-8") as f:
    reader = csv.reader(f, delimiter="\t")
    next(reader)

    for s1, matches in reader:
        n = 0 if not matches.strip() else len(matches.split(","))
        dist[n] += 1

total = sum(dist.values())

print("========== CRITICAL MATCH SUMMARY ==========")

for n in range(0, 6):
    count = dist[n]
    print(
        f"{n} matches : {count:,} "
        f"({count/total*100:.2f}%)"
    )

five_plus = sum(v for k, v in dist.items() if k >= 5)
one_plus = total - dist[0]

print("\n1+ matches :", f"{one_plus:,}")
print("2+ matches :", f"{sum(v for k,v in dist.items() if k >= 2):,}")
print("3+ matches :", f"{sum(v for k,v in dist.items() if k >= 3):,}")
print("5+ matches :", f"{five_plus:,}")
print("Total S1   :", f"{total:,}")
```

```python
import csv
import os

SRC = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_SUBMISSION/matching_results.tsv"
OUT = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_SUBMISSION/matching_results_conservative1.tsv"

kept = 0
emptied = 0

with open(SRC, "r", encoding="utf-8") as fin, \
     open(OUT, "w", encoding="utf-8") as fout:

    reader = csv.reader(fin, delimiter="\t")
    writer = csv.writer(fout, delimiter="\t", lineterminator="\n")

    header = next(reader)
    writer.writerow(header)

    for s1, matches in reader:

        ids = [x for x in matches.split(",") if x]

        if len(ids) == 1:
            writer.writerow([s1, ids[0]])
            kept += 1
        else:
            writer.writerow([s1, ""])
            emptied += 1

print("Created:", OUT)
print("Kept single-match S1:", f"{kept:,}")
print("Emptied multi/zero S1 :", f"{emptied:,}")
print("Total:", f"{kept + emptied:,}")
```

```python
import csv
from collections import Counter

V1 = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V1_1_REPRODUCTION/matching_results.tsv"
V5 = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_SUBMISSION/matching_results.tsv"

v1 = {}
v1_dist = Counter()

print("Reading V1.1...")

with open(V1, "r", encoding="utf-8") as f:
    reader = csv.reader(f, delimiter="\t")
    next(reader)

    for s1, matches in reader:
        ids = set(x for x in matches.split(",") if x)
        v1[s1] = ids
        v1_dist[len(ids)] += 1

print("V1.1 rows:", f"{len(v1):,}")
print("V1.1 non-empty:", f"{sum(v for k,v in v1_dist.items() if k > 0):,}")

overlap_entities = 0
overlap_matches = 0
v5_nonempty = 0

print("\nComparing V5...")

with open(V5, "r", encoding="utf-8") as f:
    reader = csv.reader(f, delimiter="\t")
    next(reader)

    for s1, matches in reader:
        v5_ids = set(x for x in matches.split(",") if x)

        if v5_ids:
            v5_nonempty += 1

        common = v1.get(s1, set()) & v5_ids

        if common:
            overlap_entities += 1
            overlap_matches += len(common)

print("\n========== OVERLAP ==========")
print("V1.1 non-empty S1 :", f"{sum(v for k,v in v1_dist.items() if k > 0):,}")
print("V5 non-empty S1   :", f"{v5_nonempty:,}")
print("Common S1 entities:", f"{overlap_entities:,}")
print("Common match IDs  :", f"{overlap_matches:,}")

print("\nV1.1 distribution:")
for k in sorted(v1_dist):
    if k <= 5:
        print(f"{k} matches:", f"{v1_dist[k]:,}")
```

```python
import csv

V1 = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V1_1_REPRODUCTION/matching_results.tsv"
V5 = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_SUBMISSION/matching_results.tsv"

OUT = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_SUBMISSION/matching_results_intersection.tsv"

# Load V1.1 IDs
v1 = {}

with open(V1, "r", encoding="utf-8") as f:
    reader = csv.reader(f, delimiter="\t")
    next(reader)

    for s1, matches in reader:
        v1[s1] = set(x for x in matches.split(",") if x)

common_s1 = 0
common_ids = 0
total = 0

with open(V5, "r", encoding="utf-8") as fin, \
     open(OUT, "w", encoding="utf-8") as fout:

    reader = csv.reader(fin, delimiter="\t")
    writer = csv.writer(fout, delimiter="\t", lineterminator="\n")

    next(reader)
    writer.writerow(["source1_entity_id", "matched_entity_ids"])

    for s1, matches in reader:
        v5_ids = set(x for x in matches.split(",") if x)

        common = v1.get(s1, set()) & v5_ids

        if common:
            common_s1 += 1
            common_ids += len(common)
            writer.writerow([s1, ",".join(sorted(common))])
        else:
            writer.writerow([s1, ""])

        total += 1

print("========== INTERSECTION ==========")
print("Total S1       :", f"{total:,}")
print("Non-empty S1   :", f"{common_s1:,}")
print("Common match IDs:", f"{common_ids:,}")
print("Match rate     :", round(common_s1 / total * 100, 2), "%")
print("Output         :", OUT)
```

```python
import csv

V1 = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V1_1_REPRODUCTION/matching_results.tsv"
V5 = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_SUBMISSION/matching_results.tsv"

OUT = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_SUBMISSION/matching_results_v1plus_v5single.tsv"

v1 = {}

with open(V1, "r", encoding="utf-8") as f:
    reader = csv.reader(f, delimiter="\t")
    next(reader)

    for s1, matches in reader:
        v1[s1] = matches.strip()

added = 0
kept_v1 = 0
empty = 0
total = 0

with open(V5, "r", encoding="utf-8") as fin, \
     open(OUT, "w", encoding="utf-8") as fout:

    reader = csv.reader(fin, delimiter="\t")
    writer = csv.writer(fout, delimiter="\t", lineterminator="\n")

    next(reader)
    writer.writerow(["source1_entity_id", "matched_entity_ids"])

    for s1, v5_matches in reader:

        v1_matches = v1.get(s1, "")

        # Proven V1.1 prediction wins
        if v1_matches:
            final_matches = v1_matches
            kept_v1 += 1

        else:
            v5_ids = [x for x in v5_matches.split(",") if x]

            # Add ONLY V5 singleton predictions
            if len(v5_ids) == 1:
                final_matches = v5_ids[0]
                added += 1
            else:
                final_matches = ""
                empty += 1

        writer.writerow([s1, final_matches])
        total += 1

print("========== HYBRID ==========")
print("Total S1          :", f"{total:,}")
print("V1.1 retained     :", f"{kept_v1:,}")
print("V5 singleton added:", f"{added:,}")
print("Empty             :", f"{empty:,}")
print("Output            :", OUT)
```

```python
import pandas as pd
import re
import unicodedata
import gc

TEST_DIR = "/content/amazon_ml_challenge/student_resource/dataset/test"

HYBRID = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_SUBMISSION/matching_results_v1plus_v5single.tsv"
V1 = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V1_1_REPRODUCTION/matching_results.tsv"

def norm(x):
    if pd.isna(x):
        return ""
    x = unicodedata.normalize("NFKC", str(x).lower().strip())
    reps = {
        "&":" and ", "street":" st ", "road":" rd ",
        "avenue":" ave ", "drive":" dr ", "boulevard":" blvd ",
        "highway":" hwy ", "lane":" ln ", "parkway":" pkwy ",
        "place":" pl ", "suite":" ste "
    }
    for a,b in reps.items():
        x = x.replace(a,b)
    x = re.sub(r"[^\w\s]", " ", x)
    x = re.sub(r"\s+", " ", x)
    return x.strip()

# --------------------------------------------------
# 1. Find ONLY the 91,330 V5-added S1 -> target pairs
# --------------------------------------------------

v1_matches = {}

with open(V1, "r", encoding="utf-8") as f:
    next(f)
    for line in f:
        sid, matches = line.rstrip("\n").split("\t", 1)
        if matches:
            v1_matches[sid] = matches

added = {}

with open(HYBRID, "r", encoding="utf-8") as f:
    next(f)
    for line in f:
        sid, matches = line.rstrip("\n").split("\t", 1)

        # Hybrid added a V5 singleton only when V1 was empty
        if not v1_matches.get(sid) and matches:
            added[sid] = matches

print("V5-only additions:", f"{len(added):,}")

# --------------------------------------------------
# 2. Load ONLY the relevant S1 rows
# --------------------------------------------------

s1_info = {}

for chunk in pd.read_csv(
    f"{TEST_DIR}/test_source1.tsv",
    sep="\t",
    dtype=str,
    keep_default_na=False,
    chunksize=100_000
):
    for row in chunk.itertuples(index=False):
        sid = row.entity_id

        if sid in added:
            s1_info[sid] = (
                norm(row.business_name),
                norm(row.business_address),
                norm(row.country)
            )

    if len(s1_info) == len(added):
        break

print("Relevant S1 loaded:", f"{len(s1_info):,}")

# --------------------------------------------------
# 3. Stream S2 + S3 and verify target rows
# --------------------------------------------------

exact_name = 0
exact_address = 0
exact_both = 0
exact_country = 0
strong_exact = 0
checked = 0

for source_name in ["test_source2.tsv", "test_source3.tsv"]:

    path = f"{TEST_DIR}/{source_name}"

    print("\nScanning", source_name)

    for chunk in pd.read_csv(
        path,
        sep="\t",
        dtype=str,
        keep_default_na=False,
        chunksize=200_000
    ):
        for row in chunk.itertuples(index=False):

            target_id = row.entity_id

            # Find whether this target is one of our 91k additions
            matched_sid = None

            # Reverse lookup is expensive, so use a temporary target map
            # built below on first pass.
            if target_id in target_to_s1:
                matched_sid = target_to_s1[target_id]

            if matched_sid is None:
                continue

            n1, a1, c1 = s1_info[matched_sid]

            n2 = norm(row.business_name)
            a2 = norm(row.business_address)
            c2 = norm(row.country)

            checked += 1

            ne = bool(n1 and n1 == n2)
            ae = bool(a1 and a1 == a2)
            ce = bool(c1 and c1 == c2)

            if ne:
                exact_name += 1
            if ae:
                exact_address += 1
            if ne and ae:
                exact_both += 1
            if ce:
                exact_country += 1
            if ce and (ne or ae):
                strong_exact += 1

        del chunk
        gc.collect()

print("\n==============================")
print("V5 SINGLETON EVIDENCE CHECK")
print("==============================")
print("Additions:", f"{len(added):,}")
print("Target rows checked:", f"{checked:,}")
print("Exact name:", f"{exact_name:,}")
print("Exact address:", f"{exact_address:,}")
print("Exact both:", f"{exact_both:,}")
print("Exact country:", f"{exact_country:,}")
print("Exact name/address + country:", f"{strong_exact:,}")
```

```python
# ============================================================
# AMAZON ML CHALLENGE 2026
# ============================================================
# ONE-HOUR RAM-SAFE TRAINING AUDIT
#
# PURPOSE:
#   1. Verify cross-country behavior in actual TRAIN GT
#   2. Measure country-aware exact blocking recall
#   3. Measure country-aware address-number rescue
#   4. Measure exact-name / exact-address coverage
#   5. Measure true match cardinality
#   6. Produce evidence for the NEXT single submission
#
# IMPORTANT:
#   - NO leaderboard submission
#   - NO test prediction generation
#   - NO full S2/S3 loading
#   - Streaming/chunked only
#   - Sampled S1 validation to control runtime/RAM
# ============================================================

import os
import gc
import re
import csv
import time
import json
import math
import random
import unicodedata
from collections import defaultdict, Counter

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------

DRIVE_ROOT = "/content/drive/MyDrive/Amazon-ML-Challenge-2026"

# We intentionally use a sample for the expensive S2/S3 scans.
# 50k is large enough to give useful evidence while staying RAM-safe.
SAMPLE_S1 = 50_000

# Chunk size for all large TSV files.
CHUNK_SIZE = 100_000

# Random seed for reproducibility.
SEED = 42

# Address-number frequency limit.
# We don't want extremely common numbers such as "1", "10", etc.
NUMBER_LIMIT = 1000

# Save results here.
AUDIT_DIR = os.path.join(
    DRIVE_ROOT,
    "ML_AUDIT",
    "Submission_A_Planning"
)

os.makedirs(AUDIT_DIR, exist_ok=True)

REPORT_FILE = os.path.join(
    AUDIT_DIR,
    "training_country_number_audit.txt"
)

JSON_FILE = os.path.join(
    AUDIT_DIR,
    "training_country_number_audit.json"
)

START_TIME = time.time()

print("=" * 80)
print("AMAZON ML CHALLENGE 2026")
print("RAM-SAFE TRAINING AUDIT")
print("=" * 80)
print("Drive root:", DRIVE_ROOT)
print("Sample S1:", f"{SAMPLE_S1:,}")
print("Chunk size:", f"{CHUNK_SIZE:,}")
print("Output:", AUDIT_DIR)
print()


# ============================================================
# 1. FIND TRAINING FILES
# ============================================================

print("=" * 80)
print("STEP 1 — LOCATING TRAINING DATA")
print("=" * 80)

target_names = {
    "train_source1.tsv",
    "train_source2.tsv",
    "train_source3.tsv",
    "train_ground_truth.tsv"
}

found = {}

# We search recursively but DO NOT read any files yet.
for root, dirs, files in os.walk(DRIVE_ROOT):

    # Ignore some potentially huge irrelevant directories.
    dirs[:] = [
        d for d in dirs
        if d not in {
            ".git",
            "__pycache__",
            "node_modules"
        }
    ]

    for fname in files:
        if fname in target_names:
            path = os.path.join(root, fname)

            # Keep first valid occurrence.
            if fname not in found:
                found[fname] = path

for name in sorted(target_names):
    if name in found:
        print("FOUND:", found[name])
    else:
        print("MISSING:", name)

print()

required = [
    "train_source1.tsv",
    "train_source2.tsv",
    "train_source3.tsv",
    "train_ground_truth.tsv"
]

missing = [x for x in required if x not in found]

if missing:
    print("=" * 80)
    print("TRAINING DATA NOT AVAILABLE ON DRIVE")
    print("=" * 80)
    print("Missing:")
    for x in missing:
        print(" -", x)

    print()
    print(
        "STOPPING SAFELY. No large data was loaded and no submission was changed."
    )

    with open(REPORT_FILE, "w", encoding="utf-8") as f:
        f.write("TRAINING AUDIT COULD NOT START\n\n")
        f.write("Missing files:\n")
        for x in missing:
            f.write(f"{x}\n")

    raise SystemExit


S1_PATH = found["train_source1.tsv"]
S2_PATH = found["train_source2.tsv"]
S3_PATH = found["train_source3.tsv"]
GT_PATH = found["train_ground_truth.tsv"]

print("All training files found.")
print()


# ============================================================
# 2. NORMALIZATION FUNCTIONS
# ============================================================

print("=" * 80)
print("STEP 2 — INITIALIZING NORMALIZATION")
print("=" * 80)

def normalize_country(x):
    if x is None:
        return ""
    return str(x).strip().lower()


def normalize_text(x):
    if x is None:
        return ""

    if pd.isna(x):
        return ""

    text = str(x).lower().strip()

    text = unicodedata.normalize("NFKC", text)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()


_num_pattern = re.compile(r"\d+")


def extract_numbers(text):
    if not text:
        return set()

    return set(_num_pattern.findall(text))


def token_set(text):
    if not text:
        return set()

    return set(text.split())


# ============================================================
# 3. READ ONLY S1 + GT
# ============================================================

print("=" * 80)
print("STEP 3 — READING S1 + GROUND TRUTH")
print("=" * 80)

print("Loading S1 only...")

s1 = pd.read_csv(
    S1_PATH,
    sep="\t",
    dtype=str,
    keep_default_na=False
)

print("S1 shape:", s1.shape)

# We only keep the columns we need.
s1 = s1[
    [
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ]
].copy()

# ------------------------------------------------------------
# Random sample
# ------------------------------------------------------------

random.seed(SEED)

if len(s1) > SAMPLE_S1:
    sample_indices = sorted(
        random.sample(
            range(len(s1)),
            SAMPLE_S1
        )
    )

    s1_sample = s1.iloc[sample_indices].copy()
else:
    s1_sample = s1.copy()

print(
    "Validation sample:",
    f"{len(s1_sample):,}"
)

# ------------------------------------------------------------
# Normalize sampled S1
# ------------------------------------------------------------

s1_sample["country_n"] = (
    s1_sample["country"]
    .map(normalize_country)
)

s1_sample["name_n"] = (
    s1_sample["business_name"]
    .map(normalize_text)
)

s1_sample["address_n"] = (
    s1_sample["business_address"]
    .map(normalize_text)
)

s1_sample["address_numbers"] = (
    s1_sample["address_n"]
    .map(extract_numbers)
)

s1_sample["address_tokens"] = (
    s1_sample["address_n"]
    .map(token_set)
)

# ------------------------------------------------------------
# Create compact lookup
# ------------------------------------------------------------

s1_rows = {}

for row in s1_sample.itertuples(index=False):

    s1_rows[row.entity_id] = {
        "country": row.country_n,
        "name": row.name_n,
        "address": row.address_n,
        "numbers": row.address_numbers,
        "tokens": row.address_tokens
    }

print("S1 sample prepared.")
print()

# Free the full S1 DataFrame.
del s1
gc.collect()


# ============================================================
# 4. READ GROUND TRUTH
# ============================================================

print("=" * 80)
print("STEP 4 — READING GROUND TRUTH FOR SAMPLE")
print("=" * 80)

# We ONLY retain GT rows for sampled S1 entities.
sample_ids = set(s1_rows.keys())

gt_sample = {}

gt_rows_read = 0
gt_rows_kept = 0

for chunk in pd.read_csv(
    GT_PATH,
    sep="\t",
    dtype=str,
    keep_default_na=False,
    chunksize=CHUNK_SIZE
):

    gt_rows_read += len(chunk)

    for row in chunk.itertuples(index=False):

        sid = row.source1_entity_id

        if sid not in sample_ids:
            continue

        value = row.matched_entity_ids.strip()

        if value:
            matches = [
                x.strip()
                for x in value.split(",")
                if x.strip()
            ]
        else:
            matches = []

        gt_sample[sid] = matches
        gt_rows_kept += 1

    if gt_rows_read % 1_000_000 == 0:
        print(
            "GT rows scanned:",
            f"{gt_rows_read:,}",
            "| sample GT rows:",
            f"{gt_rows_kept:,}"
        )

print()
print("GT rows scanned:", f"{gt_rows_read:,}")
print("GT sample rows:", f"{gt_rows_kept:,}")

# Ensure every sampled S1 has a GT entry.
for sid in sample_ids:
    if sid not in gt_sample:
        gt_sample[sid] = []

print()


# ============================================================
# 5. TRUE MATCH CARDINALITY
# ============================================================

print("=" * 80)
print("STEP 5 — TRUE MATCH CARDINALITY")
print("=" * 80)

gt_cardinality = Counter()

for sid, matches in gt_sample.items():
    gt_cardinality[len(matches)] += 1

total_gt_entities = len(gt_sample)

for k in sorted(gt_cardinality):
    count = gt_cardinality[k]
    pct = (
        100.0 * count / total_gt_entities
        if total_gt_entities
        else 0
    )

    print(
        f"{k:>3} matches : "
        f"{count:>8,} "
        f"({pct:6.2f}%)"
    )

print()

# Useful categories.
gt_nonempty = sum(
    count
    for k, count in gt_cardinality.items()
    if k > 0
)

gt_empty = gt_cardinality.get(0, 0)

gt_singleton = gt_cardinality.get(1, 0)

gt_multi = sum(
    count
    for k, count in gt_cardinality.items()
    if k >= 2
)

print("GT empty:", gt_empty)
print("GT non-empty:", gt_nonempty)
print("GT singleton:", gt_singleton)
print("GT multi-match:", gt_multi)
print()


# ============================================================
# 6. PREPARE GROUND-TRUTH ID SET
# ============================================================

print("=" * 80)
print("STEP 6 — PREPARING SAMPLE MATCH IDs")
print("=" * 80)

# This set is deliberately restricted to the 50k sample.
# It should remain manageable.
gt_match_ids = set()

for matches in gt_sample.values():
    for eid in matches:
        gt_match_ids.add(eid)

print(
    "Unique matched S2/S3 IDs in sample:",
    f"{len(gt_match_ids):,}"
)

print()


# ============================================================
# 7. STREAM S2 + S3
# ============================================================

print("=" * 80)
print("STEP 7 — STREAMING SOURCE 2 + SOURCE 3")
print("=" * 80)

print(
    "This is the expensive part.",
    "It will scan S2 and S3 without loading either file."
)
print()


# ------------------------------------------------------------
# Candidate statistics
# ------------------------------------------------------------

stats = {
    "rows_s2": 0,
    "rows_s3": 0,

    "gt_match_rows_seen": 0,

    "country_consistent_gt": 0,
    "country_inconsistent_gt": 0,

    "exact_name_hits": 0,
    "exact_address_hits": 0,

    "country_exact_name_hits": 0,
    "country_exact_address_hits": 0,

    "number_token_hits": 0,

    "gt_recovered_by_exact_name": 0,
    "gt_recovered_by_exact_address": 0,
    "gt_recovered_by_country_exact_name": 0,
    "gt_recovered_by_country_exact_address": 0,

    "gt_recovered_by_number_and_token": 0,

    "gt_total_links": 0
}


# ------------------------------------------------------------
# Build compact indexes from SAMPLE S1
# ------------------------------------------------------------

print("Building sample S1 indexes...")

name_index = defaultdict(set)
address_index = defaultdict(set)
country_name_index = defaultdict(set)
country_address_index = defaultdict(set)

country_number_index = defaultdict(set)

for sid, data in s1_rows.items():

    country = data["country"]
    name = data["name"]
    address = data["address"]

    if name:
        name_index[name].add(sid)

        if country:
            country_name_index[
                (country, name)
            ].add(sid)

    if address:
        address_index[address].add(sid)

        if country:
            country_address_index[
                (country, address)
            ].add(sid)

    for num in data["numbers"]:

        # We use country + number.
        if country and num:
            country_number_index[
                (country, num)
            ].add(sid)

print(
    "Name keys:",
    f"{len(name_index):,}"
)

print(
    "Address keys:",
    f"{len(address_index):,}"
)

print(
    "Country-name keys:",
    f"{len(country_name_index):,}"
)

print(
    "Country-address keys:",
    f"{len(country_address_index):,}"
)

print(
    "Country-number keys:",
    f"{len(country_number_index):,}"
)

print()


# ============================================================
# 8. STREAMING FUNCTION
# ============================================================

def process_source(path, source_name):

    print("-" * 80)
    print("PROCESSING", source_name)
    print("-" * 80)

    processed = 0
    last_report = 0

    # --------------------------------------------------------
    # We don't store all candidates.
    # We only accumulate whether TRUE GT links were recoverable.
    # --------------------------------------------------------

    local = {
        "gt_links": 0,
        "country_consistent": 0,
        "country_inconsistent": 0,

        "exact_name": 0,
        "exact_address": 0,

        "country_exact_name": 0,
        "country_exact_address": 0,

        "number_token": 0,

        "gt_exact_name": 0,
        "gt_exact_address": 0,

        "gt_country_exact_name": 0,
        "gt_country_exact_address": 0,

        "gt_number_token": 0
    }

    for chunk in pd.read_csv(
        path,
        sep="\t",
        dtype=str,
        keep_default_na=False,
        chunksize=CHUNK_SIZE
    ):

        for row in chunk.itertuples(index=False):

            processed += 1

            eid = row.entity_id

            # ------------------------------------------------
            # We only need detailed work for IDs that are
            # actually GT matches of our S1 sample.
            # ------------------------------------------------

            is_gt = eid in gt_match_ids

            # Fast path for non-GT rows:
            # We don't need to calculate all expensive features.
            if not is_gt:
                continue

            local["gt_links"] += 1

            name = normalize_text(row.business_name)
            address = normalize_text(row.business_address)
            country = normalize_country(row.country)

            # ------------------------------------------------
            # Find the S1 entities whose GT contains this ID.
            #
            # We build this reverse mapping only for the
            # sampled GT, not the entire dataset.
            # ------------------------------------------------

            possible_s1 = gt_reverse.get(eid, [])

            # ------------------------------------------------
            # Country consistency
            # ------------------------------------------------

            for sid in possible_s1:

                s1data = s1_rows[sid]

                if (
                    s1data["country"]
                    and country
                    and s1data["country"] == country
                ):
                    local["country_consistent"] += 1
                else:
                    local["country_inconsistent"] += 1

            # ------------------------------------------------
            # Exact name
            # ------------------------------------------------

            name_candidates = (
                name_index.get(name, set())
                if name
                else set()
            )

            if name_candidates:
                local["exact_name"] += 1

            if any(
                eid in gt_sample.get(sid, [])
                for sid in name_candidates
            ):
                local["gt_exact_name"] += 1

            # ------------------------------------------------
            # Exact address
            # ------------------------------------------------

            address_candidates = (
                address_index.get(address, set())
                if address
                else set()
            )

            if address_candidates:
                local["exact_address"] += 1

            if any(
                eid in gt_sample.get(sid, [])
                for sid in address_candidates
            ):
                local["gt_exact_address"] += 1

            # ------------------------------------------------
            # Country + exact name
            # ------------------------------------------------

            country_name_candidates = (
                country_name_index.get(
                    (country, name),
                    set()
                )
                if country and name
                else set()
            )

            if country_name_candidates:
                local["country_exact_name"] += 1

            if any(
                eid in gt_sample.get(sid, [])
                for sid in country_name_candidates
            ):
                local["gt_country_exact_name"] += 1

            # ------------------------------------------------
            # Country + exact address
            # ------------------------------------------------

            country_address_candidates = (
                country_address_index.get(
                    (country, address),
                    set()
                )
                if country and address
                else set()
            )

            if country_address_candidates:
                local["country_exact_address"] += 1

            if any(
                eid in gt_sample.get(sid, [])
                for sid in country_address_candidates
            ):
                local["gt_country_exact_address"] += 1

            # ------------------------------------------------
            # Address number + token rescue
            # ------------------------------------------------

            source_numbers = extract_numbers(address)
            source_tokens = token_set(address)

            number_candidates = set()

            for num in source_numbers:

                if num not in country_number_index:
                    continue

                possible = country_number_index[
                    (country, num)
                ]

                number_candidates.update(possible)

            if number_candidates:
                local["number_token"] += 1

            # Require:
            # 1. same country
            # 2. shared address number
            # 3. shared address token

            for sid in number_candidates:

                s1data = s1_rows[sid]

                if not (
                    source_numbers
                    & s1data["numbers"]
                ):
                    continue

                if not (
                    source_tokens
                    & s1data["tokens"]
                ):
                    continue

                if eid in gt_sample.get(sid, []):
                    local["gt_number_token"] += 1
                    break

        # ----------------------------------------------------
        # Progress report
        # ----------------------------------------------------

        if processed - last_report >= 1_000_000:

            elapsed = time.time() - START_TIME

            print(
                source_name,
                "| processed:",
                f"{processed:,}",
                "| GT rows:",
                f"{local['gt_links']:,}",
                "| elapsed:",
                f"{elapsed / 60:.1f} min"
            )

            last_report = processed

        # Explicit cleanup after each chunk.
        del chunk
        gc.collect()

    print()
    print(
        source_name,
        "COMPLETE:",
        f"{processed:,}",
        "rows"
    )

    return processed, local


# ============================================================
# 9. BUILD GT REVERSE INDEX
# ============================================================

print("=" * 80)
print("BUILDING SAMPLE GT REVERSE INDEX")
print("=" * 80)

gt_reverse = defaultdict(list)

for sid, matches in gt_sample.items():
    for eid in matches:
        gt_reverse[eid].append(sid)

print(
    "GT IDs with reverse links:",
    f"{len(gt_reverse):,}"
)

print()


# ============================================================
# 10. PROCESS S2
# ============================================================

rows2, result2 = process_source(
    S2_PATH,
    "SOURCE 2"
)

gc.collect()


# ============================================================
# 11. PROCESS S3
# ============================================================

rows3, result3 = process_source(
    S3_PATH,
    "SOURCE 3"
)

gc.collect()


# ============================================================
# 12. COMBINE RESULTS
# ============================================================

print()
print("=" * 80)
print("STEP 12 — COMBINING RESULTS")
print("=" * 80)

combined = {}

for key in result2:

    combined[key] = (
        result2.get(key, 0)
        +
        result3.get(key, 0)
    )

combined["rows_s2"] = rows2
combined["rows_s3"] = rows3
combined["sample_s1"] = len(s1_rows)
combined["sample_gt_entities"] = len(gt_sample)
combined["gt_unique_match_ids"] = len(gt_match_ids)


# ============================================================
# 13. CALCULATE RECALL METRICS
# ============================================================

print()
print("=" * 80)
print("STEP 13 — CALCULATING BLOCKING RECALL")
print("=" * 80)

total_links = combined["gt_links"]

print("Total GT links in sample:", f"{total_links:,}")
print()

if total_links > 0:

    metrics = {

        "exact_name_recall":
            combined["gt_exact_name"] / total_links,

        "exact_address_recall":
            combined["gt_exact_address"] / total_links,

        "country_exact_name_recall":
            combined["gt_country_exact_name"] / total_links,

        "country_exact_address_recall":
            combined["gt_country_exact_address"] / total_links,

        "address_number_token_recall":
            combined["gt_number_token"] / total_links,
    }

else:
    metrics = {}

for name, value in metrics.items():

    print(
        f"{name:35s}: "
        f"{value * 100:7.3f}%"
    )

print()


# ============================================================
# 14. COUNTRY CONSISTENCY
# ============================================================

print("=" * 80)
print("STEP 14 — COUNTRY CONSISTENCY OF ACTUAL GT")
print("=" * 80)

country_total = (
    combined["country_consistent"]
    +
    combined["country_inconsistent"]
)

if country_total:

    consistency = (
        combined["country_consistent"]
        / country_total
    )

    inconsistency = (
        combined["country_inconsistent"]
        / country_total
    )

    print(
        "Country-consistent GT links:",
        f"{combined['country_consistent']:,}",
        f"({consistency * 100:.4f}%)"
    )

    print(
        "Cross-country GT links:",
        f"{combined['country_inconsistent']:,}",
        f"({inconsistency * 100:.4f}%)"
    )

else:

    consistency = None
    inconsistency = None

print()


# ============================================================
# 15. ESTIMATE BENEFIT OF COUNTRY BLOCKING
# ============================================================

print("=" * 80)
print("STEP 15 — COUNTRY BLOCKING INTERPRETATION")
print("=" * 80)

if consistency is not None:

    if consistency >= 0.999:

        country_conclusion = (
            "VERY STRONG EVIDENCE: sampled GT matches are "
            "country-consistent. Country-aware blocking is "
            "strongly justified for the next experiment."
        )

    elif consistency >= 0.99:

        country_conclusion = (
            "STRONG EVIDENCE: country-aware blocking should "
            "be investigated seriously."
        )

    elif consistency >= 0.95:

        country_conclusion = (
            "COUNTRY IS USEFUL but not perfectly strict. "
            "Use country as a strong feature/block, not an "
            "absolute exclusion without further analysis."
        )

    else:

        country_conclusion = (
            "CAUTION: some GT matches cross countries. "
            "Do NOT make country a hard blocker yet."
        )

else:

    country_conclusion = "Insufficient GT links."

print(country_conclusion)
print()


# ============================================================
# 16. ADDRESS-NUMBER INTERPRETATION
# ============================================================

print("=" * 80)
print("STEP 16 — ADDRESS NUMBER RESCUE")
print("=" * 80)

number_recall = metrics.get(
    "address_number_token_recall",
    0
)

address_recall = metrics.get(
    "country_exact_address_recall",
    0
)

print(
    "Country + exact address recall:",
    f"{address_recall * 100:.3f}%"
)

print(
    "Country + number + token recall:",
    f"{number_recall * 100:.3f}%"
)

if number_recall > address_recall:

    number_conclusion = (
        "Address-number + token blocking recovers additional "
        "GT links beyond exact country+address."
    )

else:

    number_conclusion = (
        "Address-number + token blocking does not show a "
        "recall advantage over exact country+address in this sample."
    )

print(number_conclusion)
print()


# ============================================================
# 17. SAVE REPORT
# ============================================================

print("=" * 80)
print("STEP 17 — SAVING AUDIT RESULTS")
print("=" * 80)

report_lines = []

report_lines.append(
    "AMAZON ML CHALLENGE 2026 — TRAINING AUDIT"
)

report_lines.append("=" * 70)

report_lines.append(
    f"Sample S1: {len(s1_rows):,}"
)

report_lines.append(
    f"GT entities: {len(gt_sample):,}"
)

report_lines.append(
    f"S2 rows scanned: {rows2:,}"
)

report_lines.append(
    f"S3 rows scanned: {rows3:,}"
)

report_lines.append("")

report_lines.append(
    "TRUE MATCH CARDINALITY"
)

for k in sorted(gt_cardinality):

    count = gt_cardinality[k]

    pct = (
        100 * count / total_gt_entities
        if total_gt_entities
        else 0
    )

    report_lines.append(
        f"{k} matches: {count:,} ({pct:.3f}%)"
    )

report_lines.append("")

report_lines.append(
    "BLOCKING RECALL"
)

for name, value in metrics.items():

    report_lines.append(
        f"{name}: {value * 100:.4f}%"
    )

report_lines.append("")

report_lines.append(
    "COUNTRY CONSISTENCY"
)

if consistency is not None:

    report_lines.append(
        f"Country-consistent: {consistency * 100:.4f}%"
    )

    report_lines.append(
        f"Cross-country: {inconsistency * 100:.4f}%"
    )

report_lines.append("")

report_lines.append(
    "COUNTRY CONCLUSION"
)

report_lines.append(
    country_conclusion
)

report_lines.append("")

report_lines.append(
    "ADDRESS NUMBER CONCLUSION"
)

report_lines.append(
    number_conclusion
)

report_lines.append("")

elapsed = time.time() - START_TIME

report_lines.append(
    f"Runtime: {elapsed / 60:.2f} minutes"
)

with open(
    REPORT_FILE,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "\n".join(report_lines)
    )


# ------------------------------------------------------------
# JSON
# ------------------------------------------------------------

json_data = {

    "sample_s1": len(s1_rows),

    "sample_gt_entities":
        len(gt_sample),

    "rows_s2":
        rows2,

    "rows_s3":
        rows3,

    "gt_cardinality":
        {
            str(k): int(v)
            for k, v in gt_cardinality.items()
        },

    "metrics":
        {
            k: float(v)
            for k, v in metrics.items()
        },

    "country_consistent":
        int(combined["country_consistent"]),

    "country_inconsistent":
        int(combined["country_inconsistent"]),

    "country_consistency":
        None
        if consistency is None
        else float(consistency),

    "country_conclusion":
        country_conclusion,

    "number_conclusion":
        number_conclusion,

    "runtime_minutes":
        float((time.time() - START_TIME) / 60)
}

with open(
    JSON_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        json_data,
        f,
        indent=2
    )


# ============================================================
# 18. FINAL SUMMARY
# ============================================================

print()
print("=" * 80)
print("✅ TRAINING AUDIT COMPLETE")
print("=" * 80)

print(
    "Sample S1:",
    f"{len(s1_rows):,}"
)

print(
    "S2 scanned:",
    f"{rows2:,}"
)

print(
    "S3 scanned:",
    f"{rows3:,}"
)

print()

print("GT CARDINALITY:")

for k in sorted(gt_cardinality):

    print(
        f"  {k} matches:",
        f"{gt_cardinality[k]:,}"
    )

print()

print("RECALL:")

for name, value in metrics.items():

    print(
        f"  {name}:",
        f"{value * 100:.3f}%"
    )

print()

print(
    "COUNTRY CONSISTENCY:",
    f"{consistency * 100:.3f}%"
    if consistency is not None
    else "N/A"
)

print()

print("COUNTRY CONCLUSION:")
print(country_conclusion)

print()

print("ADDRESS NUMBER:")
print(number_conclusion)

print()

print("REPORT SAVED:")
print(REPORT_FILE)

print()
print("JSON SAVED:")
print(JSON_FILE)

print()

print(
    "Total runtime:",
    f"{(time.time() - START_TIME) / 60:.2f} minutes"
)

print()
print("=" * 80)
print("DO NOT SUBMIT ANYTHING FROM THIS CELL.")
print("This cell is analysis only.")
print("=" * 80)

# Final cleanup.
gc.collect()
```

```python
import os
import pandas as pd
from collections import Counter

DATASET = "/content/amazon_ml_challenge/student_resource/dataset"

TRAIN_S1 = os.path.join(DATASET, "train", "train_source1.tsv")
TEST_S1  = os.path.join(DATASET, "test", "test_source1.tsv")
GT_PATH  = os.path.join(DATASET, "train", "train_ground_truth.tsv")

print("=" * 80)
print("STEP 5 — COUNTRY / FRANCE CHECK")
print("=" * 80)

# ------------------------------------------------------------
# 1. CHECK SCHEMA WITHOUT LOADING WHOLE FILE
# ------------------------------------------------------------

print("\nTRAIN S1 columns:")
train_head = pd.read_csv(
    TRAIN_S1,
    sep="\t",
    dtype=str,
    nrows=3,
    keep_default_na=False
)
print(train_head.columns.tolist())

print("\nTEST S1 columns:")
test_head = pd.read_csv(
    TEST_S1,
    sep="\t",
    dtype=str,
    nrows=3,
    keep_default_na=False
)
print(test_head.columns.tolist())

print("\nGROUND TRUTH columns:")
gt_head = pd.read_csv(
    GT_PATH,
    sep="\t",
    dtype=str,
    nrows=5,
    keep_default_na=False
)
print(gt_head.columns.tolist())

print("\nGround truth sample:")
print(gt_head.to_string(index=False))

del train_head, test_head, gt_head
import gc
gc.collect()


# ------------------------------------------------------------
# 2. TRAIN S1 COUNTRY DISTRIBUTION
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("TRAIN S1 COUNTRY DISTRIBUTION")
print("=" * 80)

train_country = Counter()
train_rows = 0

for chunk in pd.read_csv(
    TRAIN_S1,
    sep="\t",
    dtype={"country": "string"},
    usecols=["country"],
    chunksize=200_000,
    keep_default_na=False
):
    train_rows += len(chunk)

    for c in chunk["country"]:
        c = str(c).strip().lower()
        train_country[c] += 1

    del chunk

print("Training S1 rows:", f"{train_rows:,}")

for country, count in train_country.most_common():
    print(
        f"{country!r:20s} "
        f"{count:>10,} "
        f"({100*count/train_rows:6.2f}%)"
    )


# ------------------------------------------------------------
# 3. TEST S1 COUNTRY DISTRIBUTION
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("TEST S1 COUNTRY DISTRIBUTION")
print("=" * 80)

test_country = Counter()
test_rows = 0

for chunk in pd.read_csv(
    TEST_S1,
    sep="\t",
    dtype={"country": "string"},
    usecols=["country"],
    chunksize=200_000,
    keep_default_na=False
):
    test_rows += len(chunk)

    for c in chunk["country"]:
        c = str(c).strip().lower()
        test_country[c] += 1

    del chunk

print("Test S1 rows:", f"{test_rows:,}")

for country, count in test_country.most_common():
    print(
        f"{country!r:20s} "
        f"{count:>10,} "
        f"({100*count/test_rows:6.2f}%)"
    )


# ------------------------------------------------------------
# 4. FRANCE CHECK
# ------------------------------------------------------------

train_france = train_country.get("france", 0)
test_france = test_country.get("france", 0)

print("\n" + "=" * 80)
print("🇫🇷 FRANCE RESULT")
print("=" * 80)

print("France in TRAIN S1:", f"{train_france:,}")
print("France in TEST S1 :", f"{test_france:,}")

if train_france == 0 and test_france > 0:
    print("\n✅ CONFIRMED: France is unseen in TRAIN and present in TEST.")

elif train_france > 0:
    print("\n⚠️ France exists in training — this differs from the challenge description.")

else:
    print("\n⚠️ France was not found in test S1.")


# ------------------------------------------------------------
# 5. MEMORY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("RAM-SAFE CHECK COMPLETE")
print("=" * 80)

gc.collect()
```

```python
# ============================================================
# STEP 6 — TRAINING GT COUNTRY CONSISTENCY
# RAM-SAFE SAMPLED VALIDATION
# ============================================================

import os
import gc
import random
import pandas as pd
from collections import defaultdict

DATASET = "/content/amazon_ml_challenge/student_resource/dataset"

S1_PATH = os.path.join(DATASET, "train", "train_source1.tsv")
S2_PATH = os.path.join(DATASET, "train", "train_source2.tsv")
S3_PATH = os.path.join(DATASET, "train", "train_source3.tsv")
GT_PATH = os.path.join(DATASET, "train", "train_ground_truth.tsv")

SAMPLE_S1 = 50_000
CHUNK_SIZE = 200_000
SEED = 42

print("=" * 80)
print("STEP 6 — TRAINING GT COUNTRY CONSISTENCY")
print("=" * 80)

# ------------------------------------------------------------
# 1. SAMPLE S1 IDs + COUNTRY
# ------------------------------------------------------------

print("\n[1/4] Reading S1 countries...")

s1_sample = pd.read_csv(
    S1_PATH,
    sep="\t",
    dtype=str,
    usecols=["entity_id", "country"],
    keep_default_na=False
)

random.seed(SEED)

if len(s1_sample) > SAMPLE_S1:
    s1_sample = s1_sample.sample(
        n=SAMPLE_S1,
        random_state=SEED
    ).reset_index(drop=True)

s1_country = {}

for row in s1_sample.itertuples(index=False):
    s1_country[row.entity_id] = str(row.country).strip().lower()

sample_ids = set(s1_country.keys())

print("Sampled S1:", f"{len(sample_ids):,}")

del s1_sample
gc.collect()


# ------------------------------------------------------------
# 2. GET GT MATCHES FOR SAMPLED S1
# ------------------------------------------------------------

print("\n[2/4] Reading GT for sampled S1...")

# matched_id -> expected S1 country
expected_country = {}

# Keep some entity-level information too.
gt_by_s1 = defaultdict(list)

rows_scanned = 0

for chunk in pd.read_csv(
    GT_PATH,
    sep="\t",
    dtype=str,
    keep_default_na=False,
    chunksize=CHUNK_SIZE
):

    rows_scanned += len(chunk)

    for row in chunk.itertuples(index=False):

        sid = row.source1_entity_id

        if sid not in sample_ids:
            continue

        value = row.matched_entity_ids.strip()

        if not value:
            continue

        country = s1_country[sid]

        for mid in value.split(","):

            mid = mid.strip()

            if not mid:
                continue

            gt_by_s1[sid].append(mid)

            # Same source entity should normally have
            # one expected country. If duplicate GT IDs
            # appear, we retain the first observed country.
            if mid not in expected_country:
                expected_country[mid] = country

print(
    "GT rows scanned:",
    f"{rows_scanned:,}"
)

print(
    "Unique matched S2/S3 IDs in sample:",
    f"{len(expected_country):,}"
)

total_gt_links = sum(
    len(v) for v in gt_by_s1.values()
)

print(
    "Total GT links in sample:",
    f"{total_gt_links:,}"
)

del chunk
gc.collect()


# ------------------------------------------------------------
# 3. LOOK UP COUNTRIES IN S2/S3
# ------------------------------------------------------------

print("\n[3/4] Scanning S2 + S3 for those matched IDs...")
print("Only GT-relevant IDs are inspected.")

same_country = 0
different_country = 0
not_found = 0

# Keep examples of any cross-country links.
cross_country_examples = []

def inspect_source(path, source_name):

    global same_country
    global different_country
    global not_found
    global cross_country_examples

    found_ids = set()

    processed = 0
    relevant = 0

    print("\n---", source_name, "---")

    for chunk in pd.read_csv(
        path,
        sep="\t",
        dtype=str,
        usecols=["entity_id", "country"],
        keep_default_na=False,
        chunksize=CHUNK_SIZE
    ):

        processed += len(chunk)

        # IMPORTANT:
        # Filter before doing any Python-level work.
        mask = chunk["entity_id"].isin(expected_country)

        relevant_chunk = chunk.loc[
            mask,
            ["entity_id", "country"]
        ]

        relevant += len(relevant_chunk)

        for row in relevant_chunk.itertuples(index=False):

            mid = row.entity_id
            actual_country = str(row.country).strip().lower()
            expected = expected_country[mid]

            found_ids.add(mid)

            if actual_country == expected:
                same_country += 1
            else:
                different_country += 1

                if len(cross_country_examples) < 20:
                    cross_country_examples.append({
                        "source": source_name,
                        "matched_id": mid,
                        "expected_country": expected,
                        "actual_country": actual_country
                    })

        if processed % 1_000_000 == 0:
            print(
                "Processed:",
                f"{processed:,}",
                "| relevant:",
                f"{relevant:,}"
            )

        del relevant_chunk
        del chunk
        gc.collect()

    print(
        source_name,
        "complete:",
        f"{processed:,}",
        "rows | relevant:",
        f"{relevant:,}"
    )

    return found_ids


found_s2 = inspect_source(
    S2_PATH,
    "S2"
)

gc.collect()

found_s3 = inspect_source(
    S3_PATH,
    "S3"
)

gc.collect()


# ------------------------------------------------------------
# 4. FINAL REPORT
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("STEP 6 RESULT")
print("=" * 80)

found_total = len(found_s2) + len(found_s3)

print(
    "Expected matched IDs:",
    f"{len(expected_country):,}"
)

print(
    "S2 matched IDs found:",
    f"{len(found_s2):,}"
)

print(
    "S3 matched IDs found:",
    f"{len(found_s3):,}"
)

print()

print(
    "Same-country GT links:",
    f"{same_country:,}"
)

print(
    "Cross-country GT links:",
    f"{different_country:,}"
)

print(
    "Country mismatches not observed:",
    "N/A"
)

checked = same_country + different_country

if checked:
    print(
        "\nSame-country percentage:",
        f"{100 * same_country / checked:.4f}%"
    )

    print(
        "Cross-country percentage:",
        f"{100 * different_country / checked:.4f}%"
    )

print("\nCross-country examples:")

if cross_country_examples:
    for x in cross_country_examples:
        print(x)
else:
    print("NONE FOUND IN SAMPLE")

print("\n" + "=" * 80)
print("RAM-SAFE COUNTRY VALIDATION COMPLETE")
print("=" * 80)

gc.collect()
```

```python
import csv
import os

ROOT = "/content/drive/MyDrive/Amazon-ML-Challenge-2026"

V1_PATH = os.path.join(ROOT, "ml/V1_1_REPRODUCTION/matching_results.tsv")
S2_PATH = os.path.join(ROOT, "ml/V5_Final_Production_RAMSAFE/s2_results.tsv")
S3_PATH = os.path.join(ROOT, "ml/V5_Final_Production_RAMSAFE/s3_results.tsv")

S1_TEST_PATH = "/content/amazon_ml_challenge/student_resource/dataset/test/test_source1.tsv"
S2_TEST_PATH = "/content/amazon_ml_challenge/student_resource/dataset/test/test_source2.tsv"
S3_TEST_PATH = "/content/amazon_ml_challenge/student_resource/dataset/test/test_source3.tsv"

# ------------------------------------------------------------
# 1. Recover exactly the 91,330 baseline additions
# ------------------------------------------------------------

additions = {}

with open(V1_PATH, "r", encoding="utf-8", newline="") as f1, \
     open(S2_PATH, "r", encoding="utf-8", newline="") as f2, \
     open(S3_PATH, "r", encoding="utf-8", newline="") as f3:

    r1 = csv.DictReader(f1, delimiter="\t")
    r2 = csv.DictReader(f2, delimiter="\t")
    r3 = csv.DictReader(f3, delimiter="\t")

    for row1, row2, row3 in zip(r1, r2, r3):

        m1 = row1["matched_entity_ids"].strip()
        m2 = row2["matched_entity_ids"].strip()
        m3 = row3["matched_entity_ids"].strip()

        # Only V1-empty entities
        if m1:
            continue

        n2 = len(m2.split(",")) if m2 else 0
        n3 = len(m3.split(",")) if m3 else 0

        # Exactly one V5 match across S2 + S3
        if n2 + n3 != 1:
            continue

        s1_id = row1["source1_entity_id"]

        if n2 == 1:
            additions[s1_id] = ("S2", m2)
        else:
            additions[s1_id] = ("S3", m3)

print("=" * 90)
print("BASELINE ADDITIONS")
print("=" * 90)
print(f"Collected additions: {len(additions):,}")

# ------------------------------------------------------------
# 2. Read countries only for required S1 IDs
# ------------------------------------------------------------

s1_country = {}

with open(S1_TEST_PATH, "r", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f, delimiter="\t")

    for row in reader:
        sid = row["entity_id"]

        if sid in additions:
            s1_country[sid] = row["country"].strip().lower()

# ------------------------------------------------------------
# 3. Required S2 / S3 target IDs
# ------------------------------------------------------------

s2_targets = {
    matched_id
    for source_type, matched_id in additions.values()
    if source_type == "S2"
}

s3_targets = {
    matched_id
    for source_type, matched_id in additions.values()
    if source_type == "S3"
}

s2_country = {}
s3_country = {}

with open(S2_TEST_PATH, "r", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f, delimiter="\t")

    for row in reader:
        sid = row["entity_id"]

        if sid in s2_targets:
            s2_country[sid] = row["country"].strip().lower()

with open(S3_TEST_PATH, "r", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f, delimiter="\t")

    for row in reader:
        sid = row["entity_id"]

        if sid in s3_targets:
            s3_country[sid] = row["country"].strip().lower()

# ------------------------------------------------------------
# 4. Compare countries
# ------------------------------------------------------------

same_country = 0
cross_country = 0
missing_country = 0

cross_examples = []

for s1_id, (source_type, matched_id) in additions.items():

    c1 = s1_country.get(s1_id, "")

    if source_type == "S2":
        c2 = s2_country.get(matched_id, "")
    else:
        c2 = s3_country.get(matched_id, "")

    if not c1 or not c2:
        missing_country += 1
        continue

    if c1 == c2:
        same_country += 1
    else:
        cross_country += 1

        if len(cross_examples) < 20:
            cross_examples.append(
                (s1_id, source_type, matched_id, c1, c2)
            )

print("\n" + "=" * 90)
print("COUNTRY AUDIT — 91,330 BASELINE ADDITIONS")
print("=" * 90)

print(f"Same-country       : {same_country:,}")
print(f"Cross-country      : {cross_country:,}")
print(f"Missing country    : {missing_country:,}")

checked = same_country + cross_country

if checked:
    print(f"Cross-country rate : {100 * cross_country / checked:.4f}%")

print("\nCross-country examples:")
if cross_examples:
    for x in cross_examples:
        print(x)
else:
    print("NONE")
```

```python
import csv
import os
from collections import Counter

ROOT = "/content/drive/MyDrive/Amazon-ML-Challenge-2026"

V1_PATH = os.path.join(ROOT, "ml/V1_1_REPRODUCTION/matching_results.tsv")
S2_PATH = os.path.join(ROOT, "ml/V5_Final_Production_RAMSAFE/s2_results.tsv")
S3_PATH = os.path.join(ROOT, "ml/V5_Final_Production_RAMSAFE/s3_results.tsv")

# candidate-count buckets
s2_buckets = Counter()
s3_buckets = Counter()

total_additions = 0

with open(V1_PATH, "r", encoding="utf-8", newline="") as f1, \
     open(S2_PATH, "r", encoding="utf-8", newline="") as f2, \
     open(S3_PATH, "r", encoding="utf-8", newline="") as f3:

    r1 = csv.DictReader(f1, delimiter="\t")
    r2 = csv.DictReader(f2, delimiter="\t")
    r3 = csv.DictReader(f3, delimiter="\t")

    for row1, row2, row3 in zip(r1, r2, r3):

        # Only V1-empty entities
        if row1["matched_entity_ids"].strip():
            continue

        m2 = row2["matched_entity_ids"].strip()
        m3 = row3["matched_entity_ids"].strip()

        n2 = len(m2.split(",")) if m2 else 0
        n3 = len(m3.split(",")) if m3 else 0

        # Exact 91,330 baseline additions
        if n2 + n3 != 1:
            continue

        total_additions += 1

        if n2 == 1:
            candidates = row2["candidate_entity_ids"].strip()
            count = len(candidates.split(",")) if candidates else 0
            s2_buckets[count] += 1

        else:
            candidates = row3["candidate_entity_ids"].strip()
            count = len(candidates.split(",")) if candidates else 0
            s3_buckets[count] += 1

print("=" * 90)
print("CANDIDATE AMBIGUITY — 91,330 BASELINE ADDITIONS")
print("=" * 90)

print(f"Total additions: {total_additions:,}")

def print_buckets(name, buckets):
    print("\n" + name)

    ranges = [
        ("1 candidate", lambda x: x == 1),
        ("2 candidates", lambda x: x == 2),
        ("3 candidates", lambda x: x == 3),
        ("4-5 candidates", lambda x: 4 <= x <= 5),
        ("6-10 candidates", lambda x: 6 <= x <= 10),
        ("11-20 candidates", lambda x: 11 <= x <= 20),
        ("21+ candidates", lambda x: x >= 21),
    ]

    total = sum(buckets.values())

    for label, fn in ranges:
        n = sum(v for k, v in buckets.items() if fn(k))
        print(f"{label:20s}: {n:8,} ({100*n/total:.2f}%)")

print_buckets("S2", s2_buckets)
print_buckets("S3", s3_buckets)
```

```python
import csv
import random
import re
import unicodedata

TRAIN = "/content/amazon_ml_challenge/student_resource/dataset/train"

S1_PATH = f"{TRAIN}/train_source1.tsv"
S2_PATH = f"{TRAIN}/train_source2.tsv"
S3_PATH = f"{TRAIN}/train_source3.tsv"
GT_PATH = f"{TRAIN}/train_ground_truth.tsv"

SAMPLE_GT_LINKS = 20000
random.seed(42)

# ------------------------------------------------------------
# Same normalization family we used for V5.
# IMPORTANT: this is a PROXY because friend's normalize.py
# was not included.
# ------------------------------------------------------------

def normalize_proxy(x):
    if x is None:
        return ""

    x = str(x).lower().strip()
    x = unicodedata.normalize("NFKC", x)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for old, new in replacements.items():
        x = x.replace(old, new)

    x = re.sub(r"[^\w\s]", " ", x)
    x = re.sub(r"\s+", " ", x)

    return x.strip()


def make_keys(name, address, country):
    name = normalize_proxy(name)
    address = normalize_proxy(address)
    country = str(country).lower().strip()

    # Friend's 11 key structure
    name_first = name.split()[0] if name else ""
    address_tokens = address.split()
    address_first = address_tokens[0] if address_tokens else ""
    address_last = address_tokens[-1] if address_tokens else ""

    m = re.search(r"(\d+)", address)
    house_number = m.group(1) if m else ""

    remaining = re.sub(r"^\s*\d+\s*", "", address)
    street_prefix = remaining[:7]

    return {
        country + "|" + name[:5],
        country + "|" + name[:4],
        country + "|" + name[-5:],
        country + "|" + name_first,

        country + "|" + address[:15],
        country + "|" + address[:10],
        country + "|" + address[:5],
        country + "|" + address_first,
        country + "|" + address_last,

        country + "|" + house_number,
        country + "|" + house_number + "|" + street_prefix,
    }


# ------------------------------------------------------------
# 1. Sample genuine GT links
# ------------------------------------------------------------

gt_links = []

with open(GT_PATH, "r", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f, delimiter="\t")

    for row in reader:
        s1 = row["source1_entity_id"]
        matches = row["matched_entity_ids"].strip()

        if not matches:
            continue

        for mid in matches.split(","):
            gt_links.append((s1, mid))

random.shuffle(gt_links)
gt_links = gt_links[:SAMPLE_GT_LINKS]

print("=" * 90)
print("FRIEND BLOCKING RECALL TEST")
print("=" * 90)
print(f"Sample genuine links: {len(gt_links):,}")

# ------------------------------------------------------------
# 2. Determine which source each GT target belongs to
# ------------------------------------------------------------

target_ids = {mid for _, mid in gt_links}
s1_ids = {sid for sid, _ in gt_links}

s2_ids_needed = {x for x in target_ids if x.startswith("S2-")}
s3_ids_needed = {x for x in target_ids if x.startswith("S3-")}

# ------------------------------------------------------------
# 3. Retrieve only required records
# ------------------------------------------------------------

s1_rows = {}
s2_rows = {}
s3_rows = {}

with open(S1_PATH, "r", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f, delimiter="\t")

    for row in reader:
        if row["entity_id"] in s1_ids:
            s1_rows[row["entity_id"]] = row

with open(S2_PATH, "r", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f, delimiter="\t")

    for row in reader:
        if row["entity_id"] in s2_ids_needed:
            s2_rows[row["entity_id"]] = row

with open(S3_PATH, "r", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f, delimiter="\t")

    for row in reader:
        if row["entity_id"] in s3_ids_needed:
            s3_rows[row["entity_id"]] = row

print(f"S1 records retrieved: {len(s1_rows):,}")
print(f"S2 records retrieved: {len(s2_rows):,}")
print(f"S3 records retrieved: {len(s3_rows):,}")

# ------------------------------------------------------------
# 4. Test whether at least ONE friend blocking key matches
# ------------------------------------------------------------

recovered = 0
missed = 0

missed_examples = []

for s1_id, target_id in gt_links:

    s1 = s1_rows.get(s1_id)

    if target_id.startswith("S2-"):
        target = s2_rows.get(target_id)
    else:
        target = s3_rows.get(target_id)

    if not s1 or not target:
        continue

    k1 = make_keys(
        s1["business_name"],
        s1["business_address"],
        s1["country"]
    )

    k2 = make_keys(
        target["business_name"],
        target["business_address"],
        target["country"]
    )

    if k1.intersection(k2):
        recovered += 1
    else:
        missed += 1

        if len(missed_examples) < 10:
            missed_examples.append(
                (
                    s1_id,
                    target_id,
                    s1["business_name"],
                    target["business_name"],
                    s1["business_address"],
                    target["business_address"]
                )
            )

checked = recovered + missed

print("\n" + "=" * 90)
print("RESULT")
print("=" * 90)

print(f"Checked genuine links : {checked:,}")
print(f"Recovered by keys     : {recovered:,}")
print(f"Missed by keys        : {missed:,}")

if checked:
    print(f"Blocking recall       : {100 * recovered / checked:.2f}%")

print("\nFirst missed examples:")
for x in missed_examples:
    print(x)
```

```python
import csv
import os

ROOT = "/content/drive/MyDrive/Amazon-ML-Challenge-2026"

S2_PATH = os.path.join(
    ROOT, "ml/V5_Final_Production_RAMSAFE/s2_results.tsv"
)

S3_PATH = os.path.join(
    ROOT, "ml/V5_Final_Production_RAMSAFE/s3_results.tsv"
)

# ------------------------------------------------------------
# Use the SAME 20,000 GT links from the previous cell
# ------------------------------------------------------------

target_map = {}

for s1_id, target_id in gt_links:
    target_map.setdefault(s1_id, set()).add(target_id)

found = set()

# ------------------------------------------------------------
# Scan V5 S2 candidates
# ------------------------------------------------------------

with open(S2_PATH, "r", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f, delimiter="\t")

    for row in reader:

        s1_id = row["source1_entity_id"]

        if s1_id not in target_map:
            continue

        candidates = row["candidate_entity_ids"].strip()

        if not candidates:
            continue

        candidate_set = set(candidates.split(","))

        for target_id in target_map[s1_id]:
            if target_id in candidate_set:
                found.add((s1_id, target_id))

# ------------------------------------------------------------
# Scan V5 S3 candidates
# ------------------------------------------------------------

with open(S3_PATH, "r", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f, delimiter="\t")

    for row in reader:

        s1_id = row["source1_entity_id"]

        if s1_id not in target_map:
            continue

        candidates = row["candidate_entity_ids"].strip()

        if not candidates:
            continue

        candidate_set = set(candidates.split(","))

        for target_id in target_map[s1_id]:
            if target_id in candidate_set:
                found.add((s1_id, target_id))

# ------------------------------------------------------------
# RESULT
# ------------------------------------------------------------

checked = len(gt_links)
recovered = len(found)
missed = checked - recovered

print("=" * 90)
print("V5 CANDIDATE RECALL — SAME 20,000 GENUINE LINKS")
print("=" * 90)

print(f"Checked genuine links : {checked:,}")
print(f"Recovered by V5      : {recovered:,}")
print(f"Missed by V5         : {missed:,}")

print(f"\nV5 candidate recall  : {100 * recovered / checked:.2f}%")
print(f"Friend proxy recall  : 97.73%")
print(f"Difference           : {100 * recovered / checked - 97.73:.2f} percentage points")
```

```python
import csv
import re
import unicodedata
from collections import Counter

TRAIN = "/content/amazon_ml_challenge/student_resource/dataset/train"

S1_PATH = f"{TRAIN}/train_source1.tsv"
S2_PATH = f"{TRAIN}/train_source2.tsv"
S3_PATH = f"{TRAIN}/train_source3.tsv"

# ------------------------------------------------------------
# Helpers — V5 normalization / blocker logic
# ------------------------------------------------------------

def norm(x):
    if x is None:
        return ""

    x = str(x).lower().strip()
    x = unicodedata.normalize("NFKC", x)

    replacements = {
        "&": " and ",
        "street": " st ",
        "road": " rd ",
        "avenue": " ave ",
        "drive": " dr ",
        "boulevard": " blvd ",
        "highway": " hwy ",
        "lane": " ln ",
        "parkway": " pkwy ",
        "place": " pl ",
        "suite": " ste "
    }

    for a, b in replacements.items():
        x = x.replace(a, b)

    x = re.sub(r"[^\w\s]", " ", x)
    x = re.sub(r"\s+", " ", x)

    return x.strip()


def tokens(x):
    return set(x.split()) if x else set()


def numbers(x):
    return set(re.findall(r"\d+", x)) if x else set()


def sim(a, b):
    # Standard SequenceMatcher proxy for the V5 similarity feature.
    from difflib import SequenceMatcher
    return SequenceMatcher(None, a, b).ratio()


# ------------------------------------------------------------
# SAME 20,000 GT LINKS FROM PREVIOUS CELL
# ------------------------------------------------------------

sample_links = gt_links[:20000]

target_s1 = {s1 for s1, _ in sample_links}
target_s2 = {t for _, t in sample_links if t.startswith("S2-")}
target_s3 = {t for _, t in sample_links if t.startswith("S3-")}

# ------------------------------------------------------------
# Retrieve only the required training records
# ------------------------------------------------------------

s1_rows = {}
s2_rows = {}
s3_rows = {}

with open(S1_PATH, "r", encoding="utf-8", newline="") as f:
    r = csv.DictReader(f, delimiter="\t")
    for row in r:
        if row["entity_id"] in target_s1:
            s1_rows[row["entity_id"]] = row

with open(S2_PATH, "r", encoding="utf-8", newline="") as f:
    r = csv.DictReader(f, delimiter="\t")
    for row in r:
        if row["entity_id"] in target_s2:
            s2_rows[row["entity_id"]] = row

with open(S3_PATH, "r", encoding="utf-8", newline="") as f:
    r = csv.DictReader(f, delimiter="\t")
    for row in r:
        if row["entity_id"] in target_s3:
            s3_rows[row["entity_id"]] = row

print("=" * 90)
print("V5 BLOCKER RECALL — SAME TRAINING LINKS")
print("=" * 90)

print(f"S1 records: {len(s1_rows):,}")
print(f"S2 targets: {len(s2_rows):,}")
print(f"S3 targets: {len(s3_rows):,}")

# ------------------------------------------------------------
# Normalize sampled records
# ------------------------------------------------------------

def prepare(row):
    name = norm(row["business_name"])
    address = norm(row["business_address"])

    return {
        "name": name,
        "address": address,
        "country": row["country"].lower().strip(),
        "name_tokens": tokens(name),
        "address_tokens": tokens(address),
        "numbers": numbers(address),
    }


s1_p = {k: prepare(v) for k, v in s1_rows.items()}
s2_p = {k: prepare(v) for k, v in s2_rows.items()}
s3_p = {k: prepare(v) for k, v in s3_rows.items()}

# ------------------------------------------------------------
# Collect only tokens that occur in the sampled TARGET records.
# Then scan S2/S3 to obtain their global frequencies.
# This avoids building a giant 10M-row token dictionary.
# ------------------------------------------------------------

interesting_name_tokens = set()
interesting_addr_tokens = set()

for row in list(s2_p.values()) + list(s3_p.values()):
    interesting_name_tokens.update(row["name_tokens"])
    interesting_addr_tokens.update(row["address_tokens"])

name_freq = Counter()
addr_freq = Counter()

print("\nScanning training S2/S3 for frequencies of sampled target tokens...")

for path in [S2_PATH, S3_PATH]:

    with open(path, "r", encoding="utf-8", newline="") as f:
        r = csv.DictReader(f, delimiter="\t")

        for row in r:

            n = norm(row["business_name"])
            a = norm(row["business_address"])

            for t in tokens(n):
                if t in interesting_name_tokens:
                    name_freq[t] += 1

            for t in tokens(a):
                if t in interesting_addr_tokens:
                    addr_freq[t] += 1

print("Frequency scan complete.")

# ------------------------------------------------------------
# V5 pair eligibility
# ------------------------------------------------------------

def v5_candidate(s1, target):

    # V5 high-recall conditions
    exact_name = (
        s1["name"] != "" and
        s1["name"] == target["name"]
    )

    exact_address = (
        s1["address"] != "" and
        s1["address"] == target["address"]
    )

    shared_name = s1["name_tokens"] & target["name_tokens"]
    shared_addr = s1["address_tokens"] & target["address_tokens"]

    shared_name_count = len(shared_name)
    shared_addr_count = len(shared_addr)

    rare_name_shared = sum(
        1 for t in shared_name
        if name_freq[t] <= 500
    )

    rare_addr_shared = sum(
        1 for t in shared_addr
        if addr_freq[t] <= 500
    )

    # Candidate generation / blocking
    blocked = (
        exact_name
        or exact_address
        or (
            rare_name_shared >= 2
        )
        or (
            rare_addr_shared >= 2
        )
    )

    if not blocked:
        return False

    # V5 similarity filtering
    name_sim = sim(s1["name"], target["name"])
    addr_sim = sim(s1["address"], target["address"])

    passes_similarity = (
        exact_name
        or exact_address
        or (
            shared_name_count >= 2
            and name_sim >= 0.60
        )
        or (
            shared_addr_count >= 3
        )
        or (
            shared_addr_count >= 2
            and addr_sim >= 0.30
        )
    )

    return passes_similarity


# ------------------------------------------------------------
# Evaluate same 20,000 genuine links
# ------------------------------------------------------------

recovered = 0
missed = 0
missed_examples = []

for s1_id, target_id in sample_links:

    s1 = s1_p.get(s1_id)

    if target_id.startswith("S2-"):
        target = s2_p.get(target_id)
    else:
        target = s3_p.get(target_id)

    if s1 is None or target is None:
        continue

    if v5_candidate(s1, target):
        recovered += 1
    else:
        missed += 1

        if len(missed_examples) < 10:
            missed_examples.append(
                (
                    s1_id,
                    target_id,
                    s1["name"],
                    target["name"],
                    s1["address"],
                    target["address"]
                )
            )

checked = recovered + missed

print("\n" + "=" * 90)
print("V5 RESULT")
print("=" * 90)

print(f"Checked genuine links : {checked:,}")
print(f"Recovered by V5      : {recovered:,}")
print(f"Missed by V5         : {missed:,}")

if checked:
    print(f"V5 blocker recall    : {100 * recovered / checked:.2f}%")

print("\nFriend raw key recall : 97.73%")

print("\nFirst V5 misses:")
for x in missed_examples:
    print(x)
```

```python
import os

base = "/content/drive/MyDrive/Amazon-ML-Challenge-2026"

targets = {
    "matching_results.tsv",
    "matching_results_B.tsv",
    "candidate_pairs.tsv",
    "candidate_pairs_B.tsv",
    "Tech Trojans_submission.zip",
    "v5_model.cbm",
    "pipeline.py",
    "Documentation_template.md"
}

print("Searching entire Amazon-ML-Challenge-2026 folder...\n")

found = []

for root, dirs, files in os.walk(base):
    # Skip block indexes because it contains many files
    dirs[:] = [d for d in dirs if d != "block_indexes"]

    for file in files:
        if file in targets:
            path = os.path.join(root, file)
            size_mb = os.path.getsize(path) / (1024 * 1024)
            found.append((file, path, size_mb))

if found:
    for file, path, size in found:
        print(f"{file}")
        print(f"  {size:.2f} MB")
        print(f"  {path}\n")
else:
    print("NO TARGET FILES FOUND")
```

```python
import os

base = "/content/drive/MyDrive/Amazon-ML-Challenge-2026"

print("Searching for B / rescue / final submission files...\n")

for root, dirs, files in os.walk(base):
    # Skip huge block index folders
    dirs[:] = [d for d in dirs if d != "block_indexes"]

    for file in files:
        name = file.lower()

        if (
            "rescue" in name
            or "submission_b" in name
            or "final_submission" in name
            or "matching_results_b" in name
            or "candidate_pairs_b" in name
        ):
            path = os.path.join(root, file)
            size_mb = os.path.getsize(path) / (1024 * 1024)

            print(f"{file}")
            print(f"  Size: {size_mb:.2f} MB")
            print(f"  Path: {path}\n")
```

```python
import os

base = "/content/drive/MyDrive/Amazon-ML-Challenge-2026"

for folder in [
    "ml",
    "notebooks",
    "ML_AUDIT",
    "SUBMISSION_A_HYBRID"
]:
    path = os.path.join(base, folder)

    print("\n" + "="*60)
    print(folder)
    print("="*60)

    if not os.path.exists(path):
        print("NOT FOUND")
        continue

    for item in os.listdir(path):
        p = os.path.join(path, item)

        if os.path.isfile(p):
            size = os.path.getsize(p)/(1024*1024)
            print(f"FILE   {item}   ({size:.2f} MB)")
        else:
            print(f"FOLDER {item}/")
```

```python
import os

base = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml"

targets = [
    "pipeline.py",
    "README.md",
    "requirements.txt",
    "v5_model.cbm",
    "matching_results.tsv",
    "candidate_pairs.tsv",
    "matching_results_B.tsv",
    "candidate_pairs_B.tsv",
    "Documentation_template.md"
]

print("Searching V5 folders...\n")

for root, dirs, files in os.walk(base):

    # Don't search huge index folders
    dirs[:] = [
        d for d in dirs
        if d not in ["V5_full_indexes", "block_indexes"]
    ]

    # Only inspect V5-related folders
    if not any(x in root for x in [
        "V5_ML_Final",
        "V5_Final_Inference",
        "V5_Final_Production",
        "V5_Final_Production_RAMSAFE",
        "V5_Final_Validation",
        "V5_SUBMISSION"
    ]):
        continue

    for file in files:
        if file in targets:
            path = os.path.join(root, file)
            size_mb = os.path.getsize(path) / (1024 * 1024)

            print(f"FOUND: {file}")
            print(f"Size: {size_mb:.2f} MB")
            print(f"Path: {path}\n")
```

```python
import zipfile
import os

zip_path = "/content/drive/MyDrive/Amazon-ML-Challenge-2026/ml/V5_ML_Final-20260925T182640Z-1-001.zip"

extract_path = "/content/V5_GITHUB_FILES"

os.makedirs(extract_path, exist_ok=True)

with zipfile.ZipFile(zip_path, "r") as z:
    z.extractall(extract_path)

print("Extracted successfully.")

for root, dirs, files in os.walk(extract_path):
    for file in files:
        path = os.path.join(root, file)
        size_mb = os.path.getsize(path) / (1024*1024)
        print(f"{path}  ({size_mb:.2f} MB)")
```

```python
import os
import shutil

repo = "/content/Amazon_ML_Challenge_2026"
src = "/content/V5_GITHUB_FILES/V5_ML_Final"

# Create project folder
dest = os.path.join(repo, "V5_ML_Final")
os.makedirs(dest, exist_ok=True)

# Files we want in GitHub
files_to_copy = [
    "v5_inference.py",
    "v5_feature_list.txt",
    "v5_model.cbm",
    "V5_final_report.txt"
]

for file in files_to_copy:
    source = os.path.join(src, file)
    target = os.path.join(dest, file)

    shutil.copy2(source, target)
    print("Copied:", file)

print("\nDone.")
```
